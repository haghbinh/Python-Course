# پروژه‌ی ۲ — راه‌حل مرجع (فقط برای مدرس)

یک پیاده‌سازی از همه‌ی الزامات، برای مقایسه با گزارش‌ها و طرح سؤال در دفاع. تصمیم‌های پاک‌سازی و مدل‌سازی (مثلاً حد ۶۰ دقیقه برای زمان خدمت، یا کنار گذاشتن زمان‌های خدمت باجه‌ی کند <code dir="ltr">T3</code> در مدل) قابل بحث‌اند؛ دانشجو می‌تواند تصمیم دیگری بگیرد، به شرط آنکه آن را توضیح دهد. داده با <code dir="ltr">s20&#95;gen&#95;data.py</code> (seed 18) ساخته شده است.

### حقیقت پنهان داده (برای مدرس)

- ورود مشتری: در هر دقیقه از ساعت <code dir="ltr">h</code> با احتمال <code dir="ltr">RATE[h]/60</code> که <code dir="ltr">RATE = {8: 16, 9: 26, 10: 36, 11: 34, 12: 24, 13: 18}</code>؛ اوج ساعت ۱۰ تا ۱۲.
- نوع خدمت: واریز ۳۵٪ (میانگین ۲٫۵ دقیقه)، برداشت ۳۰٪ (۲)، انتقال ۲۰٪ (۴)، وام ۱۰٪ (۱۰)، افتتاح حساب ۵٪ (۱۵).
- **چهارشنبه ۲۰۲۶-۱۱-۱۸** فقط دو باجه (<code dir="ltr">T3</code> غایب): میانگین انتظار حدود ۱۶ دقیقه در برابر کمتر از ۳ دقیقه در روزهای دیگر، و **همه‌ی ۱۳ مشتری که شعبه را ترک کرده‌اند** مربوط به همین روزند.
- باجه‌ی <code dir="ltr">T3</code> حدود ۳۰٪ کندتر است (مثلاً میانگین واریز حدود ۳٫۴ دقیقه در برابر ۲٫۵).

### ایرادهای کاشته‌شده

| ایراد | تعداد |
|---|---|
| سطر خالی | ۱ |
| سطر با یک فیلد کم | ۱ |
| سطر تکراری عیناً یکسان | ۴ |
| تاریخ جمعه (شعبه تعطیل است) | ۱ |
| زمان ورود بعد از ساعت کار (<code dir="ltr">15:10:22</code>) | ۱ |
| نوع خدمت خالی | ۱ |
| پایان خدمت قبل از شروع (اشتباه در ساعت) | ۲ |
| شروع خدمت قبل از ورود | ۱ |
| زمان خدمت ۲۱۵ دقیقه‌ای (نوبت بسته نشده) | ۱ |
| قابل اصلاح: زمان بدون ثانیه، بدون صفر اول، با نقطه به جای <code dir="ltr">:</code>؛ نام باجه با حروف کوچک یا فاصله؛ <code dir="ltr">withdrawl</code>، <code dir="ltr">Deposit </code>، <code dir="ltr">TRANSFER</code>، <code dir="ltr">Loan</code> | حدود ۱۸ |
| معتبر ولی خاص: مشتری‌هایی که شعبه را ترک کرده‌اند (زمان‌های خالی) | ۱۳ |

نکته: یک افتتاح حساب واقعی ۷۱ دقیقه طول کشیده است؛ قاعده‌ی «بیش از ۶۰ دقیقه رد شود» آن را هم کنار می‌گذارد. این مثال خوبی برای پرسش در دفاع است: «قاعده‌ی شما کدام داده‌ی درست را هم حذف کرد؟»

### نتیجه‌های کلیدی (این پیاده‌سازی)

- ۱۵۴۳ رکورد سالم و ۱۴ سطر رد شده. انتظار کل: میانگین ۲٫۸ دقیقه و ۵٫۸٪ بیش از ۱۵ دقیقه؛ بدون چهارشنبه‌ی ۱۸: ۱٫۴ دقیقه و ۰٫۹٪.
- اعتبارسنجی: شبیه‌سازی ۳ باجه با صف مشترک میانگین ۱٫۸ دقیقه می‌دهد که به ۱٫۴ دقیقه‌ی روزهای عادی نزدیک است.
- سیاست‌ها (۲۰۰ روز، seed 0): صف مشترک همیشه بهتر از صف جدا (با ۳ باجه ۲٫۴٪ در برابر ۵٫۹٪ بیش از ۱۵ دقیقه). هدف ۵٪ با **۳ باجه و صف مشترک** برآورده می‌شود. باجه‌دار پاره‌وقت ۱۰ تا ۱۲ نسبت را به حدود ۰٫۳٪ می‌رساند و تقریباً به خوبی ۴ باجه‌ی تمام‌وقت است.
- با ۲ باجه: میانگین هر سه سیاست حدود ۲۰ دقیقه است، اما باجه‌ی سریع انتظار خدمت‌های کوتاه را به حدود ۷ دقیقه می‌رساند و انتظار خدمت‌های طولانی را به حدود ۵۰ دقیقه؛ این همان «برای کدام مشتری‌ها؟» در الزام ۷ (د) است.

```python
import csv
import math
import random
from collections import Counter, defaultdict, deque
from datetime import datetime, date
from statistics import mean, median



OPEN_H, CLOSE_H = 8, 14
TYPES = ["deposit", "withdrawal", "transfer", "loan", "account_opening"]
SHORT = {"deposit", "withdrawal"}
FIX_TYPE = {"withdrawl": "withdrawal"}


class DataError(Exception):
    pass


def parse_time(text):
    text = text.strip().replace(".", ":")
    for fmt in ("%H:%M:%S", "%H:%M"):
        try:
            return datetime.strptime(text, fmt)
        except ValueError:
            pass
    raise DataError(f"bad time {text!r}")


def minutes(t):
    return t.hour * 60 + t.minute + t.second / 60


def clean_log(path):
    records, rejected, seen = [], [], set()
    with open(path, newline="", encoding="utf-8") as f:
        reader = csv.reader(f)
        next(reader)
        for line_no, row in enumerate(reader, start=2):
            try:
                if not row:
                    raise DataError("blank line")
                if len(row) != 7:
                    raise DataError("wrong number of fields")
                if tuple(row) in seen:
                    raise DataError("duplicate row")
                seen.add(tuple(row))
                ticket, day, arr, start, end, teller, kind = [x.strip() for x in row]
                d = date.fromisoformat(day)
                if d.weekday() == 4:
                    raise DataError("bank closed on Friday")
                kind = FIX_TYPE.get(kind.lower(), kind.lower())
                if kind not in TYPES:
                    raise DataError("unknown service type")
                a = minutes(parse_time(arr))
                if not (OPEN_H * 60 <= a < CLOSE_H * 60):
                    raise DataError("arrival outside opening hours")
                if start == "" and end == "":
                    rec = {"ticket": ticket, "date": d, "arrival": a, "start": None, "end": None,
                           "teller": None, "type": kind, "left": True}
                else:
                    s, e = minutes(parse_time(start)), minutes(parse_time(end))
                    if s < a:
                        raise DataError("service starts before arrival")
                    if e <= s:
                        raise DataError("service ends before it starts")
                    if e - s > 60:
                        raise DataError("service longer than 60 minutes")
                    rec = {"ticket": ticket, "date": d, "arrival": a, "start": s, "end": e,
                           "teller": teller.upper(), "type": kind, "left": False}
                records.append(rec)
            except (DataError, ValueError) as err:
                rejected.append((line_no, ",".join(row), str(err)))
    return records, rejected


def fit_model(records):
    days = {r["date"] for r in records}
    per_hour = Counter(int(r["arrival"] // 60) for r in records)
    rate = {h: per_hour[h] / len(days) for h in range(OPEN_H, CLOSE_H)}
    mix = Counter(r["type"] for r in records)
    service = defaultdict(list)
    for r in records:
        if not r["left"] and r["teller"] != "T3":
            service[r["type"]].append(r["end"] - r["start"])
    return {"rate": rate, "mix": dict(mix), "service": dict(service)}


# ---------------------------------------------------------------- simulation (minute steps)
class Customer:
    def __init__(self, arrival, kind, service):
        self.arrival, self.kind, self.service = arrival, kind, service
        self.start = None


class Teller:
    def __init__(self, name):
        self.name, self.busy_until, self.busy_time, self.current = name, 0, 0.0, None

    def is_free(self, t):
        return self.busy_until <= t

    def serve(self, c, t):
        c.start = t
        self.busy_until = t + c.service
        self.busy_time += c.service


class Bank:
    """Base class: single shared queue."""
    def __init__(self, n_tellers):
        self.tellers = [Teller(f"T{i+1}") for i in range(n_tellers)]
        self.queue = deque()
        self.max_queue = 0

    def join(self, c):
        self.queue.append(c)

    def queue_len(self):
        return len(self.queue)

    def assign(self, t):
        for tl in self.tellers:
            if tl.is_free(t) and self.queue:
                tl.serve(self.queue.popleft(), t)

    def step(self, t, arrivals):
        for c in arrivals:
            self.join(c)
        self.assign(t)
        self.max_queue = max(self.max_queue, self.queue_len())

    def empty(self):
        return self.queue_len() == 0


class MultiQueueBank(Bank):
    """One queue per teller; a customer joins the shortest queue and never switches."""
    def __init__(self, n_tellers):
        super().__init__(n_tellers)
        self.queues = [deque() for _ in self.tellers]

    def join(self, c):
        min(self.queues, key=len).append(c)

    def queue_len(self):
        return sum(len(q) for q in self.queues)

    def assign(self, t):
        for tl, q in zip(self.tellers, self.queues):
            if tl.is_free(t) and q:
                tl.serve(q.popleft(), t)


class ExpressBank(Bank):
    """Teller T1 serves only short services (deposit/withdrawal); others serve everyone, express queue first? No: FIFO general."""
    def __init__(self, n_tellers):
        super().__init__(n_tellers)
        self.express = deque()

    def join(self, c):
        (self.express if c.kind in SHORT else self.queue).append(c)

    def queue_len(self):
        return len(self.queue) + len(self.express)

    def assign(self, t):
        t1, others = self.tellers[0], self.tellers[1:]
        if t1.is_free(t) and self.express:
            t1.serve(self.express.popleft(), t)
        for tl in others:
            if tl.is_free(t):
                # serve whoever has waited longest among the two queues
                cands = [q for q in (self.queue, self.express) if q]
                if cands:
                    q = min(cands, key=lambda q: q[0].arrival)
                    tl.serve(q.popleft(), t)


POLICIES = {"single": Bank, "multi": MultiQueueBank, "express": ExpressBank}


def simulate_one_day(model, policy, n_tellers, rng, extra=None):
    """extra = (start_minute, end_minute): one extra teller (single queue only) in that window."""
    bank = POLICIES[policy](n_tellers + (1 if extra else 0))
    if extra:
        ex = bank.tellers[-1]
        ex.is_free = lambda t, ex=ex: ex.busy_until <= t and extra[0] <= t < extra[1]
    kinds, weights = list(model["mix"]), list(model["mix"].values())
    customers = []
    t = OPEN_H * 60
    while t < CLOSE_H * 60 or not bank.empty() or any(tl.busy_until > t for tl in bank.tellers):
        arrivals = []
        if t < CLOSE_H * 60 and rng.random() < model["rate"][t // 60] / 60:
            kind = rng.choices(kinds, weights)[0]
            arrivals.append(Customer(t, kind, rng.choice(model["service"][kind])))
        customers += arrivals
        bank.step(t, arrivals)
        t += 1
    waits = [(c.kind, c.start - c.arrival) for c in customers]
    hours = (t - OPEN_H * 60)
    staffed = n_tellers * hours + ((extra[1] - extra[0]) if extra else 0)
    util = sum(tl.busy_time for tl in bank.tellers) / staffed
    return waits, bank.max_queue, util


def simulate(model, policy="single", n_tellers=3, n_days=200, seed=0, extra=None):
    rng = random.Random(seed)
    pairs, max_q, utils = [], [], []
    for _ in range(n_days):
        w, m, u = simulate_one_day(model, policy, n_tellers, rng, extra)
        pairs += w; max_q.append(m); utils.append(u)
    waits = sorted(w for _, w in pairs)
    by_type = defaultdict(list)
    for k, w in pairs:
        by_type[k].append(w)
    return {"mean_wait": sum(waits) / len(waits),
            "p90_wait": waits[int(0.9 * (len(waits) - 1))],
            "share_over_15": sum(w > 15 for w in waits) / len(waits),
            "max_queue": max(max_q),
            "utilization": sum(utils) / len(utils),
            "by_type": {k: sum(v) / len(v) for k, v in by_type.items()}}
```

```python
records, rejected = clean_log("data/bank_log.csv")
print(len(records), "clean records;", len(rejected), "rejected")
for r in rejected:
    print(r)
```

```python
served = [r for r in records if not r["left"]]
print("left without service:", Counter(r["date"].isoformat() for r in records if r["left"]))

def summary(waits):
    w = sorted(waits)
    return round(mean(w), 1), round(median(w), 1), round(w[int(0.9 * (len(w) - 1))], 1), f"{sum(x > 15 for x in w) / len(w):.1%}"

print("all:", summary([r["start"] - r["arrival"] for r in served]))
by_day = defaultdict(list)
for r in served:
    by_day[r["date"]].append(r["start"] - r["arrival"])
for d, w in sorted(by_day.items()):
    print(d, d.strftime("%a"), len(w), summary(w))

by_hour = defaultdict(list)
for r in served:
    by_hour[int(r["arrival"] // 60)].append(r["start"] - r["arrival"])
arrivals = Counter(int(r["arrival"] // 60) for r in records)
for h in range(OPEN_H, CLOSE_H):
    print(f"{h}:00", arrivals[h] / len(by_day), summary(by_hour[h]))

for t in ["T1", "T2", "T3"]:
    row = []
    for k in TYPES:
        s = [r["end"] - r["start"] for r in served if r["teller"] == t and r["type"] == k]
        row.append(f"{k} {mean(s):.2f}")
    print(t, Counter(r["teller"] for r in served)[t], row)
```

```python
model = fit_model(records)
print({h: round(v, 1) for h, v in model["rate"].items()})
print(model["mix"])
print({k: round(mean(v), 2) for k, v in model["service"].items()})

normal = [w for d, ws in by_day.items() if d != date(2026, 11, 18) for w in ws]
print("history, normal days:", summary(normal))
sim = simulate(model, "single", 3, 200, seed=0)
print("simulation, 3 tellers single:", {k: round(v, 3) for k, v in sim.items() if k != "by_type"})
```
