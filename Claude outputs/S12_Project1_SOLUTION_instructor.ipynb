## پیش‌زمینه: ایده‌ی Naive Bayes

سؤال ما این است: با دیدن کلمه‌های یک پیام، **احتمال** اسپم بودن آن چقدر است؟ قاعده‌ی بیز می‌گوید این احتمال متناسب است با:

<div dir="ltr">

```text
P(class | message)  ∝  P(class) × P(word_1 | class) × P(word_2 | class) × ... × P(word_n | class)
```

</div>

- <code dir="ltr">P(class)</code>: **احتمال پیشین (prior)**؛ چه کسری از پیام‌های آموزش از این کلاس‌اند.
- <code dir="ltr">P(word &#124; class)</code>: **درست‌نمایی (likelihood)**؛ در پیام‌های این کلاس، این کلمه چه سهمی از کل کلمه‌ها را دارد.

فرض «**ساده‌لوحانه (naive)**» این است که کلمه‌ها **مستقل** از هم ظاهر می‌شوند؛ برای همین احتمال‌هایشان در هم ضرب می‌شود. پیام به کلاسی نسبت داده می‌شود که این حاصل‌ضرب برایش بیشتر است.

**هموارسازی لاپلاس (Laplace smoothing).** اگر کلمه‌ای هرگز در پیام‌های یک کلاس دیده نشده باشد، احتمال آن صفر و کل حاصل‌ضرب صفر می‌شود. برای جلوگیری از این، به شمارش همه‌ی کلمه‌ها عدد کوچک <code dir="ltr">alpha</code> (پیش‌فرض <code dir="ltr">1</code>) اضافه می‌کنیم:

<div dir="ltr">

```text
P(word | class) = (count(word in class) + alpha) / (total words in class + alpha × V)
```

</div>

که <code dir="ltr">V</code> اندازه‌ی **واژگان** (تعداد کلمه‌های متمایز کل داده‌ی آموزش) است.

**کار با لگاریتم.** حاصل‌ضرب ده‌ها عدد کوچک‌تر از ۱ آن‌قدر کوچک می‌شود که به صفر می‌رسد (**underflow**؛ سلول زیر را اجرا کنید). به جای ضرب احتمال‌ها، **لگاریتم** آن‌ها را جمع می‌کنیم؛ چون <code dir="ltr">log</code> صعودی است، کلاس برنده عوض نمی‌شود:

<div dir="ltr">

```text
score(class) = log P(class) + log P(word_1 | class) + ... + log P(word_n | class)
```

</div>

کلمه‌هایی که در واژگان نیستند (در آموزش دیده نشده‌اند) در امتیاز **نادیده** گرفته می‌شوند.

</div>

````python
import math

print(0.01 ** 200)              # underflow: the true value is 1e-400
print(200 * math.log(0.01))     # the log of the same number: no problem
````

## وظیفه‌ی ۲: پیش‌پردازش

تابع <code dir="ltr">tokenize(text)</code> را بنویسید که:

1. متن را به حروف کوچک تبدیل کند،
2. هر علامت نگارشی (<code dir="ltr">string.punctuation</code>) را با **فاصله** جایگزین کند،
3. لیست کلمه‌ها را برگرداند.

</div>

````python
import string

def tokenize(text):
    text = text.lower()
    clean = []
    for ch in text:
        if ch in string.punctuation:
            clean.append(" ")
        else:
            clean.append(ch)
    return "".join(clean).split()
````

## وظیفه‌ی ۳: تقسیم آموزش و آزمون

تابع <code dir="ltr">train&#95;test&#95;split(data, test&#95;ratio=0.25, seed=0)</code> را بنویسید که:

- <code dir="ltr">random.seed(seed)</code> را صدا بزند و یک **کپی** از <code dir="ltr">data</code> را با <code dir="ltr">random.shuffle</code> به هم بریزد (لیست اصلی نباید تغییر کند)،
- <code dir="ltr">n&#95;test = int(test&#95;ratio &#42; len(data))</code> عنصر **اول** کپی را داده‌ی آزمون و بقیه را داده‌ی آموزش بگیرد،
- تاپل <code dir="ltr">(train, test)</code> را برگرداند.

</div>

````python
import random

def train_test_split(data, test_ratio=0.25, seed=0):
    random.seed(seed)
    shuffled = data.copy()
    random.shuffle(shuffled)
    n_test = int(test_ratio * len(shuffled))
    return shuffled[n_test:], shuffled[:n_test]
````

## وظیفه‌ی ۴: آموزش مدل

«آموزش» Naive Bayes فقط **شمردن** است. تابع <code dir="ltr">train(data, alpha=1.0)</code> را بنویسید که یک **دیکشنری** با دقیقاً این کلیدها برگرداند:

| کلید | مقدار |
|---|---|
| <code dir="ltr">"class&#95;counts"</code> | <code dir="ltr">{label: number of messages}</code> |
| <code dir="ltr">"word&#95;counts"</code> | <code dir="ltr">{label: {word: count}}</code>؛ تعداد تکرار هر کلمه در پیام‌های هر کلاس |
| <code dir="ltr">"total&#95;words"</code> | <code dir="ltr">{label: total number of words}</code> |
| <code dir="ltr">"vocab"</code> | **مجموعه‌ی** همه‌ی کلمه‌های دیده‌شده |
| <code dir="ltr">"n&#95;docs"</code> | تعداد کل پیام‌ها |
| <code dir="ltr">"alpha"</code> | همان <code dir="ltr">alpha</code> ورودی |

کلمه‌ها را با <code dir="ltr">tokenize</code> استخراج کنید.

</div>

````python
def train(data, alpha=1.0):
    class_counts = {}
    word_counts = {}
    total_words = {}
    vocab = set()
    for text, label in data:
        class_counts[label] = class_counts.get(label, 0) + 1
        if label not in word_counts:
            word_counts[label] = {}
            total_words[label] = 0
        for w in tokenize(text):
            word_counts[label][w] = word_counts[label].get(w, 0) + 1
            total_words[label] += 1
            vocab.add(w)
    return {"class_counts": class_counts, "word_counts": word_counts,
            "total_words": total_words, "vocab": vocab,
            "n_docs": len(data), "alpha": alpha}
````

````python
# self-check (do not edit)
m = train(toy)
assert m["class_counts"] == {"spam": 2, "ham": 2}
assert m["total_words"] == {"spam": 5, "ham": 5}
assert m["vocab"] == {"free", "cash", "call", "now", "me", "see", "you"}
assert m["word_counts"]["spam"]["free"] == 2
assert m["word_counts"]["ham"].get("free", 0) == 0
assert m["n_docs"] == 4 and m["alpha"] == 1.0
assert train(toy, alpha=0.5)["alpha"] == 0.5
print("train: all checks passed")
````

## وظیفه‌ی ۵: پیش‌بینی

سه تابع بنویسید:

- <code dir="ltr">word&#95;prob(model, word, label)</code>: احتمال هموارشده‌ی <code dir="ltr">P(word &#124; label)</code> طبق فرمول پیش‌زمینه.
- <code dir="ltr">log&#95;scores(model, text)</code>: دیکشنری <code dir="ltr">{label: score}</code> که در آن <code dir="ltr">score</code> لگاریتم احتمال پیشین به‌علاوه‌ی مجموع لگاریتم <code dir="ltr">word&#95;prob</code> برای هر توکنِ **داخل واژگان** است (توکن تکراری، هر بار حساب می‌شود).
- <code dir="ltr">predict(model, text)</code>: برچسب با **بیشترین** امتیاز. در حالت تساوی، اولین برچسب (به ترتیب کلیدهای دیکشنری) را برگردانید.

</div>

````python
def word_prob(model, word, label):
    V = len(model["vocab"])
    a = model["alpha"]
    count = model["word_counts"][label].get(word, 0)
    return (count + a) / (model["total_words"][label] + a * V)

def log_scores(model, text):
    scores = {}
    for label in model["class_counts"]:
        s = math.log(model["class_counts"][label] / model["n_docs"])
        for w in tokenize(text):
            if w in model["vocab"]:
                s += math.log(word_prob(model, w, label))
        scores[label] = s
    return scores

def predict(model, text):
    scores = log_scores(model, text)
    best = None
    for label in scores:
        if best is None or scores[label] > scores[best]:
            best = label
    return best
````

````python
# self-check (do not edit)
m = train(toy)
assert abs(word_prob(m, "free", "spam") - 3 / 12) < 1e-12
assert abs(word_prob(m, "free", "ham") - 1 / 12) < 1e-12
s = log_scores(m, "free call")
assert set(s) == {"spam", "ham"}
assert abs(s["spam"] - math.log(0.5 * 3 / 12 * 2 / 12)) < 1e-9
assert abs(log_scores(m, "hello world")["ham"] - math.log(0.5)) < 1e-9
assert predict(m, "free call") == "spam"
assert predict(m, "see me") == "ham"
assert predict(m, "FREE cash!!") == "spam"
print("word_prob / log_scores / predict: all checks passed")
````

## وظیفه‌ی ۶: احتمال پیش‌بینی

امتیازهای <code dir="ltr">log&#95;scores</code> احتمال نیستند. تابع <code dir="ltr">predict&#95;proba(model, text)</code> را بنویسید که دیکشنری <code dir="ltr">{label: probability}</code> برگرداند؛ احتمال‌ها باید متناسب با <code dir="ltr">exp(score)</code> باشند و مجموعشان ۱ شود.

**شرط:** تابع شما باید برای پیام‌های **بسیار طولانی** هم درست کار کند (امتیازهایی مثل <code dir="ltr">-5000</code>). راهنمایی: softmax پایدار (جلسه‌ی ۸).

</div>

````python
def predict_proba(model, text):
    scores = log_scores(model, text)
    m = max(scores.values())
    exps = {}
    total = 0
    for label, s in scores.items():
        exps[label] = math.exp(s - m)
        total += exps[label]
    probs = {}
    for label in exps:
        probs[label] = exps[label] / total
    return probs
````

## وظیفه‌ی ۷: ارزیابی

تابع <code dir="ltr">evaluate(model, data, positive="spam")</code> را بنویسید که برای هر پیام <code dir="ltr">data</code> پیش‌بینی کند و تاپل <code dir="ltr">(cm, accuracy, precision, recall)</code> را برگرداند:

- <code dir="ltr">cm</code>: ماتریس درهم‌ریختگی به شکل دیکشنری با کلید تاپلی <code dir="ltr">{(true&#95;label, predicted&#95;label): count}</code> (فقط زوج‌هایی که دست‌کم یک بار رخ داده‌اند).
- <code dir="ltr">accuracy</code>: کسر پیش‌بینی‌های درست.
- <code dir="ltr">precision</code>: از پیام‌هایی که مدل گفته <code dir="ltr">positive</code> هستند، چه کسری واقعاً <code dir="ltr">positive</code> بوده‌اند.
- <code dir="ltr">recall</code>: از پیام‌هایی که واقعاً <code dir="ltr">positive</code> بوده‌اند، چه کسری را مدل پیدا کرده است.

اگر مخرج precision یا recall صفر بود، مقدار <code dir="ltr">0.0</code> برگردانید.

</div>

````python
def evaluate(model, data, positive="spam"):
    cm = {}
    for text, label in data:
        pred = predict(model, text)
        cm[(label, pred)] = cm.get((label, pred), 0) + 1
    tp = cm.get((positive, positive), 0)
    fp = 0
    fn = 0
    correct = 0
    for (true, pred), n in cm.items():
        if true != positive and pred == positive:
            fp += n
        if true == positive and pred != positive:
            fn += n
        if true == pred:
            correct += n
    accuracy = correct / len(data)
    precision = tp / (tp + fp) if tp + fp > 0 else 0.0
    recall = tp / (tp + fn) if tp + fn > 0 else 0.0
    return cm, accuracy, precision, recall
````

````python
# self-check (do not edit)
m = train(toy)
toy_test = [("free cash now", "spam"), ("call me", "spam"),
            ("see you", "ham"), ("free me", "ham"), ("cash", "spam")]
cm, acc, prec, rec = evaluate(m, toy_test)
assert cm == {("spam", "spam"): 2, ("spam", "ham"): 1, ("ham", "ham"): 1, ("ham", "spam"): 1}
assert abs(acc - 3 / 5) < 1e-12
assert abs(prec - 2 / 3) < 1e-12
assert abs(rec - 2 / 3) < 1e-12
cm, acc, prec, rec = evaluate(m, [("see you", "ham")])
assert prec == 0.0 and rec == 0.0 and acc == 1.0
print("evaluate: all checks passed")
````
