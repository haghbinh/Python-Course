## ۱. انواع خطا

### 💡 خطای نحوی و استثنا

از جلسه‌ی ۱ با پیام‌های خطا آشنا هستید. خطاها دو دسته‌اند:

- **خطای نحوی (<code dir="ltr">SyntaxError</code>):** کد اصلاً پایتونِ درست نیست (پرانتز جاافتاده، <code dir="ltr">:</code> فراموش‌شده، تورفتگی نادرست). پایتون **قبل از اجرای حتی یک خط** آن را تشخیص می‌دهد و تا اصلاحش نکنید هیچ چیز اجرا نمی‌شود.
- **استثنا (exception):** کد از نظر نحوی درست است، اما **در حین اجرا** اتفاقی می‌افتد که ادامه را ناممکن می‌کند: تقسیم بر صفر، کلیدی که در دیکشنری نیست، رشته‌ای که عدد نیست. استثناها را می‌توان **گرفت و مدیریت کرد**؛ موضوع این جلسه.

</div>

````python
print("start")
print(10 / 0)
print("end")
````

### 💡 خواندن traceback

وقتی استثنا در عمق چند تابع رخ می‌دهد، پایتون مسیر کامل فراخوانی‌ها را چاپ می‌کند. این گزارش را **traceback** می‌گویند:

</div>

````python
def mean(values):
    return sum(values) / len(values)

def normalize(values):
    m = mean(values)
    return [v - m for v in values]

normalize([])
````

### 🔍 کد را بخوان

traceback را **از پایین به بالا** بخوانید:

- **خط آخر** مهم‌ترین خط است: **نوع** خطا (<code dir="ltr">ZeroDivisionError</code>) و **پیام** آن (<code dir="ltr">division by zero</code>).
- خط‌های بالاتر، **زنجیره‌ی فراخوانی** را نشان می‌دهند: سلول، <code dir="ltr">normalize</code> را صدا زد؛ <code dir="ltr">normalize</code>، <code dir="ltr">mean</code> را؛ و خطا در <code dir="ltr">mean</code> رخ داد.
- خطا جایی **رخ داده** که پایین‌ترین فلش نشان می‌دهد، اما **علت** آن ممکن است بالاتر باشد: اینجا مشکل واقعی، فرستادن لیست خالی به <code dir="ltr">normalize</code> است، نه کد <code dir="ltr">mean</code>.

## ۲. <code dir="ltr">try</code> و <code dir="ltr">except</code>

### 💡 مفهوم

کدی را که ممکن است خطا بدهد داخل <code dir="ltr">try</code> می‌گذاریم. اگر خطایی از نوع مشخص‌شده رخ دهد، اجرای <code dir="ltr">try</code> همان‌جا قطع و بلوک <code dir="ltr">except</code> اجرا می‌شود و برنامه **ادامه پیدا می‌کند**:

<div dir="ltr">

```python
try:
    risky code
except SomeError:
    what to do instead
```

</div>

</div>

````python
values = ["12", "7.5", "abc", "-3"]
numbers = []
for v in values:
    try:
        numbers.append(float(v))
    except ValueError:
        print("skipped:", v)
print(numbers)
````

<div dir="rtl">

### 🔍 کد را بخوان

- برای <code dir="ltr">"abc"</code>، <code dir="ltr">float</code> استثنای <code dir="ltr">ValueError</code> ایجاد می‌کند؛ <code dir="ltr">append</code> هرگز اجرا نمی‌شود و کنترل به <code dir="ltr">except</code> می‌رود.
- بعد از <code dir="ltr">except</code> حلقه **ادامه می‌یابد**؛ یک مقدار خراب، کل پردازش را متوقف نمی‌کند. این الگوی اصلی پاک‌سازی داده‌های واقعی است.

### 💡 گرفتن چند نوع خطا و پیام خطا

- چند <code dir="ltr">except</code> پشت سر هم: هر نوع خطا رفتار خودش را دارد.
- یک **تاپل** از انواع: <code dir="ltr">except (ValueError, TypeError):</code>
- <code dir="ltr">as e</code>: خود شیء خطا را در متغیر <code dir="ltr">e</code> نگه می‌دارد تا پیامش را چاپ یا ثبت کنید.

</div>

````python
def safe_ratio(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("division by zero, returning 0")
        return 0.0
    except TypeError as e:
        print("bad types:", e)
        return None

print(safe_ratio(6, 3))
print(safe_ratio(6, 0))
print(safe_ratio("6", 3))
````

<div dir="rtl">

### 💡 «آیا این رشته عدد است؟» — راه درست

در جلسه‌ی ۱۱ دیدید که <code dir="ltr">isdigit</code> برای <code dir="ltr">"-12"</code> و <code dir="ltr">"3.5"</code> نادرست برمی‌گرداند و قول دادیم راه درست را اینجا ببینیم: به جای اینکه همه‌ی حالت‌ها را **قبل از** تبدیل بررسی کنیم، **تبدیل را امتحان می‌کنیم** و شکستش را می‌گیریم:

</div>

````python
def is_number(s):
    try:
        float(s)
        return True
    except ValueError:
        return False

for s in ["2026", "-12", "3.5", " 42 ", "1e-3", "", "abc", "12,5"]:
    print(repr(s), is_number(s))
````

<div dir="rtl">

🔍 این تابع دقیقاً همان قواعدی را به کار می‌برد که خود <code dir="ltr">float</code> دارد: علامت منفی، نقطه‌ی اعشار، نماد علمی (<code dir="ltr">1e-3</code>) و فاصله‌های دو طرف پذیرفته می‌شوند؛ رشته‌ی خالی و ویرگول اعشاری (<code dir="ltr">12,5</code>) نه.

این سبک را **EAFP** می‌گویند: *«بخشش خواستن آسان‌تر از اجازه گرفتن است»*؛ یعنی کار را انجام بده و اگر نشد، خطا را مدیریت کن. سبک مقابل، **LBYL** (*«قبل از پریدن نگاه کن»*) است: اول با <code dir="ltr">if</code> همه‌ی شرط‌ها را بررسی کن. هر دو جای خود را دارند، اما وقتی قواعد پیچیده‌اند (مثل «عدد معتبر بودن») EAFP ساده‌تر و دقیق‌تر است.

### ⚠️ اشتباه رایج: <code dir="ltr">except:</code> خالی

<code dir="ltr">except:</code> بدون نوع (یا <code dir="ltr">except Exception:</code> در جایی که منظورتان یک خطای خاص است) **همه‌ی** خطاها را می‌بلعد؛ از جمله غلط تایپی در نام متغیرها (<code dir="ltr">NameError</code>) و حتی توقف برنامه با کلید Ctrl+C. نتیجه: برنامه بی‌صدا نتیجه‌ی غلط می‌دهد و شما هرگز نمی‌فهمید چرا. **فقط خطاهایی را بگیرید که انتظارشان را دارید و می‌دانید با آن‌ها چه کنید.**

### ⚠️ اشتباه رایج: <code dir="ltr">try</code> خیلی بزرگ

هرچه کد داخل <code dir="ltr">try</code> بیشتر باشد، احتمال اینکه خطایی **غیر از** خطای مورد انتظار هم گرفته شود بیشتر است. فقط خطی را که ممکن است خطای مورد نظر را بدهد داخل <code dir="ltr">try</code> بگذارید.

### ❓ پیش‌بینی کن

</div>

````python
data = {"a": "1", "b": "x"}
for key in ["a", "b", "c"]:
    try:
        print(key, int(data[key]))
    except ValueError:
        print(key, "not a number")
    except KeyError as e:
        print(key, "missing:", e)
````

<div dir="rtl">

🔍 <code dir="ltr">a 1</code>، سپس <code dir="ltr">b not a number</code>، سپس <code dir="ltr">c missing: 'c'</code>. پیام یک <code dir="ltr">KeyError</code> خودِ کلید ناموجود است.

### 💡 سلسله‌مراتب استثناها

استثناها خانواده‌اند: مثلاً <code dir="ltr">KeyError</code> و <code dir="ltr">IndexError</code> هر دو زیرمجموعه‌ی <code dir="ltr">LookupError</code> هستند و تقریباً همه‌ی استثناها زیرمجموعه‌ی <code dir="ltr">Exception</code>. یک <code dir="ltr">except</code> همه‌ی **زیرمجموعه‌های** نوع خودش را هم می‌گیرد؛ پس <code dir="ltr">except LookupError:</code> هر دو خطا را می‌گیرد.

⚠️ پایتون <code dir="ltr">except</code>ها را **به ترتیب** بررسی می‌کند و اولین مورد منطبق را اجرا می‌کند. اگر <code dir="ltr">except Exception:</code> را **اول** بنویسید، <code dir="ltr">except</code>های خاص‌تر بعدی هرگز اجرا نمی‌شوند. همیشه خاص‌ترین نوع را اول بگذارید.

</div>

## ۳. <code dir="ltr">else</code> و <code dir="ltr">finally</code>

### 💡 مفهوم

دستور <code dir="ltr">try</code> دو بخش اختیاری دیگر هم دارد:

- <code dir="ltr">else</code>: فقط وقتی اجرا می‌شود که در <code dir="ltr">try</code> **هیچ خطایی** رخ نداده باشد.
- <code dir="ltr">finally</code>: **همیشه** اجرا می‌شود؛ چه خطا رخ دهد چه نه، چه گرفته شود چه نه، حتی اگر داخل <code dir="ltr">try</code> یا <code dir="ltr">except</code> دستور <code dir="ltr">return</code> اجرا شده باشد. جای کارهای «جمع‌وجور کردن» است: بستن فایل، آزاد کردن منابع، ثبت پایان کار.

</div>

<p align="center"><img src="images/s15_try_flow.png" width="620" alt="s15_try_flow.png"></p>

<p align="center" dir="rtl"><i>مسیر اجرای <code dir="ltr">try</code> / <code dir="ltr">except</code> / <code dir="ltr">else</code> / <code dir="ltr">finally</code>.</i></p>

````python
def parse(s):
    try:
        value = float(s)
    except ValueError:
        print("  except: bad value", repr(s))
        value = None
    else:
        print("  else: parsed", value)
    finally:
        print("  finally: done with", repr(s))
    return value

for s in ["2.5", "oops"]:
    print("result:", parse(s))
````

<div dir="rtl">

### 🔍 کد را بخوان

- برای <code dir="ltr">"2.5"</code>: <code dir="ltr">try</code> موفق است؛ پس <code dir="ltr">else</code> و بعد <code dir="ltr">finally</code> اجرا می‌شوند.
- برای <code dir="ltr">"oops"</code>: <code dir="ltr">try</code> شکست می‌خورد؛ <code dir="ltr">except</code> و بعد <code dir="ltr">finally</code> اجرا می‌شوند و <code dir="ltr">else</code> نه.

**چرا <code dir="ltr">else</code>؟** چرا کد بلوک <code dir="ltr">else</code> را در انتهای <code dir="ltr">try</code> ننویسیم؟ چون در آن صورت خطاهای **آن** کد هم توسط <code dir="ltr">except</code> گرفته می‌شد. <code dir="ltr">else</code> کمک می‌کند <code dir="ltr">try</code> فقط شامل خط پرخطر بماند (اشتباه رایج بخش ۲).

### ❓ پیش‌بینی کن

اگر خطایی رخ دهد که **هیچ** <code dir="ltr">except</code> منطبقی ندارد، چه اتفاقی می‌افتد؟

</div>

````python
def risky():
    try:
        print("working")
        return [1, 2, 3][10]
    except KeyError:
        print("key problem")
    finally:
        print("cleanup")

risky()
````

<div dir="rtl">

🔍 ابتدا <code dir="ltr">working</code>، سپس <code dir="ltr">IndexError</code> رخ می‌دهد که با <code dir="ltr">except KeyError</code> منطبق نیست. با این حال **<code dir="ltr">cleanup</code> چاپ می‌شود** و **بعد** خطا به بیرون منتقل می‌شود و traceback را می‌بینید. <code dir="ltr">finally</code> جلوی خطا را نمی‌گیرد؛ فقط تضمین می‌کند کار پایانی انجام شود.

### ⚠️ اشتباه رایج: <code dir="ltr">return</code> داخل <code dir="ltr">finally</code>

اگر در <code dir="ltr">finally</code> دستور <code dir="ltr">return</code> بنویسید، مقدار بازگشتی <code dir="ltr">try</code> را **جایگزین** می‌کند و حتی خطای در حال انتقال را **بی‌صدا از بین می‌برد**. در <code dir="ltr">finally</code> فقط کار پایانی انجام دهید، نه <code dir="ltr">return</code>.

</div>

## ۴. <code dir="ltr">raise</code>

### 💡 ایجاد خطا در توابع خودتان

تا اینجا خطاها را **گرفتیم**. اما وقتی تابعی می‌نویسید که ورودی نامعتبر گرفته، بهترین کار اغلب **ایجاد** خطا با پیام روشن است، نه برگرداندن یک مقدار عجیب یا ادامه‌ی بی‌صدا:

</div>

````python
def dot(a, b):
    if len(a) != len(b):
        raise ValueError(f"length mismatch: {len(a)} != {len(b)}")
    return sum(x * y for x, y in zip(a, b))

print(dot([1, 2, 3], [4, 5, 6]))
print(dot([1, 2, 3], [4, 5]))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">raise ValueError("...")</code> یک شیء خطا با پیام دلخواه می‌سازد و آن را «پرتاب» می‌کند؛ اجرای تابع همان‌جا تمام می‌شود و خطا به فراخواننده می‌رسد، که می‌تواند آن را با <code dir="ltr">try</code> بگیرد یا اجازه دهد برنامه متوقف شود.
- بدون این بررسی، <code dir="ltr">zip</code> بی‌صدا لیست بلندتر را کوتاه می‌کرد (جلسه‌ی ۸) و تابع یک **عدد غلط** برمی‌گرداند. خطای صریح بسیار بهتر از جواب غلطِ بی‌صداست؛ وقتی در NumPy دو بردار با طول متفاوت را ضرب داخلی می‌کنید، دقیقاً چنین خطایی می‌گیرید.
- قرارداد رایج: نوع نادرست، <code dir="ltr">TypeError</code>؛ نوع درست اما مقدار نامعتبر، <code dir="ltr">ValueError</code>.

### 💡 گرفتن، کاری کردن و دوباره پرتاب کردن

گاهی می‌خواهید خطا را ببینید (مثلاً ثبت کنید) اما تصمیمش را به فراخواننده بسپارید. <code dir="ltr">raise</code> **به‌تنهایی** داخل <code dir="ltr">except</code>، همان خطا را دوباره پرتاب می‌کند:

</div>

````python
def load_value(s):
    try:
        return float(s)
    except ValueError:
        print("log: could not parse", repr(s))
        raise

try:
    load_value("12,5")
except ValueError as e:
    print("caller got:", e)
````

### 💡 <code dir="ltr">next</code> و <code dir="ltr">StopIteration</code>

### سؤال ۴ — لگاریتم ایمن: تابع زیان آنتروپی متقاطع

در دسته‌بندی دودویی، **log loss** (آنتروپی متقاطع) میزان خطای احتمال‌های پیش‌بینی‌شده را می‌سنجد: برای برچسب واقعی <code dir="ltr">y</code> (صفر یا یک) و احتمال پیش‌بینی‌شده‌ی <code dir="ltr">p</code>، زیان برابر <code dir="ltr">-(y·log(p) + (1-y)·log(1-p))</code> است و میانگین آن روی همه‌ی نمونه‌ها گزارش می‌شود.

</div>

```python
import math

def log_loss(y_true, p_pred, eps=1e-15):
    if len(y_true) != len(p_pred):
        raise ValueError("length mismatch")
    total = 0.0
    for y, p in zip(y_true, p_pred):
        if not 0 <= p <= 1:
            raise ValueError(f"probability out of range: {p}")
        p = min(max(p, eps), 1 - eps)
        total += -(y * math.log(p) + (1 - y) * math.log(1 - p))
    return total / len(y_true)
```

<div dir="rtl">

الف) <code dir="ltr">log&#95;loss([1, 0, 1], [0.9, 0.2, 0.6])</code> را تا چهار رقم اعشار حساب کنید.

ب) اگر خط <code dir="ltr">p = min(max(p, eps), 1 - eps)</code> حذف شود، <code dir="ltr">log&#95;loss([1, 0], [1.0, 0.0])</code> چه می‌شود؟ با این خط چه مقداری برمی‌گرداند؟

ج) <code dir="ltr">log&#95;loss([1], [0.0])</code> با این خط تقریباً چقدر است؟ این عدد درباره‌ی مدلی که با اطمینان کامل اشتباه می‌کند چه می‌گوید؟

د) چرا برای <code dir="ltr">p = 1.2</code> خطا ایجاد می‌کنیم اما <code dir="ltr">p = 0</code> را بی‌صدا «اصلاح» می‌کنیم؟

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">-(log 0.9 + log 0.8 + log 0.6) / 3 = (0.1054 + 0.2231 + 0.5108) / 3 ≈ 0.2798</code>. (برای نمونه‌ی دوم <code dir="ltr">y = 0</code>، پس جمله‌ی <code dir="ltr">log(1 - 0.2) = log 0.8</code> حساب می‌شود.)

ب) بدون آن خط، برای نمونه‌ی اول جمله‌ی <code dir="ltr">(1 - y)·log(1 - p)</code> برابر <code dir="ltr">0 · log(0)</code> است و <code dir="ltr">math.log(0)</code> خطای <code dir="ltr">ValueError: math domain error</code> می‌دهد؛ حتی با اینکه ضریبش صفر است، پایتون اول <code dir="ltr">log</code> را حساب می‌کند. با آن خط، <code dir="ltr">p</code> به <code dir="ltr">1 - 1e-15</code> محدود می‌شود و نتیجه تقریباً <code dir="ltr">0</code> است (حدود <code dir="ltr">1e-15</code>): پیش‌بینی‌های کاملاً درست، زیان صفر دارند.

### سؤال ۵ — تلاش دوباره (retry)

وقتی داده را از یک سرور یا پایگاه داده می‌گیریم، گاهی اتصال موقتاً شکست می‌خورد. تابع <code dir="ltr">fetch</code> زیر این را شبیه‌سازی می‌کند (<code dir="ltr">outcomes</code> تعیین می‌کند هر تلاش موفق است یا نه):

</div>

```python
outcomes = iter([False, False, True])

def fetch():
    if next(outcomes):
        return "data"
    raise ConnectionError("server busy")

def retry(func, attempts=3):
    for i in range(1, attempts + 1):
        try:
            result = func()
        except ConnectionError as e:
            print(f"attempt {i} failed: {e}")
            if i == attempts:
                raise
        else:
            print(f"attempt {i} ok")
            return result

print(retry(fetch))
```

<div dir="rtl">

الف) خروجی را بنویسید.

ب) اگر <code dir="ltr">outcomes = iter([False, False, False, True])</code> باشد، خروجی <code dir="ltr">retry(fetch)</code> چیست و برنامه چطور تمام می‌شود؟ با <code dir="ltr">retry(fetch, 5)</code> چطور؟

ج) نقش <code dir="ltr">raise</code> تنها در خط <code dir="ltr">if i == attempts:</code> چیست؟ اگر آن دو خط را حذف کنیم، تابع در حالت شکست کامل چه برمی‌گرداند و چرا این بد است؟

د) چرا فقط <code dir="ltr">ConnectionError</code> را می‌گیریم و نه <code dir="ltr">Exception</code> را؟

<details>
<summary><b>پاسخ</b></summary>

```python
def average_score(rows, column="score"):
    total = 0.0
    count = 0
    for row in rows:
        value = row.get(column)
        if value is None:
            continue
        try:
            total += float(value)
        except ValueError:
            continue
        count += 1
    if count == 0:
        raise ValueError(f"no valid values in column {column!r}")
    return total / count
```

def h(x):
    print("h start")
    return int(x)

def g(x):
    print("g start")
    try:
        v = h(x)
    except TypeError:
        print("g caught TypeError")
        v = 0
    print("g end")
    return v * 2

def f(x):
    try:
        return g(x)
    except ValueError as e:
        print("f caught:", e)
        return None
    finally:
        print("f finally")

for x in ["21", "abc", None]:
    print("->", f(x))

## خلاصه‌ی جلسه

- خطای نحوی قبل از اجرا و استثنا در حین اجرا رخ می‌دهد؛ traceback را از پایین به بالا بخوانید.
- با <code dir="ltr">try</code> / <code dir="ltr">except</code> فقط خطاهای **مشخص و مورد انتظار** را بگیرید؛ <code dir="ltr">except:</code> خالی و <code dir="ltr">try</code> بزرگ باگ‌ها را پنهان می‌کنند.
- <code dir="ltr">else</code> فقط در صورت موفقیت و <code dir="ltr">finally</code> همیشه اجرا می‌شود؛ در <code dir="ltr">finally</code> دستور <code dir="ltr">return</code> ننویسید.
- در توابع خودتان، ورودی نامعتبر را با <code dir="ltr">raise ValueError(...)</code> یا <code dir="ltr">raise TypeError(...)</code> و پیام روشن رد کنید؛ خطای صریح بهتر از جواب غلطِ بی‌صداست.
- «تبدیل بدون خطا» به معنی «مقدار معتبر» نیست؛ پاک‌سازی داده یعنی بررسی قالب **و** معنی.

**جلسه‌ی بعد:** کار با فایل؛ خواندن و نوشتن فایل‌های متنی، CSV و JSON، جایی که <code dir="ltr">FileNotFoundError</code> و داده‌ی خراب به‌وفور پیش می‌آیند.
