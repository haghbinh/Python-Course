## ۱. پایتون چیست و چرا برای علوم داده؟

### 💡 مفهوم

**پایتون (Python)** یک زبان برنامه‌نویسی سطح بالا است که خواندن کد آن تقریباً مثل خواندن یک متن انگلیسی ساده است. پایتون یک زبان **مفسری (interpreted)** است: برنامه‌ای به نام **مفسر پایتون** کد شما را خط به خط می‌خواند و همان لحظه اجرا می‌کند.

<div dir="ltr">

```text
  your code (.py / .ipynb)  --->  Python interpreter  --->  result
```

</div>

### چرا پایتون در علوم داده؟

- **خوانایی:** کد کوتاه و قابل فهم است؛ پس تمرکز شما روی مسئله است، نه روی جزئیات زبان.
- **کتابخانه‌های قدرتمند:** تقریباً هر کاری در علوم داده یک کتابخانه‌ی آماده در پایتون دارد:

| کار | کتابخانه | کجای این دوره؟ |
|---|---|---|
| محاسبات عددی و ماتریسی | NumPy | بخش ۳ |
| کار با جدول داده | Pandas | بخش ۳ |
| رسم نمودار | Matplotlib, Seaborn | بخش ۳ |
| یادگیری ماشین | scikit-learn | درس‌های بعدی |
| یادگیری عمیق | PyTorch, TensorFlow | درس‌های بعدی |

- **جامعه‌ی بزرگ:** برای تقریباً هر خطایی که ببینید، کسی قبلاً همان سؤال را پرسیده و جواب گرفته است.
- **رایگان و چندسکویی:** مفسر پایتون و تقریباً همه‌ی کتابخانه‌های آن رایگان‌اند و روی ویندوز، مک و لینوکس اجرا می‌شوند.

نمودار زیر نتیجه‌ی یک نظرسنجی از توسعه‌دهندگان پایتون است. ببینید کدام نقش شغلی بیشترین سهم را دارد:

</div>

<p align="center"><img src="images/python_roles.png" width="600" alt="python_roles.png"></p>

<p align="center" dir="rtl"><i>نقش‌های شغلی توسعه‌دهندگان پایتون؛ «تحلیلگر/دانشمند داده» با فاصله در صدر است.</i></p>

<div dir="rtl">

این دوره شما را تا جایی می‌برد که بتوانید وارد یادگیری ماشین و یادگیری عمیق شوید.

</div>

<div dir="rtl">

### 🔍 کد را بخوان

هنوز هیچ دستوری از پایتون یاد نگرفته‌اید؛ با این حال سعی کنید کد زیر را **بخوانید** و حدس بزنید چه کار می‌کند. این همان مهارتی است که در کل دوره تمرین می‌کنیم.

</div>

````python
scores = [17, 12.5, 19, 15]
average = sum(scores) / len(scores)
print("Average:", average)
````

<div dir="rtl">

- خط اول: چهار نمره را در یک **لیست** به نام <code dir="ltr">scores</code> نگه می‌دارد.
- خط دوم: مجموع نمره‌ها (<code dir="ltr">sum</code>) را بر تعداد آن‌ها (<code dir="ltr">len</code>) تقسیم می‌کند؛ یعنی **میانگین** را حساب می‌کند و در <code dir="ltr">average</code> می‌گذارد.
- خط سوم: متن <code dir="ltr">Average:</code> و مقدار میانگین را چاپ می‌کند.

### ❓ پیش‌بینی کن

اگر عدد <code dir="ltr">19</code> را در خط اول به <code dir="ltr">11</code> تغییر دهیم، خروجی چه می‌شود؟ اول حساب کنید، بعد کد بالا را تغییر دهید و اجرا کنید.

</div>

## ۳. Jupyter Notebook

### 💡 مفهوم

**نوت‌بوک (Notebook)** فایلی با پسوند <code dir="ltr">.ipynb</code> است که از تعدادی **سلول (cell)** ساخته شده است. همین جزوه‌ای که می‌خوانید یک نوت‌بوک است. دو نوع سلول اصلی داریم:

| نوع سلول | محتوا | با اجرا چه می‌شود؟ |
|---|---|---|
| **Code** | کد پایتون | کد اجرا و خروجی زیر سلول نمایش داده می‌شود. |
| **Markdown** | متن، جدول، فرمول | متن قالب‌بندی‌شده نمایش داده می‌شود. |

پشت هر نوت‌بوک یک **کرنل (kernel)** وجود دارد: یک مفسر پایتون که روشن می‌ماند و **همه‌ی متغیرها را در حافظه نگه می‌دارد**. به همین دلیل متغیری که در یک سلول ساخته‌اید، در سلول‌های دیگر هم در دسترس است.

نوت‌بوک برای علوم داده بسیار مناسب است، چون می‌توانید کد، خروجی، نمودار و توضیح را کنار هم داشته باشید:

</div>

<div dir="rtl">

### نصب و اجرا (روی ویندوز)

</div>

```text
pip install notebook
```

<div dir="rtl">

</div>

### 🔍 کد را بخوان: حافظه‌ی مشترک سلول‌ها

دو سلول زیر را به ترتیب اجرا کنید.

</div>

````python
x = 10
````

````python
x * 2
````

- سلول اول متغیر <code dir="ltr">x</code> را می‌سازد. خروجی ندارد، چون فقط یک **انتساب** است.

- در نوت‌بوک، اگر **آخرین خط** یک سلول یک عبارت باشد (مثل <code dir="ltr">x * 2</code>)، مقدار آن خودکار نمایش داده می‌شود؛ حتی بدون <code dir="ltr">print</code>.

به عدد داخل کروشه‌ی کنار هر سلول (مثل <code dir="ltr">[3]</code>) دقت کنید: این عدد **ترتیب اجرا** را نشان می‌دهد، نه ترتیب قرار گرفتن سلول در صفحه.

### ❓ پیش‌بینی کن

سلول اول زیر را یک بار و سلول دوم را **دو بار پشت سر هم** اجرا کنید. خروجی بار دوم چیست؟

## ۴. VS Code

**Visual Studio Code (VS Code)** یک ویرایشگر کد رایگان از مایکروسافت است. هم فایل‌های نوت‌بوک (<code dir="ltr">.ipynb</code>) را باز می‌کند و هم فایل‌های اسکریپت پایتون (<code dir="ltr">.py</code>) را. امکاناتی مثل **تکمیل خودکار کد**، **نمایش خطاها** هنگام نوشتن و **اشکال‌زدایی (debugging)** دارد و برای پروژه‌های بزرگ‌تر، ابزار اصلی بیشتر برنامه‌نویس‌هاست.

### نصب و آماده‌سازی (روی ویندوز)

1. از سایت [code.visualstudio.com](https://code.visualstudio.com) نسخه‌ی ویندوز را دانلود و نصب کنید.
2. VS Code را باز کنید و از نوار سمت چپ وارد بخش **Extensions** شوید (<kbd>Ctrl</kbd> + <kbd>Shift</kbd> + <kbd>X</kbd>).
3. عبارت <code dir="ltr">python</code> را جست‌وجو کنید و این دو افزونه از **Microsoft** را نصب کنید: **Python** و **Jupyter**.

</div>

<p align="center"><img src="images/vscode_extensions.png" width="700" alt="vscode_extensions.png"></p>

<div dir="rtl">

4. یک فایل <code dir="ltr">.ipynb</code> باز کنید. بالا سمت راست روی **Select Kernel** کلیک کنید و پایتونی را که نصب کرده‌اید انتخاب کنید.
5. اگر VS Code پیشنهاد نصب بسته‌ی **ipykernel** را داد، **Install** را بزنید.

### ⚠️ اشتباه رایج: نصب افزونه‌ی اشتباه

با جست‌وجوی <code dir="ltr">python</code> ده‌ها افزونه با اسم‌های مشابه ظاهر می‌شود. فقط افزونه‌ای را نصب کنید که زیر نامش **Microsoft** نوشته شده است.

### نوت‌بوک یا اسکریپت؟

| | نوت‌بوک <code dir="ltr">.ipynb</code> | اسکریپت <code dir="ltr">.py</code> |
|---|---|---|
| ساختار | سلول به سلول | یک فایل متنی، از بالا به پایین |
| اجرا | هر سلول جدا | کل فایل یک‌جا |
| مناسب برای | تحلیل داده، آموزش، آزمایش | برنامه‌های کامل، پروژه‌ها |

برای امتحان اسکریپت: در VS Code یک فایل جدید به نام <code dir="ltr">hello.py</code> بسازید، خط <code dir="ltr">print("Hello from a script")</code> را در آن بنویسید، ذخیره کنید و دکمه‌ی ▶ (Run) بالا سمت راست را بزنید. خروجی در پنجره‌ی **Terminal** پایین صفحه نمایش داده می‌شود.

### کدام محیط را انتخاب کنم؟

## ۶. اولین برنامه و خواندن پیام خطا

### 💡 مفهوم

تابع <code dir="ltr">print</code> هر چیزی را که داخل پرانتز به آن بدهید روی خروجی نمایش می‌دهد. متن باید داخل **نقل‌قول** (<code dir="ltr">"..."</code> یا <code dir="ltr">'...'</code>) باشد.

</div>

````python
print("Hello, World!")
````

<div dir="rtl">

### 🔍 کد را بخوان

در کد زیر، هر چیزی که بعد از علامت <code dir="ltr">#</code> بیاید یک **توضیح (comment)** است و پایتون آن را اجرا نمی‌کند. توضیح‌ها برای انسان‌ها نوشته می‌شوند، نه برای کامپیوتر.

</div>

````python
# print can take several values, separated by commas
print("Python", "for", "Data Science")

# the values are joined with a space by default;
# sep changes the separator
print("2026", "10", "30", sep="-")

print(2 + 3 * 4)   # Python computes the expression first, then prints it
````

<div dir="rtl">

- خط اول کد سه متن را می‌گیرد و با یک **فاصله** بینشان چاپ می‌کند.
- با <code dir="ltr">sep="-"</code> به جای فاصله، علامت <code dir="ltr">-</code> بین مقدارها گذاشته می‌شود.
- در خط آخر، پایتون اول حاصل عبارت را حساب می‌کند (ضرب قبل از جمع) و بعد آن را چاپ می‌کند.

### ❓ پیش‌بینی کن

خروجی هر یک از دو خط زیر چیست؟ (به تفاوت <code dir="ltr">"2 + 3"</code> و <code dir="ltr">2 + 3</code> دقت کنید.)

</div>

````python
print("2 + 3")
print(2 + 3)
````

<div dir="rtl">

### 💡 خواندن پیام خطا

خطا دیدن بخش عادی برنامه‌نویسی است. مهم این است که **پیام خطا را بخوانید**. پیام خطای پایتون را **از پایین به بالا** بخوانید:

- **خط آخر** مهم‌ترین خط است: **نوع خطا** و **توضیح** آن.
- خط‌های بالاتر نشان می‌دهند خطا **کجای کد** رخ داده است.

دو سلول زیر عمداً خطا دارند. اجرایشان کنید و پیام خطا را بخوانید.

</div>

````python
pritn("Hello")
````

<div dir="rtl">

🔍 خط آخر خطا <code dir="ltr">NameError: name 'pritn' is not defined</code> است؛ یعنی پایتون چیزی به نام <code dir="ltr">pritn</code> نمی‌شناسد. علت: **غلط املایی** در <code dir="ltr">print</code>.

</div>

````python
print("Hello"))
````

<div dir="rtl">

🔍 این بار خط آخر خطا <code dir="ltr">SyntaxError: unmatched ')'</code> است؛ یعنی **دستور زبان** کد غلط است: یک پرانتز بسته‌ی اضافه وجود دارد که هیچ پرانتز بازی برایش نیست. در خطای <code dir="ltr">SyntaxError</code> پایتون حتی شروع به اجرای کد نمی‌کند؛ علامت <code dir="ltr">^</code> در پیام خطا جای مشکل را نشان می‌دهد.

### ⚠️ اشتباه رایج

- فراموش کردن نقل‌قول دور متن: <code dir="ltr">print(Hello)</code> ← پایتون <code dir="ltr">Hello</code> را اسم یک متغیر فرض می‌کند و <code dir="ltr">NameError</code> می‌دهد.
- بزرگ و کوچک بودن حروف مهم است: <code dir="ltr">Print</code> با <code dir="ltr">print</code> فرق دارد.
- علامت نقل‌قول باید انگلیسی باشد (<code dir="ltr">"</code>) و نه علامت‌هایی مثل <code dir="ltr">«</code> یا <code dir="ltr">“</code> که از کیبورد فارسی یا Word کپی می‌شوند.

</div>

- تفاوت Jupyter و VS Code و تفاوت نوت‌بوک <code dir="ltr">.ipynb</code> و اسکریپت <code dir="ltr">.py</code> را بدانید.

### سؤال ۲ — ترتیب اجرای سلول‌ها

یک نوت‌بوک سه سلول دارد:

| سلول | کد |
|:---:|---|
| A | <code dir="ltr">a = 1</code> |
| B | <code dir="ltr">a = a + 10</code> |
| C | <code dir="ltr">print(a)</code> |

دانشجویی سلول‌ها را به این ترتیب اجرا کرده است: **A، B، B، C**. چه عددی چاپ می‌شود؟ اگر بعد از آن فقط **Restart & Run All** را بزند، چه عددی چاپ می‌شود؟

<details>
<summary><b>پاسخ</b></summary>

بار اول **21** چاپ می‌شود: بعد از A مقدار <code dir="ltr">a</code> برابر 1 است، هر بار اجرای B ده واحد اضافه می‌کند (اول 1، بعد 11، بعد 21).

بعد از Restart & Run All حافظه پاک می‌شود و سلول‌ها یک بار به ترتیب A، B، C اجرا می‌شوند؛ پس **11** چاپ می‌شود.

</details>
