## ۱. صورت مسئله

معاون آموزشی یک دانشکده، نمره‌های پایان ترم سه ورودی (۱۴۰۰، ۱۴۰۱ و ۱۴۰۲) را از سامانه‌ی آموزشی بیرون کشیده است. بخشی از نمره‌ها را کارمندان دستی وارد کرده‌اند و داده **بی‌نقص نیست**: سطرهای تکراری، نمره‌های نامعتبر، سطرهای ناقص و نوشتار ناهمسان در آن دیده می‌شود.

معاون از شما خواسته است:

- داده را **پاک‌سازی** کنید و دقیقاً بگویید با هر ایراد چه کرده‌اید،
- وضعیت تحصیلی هر دانشجو را مشخص کنید (معدل، مشروطی، ممتاز بودن، رتبه)،
- عملکرد هر **درس** و هر **ورودی** را تحلیل کنید و هر چیز غیرعادی را که می‌بینید گزارش دهید،
- و به یک **پرسش سیاستی** پاسخ دهید (الزام ۷).

همه‌ی نتایج را در یک **گزارش** ارائه می‌کنید که معاون، بدون دانستن برنامه‌نویسی، بتواند بر اساس آن تصمیم بگیرد.

## ۳. الزامات برنامه

records, rejected = clean_data(raw_text)

# rejected: list of (line_number, line, reason) tuples

1. **پاک‌سازی:** <code dir="ltr">clean&#95;data</code> باید ایرادهای داده را پیدا کند. سطری که با اطمینان قابل اصلاح است (مثلاً فاصله‌ی اضافه یا حروف بزرگ و کوچک ناهمسان) اصلاح شود؛ سطری که قابل اصلاح نیست، با شماره‌ی سطر و **دلیل** در <code dir="ltr">rejected</code> ثبت شود. هیچ سطری نباید بی‌صدا حذف شود. نام‌ها را به یک شکل یکسان (مثلاً <code dir="ltr">Parisa Abbasi</code>) و نام درس‌ها را دقیقاً به شکل جدول بالا برگردانید. در گزارش، برای هر نوع ایراد بگویید چند بار دیدید و چه تصمیمی گرفتید و چرا.
2. **وضعیت هر دانشجو:** تعداد واحد اخذشده، تعداد واحد قبول‌شده، معدل و وضعیت (مشروط، عادی، ممتاز).
3. **رتبه‌بندی:** رتبه‌ی همه‌ی دانشجویان بر اساس معدل. دانشجویان با معدل برابر رتبه‌ی **برابر** می‌گیرند و رتبه‌ی بعدی به تعداد آن‌ها جلو می‌رود (مثلاً <code dir="ltr">1, 2, 2, 4</code>). ده نفر اول را گزارش کنید.
4. **آمار هر درس:** تعداد نمره، میانگین، میانه، انحراف معیار، کمینه، بیشینه و درصد مردودی. توزیع نمره‌های هر درس را با یک **نمودار میله‌ای متنی** (با کاراکتر <code dir="ltr">#</code>) در بازه‌های <code dir="ltr">[0,10)</code>، <code dir="ltr">[10,12)</code>، <code dir="ltr">[12,14)</code>، <code dir="ltr">[14,16)</code>، <code dir="ltr">[16,18)</code> و <code dir="ltr">[18,20]</code> نمایش دهید.
5. **مقایسه‌ی ورودی‌ها:** برای هر ورودی، تعداد دانشجو، میانگین معدل، تعداد و درصد مشروطی.
6. **موارد غیرعادی:** هر درس، ورودی یا الگوی غیرعادی را که در داده می‌بینید (پس از پاک‌سازی) گزارش و با عدد مستند کنید. توضیح دهید چه چیزی آن را غیرعادی می‌کند و معاون باید درباره‌اش چه کند.
7. **پرسش سیاستی:** معاون پیشنهاد کرده است که به نمره‌های درسی که **بیشترین درصد مردودی** را دارد، <code dir="ltr">2</code> نمره اضافه شود (سقف <code dir="ltr">20</code>). حساب کنید این کار چند دانشجو را از مشروطی خارج می‌کند، درصد مردودی آن درس چقدر می‌شود و رتبه‌های بالای جدول چه تغییری می‌کنند. نظر خود را درباره‌ی این پیشنهاد، با دلیل، بنویسید.
8. **فهرست‌های نهایی:** فهرست دانشجویان مشروط (از کمترین معدل) و فهرست دانشجویان ممتاز (از بیشترین معدل)، با شماره‌ی دانشجویی، نام و معدل.
