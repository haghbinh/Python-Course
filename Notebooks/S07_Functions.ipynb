## ۱. چرا تابع؟ تعریف و فراخوانی

### 💡 مفهوم

فرض کنید می‌خواهید جمع اعداد ۱ تا ۲۵، بعد ۵۰ تا ۷۶ و بعد ۹۰ تا ۱۰۱ را حساب کنید. با چیزهایی که تا اینجا یاد گرفته‌اید، باید یک حلقه را سه بار با تغییرات کوچک تکرار کنید:

</div>

````python
total = 0
for i in range(1, 26):
    total += i
print("Sum of integers from 1 to 25 is:", total)

total = 0
for i in range(50, 77):
    total += i
print("Sum of integers from 50 to 76 is:", total)

total = 0
for i in range(90, 102):
    total += i
print("Sum of integers from 90 to 101 is:", total)
````

<div dir="rtl">

این کد سه ایراد دارد: **طولانی** است، اگر بخواهید روش محاسبه را عوض کنید باید **سه جا** را درست کنید، و احتمال **اشتباه کپی** زیاد است. **تابع (function)** یک بلوک کد **نام‌دار** است که یک بار تعریف می‌شود و هر چند بار لازم باشد **فراخوانی (call)** می‌شود:

</div>

````python
def range_sum(x, y):
    total = 0
    for i in range(x, y + 1):
        total += i
    return total

print("Sum of integers from 1 to 25 is:", range_sum(1, 25))
print("Sum of integers from 50 to 76 is:", range_sum(50, 76))
print("Sum of integers from 90 to 101 is:", range_sum(90, 101))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">def</code> یعنی «تعریف تابع». بعد از آن **نام** تابع، داخل پرانتز **پارامترها** (<code dir="ltr">x</code> و <code dir="ltr">y</code>) و در آخر دونقطه می‌آید.
- بدنه با تورفتگی مشخص می‌شود (مثل <code dir="ltr">if</code> و <code dir="ltr">for</code>).
- <code dir="ltr">return total</code> نتیجه را به جایی که تابع فراخوانی شده **برمی‌گرداند**.
- <code dir="ltr">range&#95;sum(1, 25)</code> یعنی «تابع را با <code dir="ltr">x = 1</code> و <code dir="ltr">y = 25</code> اجرا کن و حاصلش را بده».
- چون <code dir="ltr">range</code> عدد <code dir="ltr">stop</code> را شامل نمی‌شود (جلسه‌ی ۶)، داخل تابع <code dir="ltr">y + 1</code> نوشته‌ایم تا خود <code dir="ltr">y</code> هم جمع شود. این جزئیات حالا **فقط یک جا** نوشته شده است.

### ⚠️ اشتباه رایج

- **تعریف کردن، اجرا کردن نیست:** اجرای سلولی که فقط <code dir="ltr">def</code> دارد هیچ خروجی‌ای ندارد؛ تابع فقط وقتی اجرا می‌شود که **فراخوانی** شود.
- **فراخوانی قبل از تعریف:** پایتون کد را از بالا به پایین اجرا می‌کند؛ اگر تابعی را قبل از اجرای <code dir="ltr">def</code> آن صدا بزنید، <code dir="ltr">NameError</code> می‌گیرید. در نوت‌بوک یعنی: اول سلول تعریف را اجرا کنید.
- **فراموش کردن پرانتز:** <code dir="ltr">range&#95;sum</code> (بدون پرانتز) خود تابع است، نه نتیجه‌ی اجرای آن.

</div>

## ۲. پارامتر و مقدار بازگشتی

### 💡 مفهوم

- **پارامتر (parameter):** نامی که در تعریف تابع داخل پرانتز می‌آید (<code dir="ltr">def add(a, b)</code>).
- **آرگومان (argument):** مقداری که هنگام فراخوانی به تابع داده می‌شود (<code dir="ltr">add(5, 7)</code>).
- **<code dir="ltr">return</code>:** اجرای تابع را **تمام می‌کند** و یک مقدار را به فراخوانی‌کننده برمی‌گرداند.

نمودار زیر اتفاقی را که هنگام اجرای <code dir="ltr">result = add(5, 7)</code> می‌افتد نشان می‌دهد:

</div>

<p align="center"><img src="images/s07_call_flow.png" width="680" alt="s07_call_flow.png"></p>

<p align="center" dir="rtl"><i>۱) فراخوانی: آرگومان‌ها به پارامترها داده می‌شوند. ۲) <code dir="ltr">return</code>: مقدار حاصل به جای فراخوانی قرار می‌گیرد.</i></p>

````python
def add(a, b):
    return a + b

result = add(5, 7)
print("Sum is:", result)
print(add(2, 3) * 10)
````

<div dir="rtl">

🔍 یک فراخوانی مثل <code dir="ltr">add(2, 3)</code> را می‌توانید مثل یک **مقدار** در نظر بگیرید: پایتون اول تابع را اجرا می‌کند و بعد **مقدار برگشتی** (<code dir="ltr">5</code>) را به جای آن می‌گذارد. پس <code dir="ltr">add(2, 3) &#42; 10</code> یعنی <code dir="ltr">5 &#42; 10</code>.

### 💡 <code dir="ltr">print</code> یا <code dir="ltr">return</code>؟

این مهم‌ترین نکته‌ی این جلسه است. در جلسه‌ی ۴ دیدید که <code dir="ltr">print</code> فقط **نمایش** می‌دهد و حاصلش <code dir="ltr">None</code> است. تابعی که به جای <code dir="ltr">return</code> از <code dir="ltr">print</code> استفاده کند، چیزی **برنمی‌گرداند**:

### ❓ پیش‌بینی کن

قبل از اجرا بنویسید هر سلول چه خروجی‌ای دارد.

</div>

````python
def double_print(x):
    print(x * 2)

z = double_print(3)
print("z =", z)
````

````python
def double_return(x):
    return x * 2

z = double_return(3)
print("z =", z)
````

<div dir="rtl">

🔍 در سلول اول، <code dir="ltr">6</code> **داخل تابع** چاپ می‌شود، اما تابع <code dir="ltr">return</code> ندارد؛ پس مقدار برگشتی‌اش <code dir="ltr">None</code> است و <code dir="ltr">z = None</code> چاپ می‌شود. در سلول دوم، تابع چیزی چاپ نمی‌کند ولی <code dir="ltr">6</code> را **برمی‌گرداند**؛ پس <code dir="ltr">z = 6</code>.

**قاعده:** تابعی که **محاسبه** می‌کند باید نتیجه را <code dir="ltr">return</code> کند تا بتوان از آن در ادامه استفاده کرد (جمع کرد، مقایسه کرد، در جدول گذاشت). <code dir="ltr">print</code> را برای **گزارش به کاربر** نگه دارید.

### 💡 <code dir="ltr">return</code> اجرای تابع را تمام می‌کند

تابع می‌تواند چند <code dir="ltr">return</code> داشته باشد؛ **اولین** <code dir="ltr">return</code> که اجرا شود، تابع را تمام می‌کند و خط‌های بعدی اجرا نمی‌شوند:

</div>

````python
def grade(score):
    """Return the letter grade for a score between 0 and 20."""
    if score >= 17:
        return "A"
    if score >= 14:
        return "B"
    if score >= 10:
        return "C"
    return "F"

print(grade(18), grade(15), grade(11), grade(4))
````

<div dir="rtl">

### 🔍 کد را بخوان

- این همان منطق رتبه‌بندی جلسه‌ی ۵ است، اما حالا یک **ابزار قابل استفاده‌ی مجدد** شده است.
- چون هر <code dir="ltr">return</code> تابع را تمام می‌کند، به <code dir="ltr">elif</code> نیازی نیست: اگر <code dir="ltr">score &gt;= 17</code> باشد، تابع همان‌جا <code dir="ltr">"A"</code> را برمی‌گرداند و به خط‌های بعدی نمی‌رسد.
- رشته‌ی سه‌نقل‌قولی زیر <code dir="ltr">def</code> یک **docstring** است: توضیح کوتاهی درباره‌ی کار تابع. با <code dir="ltr">help(grade)</code> می‌توانید آن را ببینید. عادت کنید برای تابع‌هایتان docstring بنویسید.

### 💡 برگرداندن چند مقدار

تابع می‌تواند چند مقدار را با کاما برگرداند. هنگام فراخوانی، می‌توان آن‌ها را مستقیم در چند متغیر گرفت (همان انتساب چندگانه‌ی جلسه‌ی ۲):

</div>

````python
import math

def circle(r):
    area = round(math.pi * r ** 2, 4)
    perimeter = round(2 * math.pi * r, 4)
    return area, perimeter

a, p = circle(3)
print(f"Area = {a}, Perimeter = {p}")

both = circle(3)
print(both, type(both))
````

<div dir="rtl">

🔍 <code dir="ltr">return area, perimeter</code> در واقع یک **تاپل** (جلسه‌ی ۹) برمی‌گرداند. اگر آن را در یک متغیر بگیرید (<code dir="ltr">both</code>)، کل تاپل را دارید؛ اگر در دو متغیر بگیرید، باز (unpack) می‌شود.

در این مثال از ثابت <code dir="ltr">math.pi</code> در ماژول <code dir="ltr">math</code> استفاده کردیم. این ماژول را در بخش بعد معرفی می‌کنیم.

### ⚠️ اشتباه رایج

- **<code dir="ltr">return</code> داخل حلقه:** اگر <code dir="ltr">return</code> را با تورفتگی اشتباه داخل حلقه بگذارید، تابع در **اولین دور** تمام می‌شود.
- **برگرداندن نوع‌های مختلف:** تابعی که گاهی عدد و گاهی پیام متنی برمی‌گرداند (مثلاً <code dir="ltr">"Both numbers are equal"</code>) استفاده از نتیجه را سخت می‌کند. سعی کنید هر تابع همیشه یک **نوع** را برگرداند.

</div>

<div dir="rtl">

### 💡 تابع هم یک مقدار است

نام یک تابع (بدون پرانتز) به خود تابع اشاره می‌کند؛ پس می‌توان تابع را مثل هر مقدار دیگری به یک تابع دیگر **داد**:

</div>

````python
def square(x):
    return x * x

def apply_twice(f, x):
    return f(f(x))

print(apply_twice(square, 3))
````

<div dir="rtl">

🔍 <code dir="ltr">apply&#95;twice</code> تابع <code dir="ltr">square</code> را به‌عنوان پارامتر <code dir="ltr">f</code> می‌گیرد و دو بار روی <code dir="ltr">3</code> اعمال می‌کند: <code dir="ltr">square(square(3))</code> یعنی <code dir="ltr">square(9)</code> که <code dir="ltr">81</code> است. دقت کنید هنگام دادن تابع **پرانتز نمی‌گذاریم**: <code dir="ltr">square</code> خود تابع است، ولی <code dir="ltr">square(3)</code> نتیجه‌ی اجرای آن (<code dir="ltr">9</code>).

این ایده در یادگیری ماشین همه‌جا هست: الگوریتم‌های آموزش، **تابع خطا** یا **تابع فعال‌سازی** را به‌عنوان ورودی می‌گیرند. در جلسه‌ی ۱۳ بیشتر به آن می‌پردازیم.

</div>

## ۳. توابع آماده: <code dir="ltr">math</code> و <code dir="ltr">random</code>

### 💡 مفهوم

لازم نیست همه‌ی تابع‌ها را خودتان بنویسید. پایتون دو منبع تابع آماده دارد:

- **تابع‌های داخلی (built-in):** همیشه در دسترس‌اند و <code dir="ltr">import</code> لازم ندارند. تا اینجا خیلی از آن‌ها را دیده‌اید: <code dir="ltr">print</code>، <code dir="ltr">len</code>، <code dir="ltr">type</code>، <code dir="ltr">int</code>، <code dir="ltr">round</code>، <code dir="ltr">abs</code>، <code dir="ltr">min</code>، <code dir="ltr">max</code>، <code dir="ltr">sum</code>.
- **کتابخانه‌ی استاندارد:** مجموعه‌ای از **ماژول**‌ها که همراه پایتون نصب می‌شوند. برای استفاده از هر ماژول، اول آن را با <code dir="ltr">import</code> وارد می‌کنید و بعد تابع‌هایش را با نام ماژول و یک نقطه صدا می‌زنید: <code dir="ltr">math.sqrt(16)</code>.

سازوکار کامل ماژول‌ها (و ساختن ماژول خودتان) را در جلسه‌ی ۱۷ می‌بینیم. در این جلسه با دو ماژول پرکاربرد آشنا می‌شوید. هر تابع آماده هم دقیقاً مثل تابع‌های خودتان است: **آرگومان** می‌گیرد، کاری انجام می‌دهد و **مقدار برمی‌گرداند**.

### ماژول <code dir="ltr">math</code>

| نام | کار | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">math.sqrt(x)</code> | جذر | <code dir="ltr">math.sqrt(16)</code> | <code dir="ltr">4.0</code> |
| <code dir="ltr">math.exp(x)</code> | <code dir="ltr">e</code> به توان <code dir="ltr">x</code> | <code dir="ltr">math.exp(1)</code> | <code dir="ltr">2.718...</code> |
| <code dir="ltr">math.log(x)</code> | لگاریتم طبیعی | <code dir="ltr">math.log(math.e)</code> | <code dir="ltr">1.0</code> |
| <code dir="ltr">math.log(x, base)</code> | لگاریتم در مبنای دلخواه | <code dir="ltr">math.log(8, 2)</code> | <code dir="ltr">3.0</code> |
| <code dir="ltr">math.floor(x)</code> | بزرگ‌ترین عدد صحیح کوچک‌تر یا مساوی | <code dir="ltr">math.floor(-2.5)</code> | <code dir="ltr">-3</code> |
| <code dir="ltr">math.ceil(x)</code> | کوچک‌ترین عدد صحیح بزرگ‌تر یا مساوی | <code dir="ltr">math.ceil(-2.5)</code> | <code dir="ltr">-2</code> |
| <code dir="ltr">math.isclose(a, b)</code> | مقایسه‌ی امن دو <code dir="ltr">float</code> | <code dir="ltr">math.isclose(0.1 + 0.2, 0.3)</code> | <code dir="ltr">True</code> |
| <code dir="ltr">math.pi</code>، <code dir="ltr">math.e</code> | ثابت‌ها (تابع نیستند، پرانتز ندارند) | <code dir="ltr">math.pi</code> | <code dir="ltr">3.14159...</code> |

<code dir="ltr">exp</code> و <code dir="ltr">log</code> در یادگیری ماشین همه‌جا هستند: در تابع سیگموید، softmax و تابع خطای cross-entropy.

</div>

````python
import math

print(math.sqrt(2))
print(math.exp(0), math.log(1))
print(math.log(1024, 2))
print(math.floor(2.7), math.ceil(2.1), int(-2.7), math.floor(-2.7))
````

<div dir="rtl">

🔍 به خط آخر دقت کنید: برای اعداد منفی، <code dir="ltr">int</code> (حذف قسمت اعشاری، جلسه‌ی ۲) با <code dir="ltr">math.floor</code> (گرد کردن به پایین) فرق دارد: <code dir="ltr">int(-2.7)</code> برابر <code dir="ltr">-2</code> و <code dir="ltr">math.floor(-2.7)</code> برابر <code dir="ltr">-3</code> است.

### ماژول <code dir="ltr">random</code>

| تابع | کار |
|---|---|
| <code dir="ltr">random.random()</code> | عدد اعشاری تصادفی در بازه‌ی <code dir="ltr">[0, 1)</code> |
| <code dir="ltr">random.uniform(a, b)</code> | عدد اعشاری تصادفی بین <code dir="ltr">a</code> و <code dir="ltr">b</code> |
| <code dir="ltr">random.randint(a, b)</code> | عدد صحیح تصادفی بین <code dir="ltr">a</code> و <code dir="ltr">b</code>، **هر دو شامل** |
| <code dir="ltr">random.gauss(mu, sigma)</code> | عدد تصادفی از توزیع نرمال با میانگین <code dir="ltr">mu</code> و انحراف معیار <code dir="ltr">sigma</code> |
| <code dir="ltr">random.choice(seq)</code> | یک عضو تصادفی از یک دنباله |
| <code dir="ltr">random.sample(seq, k)</code> | <code dir="ltr">k</code> عضو تصادفی **بدون تکرار** (لیست جدید) |
| <code dir="ltr">random.shuffle(lst)</code> | ترتیب اعضای لیست را **در جا** به هم می‌ریزد |
| <code dir="ltr">random.seed(n)</code> | نقطه‌ی شروع تولید اعداد تصادفی را ثابت می‌کند |

</div>

````python
import random

print(random.random())
print(random.uniform(-1, 1))
print(random.randint(1, 6))          # a dice roll
print(random.gauss(0, 1))            # a standard normal sample
print(random.choice(["rock", "paper", "scissors"]))
````

<div dir="rtl">

این سلول را چند بار اجرا کنید؛ هر بار عددهای متفاوتی می‌بینید.

### 💡 <code dir="ltr">seed</code>: تصادفی، اما تکرارپذیر

کامپیوتر در واقع اعداد «شبه‌تصادفی» تولید می‌کند: دنباله‌ای که از یک **نقطه‌ی شروع (seed)** با یک فرمول ثابت ساخته می‌شود. اگر نقطه‌ی شروع را با <code dir="ltr">random.seed</code> ثابت کنید، **همان دنباله** دوباره تولید می‌شود.

### ❓ پیش‌بینی کن

آیا دو خط چاپ‌شده یکسان‌اند؟ اگر خط <code dir="ltr">random.seed(42)</code> دوم را حذف کنید چطور؟

</div>

````python
random.seed(42)
first = [random.randint(1, 100), random.randint(1, 100), random.random()]

random.seed(42)
second = [random.randint(1, 100), random.randint(1, 100), random.random()]

print(first)
print(second)
print(first == second)
````

🔍 با seed یکسان، دنباله‌ی یکسانی تولید می‌شود؛ پس <code dir="ltr">first == second</code> برابر <code dir="ltr">True</code> است. بدون seed دوم، دنباله از جایی که قبلاً رسیده بود ادامه پیدا می‌کند و اعداد فرق می‌کنند.

**چرا در علوم داده مهم است؟** تقسیم تصادفی داده، مقداردهی اولیه‌ی وزن‌های یک شبکه‌ی عصبی و بسیاری از الگوریتم‌ها به اعداد تصادفی وابسته‌اند. بدون seed ثابت، هر بار اجرا نتیجه‌ی کمی متفاوت می‌دهد و نمی‌توانید نتایجتان را تکرار یا با دیگران مقایسه کنید. در scikit-learn همین مفهوم با پارامتری به نام <code dir="ltr">random&#95;state</code> دیده می‌شود.

### ⚠️ اشتباه رایج

- **<code dir="ltr">randint</code> با <code dir="ltr">range</code> فرق دارد:** <code dir="ltr">random.randint(1, 6)</code> می‌تواند خود <code dir="ltr">6</code> را هم برگرداند، اما <code dir="ltr">range(1, 6)</code> شامل <code dir="ltr">6</code> نیست.
- **<code dir="ltr">random.shuffle</code> چیزی برنمی‌گرداند:** لیست را **در جا** تغییر می‌دهد و حاصلش <code dir="ltr">None</code> است (مثل <code dir="ltr">sort</code> در جلسه‌ی ۸). <code dir="ltr">x = random.shuffle(x)</code> لیست را از دست می‌دهد.
- **نام فایل خودتان را <code dir="ltr">random.py</code> یا <code dir="ltr">math.py</code> نگذارید:** در این صورت <code dir="ltr">import random</code> به جای ماژول پایتون، فایل شما را وارد می‌کند و خطاهای عجیبی مثل <code dir="ltr">AttributeError: module 'random' has no attribute 'randint'</code> می‌گیرید.

</div>

## ۵. <code dir="ltr">&#42;args</code> و <code dir="ltr">&#42;&#42;kwargs</code>

### 💡 مفهوم

گاهی تعداد ورودی‌ها از قبل معلوم نیست. مثلاً <code dir="ltr">print</code> هر تعداد مقداری را که بدهید می‌پذیرد. پایتون برای این حالت دو شکل ویژه دارد:

| شکل | چه چیزی را جمع می‌کند | داخل تابع از چه نوعی است |
|---|---|---|
| <code dir="ltr">&#42;args</code> | همه‌ی آرگومان‌های **موقعیتی** اضافه | تاپل |
| <code dir="ltr">&#42;&#42;kwargs</code> | همه‌ی آرگومان‌های **نام‌دار** اضافه | دیکشنری |

اسم‌های <code dir="ltr">args</code> و <code dir="ltr">kwargs</code> فقط قرارداد هستند؛ چیزی که اهمیت دارد علامت <code dir="ltr">&#42;</code> و <code dir="ltr">&#42;&#42;</code> است.

</div>

````python
def mean(*values):
    print("values =", values)
    return sum(values) / len(values)

print(mean(17, 12.5, 19, 15))
print(mean(10, 20))
````

````python
def describe(name, **info):
    print("name:", name)
    for key in info:
        print(" ", key, "=", info[key])

describe("Ali", age=21, city="Bushehr", major="Statistics")
````

<div dir="rtl">

### 🔍 کد را بخوان

- در <code dir="ltr">mean</code>، همه‌ی آرگومان‌ها داخل یک **تاپل** به نام <code dir="ltr">values</code> جمع می‌شوند؛ خروجی خط اول این را نشان می‌دهد.
- در <code dir="ltr">describe</code>، آرگومان اول به <code dir="ltr">name</code> می‌رسد و بقیه‌ی آرگومان‌های نام‌دار داخل **دیکشنری** <code dir="ltr">info</code> جمع می‌شوند. (دیکشنری‌ها را در جلسه‌ی ۱۰ می‌خوانیم؛ فعلاً بدانید <code dir="ltr">info[key]</code> مقدار مربوط به هر نام را می‌دهد.)

لازم نیست این دو شکل را زیاد بنویسید، اما در **مستندات کتابخانه‌ها** مدام آن‌ها را می‌بینید؛ مثلاً امضای بسیاری از تابع‌های Pandas با <code dir="ltr">&#42;&#42;kwargs</code> تمام می‌شود، یعنی «پارامترهای اضافه را به تابع دیگری منتقل می‌کنم».

</div>

<div dir="rtl">

## ۶. دامنه‌ی متغیرها

### 💡 مفهوم

**دامنه (scope)** یعنی یک نام در کدام قسمت برنامه **شناخته می‌شود**:

- **متغیر محلی (local):** متغیری که **داخل** تابع ساخته می‌شود (از جمله پارامترها). فقط داخل همان تابع و فقط هنگام اجرای آن وجود دارد.
- **متغیر سراسری (global):** متغیری که **بیرون** از همه‌ی تابع‌ها ساخته می‌شود. داخل تابع‌ها هم قابل **خواندن** است.

اگر داخل تابع نامی را **انتساب** کنید، یک متغیر **محلی جدید** ساخته می‌شود؛ حتی اگر متغیری سراسری با همان نام وجود داشته باشد.

</div>

````python
def f(x):
    y = 3          # local variable
    return x * y

y = 1              # global variable
z = f(2)
print("z =", z)
print("y =", y)
````

<div dir="rtl">

### 🔍 کد را بخوان

- داخل <code dir="ltr">f</code>، انتساب <code dir="ltr">y = 3</code> یک <code dir="ltr">y</code> **محلی** می‌سازد. <code dir="ltr">x &#42; y</code> از همین <code dir="ltr">y</code> محلی استفاده می‌کند: <code dir="ltr">2 &#42; 3 = 6</code>.
- <code dir="ltr">y</code> سراسری دست‌نخورده باقی می‌ماند: <code dir="ltr">y = 1</code>.
- بعد از پایان تابع، <code dir="ltr">x</code> و <code dir="ltr">y</code> محلی از بین می‌روند؛ اگر بیرون از تابع <code dir="ltr">print(x)</code> بنویسید، <code dir="ltr">NameError</code> می‌گیرید.

### ❓ پیش‌بینی کن

خروجی هر سلول چیست؟

</div>

````python
def f(x):
    return x * y      # y is not defined inside f

y = 1
print(f(2))
y = 10
print(f(2))
````

````python
p = 20

def demo():
    p = 10
    print("inside: p =", p)

demo()
print("outside: p =", p)
````

<div dir="rtl">

🔍 در سلول اول، <code dir="ltr">f</code> خودش <code dir="ltr">y</code> ندارد؛ پس <code dir="ltr">y</code> **سراسری** را در **لحظه‌ی اجرا** می‌خواند: بار اول <code dir="ltr">2</code> و بار دوم <code dir="ltr">20</code>. در سلول دوم، <code dir="ltr">p = 10</code> یک <code dir="ltr">p</code> **محلی** می‌سازد و <code dir="ltr">p</code> سراسری همان <code dir="ltr">20</code> می‌ماند.

### ⚠️ اشتباه رایج: <code dir="ltr">UnboundLocalError</code>

سلول زیر را اجرا کنید و پیام خطا را بخوانید:

</div>

````python
counter = 0

def increment():
    counter = counter + 1
    return counter

increment()
````

<div dir="rtl">

🔍 چون داخل تابع به <code dir="ltr">counter</code> **انتساب** شده، پایتون آن را در **کل تابع** محلی فرض می‌کند. پس در سمت راست <code dir="ltr">counter + 1</code>، <code dir="ltr">counter</code> محلی هنوز مقداری ندارد و خطای <code dir="ltr">UnboundLocalError: cannot access local variable 'counter'</code> رخ می‌دهد.

با کلمه‌ی <code dir="ltr">global</code> می‌توان این را حل کرد، ولی **راه بهتر** این است که مقدار را به‌عنوان **پارامتر** بگیرید و نتیجه را **برگردانید**:

<div dir="ltr">

```python
def increment(counter):
    return counter + 1

counter = increment(counter)
```

</div>

**قاعده‌ی طلایی:** یک تابع خوب همه‌ی چیزی را که لازم دارد از **پارامترها** می‌گیرد و نتیجه را با **<code dir="ltr">return</code>** برمی‌گرداند. چنین تابعی مستقل است و به سادگی خوانده و آزمایش می‌شود.

> **نکته برای آینده:** اگر یک **لیست** یا **دیکشنری** سراسری را داخل تابع **تغییر** دهید (مثلاً با <code dir="ltr">append</code>)، بدون <code dir="ltr">global</code> هم تغییر بیرون از تابع دیده می‌شود؛ چون به همان شیء اشاره می‌کنید (یادتان هست شکل «نام‌ها و مقدارها» در جلسه‌ی ۲؟). در جلسه‌ی ۸ این موضوع را مفصل می‌بینیم.

</div>

<div dir="rtl">

## 📌 آنچه باید بلد باشید

- تعریف تابع با <code dir="ltr">def</code>، فراخوانی آن، و اینکه تعریف کردن تابع آن را اجرا نمی‌کند.
- تفاوت پارامتر و آرگومان.
- **تفاوت <code dir="ltr">print</code> و <code dir="ltr">return</code>**؛ مقدار برگشتی تابعی که <code dir="ltr">return</code> ندارد <code dir="ltr">None</code> است.
- <code dir="ltr">return</code> اجرای تابع را فوراً تمام می‌کند؛ اثر <code dir="ltr">return</code> با تورفتگی اشتباه داخل حلقه.
- برگرداندن چند مقدار و باز کردن آن‌ها در چند متغیر.
- برای هر فراخوانی بگویید هر آرگومان به کدام پارامتر می‌رسد (موقعیتی، نام‌دار، پیش‌فرض) و کدام فراخوانی خطا می‌دهد.
- تابع‌های پرکاربرد <code dir="ltr">math</code> و <code dir="ltr">random</code>؛ <code dir="ltr">randint</code> هر دو سر بازه را شامل می‌شود؛ <code dir="ltr">shuffle</code> در جا تغییر می‌دهد؛ <code dir="ltr">seed</code> نتایج تصادفی را تکرارپذیر می‌کند.
- <code dir="ltr">&#42;args</code> تاپل و <code dir="ltr">&#42;&#42;kwargs</code> دیکشنری است.
- تابع را می‌توان بدون پرانتز به تابع دیگر داد؛ تفاوت <code dir="ltr">f</code> و <code dir="ltr">f(x)</code>.
- مقدار پیش‌فرض پارامتر در لحظه‌ی **تعریف** تابع حساب می‌شود.
- متغیر محلی و سراسری؛ انتساب داخل تابع یک متغیر محلی می‌سازد؛ علت <code dir="ltr">UnboundLocalError</code>.

### سؤال ۱ — فراخوانی‌های تودرتو و مقدار پیش‌فرض

خروجی کد زیر را دقیقاً بنویسید. برای هر فراخوانی، مقدار هر پارامتر را مشخص کنید.

```python
rate = 2

def scale(x, factor=rate):
    return x * factor

def shift(x, by=1):
    return scale(x) + by

rate = 10
print(shift(3), shift(3, by=scale(1, rate)), scale(shift(0), factor=shift(1)))
```

- **نکته‌ی کلیدی:** مقدار پیش‌فرض <code dir="ltr">factor=rate</code> در **لحظه‌ی تعریف** تابع حساب می‌شود، نه در لحظه‌ی فراخوانی. در آن لحظه <code dir="ltr">rate</code> برابر <code dir="ltr">2</code> بود؛ تغییر بعدی آن به <code dir="ltr">10</code> روی پیش‌فرض اثری ندارد.
- <code dir="ltr">shift(3)</code> یعنی <code dir="ltr">scale(3) + 1</code> که برابر است با <code dir="ltr">3 &#42; 2 + 1 = 7</code>.
- <code dir="ltr">scale(1, rate)</code> یعنی <code dir="ltr">1 &#42; 10 = 10</code> (این بار <code dir="ltr">rate</code> صریحاً و با مقدار فعلی‌اش داده شده). پس <code dir="ltr">shift(3, by=10)</code> برابر <code dir="ltr">6 + 10 = 16</code> است.
- <code dir="ltr">shift(0)</code> برابر <code dir="ltr">0 + 1 = 1</code> و <code dir="ltr">shift(1)</code> برابر <code dir="ltr">2 + 1 = 3</code> است؛ پس <code dir="ltr">scale(1, factor=3)</code> برابر <code dir="ltr">3</code> است.

### سؤال ۲ — تابع‌های فعال‌سازی

در شبکه‌های عصبی، خروجی هر نورون از یک **تابع فعال‌سازی (activation function)** عبور می‌کند. سه تابع رایج را در نظر بگیرید:

</div>

```python
import math

def step(z):
    return 1 if z >= 0 else 0

def relu(z):
    return z if z > 0 else 0

def sigmoid(z):
    return 1 / (1 + math.exp(-z))
```

<div dir="rtl">

الف) جدول را کامل کنید (اعداد اعشاری را تا ۳ رقم گرد کنید؛ راهنمایی: <code dir="ltr">math.exp(-2)</code> تقریباً <code dir="ltr">0.135</code> است).

| عبارت | مقدار؟ |
|---|---|
| <code dir="ltr">step(0)</code> | |
| <code dir="ltr">relu(-2.5)</code> | |
| <code dir="ltr">relu(step(-1) - 3)</code> | |
| <code dir="ltr">sigmoid(0)</code> | |
| <code dir="ltr">sigmoid(2)</code> | |
| <code dir="ltr">sigmoid(-2) + sigmoid(2)</code> | |

ب) چرا <code dir="ltr">sigmoid(-2) + sigmoid(2)</code> این مقدار را دارد؟ آیا برای هر عدد <code dir="ltr">z</code> همین نتیجه برقرار است؟

ج) اگر در تعریف <code dir="ltr">relu</code> به جای <code dir="ltr">return</code> از <code dir="ltr">print</code> استفاده شده بود، عبارت <code dir="ltr">relu(step(-1) - 3) + 1</code> چه نتیجه‌ای داشت؟

<details>
<summary><b>پاسخ</b></summary>

الف)

| عبارت | مقدار |
|---|---|
| <code dir="ltr">step(0)</code> | <code dir="ltr">1</code> |
| <code dir="ltr">relu(-2.5)</code> | <code dir="ltr">0</code> |
| <code dir="ltr">relu(step(-1) - 3)</code> | <code dir="ltr">relu(0 - 3)</code> یعنی <code dir="ltr">0</code> |
| <code dir="ltr">sigmoid(0)</code> | <code dir="ltr">0.5</code> |
| <code dir="ltr">sigmoid(2)</code> | <code dir="ltr">1 / 1.135</code> یعنی حدود <code dir="ltr">0.881</code> |
| <code dir="ltr">sigmoid(-2) + sigmoid(2)</code> | <code dir="ltr">1</code> (پایتون به خاطر دقت محدود اعشاری ممکن است <code dir="ltr">0.9999999999999999</code> چاپ کند) |

ب) برای هر <code dir="ltr">z</code> داریم <code dir="ltr">sigmoid(-z) = 1 - sigmoid(z)</code>، پس مجموع همیشه <code dir="ltr">1</code> است. اثبات: <code dir="ltr">1 / (1 + e^z)</code> را در صورت و مخرج در <code dir="ltr">e^-z</code> ضرب کنید تا <code dir="ltr">e^-z / (1 + e^-z)</code> به دست آید که برابر <code dir="ltr">1 - sigmoid(z)</code> است. (این تقارن در رگرسیون لجستیک استفاده می‌شود: اگر احتمال کلاس ۱ برابر <code dir="ltr">sigmoid(z)</code> باشد، احتمال کلاس ۰ برابر <code dir="ltr">sigmoid(-z)</code> است.)

ج) <code dir="ltr">relu</code> مقدار <code dir="ltr">0</code> را **چاپ** می‌کرد ولی <code dir="ltr">None</code> برمی‌گرداند؛ در نتیجه <code dir="ltr">None + 1</code> خطای <code dir="ltr">TypeError</code> می‌داد. تابع‌هایی که قرار است با هم **ترکیب** شوند (مثل لایه‌های یک شبکه) حتماً باید <code dir="ltr">return</code> کنند.

</details>

### سؤال ۳ — پرسپترون

ساده‌ترین نورون مصنوعی (**پرسپترون**) مجموع وزن‌دار ورودی‌ها را حساب می‌کند و آن را از تابع <code dir="ltr">step</code> عبور می‌دهد:

</div>

```python
def neuron(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b
    return 1 if z > 0 else 0
```

<div dir="rtl">

الف) برای <code dir="ltr">w1 = w2 = 1</code> و <code dir="ltr">b = -1.5</code>، خروجی را برای هر چهار ترکیب <code dir="ltr">x1, x2</code> از مجموعه‌ی <code dir="ltr">{0, 1}</code> حساب کنید. این نورون کدام **دروازه‌ی منطقی** (<code dir="ltr">and</code>، <code dir="ltr">or</code>، ...) را پیاده‌سازی می‌کند؟

ب) فقط با تغییر <code dir="ltr">b</code> (و ثابت نگه داشتن وزن‌ها) نورون را به دروازه‌ی <code dir="ltr">or</code> تبدیل کنید. چه بازه‌ای از مقدارها برای <code dir="ltr">b</code> جواب می‌دهد؟

ج) وزن‌ها و بایاسی پیدا کنید که خروجی برابر <code dir="ltr">not x1</code> باشد (<code dir="ltr">x2</code> تأثیری نداشته باشد).

<details>
<summary><b>پاسخ</b></summary>

الف)

| <code dir="ltr">x1</code> | <code dir="ltr">x2</code> | <code dir="ltr">z</code> | خروجی |
|:---:|:---:|:---:|:---:|
| 0 | 0 | <code dir="ltr">-1.5</code> | 0 |
| 0 | 1 | <code dir="ltr">-0.5</code> | 0 |
| 1 | 0 | <code dir="ltr">-0.5</code> | 0 |
| 1 | 1 | <code dir="ltr">0.5</code> | 1 |

این دروازه‌ی **<code dir="ltr">and</code>** است.

ب) برای <code dir="ltr">or</code>، حالت <code dir="ltr">(0, 0)</code> باید <code dir="ltr">0</code> شود (<code dir="ltr">b &lt;= 0</code>، چون شرط <code dir="ltr">z &gt; 0</code> است) و حالت‌های با یک ورودی <code dir="ltr">1</code> باید <code dir="ltr">1</code> شوند (<code dir="ltr">1 + b &gt; 0</code> یعنی <code dir="ltr">b &gt; -1</code>). پس هر <code dir="ltr">b</code> در بازه‌ی <code dir="ltr">-1 &lt; b &lt;= 0</code>؛ مثلاً <code dir="ltr">b = -0.5</code>.

ج) مثلاً <code dir="ltr">w1 = -1</code>، <code dir="ltr">w2 = 0</code>، <code dir="ltr">b = 0.5</code>: برای <code dir="ltr">x1 = 0</code> مقدار <code dir="ltr">z = 0.5</code> و خروجی <code dir="ltr">1</code>؛ برای <code dir="ltr">x1 = 1</code> مقدار <code dir="ltr">z = -0.5</code> و خروجی <code dir="ltr">0</code>.

(یادگیری در شبکه‌ی عصبی یعنی همین: پیدا کردن خودکار وزن‌ها و بایاس‌هایی که خروجی درست بدهند.)

</details>

### سؤال ۴ — تابع خطا

برای سنجیدن کیفیت یک مدل، **خطا**ی هر پیش‌بینی (<code dir="ltr">y&#95;true - y&#95;pred</code>) را حساب می‌کنیم و آن‌ها را در یک عدد خلاصه می‌کنیم. دانشجویی تابع **میانگین مربعات خطا (MSE)** را این‌طور نوشته است:

</div>

```python
def mse(errors):
    total = 0
    for e in errors:
        total += e
    return total ** 2 / len(errors)

print(mse([2, -2, 1, -1]))
```

<div dir="rtl">

الف) خروجی چیست؟ چرا این عدد نشان می‌دهد تابع اشتباه است؟

ب) تابع را درست کنید. خروجی درست برای همین ورودی چقدر است؟

ج) تابع <code dir="ltr">mae(errors)</code> (میانگین **قدر مطلق** خطاها) را بنویسید. برای ورودی <code dir="ltr">[2, -2, 1, -1]</code> و ورودی <code dir="ltr">[0, 0, 0, 6]</code>، مقدار MSE و MAE را حساب کنید و توضیح دهید کدام معیار به **یک خطای بزرگ** حساس‌تر است.

ج)

```python
    for e in errors:
        total += abs(e)

```

|---|---|---|

### سؤال ۵ — گرادیان کاهشی

**گرادیان کاهشی (gradient descent)** الگوریتم اصلی آموزش تقریباً همه‌ی مدل‌های یادگیری ماشین و شبکه‌های عصبی است: در هر قدم، <code dir="ltr">x</code> را کمی **خلاف جهت مشتق** حرکت می‌دهیم تا به کمینه‌ی تابع نزدیک شویم. کد زیر کمینه‌ی تابع <code dir="ltr">f(x) = (x - 3) &#42;&#42; 2</code> را پیدا می‌کند:

</div>

```python
def grad(x):
    return 2 * (x - 3)      # derivative of (x - 3) ** 2

x = 0.0
lr = 0.25                    # learning rate
for step in range(3):
    x = x - lr * grad(x)
    print(step, x)
```

<div dir="rtl">

الف) جدول دنبال کردن را بنویسید (مقدار <code dir="ltr">grad(x)</code> و <code dir="ltr">x</code> بعد از هر قدم) و خروجی را بنویسید. اگر حلقه خیلی بیشتر تکرار شود، <code dir="ltr">x</code> به چه عددی نزدیک می‌شود؟

ب) با <code dir="ltr">lr = 1.0</code> سه مقدار اول <code dir="ltr">x</code> چیست؟ با <code dir="ltr">lr = 1.25</code> چطور؟ در هر حالت چه اتفاقی می‌افتد؟

ج) فاصله‌ی <code dir="ltr">x</code> تا جواب (<code dir="ltr">x - 3</code>) در هر قدم در چه عددی ضرب می‌شود؟ با استفاده از آن توضیح دهید برای چه بازه‌ای از <code dir="ltr">lr</code> الگوریتم همگرا می‌شود.

<details>
<summary><b>پاسخ</b></summary>

الف)

| قدم | <code dir="ltr">x</code> قبل | <code dir="ltr">grad(x)</code> | <code dir="ltr">x</code> بعد |
|:---:|:---:|:---:|:---:|
| 0 | <code dir="ltr">0.0</code> | <code dir="ltr">-6.0</code> | <code dir="ltr">1.5</code> |
| 1 | <code dir="ltr">1.5</code> | <code dir="ltr">-3.0</code> | <code dir="ltr">2.25</code> |
| 2 | <code dir="ltr">2.25</code> | <code dir="ltr">-1.5</code> | <code dir="ltr">2.625</code> |

خروجی: <code dir="ltr">0 1.5</code>، <code dir="ltr">1 2.25</code>، <code dir="ltr">2 2.625</code>. با تکرار بیشتر، <code dir="ltr">x</code> به <code dir="ltr">3</code> (نقطه‌ی کمینه) نزدیک می‌شود.

ب) با <code dir="ltr">lr = 1.0</code>: <code dir="ltr">6.0</code>، <code dir="ltr">0.0</code>، <code dir="ltr">6.0</code>؛ <code dir="ltr">x</code> بین ۰ و ۶ **نوسان** می‌کند و هرگز همگرا نمی‌شود. با <code dir="ltr">lr = 1.25</code>: <code dir="ltr">7.5</code>، <code dir="ltr">-3.75</code>، <code dir="ltr">13.125</code>؛ <code dir="ltr">x</code> هر بار دورتر می‌شود و الگوریتم **واگرا** است.

ج) <code dir="ltr">x&#95;new - 3 = (x - 3) - lr &#42; 2 &#42; (x - 3) = (1 - 2 &#42; lr) &#42; (x - 3)</code>. پس فاصله در هر قدم در <code dir="ltr">(1 - 2 &#42; lr)</code> ضرب می‌شود. برای همگرایی باید قدر مطلق این عدد کمتر از ۱ باشد: <code dir="ltr">0 &lt; lr &lt; 1</code>. (برای <code dir="ltr">lr = 0.5</code> در یک قدم به جواب می‌رسیم.) این همان دلیلی است که در یادگیری عمیق، انتخاب **نرخ یادگیری** این‌قدر مهم است.

</details>

### سؤال ۶ — تابع به‌عنوان ورودی: مشتق عددی

در عمل همیشه فرمول مشتق را نداریم. می‌توان مشتق را **به‌طور عددی** تقریب زد:

</div>

```python
def numeric_grad(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

def loss(w):
    return (w - 3) ** 2

print(round(numeric_grad(loss, 0), 4))
print(round(numeric_grad(loss, 3), 4))
```

<div dir="rtl">

الف) خروجی دو خط آخر چیست؟ آن‌ها را با <code dir="ltr">grad</code> در سؤال ۵ مقایسه کنید.

ب) چرا در فراخوانی نوشته‌ایم <code dir="ltr">numeric&#95;grad(loss, 0)</code> و نه <code dir="ltr">numeric&#95;grad(loss(0), 0)</code>؟ دومی چه خطایی می‌دهد؟

ج) با استفاده از <code dir="ltr">numeric&#95;grad</code>، مشتق <code dir="ltr">sigmoid</code> (از سؤال ۲) را در <code dir="ltr">z = 0</code> تقریب بزنید. جواب را با فرمول <code dir="ltr">sigmoid(z) &#42; (1 - sigmoid(z))</code> مقایسه کنید.

### سؤال ۸ — جست‌وجوی تصادفی وزن‌ها

ساده‌ترین روش «آموزش» یک مدل، **جست‌وجوی تصادفی (random search)** است: وزن‌های تصادفی امتحان می‌کنیم تا به وزن‌هایی برسیم که خروجی درست می‌دهند. کد زیر از تابع <code dir="ltr">neuron</code> سؤال ۳ استفاده می‌کند:

</div>

```python
import random

def is_and(w1, w2, b):
    return (neuron(0, 0, w1, w2, b) == 0 and neuron(0, 1, w1, w2, b) == 0
            and neuron(1, 0, w1, w2, b) == 0 and neuron(1, 1, w1, w2, b) == 1)

def random_search(seed, max_tries=1000):
    random.seed(seed)
    for t in range(max_tries):
        w1 = random.uniform(-1, 1)
        w2 = random.uniform(-1, 1)
        b = random.uniform(-1, 1)
        if is_and(w1, w2, b):
            return w1, w2, b, t + 1
    return None

print(random_search(1))
print(random_search(1))
print(random_search(2))
```

<div dir="rtl">

الف) آیا دو خط اول خروجی یکسان‌اند؟ خط سوم چطور؟ چرا؟

ب) اگر خط <code dir="ltr">random.seed(seed)</code> را حذف کنیم، چه چیزی تغییر می‌کند؟ چرا در گزارش یک آزمایش یادگیری ماشین باید seed را هم گزارش کرد؟

ج) هر جواب پیدا شده در چه شرایطی صدق می‌کند؟ با کمک جواب سؤال ۳، توضیح دهید چرا مثلاً <code dir="ltr">w1 = w2 = 0.6</code> و <code dir="ltr">b = -0.9</code> قابل قبول است. آیا جواب این مسئله یکتاست؟

د) تابع <code dir="ltr">is&#95;xor</code> را مشابه <code dir="ltr">is&#95;and</code> برای دروازه‌ی **XOR** بنویسید (خروجی ۱ فقط وقتی دقیقاً یکی از ورودی‌ها ۱ باشد) و <code dir="ltr">random&#95;search</code> را با آن اجرا کنید. حتی با <code dir="ltr">max&#95;tries=100000</code> هم <code dir="ltr">None</code> برمی‌گردد. با رسم چهار نقطه‌ی ورودی روی صفحه، توضیح دهید چرا **هیچ** وزنی برای یک نورون تنها وجود ندارد.

<details>
<summary><b>پاسخ</b></summary>

الف) دو خط اول **دقیقاً یکسان‌اند**: هر بار <code dir="ltr">random&#95;search(1)</code> اجرا می‌شود، seed دوباره روی <code dir="ltr">1</code> تنظیم می‌شود و همان دنباله‌ی اعداد تصادفی، همان وزن‌ها و همان تعداد تلاش را تولید می‌کند. خط سوم با seed دیگری شروع می‌کند؛ پس وزن‌ها و تعداد تلاش‌ها (به احتمال زیاد) فرق دارند.

ب) هر اجرا وزن‌ها و تعداد تلاش‌های متفاوتی می‌دهد و نتیجه **تکرارپذیر** نیست. اگر seed گزارش نشود، کس دیگری (یا خود شما در آینده) نمی‌تواند همان نتیجه را بازتولید کند.

ج) شرط‌ها (با <code dir="ltr">z = w1 &#42; x1 + w2 &#42; x2 + b</code> و خروجی ۱ وقتی <code dir="ltr">z &gt; 0</code>): <code dir="ltr">b &lt;= 0</code>، <code dir="ltr">w2 + b &lt;= 0</code>، <code dir="ltr">w1 + b &lt;= 0</code> و <code dir="ltr">w1 + w2 + b &gt; 0</code>. برای <code dir="ltr">0.6، 0.6، -0.9</code>: <code dir="ltr">-0.9</code>، <code dir="ltr">-0.3</code>، <code dir="ltr">-0.3</code> همگی منفی‌اند و <code dir="ltr">0.3</code> مثبت است؛ پس قابل قبول است. جواب **یکتا نیست**: بی‌نهایت ترکیب این شرط‌ها را برآورده می‌کنند. (مدل‌های یادگیری ماشین هم معمولاً بی‌نهایت جواب «خوب» دارند و الگوریتم آموزش یکی از آن‌ها را پیدا می‌کند.)

د)

<div dir="ltr">

```python
def is_xor(w1, w2, b):
    return (neuron(0, 0, w1, w2, b) == 0 and neuron(0, 1, w1, w2, b) == 1
            and neuron(1, 0, w1, w2, b) == 1 and neuron(1, 1, w1, w2, b) == 0)
```

</div>

یک نورون تنها صفحه را با **یک خط راست** (<code dir="ltr">w1 &#42; x1 + w2 &#42; x2 + b = 0</code>) به دو نیم تقسیم می‌کند. در XOR، نقطه‌های <code dir="ltr">(0, 1)</code> و <code dir="ltr">(1, 0)</code> باید یک طرف خط و نقطه‌های <code dir="ltr">(0, 0)</code> و <code dir="ltr">(1, 1)</code> طرف دیگر باشند؛ اما این دو جفت روی دو قطر یک مربع‌اند و هیچ خط راستی آن‌ها را جدا نمی‌کند (داده **به‌طور خطی جدایی‌پذیر نیست**). جبری: از شرط‌ها <code dir="ltr">w1 + b &gt; 0</code> و <code dir="ltr">w2 + b &gt; 0</code> با هم جمع کنید: <code dir="ltr">w1 + w2 + 2b &gt; 0</code>؛ از <code dir="ltr">b &lt;= 0</code> نتیجه می‌شود <code dir="ltr">w1 + w2 + b &gt; -b &gt;= 0</code>، که با شرط <code dir="ltr">w1 + w2 + b &lt;= 0</code> تناقض دارد. همین مشکل، تاریخی‌ترین دلیل نیاز به شبکه‌های عصبی **چندلایه** است.

## خلاصه‌ی جلسه

- تابع یک بلوک کد نام‌دار است که با <code dir="ltr">def</code> تعریف و با <code dir="ltr">name(...)</code> فراخوانی می‌شود؛ کد تکراری را حذف می‌کند.
- **<code dir="ltr">return</code>** مقدار را برمی‌گرداند و تابع را تمام می‌کند؛ **<code dir="ltr">print</code>** فقط نمایش می‌دهد. تابع بدون <code dir="ltr">return</code> مقدار <code dir="ltr">None</code> برمی‌گرداند.
- آرگومان‌ها به ترتیب (موقعیتی) یا با نام (نام‌دار) به پارامترها می‌رسند؛ پارامترها می‌توانند مقدار پیش‌فرض داشته باشند.
- تابع‌های آماده‌ی <code dir="ltr">math</code> و <code dir="ltr">random</code> را با <code dir="ltr">import</code> وارد کنید؛ برای تکرارپذیری، <code dir="ltr">random.seed</code> را ثابت کنید.
- <code dir="ltr">&#42;args</code> آرگومان‌های اضافه را در یک تاپل و <code dir="ltr">&#42;&#42;kwargs</code> در یک دیکشنری جمع می‌کند.
- متغیرهای داخل تابع محلی‌اند؛ تابع خوب ورودی را از پارامتر می‌گیرد و خروجی را <code dir="ltr">return</code> می‌کند.

**جلسه‌ی بعد:** لیست‌ها، اولین ساختار داده‌ی پایتون.
