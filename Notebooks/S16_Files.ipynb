## ۱. چرا فایل؟ مسیرها و دستور <code dir="ltr">with</code>

### 💡 مفهوم

همه‌ی متغیرهایی که تا حالا ساخته‌اید در **حافظه‌ی موقت (RAM)** هستند و با بسته شدن برنامه از بین می‌روند. داده‌ی واقعی (جدول نمره‌ها، متن نظرها، وزن‌های یک مدل آموزش‌دیده) در **فایل** روی دیسک نگه‌داری می‌شود. هر کار علوم داده با «خواندن فایل» شروع و اغلب با «نوشتن فایل» تمام می‌شود.

**مسیر (path)** آدرس یک فایل است:

| نوع مسیر | مثال (ویندوز) | نسبت به چه؟ |
|---|---|---|
| مطلق (absolute) | <code dir="ltr">C:&#92;Users&#92;Ali&#92;data&#92;scores.csv</code> | از ریشه‌ی دیسک |
| نسبی (relative) | <code dir="ltr">data&#92;scores.csv</code> یا <code dir="ltr">data/scores.csv</code> | از **پوشه‌ی کاری فعلی** (در Jupyter معمولاً پوشه‌ی خود نوت‌بوک) |

در کد، مسیر نسبی را ترجیح دهید تا برنامه روی کامپیوتر دیگران هم کار کند. در پایتون می‌توانید حتی در ویندوز از <code dir="ltr">/</code> استفاده کنید. اگر <code dir="ltr">&#92;</code> می‌نویسید، رشته‌ی خام (<code dir="ltr">r"..."</code>، جلسه‌ی ۱۱) یادتان نرود.

</div>

````python
print("data\new_scores.csv")      # \n becomes a newline!
print(r"data\new_scores.csv")     # raw string: backslash is a normal character
print("data/new_scores.csv")      # forward slash also works on Windows
````

<div dir="rtl">

### 💡 باز کردن فایل با <code dir="ltr">with</code>

تابع <code dir="ltr">open(path, mode, encoding=...)</code> یک **شیء فایل** برمی‌گرداند. فایل بازشده باید **بسته** شود؛ وگرنه ممکن است داده‌ی نوشته‌شده کامل روی دیسک ذخیره نشود یا فایل برای برنامه‌های دیگر قفل بماند. دستور <code dir="ltr">with</code> این کار را خودکار انجام می‌دهد: به محض خروج از بلوک، **حتی اگر خطا رخ دهد**، فایل بسته می‌شود.

| حالت | معنی | اگر فایل وجود نداشته باشد | اگر وجود داشته باشد |
|---|---|---|---|
| <code dir="ltr">"r"</code> (پیش‌فرض) | خواندن | <code dir="ltr">FileNotFoundError</code> | از اول خوانده می‌شود |
| <code dir="ltr">"w"</code> | نوشتن | ساخته می‌شود | **محتوای قبلی پاک می‌شود** |
| <code dir="ltr">"a"</code> | افزودن به انتها | ساخته می‌شود | به انتهای آن اضافه می‌شود |
| <code dir="ltr">"x"</code> | ساختن فایل جدید | ساخته می‌شود | <code dir="ltr">FileExistsError</code> |

همه‌ی فایل‌های این جلسه در پوشه‌ی <code dir="ltr">s16&#95;data</code> کنار همین نوت‌بوک ساخته می‌شوند. سلول زیر این پوشه را می‌سازد (جزئیات <code dir="ltr">Path</code> را در بخش ۵ می‌بینید):

</div>

````python
from pathlib import Path
Path("s16_data").mkdir(exist_ok=True)

with open("s16_data/hello.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")
    f.write("second line\n")
    print("inside:", f.closed)

print("outside:", f.closed)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">open(..., "w", ...)</code> فایل را برای نوشتن باز می‌کند و شیء فایل را در نام <code dir="ltr">f</code> قرار می‌دهد.
- <code dir="ltr">f.write</code> **خودش خط جدید اضافه نمی‌کند**؛ <code dir="ltr">&#92;n</code> را باید خودتان بنویسید.
- بعد از بلوک <code dir="ltr">with</code>، ویژگی <code dir="ltr">f.closed</code> برابر <code dir="ltr">True</code> است: فایل بسته شده، بدون اینکه <code dir="ltr">f.close()</code> بنویسیم.

**چرا <code dir="ltr">with</code>؟** معادل بدون <code dir="ltr">with</code> این است که در جلسه‌ی ۱۵ دیدید: <code dir="ltr">f = open(...)</code>، کار با فایل در <code dir="ltr">try</code> و <code dir="ltr">f.close()</code> در <code dir="ltr">finally</code>. دستور <code dir="ltr">with</code> همین الگو را در دو خط خلاصه می‌کند.

**چرا <code dir="ltr">encoding="utf-8"</code>؟** فایل روی دیسک دنباله‌ای از **بایت**‌هاست؛ <code dir="ltr">encoding</code> تعیین می‌کند کاراکترها چطور به بایت تبدیل شوند. پیش‌فرض ویندوز معمولاً یک کدگذاری قدیمی است که حروف فارسی را پشتیبانی نمی‌کند؛ نتیجه یا <code dir="ltr">UnicodeEncodeError</code> است یا متن به‌هم‌ریخته («کاراکترهای عجیب»). همیشه <code dir="ltr">encoding="utf-8"</code> بنویسید تا متن فارسی و انگلیسی در همه‌ی سیستم‌ها یکسان خوانده شود.

### ❓ پیش‌بینی کن

خروجی سلول زیر چیست؟ حالت <code dir="ltr">"x"</code> در اجرای **دوم** چه می‌کند؟

</div>

````python
with open("s16_data/modes.txt", "w", encoding="utf-8") as f:
    f.write("A\n")
with open("s16_data/modes.txt", "w", encoding="utf-8") as f:
    f.write("B\n")
with open("s16_data/modes.txt", "a", encoding="utf-8") as f:
    f.write("C\n")

with open("s16_data/modes.txt", encoding="utf-8") as f:
    print(f.read())

with open("s16_data/modes.txt", "x", encoding="utf-8") as f:
    f.write("D\n")
````

<div dir="rtl">

🔍 دومین <code dir="ltr">"w"</code> محتوای قبلی (<code dir="ltr">A</code>) را پاک می‌کند و <code dir="ltr">"a"</code> به انتها اضافه می‌کند؛ پس فایل <code dir="ltr">B</code> و <code dir="ltr">C</code> دارد. حالت <code dir="ltr">"x"</code> چون فایل از قبل وجود دارد <code dir="ltr">FileExistsError</code> می‌دهد؛ <code dir="ltr">"x"</code> برای وقتی است که **نمی‌خواهید** اشتباهاً فایل مهمی را بازنویسی کنید (مثلاً نتیجه‌ی یک آزمایش طولانی).

### ⚠️ اشتباه رایج: <code dir="ltr">"w"</code> به جای <code dir="ltr">"a"</code>

باز کردن یک فایل با <code dir="ltr">"w"</code> فوراً آن را خالی می‌کند، حتی اگر هیچ چیزی ننویسید. اگر می‌خواهید هر بار اجرای برنامه یک خط به فایل گزارش (log) اضافه کند، حالت <code dir="ltr">"a"</code> لازم است.

</div>

## ۲. فایل‌های متنی (txt)

### 💡 مفهوم

| روش خواندن | نتیجه |
|---|---|
| <code dir="ltr">f.read()</code> | کل فایل به‌صورت **یک رشته** |
| <code dir="ltr">f.read(n)</code> | <code dir="ltr">n</code> کاراکتر بعدی |
| <code dir="ltr">f.readline()</code> | **یک خط** بعدی، همراه با <code dir="ltr">&#92;n</code> انتهایش (در انتهای فایل: <code dir="ltr">""</code>) |
| <code dir="ltr">f.readlines()</code> | لیست همه‌ی خط‌ها، هر کدام با <code dir="ltr">&#92;n</code> |
| <code dir="ltr">for line in f:</code> | خط به خط، **بدون** بارگذاری کل فایل در حافظه |

شیء فایل یک **مکان‌نما (cursor)** دارد: هر خواندن از جایی ادامه می‌دهد که خواندن قبلی تمام شده است.

</div>

````python
lines_to_write = ["Ali,17.5\n", "Sara,19\n", "Reza,12.25\n"]
with open("s16_data/scores.txt", "w", encoding="utf-8") as f:
    f.writelines(lines_to_write)

with open("s16_data/scores.txt", encoding="utf-8") as f:
    first = f.readline()
    rest = f.readlines()
print(repr(first))
print(rest)

with open("s16_data/scores.txt", encoding="utf-8") as f:
    for line in f:
        name, score = line.strip().split(",")
        print(f"{name:<5} {float(score):6.2f}")
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">writelines</code> لیستی از رشته‌ها را پشت سر هم می‌نویسد و مثل <code dir="ltr">write</code> **خط جدید اضافه نمی‌کند**؛ برای همین هر عنصر لیست با <code dir="ltr">&#92;n</code> تمام شده است.
- <code dir="ltr">readline</code> خط اول را برداشت؛ پس <code dir="ltr">readlines</code> فقط خط‌های **باقی‌مانده** را برمی‌گرداند (مکان‌نما).
- در حلقه‌ی <code dir="ltr">for</code>، هر <code dir="ltr">line</code> با <code dir="ltr">&#92;n</code> تمام می‌شود؛ <code dir="ltr">strip()</code> آن را حذف می‌کند و <code dir="ltr">split(",")</code> فیلدها را جدا می‌کند (الگوی جلسه‌ی ۱۱).

### 💡 فایل‌های بزرگ: خط به خط

شیء فایل یک **iterator** است (جلسه‌ی ۱۴): حلقه‌ی <code dir="ltr">for line in f</code> در هر لحظه فقط یک خط را در حافظه نگه می‌دارد. فایلی چند گیگابایتی را می‌توان این‌طور پردازش کرد، اما <code dir="ltr">f.read()</code> یا <code dir="ltr">f.readlines()</code> کل آن را یک‌جا در RAM بار می‌کنند. ترکیب این ویژگی با یک تابع generator، یک **جریان (stream)** از رکوردها می‌سازد:

</div>

````python
def read_scores(path):
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                name, score = line.split(",")
                yield name, float(score)

total = 0
count = 0
for name, score in read_scores("s16_data/scores.txt"):
    total += score
    count += 1
print("mean:", round(total / count, 3))
````

<div dir="rtl">

🔍 <code dir="ltr">read&#95;scores</code> چیزی را از قبل نمی‌خواند؛ هر بار که حلقه یک رکورد می‌خواهد، یک خط خوانده و تبدیل می‌شود. خط‌های خالی (مثلاً خط خالی انتهای فایل) با <code dir="ltr">if line:</code> کنار گذاشته می‌شوند. فایل هم پس از تمام شدن generator بسته می‌شود، چون <code dir="ltr">yield</code> داخل بلوک <code dir="ltr">with</code> است.

### 💡 نوشتن با <code dir="ltr">print(..., file=f)</code>

<code dir="ltr">print</code> یک آرگومان <code dir="ltr">file</code> دارد و برخلاف <code dir="ltr">write</code>، خودش فاصله بین آرگومان‌ها و <code dir="ltr">&#92;n</code> انتهای خط را اضافه می‌کند و عدد را هم به رشته تبدیل می‌کند:

</div>

````python
with open("s16_data/report.txt", "w", encoding="utf-8") as f:
    print("epoch", "loss", file=f)
    for epoch, loss in [(1, 0.92), (2, 0.61), (3, 0.47)]:
        print(epoch, loss, file=f)

with open("s16_data/report.txt", encoding="utf-8") as f:
    print(f.read())
````

<div dir="rtl">

### ❓ پیش‌بینی کن

خروجی سلول زیر را بنویسید. (<code dir="ltr">f.write</code> چه چیزی **برمی‌گرداند**؟)

</div>

````python
with open("s16_data/q.txt", "w", encoding="utf-8") as f:
    n = f.write("data\nscience")
print(n)

with open("s16_data/q.txt", encoding="utf-8") as f:
    print(f.readlines())
    print(repr(f.readline()))
````

<div dir="rtl">

🔍 <code dir="ltr">write</code> تعداد کاراکترهای نوشته‌شده را برمی‌گرداند: <code dir="ltr">4 + 1 + 7 = 12</code>. <code dir="ltr">readlines</code> دو خط می‌دهد و خط آخر <code dir="ltr">&#92;n</code> ندارد، چون ننوشته‌ایم. بعد از <code dir="ltr">readlines</code> مکان‌نما در انتهای فایل است و <code dir="ltr">readline</code> رشته‌ی خالی <code dir="ltr">''</code> برمی‌گرداند (نه خطا).

### ⚠️ اشتباه رایج: خواندن دوباره‌ی یک فایل در همان <code dir="ltr">with</code>

بعد از <code dir="ltr">f.read()</code>، فراخوانی دوباره‌ی <code dir="ltr">f.read()</code> رشته‌ی خالی می‌دهد. یا نتیجه‌ی اول را در یک متغیر نگه دارید، یا فایل را دوباره باز کنید.

## ۳. فایل‌های CSV

### 💡 مفهوم

در جلسه‌ی ۱۱ (سؤال ۴) دیدید که <code dir="ltr">split(",")</code> با مقداری مثل <code dir="ltr">"Tehran, Iran"</code> که خودش کاما دارد، خراب می‌شود. استاندارد CSV برای این حالت مقدار را در **نقل‌قول** می‌گذارد. ماژول <code dir="ltr">csv</code> از کتابخانه‌ی استاندارد همه‌ی این قاعده‌ها را می‌داند:

| ابزار | کار |
|---|---|
| <code dir="ltr">csv.writer(f)</code> و <code dir="ltr">.writerow(row)</code> / <code dir="ltr">.writerows(rows)</code> | نوشتن یک یا چند سطر (لیست) |
| <code dir="ltr">csv.reader(f)</code> | خواندن؛ هر سطر یک **لیست از رشته‌ها** |
| <code dir="ltr">csv.DictWriter(f, fieldnames)</code> و <code dir="ltr">.writeheader()</code> | نوشتن سطرها از دیکشنری |
| <code dir="ltr">csv.DictReader(f)</code> | خواندن؛ هر سطر یک **دیکشنری** با کلیدهای سطر عنوان |

فایل CSV را همیشه با <code dir="ltr">newline=""</code> باز کنید؛ وگرنه در ویندوز بین سطرها خط خالی اضافه می‌شود.

</div>

````python
import csv

students = [
    ["name", "city", "score"],
    ["Ali", "Bushehr", 17.5],
    ["Sara", "Tehran, Iran", 19],
    ["Reza", "Shiraz", 12.25],
]
with open("s16_data/students.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(students)

with open("s16_data/students.csv", encoding="utf-8") as f:
    print(f.read())

with open("s16_data/students.csv", newline="", encoding="utf-8") as f:
    for row in csv.reader(f):
        print(row)
````

<div dir="rtl">

### 🔍 کد را بخوان

- در فایل نوشته‌شده، <code dir="ltr">writer</code> خودش مقدار <code dir="ltr">"Tehran, Iran"</code> را در نقل‌قول گذاشته است.
- <code dir="ltr">csv.reader</code> هنگام خواندن نقل‌قول را می‌فهمد و این مقدار را **یک فیلد** برمی‌گرداند؛ هر سطر دقیقاً سه فیلد دارد.
- همه‌ی مقدارهای خوانده‌شده **رشته**‌اند (<code dir="ltr">'17.5'</code>، نه <code dir="ltr">17.5</code>). CSV نوع داده را نگه نمی‌دارد؛ تبدیل نوع با شماست.

### 💡 <code dir="ltr">DictReader</code> و <code dir="ltr">DictWriter</code>

با <code dir="ltr">DictReader</code> به جای <code dir="ltr">row[2]</code> می‌نویسید <code dir="ltr">row["score"]</code>؛ کد خواناتر می‌شود و اگر ترتیب ستون‌های فایل عوض شود، خراب نمی‌شود:

</div>

````python
with open("s16_data/students.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(rows[1])

passed = [r for r in rows if float(r["score"]) >= 14]

with open("s16_data/passed.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"], extrasaction="ignore")
    writer.writeheader()
    writer.writerows(passed)

with open("s16_data/passed.csv", encoding="utf-8") as f:
    print(f.read())
````

<div dir="rtl">

🔍 <code dir="ltr">DictReader</code> سطر اول را عنوان ستون‌ها حساب می‌کند. <code dir="ltr">DictWriter</code> فقط ستون‌های <code dir="ltr">fieldnames</code> را می‌نویسد؛ <code dir="ltr">extrasaction="ignore"</code> می‌گوید کلید اضافه (<code dir="ltr">city</code>) را نادیده بگیر (بدون آن، <code dir="ltr">ValueError</code> می‌گیرید). خط <code dir="ltr">passed = [...]</code> یک list comprehension است (جلسه‌ی ۱۳).

### ❓ پیش‌بینی کن

فایل زیر یک مقدار گمشده و یک فاصله‌ی اضافه دارد. خروجی چیست؟

</div>

## ۴. فایل‌های JSON

### 💡 مفهوم

**JSON** قالب متنی استاندارد برای ذخیره‌ی داده‌ی **تودرتو** است: دیکشنری‌هایی که لیست دارند، لیست‌هایی از دیکشنری‌ها و ... . فایل‌های تنظیمات، پاسخ APIهای وب، واژگان یک مدل زبانی و ابرپارامترهای یک آزمایش معمولاً JSON هستند. JSON تقریباً شبیه دیکشنری پایتون نوشته می‌شود.

| تابع | کار |
|---|---|
| <code dir="ltr">json.dump(obj, f)</code> | نوشتن شیء پایتون در فایل |
| <code dir="ltr">json.load(f)</code> | خواندن فایل و ساختن شیء پایتون |
| <code dir="ltr">json.dumps(obj)</code> | تبدیل به **رشته**‌ی JSON (<code dir="ltr">s</code> یعنی string) |
| <code dir="ltr">json.loads(text)</code> | ساختن شیء از رشته‌ی JSON |

| پایتون | JSON |
|---|---|
| <code dir="ltr">dict</code> | object <code dir="ltr">{...}</code> (کلیدها **فقط رشته**) |
| <code dir="ltr">list</code> و <code dir="ltr">tuple</code> | array <code dir="ltr">[...]</code> |
| <code dir="ltr">str</code> | string (فقط با <code dir="ltr">"</code>) |
| <code dir="ltr">int</code> و <code dir="ltr">float</code> | number |
| <code dir="ltr">True</code> / <code dir="ltr">False</code> / <code dir="ltr">None</code> | <code dir="ltr">true</code> / <code dir="ltr">false</code> / <code dir="ltr">null</code> |

</div>

````python
import json

experiment = {
    "model": "logistic_regression",
    "learning_rate": 0.01,
    "epochs": 50,
    "layers": [64, 32],
    "use_bias": True,
    "notes": None,
}
with open("s16_data/config.json", "w", encoding="utf-8") as f:
    json.dump(experiment, f, indent=2)

with open("s16_data/config.json", encoding="utf-8") as f:
    print(f.read())

with open("s16_data/config.json", encoding="utf-8") as f:
    loaded = json.load(f)
print(loaded == experiment, loaded["layers"][0])
````

- در فایل، <code dir="ltr">True</code> به <code dir="ltr">true</code> و <code dir="ltr">None</code> به <code dir="ltr">null</code> تبدیل شده‌اند.
- <code dir="ltr">json.load</code> دقیقاً همان ساختار را برمی‌گرداند؛ به همین سادگی می‌توان تنظیمات یک آزمایش را ذخیره و در اجرای بعدی بازسازی کرد.

به‌طور پیش‌فرض <code dir="ltr">json</code> همه‌ی کاراکترهای غیر ASCII را به شکل کد <code dir="ltr">&#92;uXXXX</code> می‌نویسد. فایل درست است، اما برای انسان خوانا نیست. برای متن فارسی (یا هر زبان دیگر) <code dir="ltr">ensure&#95;ascii=False</code> بگذارید و فایل را با <code dir="ltr">encoding="utf-8"</code> باز کنید:

## ۵. ماژول <code dir="ltr">pathlib</code>

### 💡 مفهوم

تا اینجا مسیرها را به شکل رشته نوشتیم. ماژول <code dir="ltr">pathlib</code> مسیر را یک **شیء** می‌کند که متدهای مفید دارد و روی ویندوز، لینوکس و مک یکسان کار می‌کند. مهم‌ترین ابزارها:

| عبارت | نتیجه |

```python
def batches(path, size):
    batch = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            batch.append(line.rstrip("\n"))
            if len(batch) == size:
                yield batch
                batch = []
    if batch:
        yield batch
```

</div>

### سؤال ۶ — بارگذاری مقاوم تنظیمات

```python
DEFAULTS = {"lr": 0.01, "epochs": 10, "batch_size": 32}

def load_settings(path):
    settings = DEFAULTS.copy()

        with open(path, encoding="utf-8") as f:
            user = json.load(f)
    except FileNotFoundError:

    else:
        settings.update(user)
    return settings
```

الف) خروجی تابع را در سه حالت بنویسید: فایل وجود ندارد؛ فایل محتوای <code dir="ltr">{"lr": 0.1, "epochs": 50}</code> دارد؛ فایل محتوای <code dir="ltr">{"lr": 0.1, "epochs": }</code> دارد.

ج) چرا <code dir="ltr">DEFAULTS.copy()</code> لازم است؟ اگر بنویسیم <code dir="ltr">settings = DEFAULTS</code> چه اشکالی در فراخوانی‌های بعدی پیش می‌آید؟

<summary><b>پاسخ</b></summary>

### سؤال ۷ — پردازش یک پوشه با <code dir="ltr">pathlib</code>

```python
from pathlib import Path
import csv

folder = Path("sales")
totals = {}
for path in sorted(folder.glob("*.csv")):
    with open(path, newline="", encoding="utf-8") as f:
        totals[path.stem] = sum(float(r["amount"]) for r in csv.DictReader(f))

with open(folder / "summary.csv", "w", newline="", encoding="utf-8") as f:
```

## خلاصه‌ی جلسه

- داده‌ی ماندگار در فایل است؛ مسیر نسبی را ترجیح دهید و در مسیرهای ویندوز از <code dir="ltr">/</code> یا رشته‌ی خام استفاده کنید.
- فایل را همیشه با <code dir="ltr">with</code> و <code dir="ltr">encoding="utf-8"</code> باز کنید؛ حالت را با دقت انتخاب کنید (<code dir="ltr">"w"</code> پاک می‌کند، <code dir="ltr">"a"</code> اضافه می‌کند، <code dir="ltr">"x"</code> بازنویسی نمی‌کند).
- <code dir="ltr">write</code> و <code dir="ltr">writelines</code> خط جدید اضافه نمی‌کنند؛ هنگام خواندن، <code dir="ltr">&#92;n</code> انتهای خط را با <code dir="ltr">strip</code> حذف کنید؛ فایل بزرگ را خط به خط یا با generator بخوانید.
- برای CSV از ماژول <code dir="ltr">csv</code> استفاده کنید (<code dir="ltr">newline=""</code>)؛ مقدارها رشته‌اند و تبدیل نوع و پاک‌سازی با شماست.
- JSON برای داده‌ی تودرتو و تنظیمات است؛ تاپل به لیست و کلید عددی به رشته تبدیل می‌شود و <code dir="ltr">set</code> ذخیره نمی‌شود.
- <code dir="ltr">pathlib</code> مسیرها را شیء می‌کند: <code dir="ltr">/</code>، <code dir="ltr">glob</code>، <code dir="ltr">mkdir(exist&#95;ok=True)</code>، <code dir="ltr">read&#95;text</code>.

**جلسه‌ی بعد:** ماژول‌ها و کتابخانه‌ی استاندارد؛ ساختن ماژول خودتان و آشنایی با <code dir="ltr">datetime</code> و <code dir="ltr">collections</code>.

</div>
