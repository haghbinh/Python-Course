## ۱. صورت مسئله

اداره‌ی محیط زیست یک شهر، داده‌ی یک سال کامل (۲۰۲۵) از **چهار ایستگاه سنجش آلودگی هوا** را در اختیار شما گذاشته است: غلظت ساعتی ذرات معلق ریز (<code dir="ltr">PM2.5</code>) و دی‌اکسید نیتروژن (<code dir="ltr">NO2</code>)، همراه با داده‌ی روزانه‌ی هواشناسی و فهرست روزهایی که مدارس و ادارات به دلیل آلودگی تعطیل شده‌اند.

داده از سامانه‌ی ثبت خودکار ایستگاه‌ها آمده و **بی‌نقص نیست**: حسگرهایی که خراب شده‌اند، ساعت‌هایی که ثبت نشده‌اند، کدهای ناهمسان، واحدهای نادرست و سطرهای تکراری در آن دیده می‌شود.

مدیر اداره از شما خواسته است:

- داده را **پاک‌سازی** کنید و دقیقاً بگویید با هر ایراد چه کرده‌اید،
- **وضعیت کیفیت هوای شهر** را در این سال توصیف کنید: چند روز هوا ناسالم بوده، کدام ایستگاه آلوده‌تر است و بدترین روزها کدام‌اند،
- **الگوهای زمانی** آلودگی را پیدا کنید: در کدام فصل‌ها، کدام ساعت‌های روز و کدام روزهای هفته هوا آلوده‌تر است،
- **رابطه‌ی آلودگی با هوا** (باد، باران، دما) را بررسی کنید،
- و به یک **پرسش سیاستی** پاسخ دهید: آیا تعطیلی مدارس و ادارات واقعاً هوا را بهتر کرده است؟ (الزام ۷)

## ۳. الزامات برنامه

جزئیات طراحی با خودتان است، اما **سه تابع زیر باید دقیقاً با همین امضا وجود داشته باشند**، چون در جلسه‌ی دفاع با داده‌ی تازه (مثلاً سال یا ایستگاه دیگر) آزمایش می‌شوند. پس کد شما نباید به سال ۲۰۲۵ یا نام این چهار ایستگاه وابسته باشد.

1. **پاک‌سازی:** همه‌ی ایرادهای داده را پیدا کنید: کدهای ناهمسان ایستگاه، قالب‌های ناهمسان تاریخ و عدد، سطرهای تکراری (یکسان یا متناقض)، مقدارهای گمشده و کدهای قراردادی، مقدارهای ناممکن، **حسگری که گیر کرده** (یک مقدار ثابت برای چند ساعت پشت سر هم)، و **بازه‌ای که واحد آن اشتباه است**. ساعت‌هایی که اصلاً ثبت نشده‌اند باید در <code dir="ltr">hourly</code> به‌صورت سطر با مقدار <code dir="ltr">NaN</code> دیده شوند. جای خالی‌ها را با حدس پر نکنید. در گزارش، برای هر نوع ایراد بگویید چند مقدار را درگیر کرده و چه تصمیمی گرفتید و چرا. داده‌ی هواشناسی و فهرست تعطیلی‌ها را هم پاک‌سازی کنید.
2. **ساختار کد:** تابع‌های قابل استفاده‌ی دوباره (دست‌کم سه تابع بالا) را در یک **ماژول** به نام <code dir="ltr">airtools.py</code> بنویسید و در نوت‌بوک <code dir="ltr">import</code> کنید (جلسه‌ی ۱۷). بررسی‌های کیفیت داده را در قالب دست‌کم یک **کلاس** پیاده کنید (مثلاً کلاسی که هر بررسی یک متد آن است و گزارش ایرادها را جمع می‌کند؛ جلسه‌های ۱۸ و ۱۹).
3. **وضعیت کیفیت هوا در سال:**
   - میانگین سالانه‌ی هر آلاینده در هر ایستگاه، و تعداد روزهای «نامعلوم» هر ایستگاه،
   - تعداد روزهای هر دسته‌ی کیفیت هوا برای شهر،
   - ده روز آلوده‌ی سال (بر اساس <code dir="ltr">PM2.5</code> شهر).
4. **الگوهای زمانی:**
   - روند ماهانه و تفاوت فصل‌ها،
   - الگوی ساعتی شبانه‌روز برای هر آلاینده، جداگانه برای روزهای کاری و جمعه،
   - یک نقشه‌ی گرمایی (heatmap) از میانگین <code dir="ltr">PM2.5</code> بر حسب ماه و ساعت.
5. **آلودگی و هوا:** داده‌ی روزانه‌ی شهر را با داده‌ی هواشناسی <code dir="ltr">merge</code> کنید. همبستگی آلاینده‌ها با باد، باران، دما و رطوبت را حساب و تفسیر کنید. آیا روز **بعد** از یک باران شدید هوا پاک‌تر است؟
6. **ایستگاه‌ها:** ایستگاه‌ها را با هم مقایسه کنید و همبستگی سری‌های روزانه‌ی آن‌ها را حساب کنید. این همبستگی درباره‌ی منشأ آلودگی (عامل‌های سراسری شهر در برابر منبع‌های محلی) چه می‌گوید؟
7. **پرسش سیاستی، آیا تعطیلی‌ها مؤثر بوده‌اند؟**
   - **الف)** میانگین آلاینده‌ها را در روزهای تعطیلی با بقیه‌ی روزها مقایسه کنید. نتیجه چه می‌گوید و چرا این مقایسه‌ی ساده **گمراه‌کننده** است؟
   - **ب)** یک مقایسه‌ی **منصفانه‌تر** طراحی کنید (مثلاً مقایسه با روزهای غیرتعطیلی که روز قبلشان به همان اندازه آلوده بوده، یا مقایسه‌ی تغییر نسبت به روز قبل) و نتیجه را برای <code dir="ltr">PM2.5</code> و <code dir="ltr">NO2</code> جداگانه گزارش کنید. چرا اثر تعطیلی روی این دو آلاینده یکسان نیست؟
   - **ج)** نتیجه‌گیری شما برای مدیر چیست و این داده چه چیزی را **نمی‌تواند** ثابت کند؟
8. **نمودارها و خروجی‌ها:** دست‌کم این شش نمودار را بسازید و به‌صورت فایل PNG در پوشه‌ی <code dir="ltr">figures</code> ذخیره کنید (برچسب‌ها انگلیسی): سری زمانی روزانه‌ی <code dir="ltr">PM2.5</code> شهر با خط‌های مرز دسته‌ها، نمودار جعبه‌ای ماهانه، الگوی ساعتی (کاری و جمعه)، نقشه‌ی گرمایی ماه و ساعت، نمودار پراکنش <code dir="ltr">PM2.5</code> در برابر سرعت باد، و نمودار مقایسه‌ی تعطیلی‌ها. داده‌ی ساعتی پاک‌شده و خلاصه‌ی روزانه را هم در دو فایل CSV ذخیره کنید.
