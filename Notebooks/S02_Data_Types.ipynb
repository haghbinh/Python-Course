## ۱. متغیرها و نام‌گذاری

### 💡 مفهوم

**متغیر (variable)** یک **نام** است که به یک **مقدار** در حافظه اشاره می‌کند. با دستور **انتساب** (<code dir="ltr">=</code>) یک نام به یک مقدار وصل می‌شود:

<div dir="ltr">

```python
x = 5
```

</div>

این خط را این‌طور بخوانید: «مقدار <code dir="ltr">5</code> را بساز و نام <code dir="ltr">x</code> را به آن وصل کن». علامت <code dir="ltr">=</code> در پایتون یعنی **انتساب**، نه «مساوی است با».

در پایتون لازم نیست نوع متغیر را از قبل اعلام کنید؛ نوع از روی **مقدار** مشخص می‌شود.

</div>

````python
x = 5          # an integer (int)
pi = 3.14      # a decimal number (float)
name = "Ali"   # text (str)
flag = True    # a boolean value (bool)

print(x, pi, name, flag)
````

<div dir="rtl">

### 🔍 کد را بخوان

- هر خط یک **انتساب** است: سمت راست <code dir="ltr">=</code> یک مقدار است و سمت چپ نامی که به آن وصل می‌شود.
- چهار مقدار از چهار **نوع** مختلف ساخته شده‌اند: عدد صحیح، عدد اعشاری، متن و مقدار درست/نادرست.
- <code dir="ltr">print</code> هر چهار مقدار را با یک فاصله بینشان چاپ می‌کند.

### ترتیب اجرای انتساب

پایتون همیشه **اول سمت راست** <code dir="ltr">=</code> را حساب می‌کند و **بعد** نام سمت چپ را به نتیجه وصل می‌کند. برای همین خط زیر کاملاً درست است، با اینکه از نظر ریاضی بی‌معنی است:

</div>

````python
count = 10
count = count + 1
print(count)
````

<div dir="rtl">

🔍 خط دوم: اول <code dir="ltr">count + 1</code> حساب می‌شود (یعنی <code dir="ltr">11</code>)، بعد نام <code dir="ltr">count</code> به این مقدار جدید وصل می‌شود.

### ❓ پیش‌بینی کن

قبل از اجرای سلول زیر، روی کاغذ بنویسید چه چیزی چاپ می‌شود.

</div>

````python
a = 3
b = a
a = 7
print(b)
````

<div dir="rtl">

### 🔍 توضیح: نام‌ها و مقدارها

خط <code dir="ltr">b = a</code> یعنی «<code dir="ltr">b</code> را به **همان مقداری** وصل کن که <code dir="ltr">a</code> الان به آن اشاره می‌کند». بعد از آن، <code dir="ltr">a = 7</code> فقط نام <code dir="ltr">a</code> را به مقدار جدید وصل می‌کند و **کاری به <code dir="ltr">b</code> ندارد**. پس <code dir="ltr">b</code> همچنان <code dir="ltr">3</code> است.

</div>

<p align="center"><img src="images/s02_names_objects.png" width="720" alt="s02_names_objects.png"></p>

<p align="center" dir="rtl"><i>سمت چپ: بعد از <code dir="ltr">a = 3</code> و <code dir="ltr">b = a</code> هر دو نام به یک مقدار اشاره می‌کنند. سمت راست: بعد از <code dir="ltr">a = 7</code> فقط پیکان <code dir="ltr">a</code> جابه‌جا می‌شود.</i></p>

<div dir="rtl">

این تصویر «نام‌ها و مقدارها» را به خاطر بسپارید؛ در جلسه‌ی لیست‌ها دوباره به آن برمی‌گردیم و آنجا نتیجه‌ی غافلگیرکننده‌تری می‌بینیم.

### انتساب چندگانه و جابه‌جایی

در پایتون می‌توانید در یک خط به چند نام مقدار بدهید. یک کاربرد جالب، جابه‌جا کردن مقدار دو متغیر است:

</div>

````python
x, y = 5, 10
print(x, y)

x, y = y, x    # swap
print(x, y)
````

<div dir="rtl">

🔍 در خط <code dir="ltr">x, y = y, x</code> باز هم **اول سمت راست** حساب می‌شود: مقدارهای فعلی <code dir="ltr">y</code> و <code dir="ltr">x</code> (یعنی <code dir="ltr">10</code> و <code dir="ltr">5</code>) برداشته می‌شوند و بعد به ترتیب به <code dir="ltr">x</code> و <code dir="ltr">y</code> داده می‌شوند.

### قواعد نام‌گذاری

| قاعده | مثال درست | مثال نادرست |
|---|---|---|
| فقط حروف انگلیسی، رقم و <code dir="ltr">&#95;</code> | <code dir="ltr">total&#95;sales</code> | <code dir="ltr">total-sales</code> |
| نباید با رقم شروع شود | <code dir="ltr">value2</code> | <code dir="ltr">2value</code> |
| نباید یکی از **کلمات کلیدی** پایتون باشد | <code dir="ltr">class&#95;name</code> | <code dir="ltr">class</code> |
| حروف بزرگ و کوچک فرق دارند | <code dir="ltr">Name</code> و <code dir="ltr">name</code> دو متغیر متفاوت‌اند | |

**قرارداد رایج در پایتون:** نام متغیرها با حروف کوچک و کلمات با <code dir="ltr">&#95;</code> جدا شوند (به این سبک **snake_case** می‌گویند)؛ مثل <code dir="ltr">student&#95;count</code>. نام باید **معنی‌دار** باشد: <code dir="ltr">average&#95;score</code> خیلی بهتر از <code dir="ltr">a2</code> است.

**کلمات کلیدی** کلمه‌هایی هستند که پایتون برای خودش رزرو کرده است. فهرست کامل آن‌ها را این‌طور ببینید:

</div>

````python
import keyword
print(keyword.kwlist)
````

### ⚠️ اشتباه رایج

- **استفاده از متغیری که هنوز ساخته نشده:** اگر قبل از انتساب از یک نام استفاده کنید، خطای <code dir="ltr">NameError</code> می‌گیرید (همان خطایی که در جلسه‌ی ۱ دیدید).
- **گذاشتن اسم توابع آماده روی متغیر:** نام‌هایی مثل <code dir="ltr">print</code>، <code dir="ltr">type</code>، <code dir="ltr">sum</code>، <code dir="ltr">list</code> یا <code dir="ltr">str</code> کلمه‌ی کلیدی نیستند و پایتون اجازه می‌دهد از آن‌ها استفاده کنید؛ اما بعد از <code dir="ltr">sum = 10</code> دیگر تابع <code dir="ltr">sum</code> کار نمی‌کند! این اسم‌ها را برای متغیرها به کار نبرید.

</div>

## ۲. نوع داده و تابع <code dir="ltr">type</code>

### 💡 مفهوم

هر مقدار در پایتون یک **نوع (type)** دارد. نوع تعیین می‌کند آن مقدار **چه اطلاعاتی** نگه می‌دارد و **چه کارهایی** روی آن مجاز است. مثلاً دو عدد را می‌شود تقسیم کرد، ولی دو متن را نه.

تابع <code dir="ltr">type</code> نوع هر مقدار را نشان می‌دهد:

</div>

````python
print(type(5))
print(type(3.14))
print(type("Ali"))
print(type(True))
````

<div dir="rtl">

### چهار نوع پایه‌ی این جلسه

| نوع | نام کامل | مثال | کاربرد در علوم داده |
|---|---|---|---|
| <code dir="ltr">int</code> | integer (عدد صحیح) | <code dir="ltr">42</code>، <code dir="ltr">-7</code>، <code dir="ltr">0</code> | تعداد، شمارنده، سال |
| <code dir="ltr">float</code> | floating point (عدد اعشاری) | <code dir="ltr">3.14</code>، <code dir="ltr">-0.5</code>، <code dir="ltr">2.0</code> | اندازه‌گیری، میانگین، احتمال |
| <code dir="ltr">str</code> | string (رشته، متن) | <code dir="ltr">"Ali"</code>، <code dir="ltr">'Tehran'</code> | نام، دسته، متن |
| <code dir="ltr">bool</code> | boolean (منطقی) | <code dir="ltr">True</code>، <code dir="ltr">False</code> | نتیجه‌ی مقایسه، فیلتر کردن داده |

### نوع پویا (dynamic typing)

نوع به **مقدار** تعلق دارد، نه به **نام**. پس یک نام می‌تواند در طول برنامه به مقدارهایی از نوع‌های مختلف اشاره کند:

</div>

````python
x = 42
print(type(x))

x = "Now I am a string!"
print(type(x))
````

<div dir="rtl">

🔍 خط اول <code dir="ltr">x</code> را به یک <code dir="ltr">int</code> وصل می‌کند. بعد همان نام به یک <code dir="ltr">str</code> وصل می‌شود. پایتون اعتراضی نمی‌کند؛ اما در برنامه‌های واقعی این کار باعث سردرگمی می‌شود. **یک نام را برای یک نوع داده نگه دارید.**

</div>

<div dir="rtl">

## ۳. اعداد: <code dir="ltr">int</code> و <code dir="ltr">float</code>

### 💡 مفهوم

- **<code dir="ltr">int</code>** عدد صحیح است و در پایتون **محدودیت اندازه ندارد**؛ هر قدر حافظه اجازه دهد می‌تواند بزرگ باشد.
- **<code dir="ltr">float</code>** عدد اعشاری است. هر عددی که **نقطه‌ی اعشار** داشته باشد <code dir="ltr">float</code> است، حتی <code dir="ltr">2.0</code>.
- اعداد خیلی بزرگ یا خیلی کوچک را می‌توان با **نماد علمی** نوشت: <code dir="ltr">1.5e3</code> یعنی 1.5 × 10³ و <code dir="ltr">2e-4</code> یعنی 0.0002 (هر دو <code dir="ltr">float</code> هستند).
- پایتون نوع سومی هم برای اعداد مختلط دارد (<code dir="ltr">complex</code>، مثل <code dir="ltr">1 + 2j</code>) که در این دوره کمتر به آن نیاز داریم.

</div>

````python
population = 88_550_570      # underscores make big numbers readable
print(population, type(population))

big = 2 ** 100               # int has no size limit
print(big)

distance = 1.5e3
tiny = 2e-4
print(distance, tiny, type(distance))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">&#95;</code> داخل عدد فقط برای خوانایی است و پایتون آن را نادیده می‌گیرد.
- <code dir="ltr">&#42;&#42;</code> یعنی **توان**؛ <code dir="ltr">2 &#42;&#42; 100</code> یک عدد ۳۱ رقمی است و پایتون بدون مشکل آن را نگه می‌دارد.
- <code dir="ltr">1.5e3</code> با اینکه مقدارش <code dir="ltr">1500</code> است، از نوع <code dir="ltr">float</code> است.

### ❓ پیش‌بینی کن

نوع حاصل هر عبارت را پیش‌بینی کنید. (راهنمایی: به خروجی تقسیم <code dir="ltr">/</code> دقت کنید.)

</div>

````python
print(type(3 + 4))
print(type(3 + 4.0))
print(type(8 / 2))
print(8 / 2)
````

<div dir="rtl">

🔍 دو قاعده‌ی مهم:

1. اگر در یک عبارت <code dir="ltr">int</code> و <code dir="ltr">float</code> با هم باشند، نتیجه **<code dir="ltr">float</code>** است.
2. تقسیم با <code dir="ltr">/</code> **همیشه** <code dir="ltr">float</code> می‌دهد، حتی وقتی تقسیم بدون باقی‌مانده باشد (<code dir="ltr">8 / 2</code> برابر <code dir="ltr">4.0</code> است، نه <code dir="ltr">4</code>).

### ⚠️ اشتباه رایج: دقت اعداد اعشاری

کامپیوتر اعداد اعشاری را در **مبنای ۲** ذخیره می‌کند و بعضی اعداد ساده مثل <code dir="ltr">0.1</code> در مبنای ۲ دقیقاً قابل نمایش نیستند. نتیجه را ببینید:

</div>

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

<div dir="rtl">

🔍 حاصل <code dir="ltr">0.1 + 0.2</code> عددی **بسیار نزدیک** به <code dir="ltr">0.3</code> است، ولی دقیقاً خودش نیست؛ برای همین مقایسه‌ی <code dir="ltr">==</code> نتیجه‌ی <code dir="ltr">False</code> می‌دهد. این باگ پایتون نیست؛ در همه‌ی زبان‌های برنامه‌نویسی همین‌طور است.

**درس مهم برای علوم داده:** هیچ‌وقت دو <code dir="ltr">float</code> را مستقیم با <code dir="ltr">==</code> مقایسه نکنید. به جای آن بررسی کنید که **خیلی به هم نزدیک** باشند:

</div>

````python
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 2))
````

## ۴. رشته‌ها: <code dir="ltr">str</code>

### 💡 مفهوم

**رشته (string)** یک دنباله از کاراکترهاست و برای نگهداری متن استفاده می‌شود. رشته را می‌توان بین <code dir="ltr">'...'</code> یا <code dir="ltr">"..."</code> نوشت؛ هر دو یکسان‌اند. برای متن چندخطی از سه علامت نقل‌قول (<code dir="ltr">"""..."""</code>) استفاده می‌شود.

در این جلسه فقط با اصول رشته آشنا می‌شویم؛ جلسه‌ی ۱۱ کاملاً به رشته‌ها اختصاص دارد.

</div>

````python
text = "Hello"
print(text)
print(len(text))      # number of characters
print(text[0])        # first character (counting starts from 0)
print(text.upper())   # a NEW string in upper case
print(text)           # the original is unchanged

message = """This is
a multi-line
string."""
print(message)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">len(text)</code> تعداد کاراکترها را می‌دهد: <code dir="ltr">5</code>.
- <code dir="ltr">text[0]</code> **اولین** کاراکتر است؛ در پایتون شماره‌گذاری از **صفر** شروع می‌شود.
- <code dir="ltr">text.upper()</code> یک رشته‌ی **جدید** با حروف بزرگ می‌سازد. خود <code dir="ltr">text</code> تغییر نمی‌کند؛ این را در خط بعد می‌بینید.

### ❓ پیش‌بینی کن

خروجی هر خط چیست؟ (به تفاوت عدد و رشته‌ای که شبیه عدد است دقت کنید.)

</div>

````python
print("5" + "3")
print(5 + 3)
print("ab" * 3)
print(len("Data Science"))
````

<div dir="rtl">

🔍 عملگر <code dir="ltr">+</code> روی رشته‌ها یعنی **چسباندن** (concatenation)، نه جمع. پس <code dir="ltr">"5" + "3"</code> می‌شود <code dir="ltr">"53"</code>. عملگر <code dir="ltr">&#42;</code> بین رشته و عدد صحیح، رشته را **تکرار** می‌کند. فاصله هم یک کاراکتر است؛ پس طول <code dir="ltr">"Data Science"</code> برابر <code dir="ltr">12</code> است.

### ⚠️ اشتباه رایج: ترکیب رشته و عدد

پایتون رشته و عدد را خودبه‌خود به هم تبدیل نمی‌کند:

</div>

````python
age = 25
print("Age: " + age)
````

<div dir="rtl">

🔍 خط آخر پیام خطا <code dir="ltr">TypeError: can only concatenate str (not "int") to str</code> است؛ یعنی «فقط می‌توانم رشته را به رشته بچسبانم، نه عدد صحیح را». راه حل‌ها: <code dir="ltr">print("Age: " + str(age))</code> یا <code dir="ltr">print("Age:", age)</code>. (روش بهتر، یعنی **f-string**، را در جلسه‌ی ۴ می‌بینید.)

</div>

## ۵. مقادیر بولی: <code dir="ltr">bool</code> و مقدار <code dir="ltr">None</code>

### 💡 مفهوم

نوع **<code dir="ltr">bool</code>** فقط دو مقدار دارد: <code dir="ltr">True</code> (درست) و <code dir="ltr">False</code> (نادرست). حرف اول هر دو **بزرگ** است. نتیجه‌ی هر **مقایسه** یک مقدار بولی است و در جلسه‌ی شرط‌ها (جلسه‌ی ۵) از آن برای تصمیم‌گیری استفاده می‌کنیم.

</div>

````python
a = 10
b = 5
print(a > b)
print(a == b)
print(type(a > b))
````

<div dir="rtl">

🔍 <code dir="ltr">a &gt; b</code> یک سؤال است: «آیا <code dir="ltr">a</code> از <code dir="ltr">b</code> بزرگ‌تر است؟» و جوابش <code dir="ltr">True</code> است. دقت کنید <code dir="ltr">==</code> (دو علامت) یعنی **مقایسه**، ولی <code dir="ltr">=</code> (یک علامت) یعنی **انتساب**.

### 💡 بولی‌ها در واقع عدد هستند

در پایتون <code dir="ltr">True</code> مثل عدد <code dir="ltr">1</code> و <code dir="ltr">False</code> مثل عدد <code dir="ltr">0</code> رفتار می‌کند. این نکته در علوم داده بسیار کاربردی است: **جمع چند مقدار بولی، تعداد <code dir="ltr">True</code>ها را می‌دهد.**

### ❓ پیش‌بینی کن

</div>

````python
passed_1 = True
passed_2 = False
passed_3 = True
print(passed_1 + passed_2 + passed_3)
````

<div dir="rtl">

🔍 حاصل <code dir="ltr">1 + 0 + 1</code> یعنی <code dir="ltr">2</code>: تعداد دانشجویانی که قبول شده‌اند. بعدها در Pandas دقیقاً با همین ایده می‌شماریم چند ردیف از داده یک شرط را دارند.

## ۶. تبدیل نوع

### 💡 مفهوم

گاهی داده با نوع نامناسب به دست ما می‌رسد. مثلاً عددی که از یک فایل متنی یا از ورودی کاربر خوانده می‌شود، **همیشه رشته** است (<code dir="ltr">"25"</code> و نه <code dir="ltr">25</code>). برای تبدیل، از **نام نوع** مثل یک تابع استفاده می‌کنیم:

| تابع | کار | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">int()</code> | تبدیل به عدد صحیح | <code dir="ltr">int("25")</code> | <code dir="ltr">25</code> |
| <code dir="ltr">float()</code> | تبدیل به عدد اعشاری | <code dir="ltr">float("3.5")</code> | <code dir="ltr">3.5</code> |
| <code dir="ltr">str()</code> | تبدیل به رشته | <code dir="ltr">str(10)</code> | <code dir="ltr">"10"</code> |
| <code dir="ltr">bool()</code> | تبدیل به بولی | <code dir="ltr">bool(0)</code> | <code dir="ltr">False</code> |

</div>

````python
score_text = "17.5"            # e.g. read from a file
score = float(score_text)
print(score + 1, type(score))

print(int(3.99))               # int() cuts the decimal part, it does NOT round
print(round(3.99))             # use round() for rounding
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">float("17.5")</code> رشته را به عدد اعشاری <code dir="ltr">17.5</code> تبدیل می‌کند و حالا می‌شود با آن حساب کرد.
- <code dir="ltr">int(3.99)</code> برابر <code dir="ltr">3</code> است: <code dir="ltr">int</code> قسمت اعشاری را **دور می‌اندازد** و گرد نمی‌کند. برای گرد کردن از <code dir="ltr">round</code> استفاده کنید.

### تبدیل به <code dir="ltr">bool</code>: چه چیزی «خالی» است؟

قاعده ساده است: **صفر و مقدارهای خالی** به <code dir="ltr">False</code> تبدیل می‌شوند و **هر چیز دیگری** به <code dir="ltr">True</code>.

### ❓ پیش‌بینی کن

قبل از اجرا، جلوی هر خط بنویسید <code dir="ltr">True</code> یا <code dir="ltr">False</code>. (به خط آخر خیلی دقت کنید!)

</div>

````python
print(bool(0))
print(bool(42))
print(bool(0.0))
print(bool(""))
print(bool(" "))
print(bool("False"))
````

<div dir="rtl">

🔍 <code dir="ltr">" "</code> یک فاصله دارد، پس خالی نیست و <code dir="ltr">True</code> است. <code dir="ltr">"False"</code> هم یک رشته‌ی **پنج‌حرفی** است، پس خالی نیست و به <code dir="ltr">True</code> تبدیل می‌شود. <code dir="ltr">bool</code> به **محتوای** رشته نگاه نمی‌کند، فقط به خالی بودن یا نبودن آن.

### ⚠️ اشتباه رایج: تبدیل غیرممکن

هر رشته‌ای را نمی‌توان به عدد تبدیل کرد. سلول زیر را اجرا کنید و پیام خطا را بخوانید:

</div>

````python
int("3.5")
````

<div dir="rtl">

🔍 خطای <code dir="ltr">ValueError: invalid literal for int() with base 10: '3.5'</code> یعنی «رشته‌ی <code dir="ltr">'3.5'</code> شکل یک عدد صحیح را ندارد». <code dir="ltr">int</code> فقط رشته‌هایی مثل <code dir="ltr">"35"</code> را می‌پذیرد. راه حل: اول به <code dir="ltr">float</code> و بعد به <code dir="ltr">int</code> تبدیل کنید: <code dir="ltr">int(float("3.5"))</code>.

دو نوع خطایی که در این جلسه دیدید را با هم مقایسه کنید:

| خطا | معنی | مثال |
|---|---|---|
| <code dir="ltr">TypeError</code> | عملیات برای این **نوع** مجاز نیست | <code dir="ltr">"Age: " + 25</code> |
| <code dir="ltr">ValueError</code> | نوع درست است، اما **مقدار** مناسب نیست | <code dir="ltr">int("3.5")</code> |

</div>

## 📌 آنچه باید بلد باشید

- <code dir="ltr">=</code> یعنی **انتساب**: اول سمت راست حساب می‌شود، بعد نام به نتیجه وصل می‌شود؛ <code dir="ltr">==</code> یعنی **مقایسه**.
- بعد از <code dir="ltr">b = a</code>، تغییر دادن <code dir="ltr">a</code> با یک انتساب جدید روی <code dir="ltr">b</code> اثری ندارد.
- نام‌های مجاز و غیرمجاز متغیر را تشخیص دهید.
- نوع هر مقدار را بدانید و نوع **حاصل یک عبارت** را پیش‌بینی کنید (مخصوصاً: ترکیب <code dir="ltr">int</code> و <code dir="ltr">float</code>، و تقسیم <code dir="ltr">/</code>).
- چرا <code dir="ltr">0.1 + 0.2 == 0.3</code> نادرست است و به جای آن چه باید کرد.
- رفتار <code dir="ltr">+</code> و <code dir="ltr">&#42;</code> روی رشته‌ها و خطای ترکیب رشته با عدد.
- <code dir="ltr">True</code> و <code dir="ltr">False</code> مثل <code dir="ltr">1</code> و <code dir="ltr">0</code> رفتار می‌کنند.
- نتیجه‌ی <code dir="ltr">int()</code>، <code dir="ltr">float()</code>، <code dir="ltr">str()</code> و <code dir="ltr">bool()</code> را برای ورودی‌های مختلف پیش‌بینی کنید.
- تفاوت <code dir="ltr">TypeError</code> و <code dir="ltr">ValueError</code>.

</div>

### سؤال ۶ — نام‌گذاری

کدام نام‌ها برای متغیر **مجاز** هستند؟ برای نام‌های غیرمجاز دلیل بیاورید. کدام نام مجاز است اما **استفاده از آن توصیه نمی‌شود**؟

<code dir="ltr">2nd&#95;value</code> — <code dir="ltr">&#95;count</code> — <code dir="ltr">my-var</code> — <code dir="ltr">total&#95;sales</code> — <code dir="ltr">class</code> — <code dir="ltr">Class</code> — <code dir="ltr">sum</code>

<details>
<summary><b>پاسخ</b></summary>

- <code dir="ltr">2nd&#95;value</code>: غیرمجاز (با رقم شروع شده)
- <code dir="ltr">&#95;count</code>: مجاز
- <code dir="ltr">my-var</code>: غیرمجاز (<code dir="ltr">-</code> مجاز نیست؛ پایتون آن را تفریق می‌فهمد)
- <code dir="ltr">total&#95;sales</code>: مجاز (و به سبک snake_case)
- <code dir="ltr">class</code>: غیرمجاز (کلمه‌ی کلیدی)
- <code dir="ltr">Class</code>: مجاز (با <code dir="ltr">class</code> فرق دارد، چون حروف بزرگ و کوچک فرق دارند)
- <code dir="ltr">sum</code>: مجاز، اما **توصیه نمی‌شود**؛ چون تابع آماده‌ی <code dir="ltr">sum</code> را از کار می‌اندازد.

</details>

## خلاصه‌ی جلسه

- متغیر یک **نام** است که به یک **مقدار** اشاره می‌کند؛ <code dir="ltr">=</code> انتساب است و اول سمت راست حساب می‌شود.
- نوع به مقدار تعلق دارد و با <code dir="ltr">type</code> دیده می‌شود. چهار نوع پایه: <code dir="ltr">int</code>، <code dir="ltr">float</code>، <code dir="ltr">str</code>، <code dir="ltr">bool</code>؛ و مقدار ویژه‌ی <code dir="ltr">None</code>.
- ترکیب <code dir="ltr">int</code> و <code dir="ltr">float</code> و همچنین تقسیم <code dir="ltr">/</code> همیشه <code dir="ltr">float</code> می‌دهد. اعداد اعشاری را با <code dir="ltr">==</code> مقایسه نکنید.
- <code dir="ltr">+</code> روی رشته‌ها یعنی چسباندن؛ رشته و عدد خودبه‌خود ترکیب نمی‌شوند.
- <code dir="ltr">True</code> و <code dir="ltr">False</code> مثل <code dir="ltr">1</code> و <code dir="ltr">0</code> رفتار می‌کنند.
- <code dir="ltr">int()</code>، <code dir="ltr">float()</code>، <code dir="ltr">str()</code> و <code dir="ltr">bool()</code> نوع را تبدیل می‌کنند؛ <code dir="ltr">TypeError</code> و <code dir="ltr">ValueError</code> را از هم تشخیص دهید.

**جلسه‌ی بعد:** عملگرها (ریاضی، مقایسه‌ای، منطقی و انتساب) و اولویت آن‌ها.
