## ۱. نمودارهای توزیع

### 💡 Seaborn چیست؟

**Seaborn** کتابخانه‌ای است که **روی** Matplotlib (جلسه‌ی ۲۷) ساخته شده است. هر نمودار Seaborn در نهایت یک شکل و محور Matplotlib است، اما Seaborn سه کار را برای شما انجام می‌دهد:

- مستقیماً با **DataFrame** کار می‌کند: نام ستون‌ها را می‌دهید (<code dir="ltr">x="body&#95;mass&#95;g"</code>)، نه آرایه‌ها را،
- آمار لازم (شمارش، میانگین، بازه‌ی اطمینان، تخمین چگالی) را **خودش** حساب می‌کند،
- با آرگومان <code dir="ltr">hue</code> یک متغیر گروهی را با رنگ وارد نمودار می‌کند و راهنما (legend) را خودکار می‌سازد.

اگر Seaborn را نصب نکرده‌اید، یک بار در Command Prompt اجرا کنید: <code dir="ltr">pip install seaborn</code> (مثل نصب Jupyter در جلسه‌ی ۱).

### 💡 داده‌ی این جلسه: پنگوئن‌های پالمر

در این جلسه از یک داده‌ی واقعی استفاده می‌کنیم: اندازه‌گیری‌های ۳۴۴ پنگوئن از سه گونه (<code dir="ltr">Adelie</code>، <code dir="ltr">Chinstrap</code> و <code dir="ltr">Gentoo</code>) روی سه جزیره در قطب جنوب (داده‌ی **Palmer Penguins**، جمع‌آوری‌شده توسط دکتر Kristen Gorman، با مجوز آزاد CC0). فایل در پوشه‌ی <code dir="ltr">data</code> کنار نوت‌بوک‌هاست؛ اگر فقط همین نوت‌بوک را دانلود کرده‌اید، کد زیر فایل را مستقیم از سایت درس می‌خواند.

</div>

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
pd.set_option("display.width", 120)

LOCAL = Path("data/penguins.csv")
URL = "https://haghbinh.github.io/Python-Course/Notebooks/data/penguins.csv"
penguins = pd.read_csv(LOCAL if LOCAL.exists() else URL)

print(penguins.shape)
penguins.head()
````

<div dir="rtl">

### 🔍 کد را بخوان

- نام مستعار رایج Seaborn، <code dir="ltr">sns</code> است (همان‌طور که <code dir="ltr">np</code> و <code dir="ltr">pd</code> را در جلسه‌های ۲۱ و ۲۳ دیدید).
- <code dir="ltr">sns.set&#95;theme(style="whitegrid")</code> ظاهر پیش‌فرض همه‌ی نمودارهای بعدی را تنظیم می‌کند (پس‌زمینه‌ی سفید با خطوط راهنما). <code dir="ltr">pd.set&#95;option("display.width", 120)</code> فقط اجازه می‌دهد جدول‌های چاپ‌شده پهن‌تر باشند تا ستون‌ها شکسته نشوند.
- <code dir="ltr">LOCAL if LOCAL.exists() else URL</code> یک عبارت شرطی یک‌خطی است (جلسه‌ی ۵): اگر فایل محلی وجود داشت همان، وگرنه نشانی اینترنتی. <code dir="ltr">pd.read&#95;csv</code> هر دو را می‌پذیرد.
- ستون‌ها: گونه، جزیره، طول و عمق منقار (<code dir="ltr">bill</code>، میلی‌متر)، طول باله (<code dir="ltr">flipper</code>، میلی‌متر)، جرم بدن (گرم)، جنس و سال. چند مقدار گمشده هم دارد (جلسه‌ی ۲۶)؛ Seaborn مقدارهای گمشده را خودش کنار می‌گذارد.

### 💡 هیستوگرام: <code dir="ltr">histplot</code>

**هیستوگرام** محور افقی را به چند بازه (bin) تقسیم می‌کند و نشان می‌دهد چند مشاهده در هر بازه افتاده است. با <code dir="ltr">kde=True</code> یک منحنی هموار هم روی آن کشیده می‌شود.

</div>

````python
fig, ax = plt.subplots(figsize=(7, 4))
sns.histplot(data=penguins, x="body_mass_g", bins=20, kde=True, ax=ax)
ax.set_xlabel("Body mass (g)")
ax.set_title("Distribution of body mass")
plt.show()
````

### 🔍 کد را بخوان

- <code dir="ltr">data=penguins, x="body&#95;mass&#95;g"</code>: به جای آرایه، DataFrame و **نام ستون** را می‌دهیم.
- <code dir="ltr">ax=ax</code>: نمودار روی محوری کشیده می‌شود که خودمان با <code dir="ltr">plt.subplots</code> ساخته‌ایم؛ پس همه‌ی ابزارهای جلسه‌ی ۲۷ (<code dir="ltr">set&#95;xlabel</code>، <code dir="ltr">set&#95;title</code>، <code dir="ltr">savefig</code>) همچنان کار می‌کنند.
- توزیع **چوله به راست** است: دنباله‌ی سمت راست (پنگوئن‌های سنگین) بلندتر از دنباله‌ی چپ است. ضریب چولگی آن حدود <code dir="ltr">0.47</code> است (جلسه‌ی ۲۹).

### 💡 آرگومان‌های مهم <code dir="ltr">histplot</code>

| آرگومان | کار | مقدار نمونه |
|---|---|---|
| <code dir="ltr">bins</code> | تعداد بازه‌ها (یا لبه‌های بازه‌ها) | <code dir="ltr">20</code> |
| <code dir="ltr">binwidth</code> | پهنای هر بازه | <code dir="ltr">100</code> |
| <code dir="ltr">stat</code> | محور عمودی چه باشد | <code dir="ltr">"count"</code>، <code dir="ltr">"probability"</code>، <code dir="ltr">"density"</code> |
| <code dir="ltr">kde</code> | منحنی چگالی هموار | <code dir="ltr">True</code> |
| <code dir="ltr">hue</code> | رنگ‌بندی بر اساس یک ستون گروهی | <code dir="ltr">"species"</code> |
| <code dir="ltr">element</code> | شکل رسم وقتی <code dir="ltr">hue</code> دارید | <code dir="ltr">"bars"</code>، <code dir="ltr">"step"</code> |
| <code dir="ltr">multiple</code> | چیدمان گروه‌ها | <code dir="ltr">"layer"</code>، <code dir="ltr">"stack"</code>، <code dir="ltr">"dodge"</code> |

### ❓ پیش‌بینی کن

هیستوگرام **طول باله** (<code dir="ltr">flipper&#95;length&#95;mm</code>) را بدون <code dir="ltr">hue</code> بکشید. به نظر شما چند قله دارد؟ اگر دو قله ببینید، چه توضیحی برایش دارید؟ قبل از اجرای سلول بعد فکر کنید.

</div>

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.histplot(data=penguins, x="flipper_length_mm", bins=25, ax=axes[0])
axes[0].set_title("All penguins")
sns.histplot(data=penguins, x="flipper_length_mm", hue="species",
             bins=25, element="step", ax=axes[1])
axes[1].set_title("Colored by species")
plt.tight_layout()
plt.show()
````

<p align="center" dir="rtl"><i>طول باله: بدون hue دو قله دیده می‌شود؛ با hue معلوم می‌شود هر قله مربوط به کدام گونه است.</i></p>

<div dir="rtl">

🔍 هیستوگرام کل داده **دو‌قله‌ای** (bimodal) است. با <code dir="ltr">hue="species"</code> توضیح آن پیدا می‌شود: قله‌ی سمت راست تقریباً فقط گونه‌ی <code dir="ltr">Gentoo</code> است (میانگین باله حدود ۲۱۷ میلی‌متر) و قله‌ی چپ ترکیب <code dir="ltr">Adelie</code> و <code dir="ltr">Chinstrap</code> (حدود ۱۹۰ و ۱۹۶). **قاعده‌ی کلی:** توزیعی که چند قله دارد معمولاً نشانه‌ی یک **متغیر گروهی پنهان** است. برای یک مدل یادگیری ماشین هم همین پیام را دارد: این ویژگی احتمالاً برای تشخیص <code dir="ltr">Gentoo</code> بسیار مفید است.

### 💡 <code dir="ltr">kdeplot</code> و <code dir="ltr">ecdfplot</code>

- **KDE** (تخمین چگالی کرنلی) یک منحنی هموار است که شکل توزیع را بدون وابستگی به مرز بازه‌ها نشان می‌دهد. پارامتر <code dir="ltr">bw&#95;adjust</code> میزان همواری را تنظیم می‌کند: عدد کوچک‌تر یعنی منحنی پرنوسان‌تر.
- **ECDF** (تابع توزیع تجمعی تجربی) برای هر مقدار <code dir="ltr">x</code> نشان می‌دهد چه کسری از داده‌ها کوچک‌تر یا مساوی <code dir="ltr">x</code> هستند. هیچ پارامتری برای تنظیم ندارد، پس برای **مقایسه‌ی دقیق** گروه‌ها مناسب است: میانه جایی است که منحنی از <code dir="ltr">0.5</code> می‌گذرد.

</div>

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.kdeplot(data=penguins, x="body_mass_g", hue="species",
            fill=True, common_norm=False, ax=axes[0])
sns.ecdfplot(data=penguins, x="body_mass_g", hue="species", ax=axes[1])
plt.tight_layout()
plt.show()
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">fill=True</code> زیر منحنی‌ها را رنگ می‌کند.
- <code dir="ltr">common&#95;norm=False</code> باعث می‌شود **هر** منحنی جداگانه نرمال شود (مساحت زیر هر کدام <code dir="ltr">1</code>). در نمودار ECDF، منحنی <code dir="ltr">Gentoo</code> کاملاً در سمت راست است: تقریباً همه‌ی <code dir="ltr">Gentoo</code>ها از تقریباً همه‌ی <code dir="ltr">Adelie</code>ها سنگین‌ترند.

### ⚠️ اشتباه رایج: <code dir="ltr">common&#95;norm</code> پیش‌فرض

وقتی <code dir="ltr">hue</code> دارید و <code dir="ltr">stat="density"</code> یا <code dir="ltr">kdeplot</code> به کار می‌برید، پیش‌فرض <code dir="ltr">common&#95;norm=True</code> است: مساحت **کل** همه‌ی منحنی‌ها با هم <code dir="ltr">1</code> می‌شود، نه هر کدام. در نتیجه گروه کوچک‌تر (مثلاً <code dir="ltr">Chinstrap</code> با ۶۸ پنگوئن) منحنی کوتاه‌تری دارد، حتی اگر شکل توزیعش همان باشد. اگر می‌خواهید **شکل** توزیع گروه‌ها را مقایسه کنید، <code dir="ltr">common&#95;norm=False</code> بگذارید (سؤال ۲).

### 💡 سطح محور یا سطح شکل؟

Seaborn دو نوع تابع دارد:

| | سطح محور (axes-level) | سطح شکل (figure-level) |
|---|---|---|
| مثال | <code dir="ltr">histplot</code>، <code dir="ltr">kdeplot</code>، <code dir="ltr">boxplot</code>، <code dir="ltr">scatterplot</code> | <code dir="ltr">displot</code>، <code dir="ltr">catplot</code>، <code dir="ltr">relplot</code>، <code dir="ltr">pairplot</code> |
| خروجی | روی یک <code dir="ltr">ax</code> می‌کشد و آن را برمی‌گرداند | یک شکل کامل (چند محور) می‌سازد و یک شیء <code dir="ltr">FacetGrid</code> برمی‌گرداند |
| قرار دادن در subplot خودتان | بله، با <code dir="ltr">ax=</code> | خیر؛ خودش شکل می‌سازد |
| اندازه | <code dir="ltr">figsize</code> در <code dir="ltr">plt.subplots</code> | <code dir="ltr">height</code> و <code dir="ltr">aspect</code> |
| تقسیم به پنل‌ها | با حلقه، دستی | با <code dir="ltr">col=</code> و <code dir="ltr">row=</code>، خودکار |

</div>

````python
g = sns.displot(data=penguins, x="flipper_length_mm", col="species",
                bins=15, height=3, aspect=1.1)
g.set_axis_labels("Flipper length (mm)", "Count")
plt.show()
````

<div dir="rtl">

🔍 <code dir="ltr">displot</code> نسخه‌ی سطح شکلِ <code dir="ltr">histplot</code> است: با <code dir="ltr">col="species"</code> برای هر گونه یک پنل جدا می‌سازد و محورها را **مشترک** نگه می‌دارد تا مقایسه آسان باشد. به جای <code dir="ltr">ax.set&#95;xlabel</code>، از متدهای شیء برگشتی (<code dir="ltr">g.set&#95;axis&#95;labels</code>) استفاده می‌کنیم.

### ⚠️ اشتباه رایج: <code dir="ltr">ax=</code> در تابع سطح شکل

نوشتن <code dir="ltr">sns.displot(..., ax=axes[0])</code> کار نمی‌کند (Seaborn هشدار می‌دهد و آرگومان را نادیده می‌گیرد)، چون <code dir="ltr">displot</code> شکل خودش را می‌سازد. برای قرار دادن یک هیستوگرام در subplotهای خودتان، از <code dir="ltr">histplot</code> استفاده کنید.

## ۲. نمودارهای رسته‌ای

### 💡 مفهوم

وقتی یک متغیر **رسته‌ای** (مثل گونه یا جزیره) و یک متغیر **عددی** (مثل جرم) داریم، سؤال اصلی معمولاً این است: «متغیر عددی در گروه‌های مختلف چه فرقی دارد؟» Seaborn چند پاسخ متفاوت به این سؤال دارد:

| تابع | چه چیزی نشان می‌دهد |
|---|---|
| <code dir="ltr">countplot</code> | فقط **تعداد** مشاهده‌ها در هر گروه (بدون متغیر عددی) |
| <code dir="ltr">barplot</code> | یک **آماره** (پیش‌فرض: میانگین) و یک **میله‌ی خطا** |
| <code dir="ltr">boxplot</code> | میانه، چارک‌ها و داده‌های پرت (جلسه‌ی ۲۶) |
| <code dir="ltr">violinplot</code> | نمودار جعبه‌ای + شکل کامل توزیع (KDE) |
| <code dir="ltr">stripplot</code> / <code dir="ltr">swarmplot</code> | **تک‌تک** مشاهده‌ها به شکل نقطه |

</div>

````python
fig, ax = plt.subplots(figsize=(6, 3.5))
sns.countplot(data=penguins, x="island", hue="species", ax=ax)
plt.show()

print(pd.crosstab(penguins["island"], penguins["species"]))
````

<div dir="rtl">

### 🔍 کد را بخوان

<code dir="ltr">countplot</code> نسخه‌ی تصویری <code dir="ltr">pd.crosstab</code> (جلسه‌ی ۲۴) است. نتیجه‌ی مهمی که دیده می‌شود: <code dir="ltr">Gentoo</code> **فقط** در جزیره‌ی <code dir="ltr">Biscoe</code> و <code dir="ltr">Chinstrap</code> **فقط** در <code dir="ltr">Dream</code> زندگی می‌کند؛ <code dir="ltr">Adelie</code> در هر سه جزیره هست. این نکته را در سؤال ۴ لازم دارید.

### 💡 میانگین، میانه و همه‌ی داده‌ها در یک نگاه

</div>

````python
fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)

sns.barplot(data=penguins, x="species", y="body_mass_g",
            errorbar=("ci", 95), ax=axes[0])
axes[0].set_title("barplot: mean + 95% CI")

sns.boxplot(data=penguins, x="species", y="body_mass_g", ax=axes[1])
axes[1].set_title("boxplot: median + quartiles")

sns.violinplot(data=penguins, x="species", y="body_mass_g",
               inner=None, color="0.9", ax=axes[2])
sns.stripplot(data=penguins, x="species", y="body_mass_g",
              size=3, alpha=0.6, ax=axes[2])
axes[2].set_title("violin + strip: every point")

plt.tight_layout()
plt.show()
````

<p align="center" dir="rtl"><i>یک داده، سه نمایش: میانگین و بازه‌ی اطمینان، نمودار جعبه‌ای، و ویولن با تک‌تک نقطه‌ها.</i></p>

<div dir="rtl">

### 🔍 کد را بخوان

- **<code dir="ltr">barplot</code>**: ارتفاع میله **میانگین** جرم هر گونه است (حدود ۳۷۰۱، ۳۷۳۳ و ۵۰۷۶ گرم). خط سیاه روی میله **بازه‌ی اطمینان ۹۵٪ میانگین** است که Seaborn با bootstrap (نمونه‌گیری دوباره، جلسه‌ی ۱۷) حساب می‌کند. این میله‌ی خطا درباره‌ی **دقت تخمین میانگین** است، نه درباره‌ی پراکندگی پنگوئن‌ها.
- **<code dir="ltr">boxplot</code>**: خط وسط جعبه **میانه**، لبه‌های جعبه چارک اول و سوم، و سبیل‌ها تا ۱٫۵ برابر دامنه‌ی میان‌چارکی (IQR) ادامه دارند؛ نقطه‌های بیرون از آن، داده‌های پرت احتمالی‌اند (جلسه‌ی ۲۶).
- **ویولن + نقطه‌ها**: دو تابع روی **یک محور** کشیده شده‌اند؛ این ترکیب، شکل توزیع و تک‌تک داده‌ها را با هم نشان می‌دهد. <code dir="ltr">inner=None</code> نمودار جعبه‌ای کوچک داخل ویولن را حذف می‌کند و <code dir="ltr">color="0.9"</code> آن را خاکستری روشن می‌کند تا نقطه‌ها دیده شوند.
- <code dir="ltr">sharey=True</code> محور عمودی هر سه پنل را یکسان می‌کند تا مقایسه منصفانه باشد.

### 💡 تغییر آماره و میله‌ی خطا

| کد | ارتفاع میله | میله‌ی خطا |
|---|---|---|
| <code dir="ltr">barplot(..., errorbar=("ci", 95))</code> | میانگین | بازه‌ی اطمینان ۹۵٪ میانگین (پیش‌فرض) |
| <code dir="ltr">barplot(..., errorbar="sd")</code> | میانگین | یک انحراف معیار از داده‌ها |
| <code dir="ltr">barplot(..., errorbar=None)</code> | میانگین | بدون میله‌ی خطا |
| <code dir="ltr">barplot(..., estimator="median")</code> | میانه | بازه‌ی اطمینان میانه |

### ❓ پیش‌بینی کن

اگر تعداد پنگوئن‌های هر گونه ۱۰۰ برابر شود ولی توزیع جرم همان بماند، کدام‌یک کوتاه‌تر می‌شود: میله‌ی خطای <code dir="ltr">errorbar=("ci", 95)</code> یا میله‌ی خطای <code dir="ltr">errorbar="sd"</code>؟ جعبه‌ی <code dir="ltr">boxplot</code> چطور؟

<details>
<summary><b>پاسخ</b></summary>

فقط بازه‌ی اطمینان کوتاه می‌شود: پهنای آن تقریباً با <code dir="ltr">sd / sqrt(n)</code> متناسب است، پس با ۱۰۰ برابر شدن <code dir="ltr">n</code>، حدود ۱۰ برابر باریک‌تر می‌شود. انحراف معیار و چارک‌ها ویژگی‌های **خود توزیع**‌اند و با بیشتر شدن داده تغییری نمی‌کنند (فقط دقیق‌تر تخمین زده می‌شوند).

</details>

دو گروه با میانگین یکسان می‌توانند توزیع‌های کاملاً متفاوتی داشته باشند (یکی متقارن، یکی دو‌قله‌ای؛ یکی پراکنده، یکی فشرده). <code dir="ltr">barplot</code> فقط یک عدد از هر گروه نشان می‌دهد. وقتی داده‌ها کم‌اند، نقطه‌ها را هم نشان دهید (<code dir="ltr">stripplot</code>)؛ وقتی زیادند، <code dir="ltr">boxplot</code> یا <code dir="ltr">violinplot</code> امن‌ترند.

### 💡 <code dir="ltr">catplot</code>: نمودار رسته‌ای با چند پنل

<code dir="ltr">catplot</code> نسخه‌ی سطح شکل همه‌ی نمودارهای رسته‌ای است؛ نوع نمودار را با <code dir="ltr">kind</code> انتخاب می‌کنید و با <code dir="ltr">col</code> پنل‌ها را می‌سازید:

## ۳. همبستگی و heatmap

### 💡 مفهوم

**ضریب همبستگی Pearson** (<code dir="ltr">r</code>) قدرت و جهت رابطه‌ی **خطی** دو متغیر عددی را با عددی بین <code dir="ltr">-1</code> و <code dir="ltr">1</code> می‌سنجد: نزدیک <code dir="ltr">1</code> یعنی با افزایش یکی، دیگری هم به‌طور خطی افزایش می‌یابد؛ نزدیک <code dir="ltr">-1</code> رابطه‌ی خطی معکوس؛ نزدیک <code dir="ltr">0</code> یعنی رابطه‌ی **خطی** ضعیف (نه لزوماً بی‌رابطه!). در جلسه‌ی ۸ شباهت کسینوسی دو بردار را دیدید؛ <code dir="ltr">r</code> در واقع شباهت کسینوسی دو ستون **پس از کم کردن میانگین** آن‌هاست.

<code dir="ltr">DataFrame.corr</code> همبستگی همه‌ی جفت ستون‌ها را یک‌جا حساب می‌کند و نتیجه یک ماتریس متقارن است:

</div>

````python
features = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
corr = penguins[features].corr()
print(corr.round(2))
````

<div dir="rtl">

### 🔍 کد را بخوان

- قطر اصلی همیشه <code dir="ltr">1</code> است (همبستگی هر ستون با خودش).
- ستون <code dir="ltr">year</code> را عمداً کنار گذاشته‌ایم: عددی است اما معنای «اندازه» ندارد. اگر کل DataFrame را بدهید، باید <code dir="ltr">numeric&#95;only=True</code> بگذارید تا ستون‌های متنی (<code dir="ltr">species</code>، <code dir="ltr">island</code>، <code dir="ltr">sex</code>) خطا ندهند.
- قوی‌ترین رابطه بین طول باله و جرم بدن است (<code dir="ltr">0.87</code>).

### 💡 نمایش با <code dir="ltr">heatmap</code>

خواندن یک جدول بزرگ از اعداد سخت است؛ <code dir="ltr">heatmap</code> هر خانه را با رنگ نشان می‌دهد. چون ماتریس متقارن است، نیمه‌ی بالایی تکراری است و می‌توان آن را با یک **ماسک** پنهان کرد:

</div>

````python
mask = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f",
            cmap="vlag", center=0, vmin=-1, vmax=1,
            square=True, linewidths=0.5, ax=ax)
ax.set_title("Pearson correlation")
plt.tight_layout()
plt.show()
````

<p align="center"><img src="images/s28_heatmap.png" width="460" alt="s28_heatmap.png"></p>

<p align="center" dir="rtl"><i>ماتریس همبستگی ویژگی‌های پنگوئن‌ها؛ نیمه‌ی بالایی با ماسک پنهان شده است.</i></p>

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">np.triu(np.ones&#95;like(corr, dtype=bool))</code> یک ماتریس بولی می‌سازد که روی قطر و بالای آن <code dir="ltr">True</code> است (جلسه‌ی ۲۱)؛ خانه‌هایی که ماسک آن‌ها <code dir="ltr">True</code> است رسم **نمی‌شوند**.
- <code dir="ltr">annot=True, fmt=".2f"</code> عدد هر خانه را با دو رقم اعشار می‌نویسد.
- <code dir="ltr">cmap="vlag"</code> یک نقشه‌ی رنگ **واگرا** (آبی، سفید، قرمز) است و <code dir="ltr">center=0</code> سفید را روی صفر می‌گذارد؛ با <code dir="ltr">vmin=-1, vmax=1</code> رنگ‌ها بین نمودارهای مختلف قابل مقایسه می‌مانند. برای همبستگی که علامت آن مهم است، همیشه نقشه‌ی رنگ واگرا انتخاب کنید.

### 💡 Pearson یا Spearman؟

**Spearman** همان Pearson است، اما روی **رتبه**‌ی داده‌ها حساب می‌شود (کوچک‌ترین مقدار رتبه‌ی ۱، بعدی رتبه‌ی ۲، ...). پس رابطه‌ی **یکنوا** (همیشه صعودی یا همیشه نزولی، حتی اگر خطی نباشد) را می‌سنجد و به داده‌های پرت حساسیت کمتری دارد.

</div>

````python
print(penguins[features].corr(method="spearman").round(2))

x = np.linspace(0, 5, 30)
demo = pd.DataFrame({"x": x, "y": np.exp(x)})
print(demo.corr(method="pearson").iloc[0, 1].round(3),
      demo.corr(method="spearman").iloc[0, 1].round(3))
````

<div dir="rtl">

🔍 برای پنگوئن‌ها دو روش تقریباً هم‌عددند. اما برای <code dir="ltr">y = exp(x)</code> رابطه کاملاً یکنواست، پس Spearman دقیقاً <code dir="ltr">1</code> می‌شود، در حالی که Pearson کمتر از <code dir="ltr">1</code> است چون رابطه خطی نیست.

### ⚠️ همبستگی، علیت نیست

همبستگی بالای طول باله و جرم بدن نمی‌گوید که «باله‌ی بلند باعث سنگینی می‌شود»؛ هر دو نتیجه‌ی **اندازه‌ی کلی بدن** و گونه هستند. همچنین <code dir="ltr">r ≈ 0</code> به معنی «بی‌رابطه» نیست؛ یک رابطه‌ی سهمی‌شکل کامل هم می‌تواند <code dir="ltr">r</code> نزدیک صفر داشته باشد (سؤال ۵). **قبل از تفسیر هر ضریب همبستگی، نمودار پراکندگی را ببینید.**

### ❓ پیش‌بینی کن

همبستگی طول منقار و عمق منقار روی **کل** داده حدود <code dir="ltr">-0.24</code> است (منفی). فکر می‌کنید اگر همین همبستگی را **جداگانه برای هر گونه** حساب کنید، علامتش چه می‌شود؟

</div>

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2), sharey=True)

sns.regplot(data=penguins, x="bill_length_mm", y="bill_depth_mm",
            scatter_kws={"s": 12, "alpha": 0.5}, color="gray", ax=axes[0])
axes[0].set_title("All penguins: negative trend")

colors = dict(zip(penguins["species"].unique(), sns.color_palette()))
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm",
                hue="species", palette=colors, s=15, ax=axes[1])
for name, group in penguins.groupby("species"):
    sns.regplot(data=group, x="bill_length_mm", y="bill_depth_mm",
                scatter=False, color=colors[name], ax=axes[1])
axes[1].set_title("Within each species: positive trend")

plt.tight_layout()
plt.show()

print(penguins.groupby("species")[["bill_length_mm", "bill_depth_mm"]]
      .corr().iloc[0::2, 1].round(2))
````

<p align="center" dir="rtl"><i>پارادوکس سیمپسون: روند کلی منفی است، اما درون هر گونه روند مثبت است.</i></p>

<div dir="rtl">

🔍 درون **هر** گونه همبستگی **مثبت** است (حدود <code dir="ltr">0.39</code>، <code dir="ltr">0.65</code> و <code dir="ltr">0.64</code>): پنگوئن بزرگ‌تر، منقار بلندتر **و** عمیق‌تری دارد، که منطقی است. اما وقتی گونه‌ها را با هم مخلوط می‌کنیم، <code dir="ltr">Gentoo</code>ها (منقار بلند و **کم‌عمق**) نقطه‌ها را به پایین و راست می‌کشند و روند کلی **منفی** می‌شود. به این وارونه شدن جهت رابطه با اضافه کردن یک متغیر گروهی، **پارادوکس سیمپسون** می‌گویند.

- <code dir="ltr">regplot</code> نمودار پراکندگی را همراه با خط رگرسیون (و بازه‌ی اطمینان آن) می‌کشد؛ با <code dir="ltr">scatter=False</code> فقط خط را.
- دیکشنری <code dir="ltr">colors</code> به هر گونه یک رنگ ثابت می‌دهد تا رنگ هر خط با رنگ نقطه‌های همان گونه یکی باشد. بدون آن، <code dir="ltr">groupby</code> گونه‌ها را به ترتیب الفبا می‌دهد و رنگ‌ها جابه‌جا می‌شوند؛ اشتباهی که در نمودار هیچ خطایی نمی‌دهد ولی خواننده را گمراه می‌کند.
- <code dir="ltr">groupby(...)[cols].corr()</code> برای هر گروه یک ماتریس ۲ در ۲ می‌سازد؛ <code dir="ltr">iloc[0::2, 1]</code> از هر ماتریس فقط خانه‌ی خارج از قطر را برمی‌دارد.

**پیام برای مدل‌سازی:** اگر متغیری مثل گونه در داده هست، مدلی که آن را نادیده بگیرد ممکن است رابطه‌ای با **جهت غلط** یاد بگیرد.

</div>

## ۴. pairplot

### 💡 مفهوم

<code dir="ltr">heatmap</code> فقط یک عدد برای هر جفت ویژگی نشان می‌دهد. <code dir="ltr">pairplot</code> برای **هر جفت** ویژگی یک نمودار پراکندگی می‌کشد و روی قطر، توزیع تک‌تک ویژگی‌ها را نشان می‌دهد. با <code dir="ltr">hue</code> می‌بینید کلاس‌ها در کدام جفت ویژگی **از هم جدا** می‌شوند؛ این همان شهودی است که قبل از ساختن یک دسته‌بند لازم دارید.

</div>

````python
g = sns.pairplot(penguins, vars=features, hue="species",
                 corner=True, diag_kind="kde", height=1.9,
                 plot_kws={"s": 12, "alpha": 0.6})
plt.show()
````

- <code dir="ltr">diag&#95;kind="kde"</code>: روی قطر به جای هیستوگرام، KDE هر گونه.

- پس انتظار داریم بهترین جفت ویژگی برای تشخیص هر سه گونه، **طول منقار + یکی از ویژگی‌هایی که Gentoo را جدا می‌کند** باشد (سؤال ۶).

### سؤال ۶ — از heatmap و pairplot تا مدل

الف) در heatmap بخش ۳، همبستگی طول باله و جرم بدن <code dir="ltr">0.87</code> است. برای مدل رگرسیون خطی که هر دو را به‌عنوان ورودی می‌گیرد، **ضریب تورم واریانس** (VIF) هر کدام در حالت دو ویژگی برابر <code dir="ltr">1 / (1 - r²)</code> است. VIF را حساب کنید. این پدیده (**هم‌خطی چندگانه**) چه مشکلی برای تفسیر ضرایب مدل ایجاد می‌کند؟ دو راه حل نام ببرید.

ب) بر اساس pairplot بخش ۴، اگر فقط **دو** ویژگی برای تشخیص سه گونه مجاز باشد، کدام جفت را انتخاب می‌کنید؟ چرا جفت «طول باله + جرم بدن» با وجود همبستگی بالا با هم، انتخاب خوبی نیست؟

ج) کد زیر یک دسته‌بند ساده‌ی «نزدیک‌ترین مرکز» (هر نقطه به گونه‌ای نسبت داده می‌شود که میانگین آن گونه به آن نزدیک‌تر است) را روی ویژگی‌های استانداردشده و برای همه‌ی جفت ویژگی‌ها اجرا می‌کند. آیا نتیجه حدس شما در «ب» را تأیید می‌کند؟

</div>

```python
from itertools import combinations

data = penguins.dropna(subset=features)
X = (data[features] - data[features].mean()) / data[features].std()
y = data["species"].to_numpy()

for a, b in combinations(features, 2):
    Z = X[[a, b]].to_numpy()
    centers = {s: Z[y == s].mean(axis=0) for s in np.unique(y)}
    labels = list(centers)
    dist = np.stack([((Z - centers[s]) ** 2).sum(axis=1) for s in labels], axis=1)
    pred = np.array(labels)[dist.argmin(axis=1)]
    print(f"{a:<18} {b:<18} {(pred == y).mean():.3f}")
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">r ≈ 0.871</code>؛ <code dir="ltr">VIF = 1 / (1 - 0.871²) ≈ 4.15</code>. یعنی واریانس تخمین هر ضریب حدود چهار برابر حالتی است که دو ویژگی مستقل بودند. وقتی دو ورودی تقریباً هم‌جهت‌اند، مدل نمی‌تواند تشخیص دهد اثر مال کدام است: ضرایب ناپایدار می‌شوند (با تغییر کوچک داده بالا و پایین می‌روند، حتی ممکن است علامتشان برعکس شود)، اگرچه **پیش‌بینی** مدل هنوز خوب است. راه‌حل‌ها: حذف یکی از دو ویژگی، ترکیب آن‌ها (مثلاً یک «اندازه‌ی بدن» واحد یا PCA)، یا تنظیم (regularization) مثل Ridge که در درس یادگیری ماشین می‌بینید.
