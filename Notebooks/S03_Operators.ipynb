## ۱. عملگرهای ریاضی

### 💡 مفهوم

**عملگر (operator)** نمادی است که روی یک یا چند مقدار (**عملوند**) کاری انجام می‌دهد و یک مقدار جدید می‌سازد. در <code dir="ltr">3 + 4</code>، علامت <code dir="ltr">+</code> عملگر و <code dir="ltr">3</code> و <code dir="ltr">4</code> عملوندها هستند.

| عملگر | نام | مثال | نتیجه |
|:---:|---|---|---|
| <code dir="ltr">+</code> | جمع | <code dir="ltr">5 + 2</code> | <code dir="ltr">7</code> |
| <code dir="ltr">-</code> | تفریق | <code dir="ltr">5 - 2</code> | <code dir="ltr">3</code> |
| <code dir="ltr">&#42;</code> | ضرب | <code dir="ltr">5 &#42; 2</code> | <code dir="ltr">10</code> |
| <code dir="ltr">/</code> | تقسیم | <code dir="ltr">5 / 2</code> | <code dir="ltr">2.5</code> |
| <code dir="ltr">//</code> | تقسیم صحیح (خارج‌قسمت) | <code dir="ltr">5 // 2</code> | <code dir="ltr">2</code> |
| <code dir="ltr">%</code> | باقی‌مانده | <code dir="ltr">5 % 2</code> | <code dir="ltr">1</code> |
| <code dir="ltr">&#42;&#42;</code> | توان | <code dir="ltr">5 &#42;&#42; 2</code> | <code dir="ltr">25</code> |

</div>

````python
a = 10
b = 3
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

<div dir="rtl">

### 🔍 کد را بخوان: <code dir="ltr">//</code> و <code dir="ltr">%</code> با هم

<code dir="ltr">//</code> و <code dir="ltr">%</code> دو نیمه‌ی یک تقسیم با باقی‌مانده هستند. وقتی <code dir="ltr">10</code> را بر <code dir="ltr">3</code> تقسیم می‌کنیم: خارج‌قسمت <code dir="ltr">3</code> و باقی‌مانده <code dir="ltr">1</code> است، چون

<code dir="ltr">10 = 3 × 3 + 1</code>

همیشه این رابطه برقرار است: <code dir="ltr">a == (a // b) &#42; b + (a % b)</code>

این دو عملگر در عمل بسیار پرکاربردند. کد زیر را بخوانید و قبل از اجرا حدس بزنید هر خط چه می‌کند:

</div>

````python
total_minutes = 135
hours = total_minutes // 60
minutes = total_minutes % 60
print(hours, "h", minutes, "min")

n = 2026
print(n % 2 == 0)     # is n even?
print(n % 10)         # last digit of n
print(n // 10)        # n without its last digit
````

<div dir="rtl">

- <code dir="ltr">135 // 60</code> تعداد ساعت‌های **کامل** است (<code dir="ltr">2</code>) و <code dir="ltr">135 % 60</code> دقیقه‌های **باقی‌مانده** (<code dir="ltr">15</code>).
- عددی زوج است که باقی‌مانده‌اش بر ۲ صفر باشد: <code dir="ltr">n % 2 == 0</code>.
- <code dir="ltr">n % 10</code> رقم یکان عدد را می‌دهد و <code dir="ltr">n // 10</code> رقم یکان را حذف می‌کند.

### ❓ پیش‌بینی کن

حاصل هر خط چیست؟ (خط‌های سوم و چهارم را با دقت فکر کنید.)

</div>

````python
print(17 // 5, 17 % 5)
print(17.0 // 5)
print(-7 // 2)
print(-7 % 2)
````

<div dir="rtl">

🔍 نکته‌ها:

- اگر یکی از عملوندها <code dir="ltr">float</code> باشد، <code dir="ltr">//</code> هم <code dir="ltr">float</code> برمی‌گرداند: <code dir="ltr">17.0 // 5</code> برابر <code dir="ltr">3.0</code> است.
- <code dir="ltr">//</code> همیشه به سمت **پایین** (منفی بی‌نهایت) گرد می‌کند، نه به سمت صفر. حاصل <code dir="ltr">-7 / 2</code> برابر <code dir="ltr">-3.5</code> است و پایین‌ترین عدد صحیح بعد از آن <code dir="ltr">-4</code> است، نه <code dir="ltr">-3</code>.
- پس باقی‌مانده باید طوری باشد که رابطه‌ی بالا برقرار بماند: <code dir="ltr">-7 == (-4) &#42; 2 + 1</code>؛ یعنی <code dir="ltr">-7 % 2</code> برابر <code dir="ltr">1</code> است.

### ⚠️ اشتباه رایج

- **<code dir="ltr">^</code> توان نیست!** در پایتون توان با <code dir="ltr">&#42;&#42;</code> نوشته می‌شود. علامت <code dir="ltr">^</code> عملگر دیگری است (XOR بیتی) و <code dir="ltr">2 ^ 3</code> برابر <code dir="ltr">1</code> است، نه <code dir="ltr">8</code>. این خطا پیام خطا نمی‌دهد و فقط جواب غلط می‌دهد؛ پس خطرناک‌تر است.
- **تقسیم بر صفر:** <code dir="ltr">5 / 0</code>، <code dir="ltr">5 // 0</code> و <code dir="ltr">5 % 0</code> همه خطای <code dir="ltr">ZeroDivisionError</code> می‌دهند.

</div>

````python
print(2 ^ 3)
print(2 ** 3)
````

## ۲. عملگرهای مقایسه‌ای

### 💡 مفهوم

عملگرهای مقایسه‌ای دو مقدار را مقایسه می‌کنند و نتیجه‌ی آن‌ها **همیشه** یک <code dir="ltr">bool</code> است (<code dir="ltr">True</code> یا <code dir="ltr">False</code>).

| عملگر | معنی | مثال | نتیجه |
|:---:|---|---|---|
| <code dir="ltr">==</code> | مساوی است | <code dir="ltr">5 == 5</code> | <code dir="ltr">True</code> |
| <code dir="ltr">!=</code> | مساوی نیست | <code dir="ltr">5 != 3</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&gt;</code> | بزرگ‌تر | <code dir="ltr">5 &gt; 3</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;</code> | کوچک‌تر | <code dir="ltr">5 &lt; 3</code> | <code dir="ltr">False</code> |
| <code dir="ltr">&gt;=</code> | بزرگ‌تر یا مساوی | <code dir="ltr">5 &gt;= 5</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;=</code> | کوچک‌تر یا مساوی | <code dir="ltr">3 &lt;= 2</code> | <code dir="ltr">False</code> |

</div>

````python
x = 5
y = 10
print(x == y)
print(x != y)
print(x < y)
````

<div dir="rtl">

### 💡 مقایسه‌ی زنجیره‌ای

در پایتون می‌توان چند مقایسه را مثل ریاضی پشت سر هم نوشت. <code dir="ltr">0 &lt;= score &lt;= 20</code> یعنی <code dir="ltr">0 &lt;= score</code> **و** <code dir="ltr">score &lt;= 20</code>:

</div>

````python
score = 17.5
print(0 <= score <= 20)     # is it a valid grade?

score = 21
print(0 <= score <= 20)
````

<div dir="rtl">

### 💡 مقایسه‌ی رشته‌ها

رشته‌ها را هم می‌توان مقایسه کرد. مقایسه **حرف به حرف** و بر اساس ترتیب الفبایی (دقیق‌تر: کد کاراکترها) انجام می‌شود؛ مثل ترتیب کلمات در فرهنگ لغت. حروف بزرگ انگلیسی **قبل از** حروف کوچک می‌آیند.

### ❓ پیش‌بینی کن

جلوی هر خط بنویسید <code dir="ltr">True</code> یا <code dir="ltr">False</code>. (خط آخر را با دقت فکر کنید.)

</div>

````python
print("apple" < "banana")
print("Ali" == "ali")
print("Z" < "a")
print(10 < 9)
print("10" < "9")
````

<div dir="rtl">

🔍 در خط آخر، دو **رشته** مقایسه می‌شوند، نه دو عدد. مقایسه از حرف اول شروع می‌شود: <code dir="ltr">"1"</code> کوچک‌تر از <code dir="ltr">"9"</code> است، پس <code dir="ltr">"10" &lt; "9"</code> درست است؛ بقیه‌ی حروف اصلاً بررسی نمی‌شوند.

**چرا برای علوم داده مهم است؟** اگر عددها به شکل رشته در داده ذخیره شده باشند (مثلاً از یک فایل CSV)، مرتب‌سازی آن‌ها نتیجه‌ی غلط می‌دهد: <code dir="ltr">"10"</code> قبل از <code dir="ltr">"9"</code> قرار می‌گیرد. قبل از مقایسه، نوع را تبدیل کنید (جلسه‌ی ۲).

### ⚠️ اشتباه رایج

- **<code dir="ltr">=</code> به جای <code dir="ltr">==</code>:** <code dir="ltr">x = 5</code> مقدار می‌دهد؛ <code dir="ltr">x == 5</code> سؤال می‌پرسد.
- **مقایسه‌ی <code dir="ltr">float</code> با <code dir="ltr">==</code>:** همان‌طور که در جلسه‌ی ۲ دیدید، <code dir="ltr">0.1 &#42; 3 == 0.3</code> نادرست است. از <code dir="ltr">math.isclose</code> استفاده کنید.

</div>

## ۳. عملگرهای منطقی

### 💡 مفهوم

عملگرهای منطقی چند شرط را با هم ترکیب می‌کنند:

| عملگر | معنی | وقتی <code dir="ltr">True</code> است که ... |
|:---:|---|---|
| <code dir="ltr">and</code> | و | **هر دو** طرف <code dir="ltr">True</code> باشند |
| <code dir="ltr">or</code> | یا | **حداقل یکی** از دو طرف <code dir="ltr">True</code> باشد |
| <code dir="ltr">not</code> | نقیض | عملوند <code dir="ltr">False</code> باشد (مقدار را برعکس می‌کند) |

جدول درستی:

| <code dir="ltr">A</code> | <code dir="ltr">B</code> | <code dir="ltr">A and B</code> | <code dir="ltr">A or B</code> | <code dir="ltr">not A</code> |
|:---:|:---:|:---:|:---:|:---:|
| <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> |
| <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> |

</div>

````python
score = 14
absences = 4
has_excuse = True

passed = score >= 10 and (absences <= 3 or has_excuse)
print(passed)
````

<div dir="rtl">

### 🔍 کد را بخوان

قانون قبولی: «نمره حداقل ۱۰ باشد **و** (غیبت حداکثر ۳ جلسه باشد **یا** دانشجو عذر موجه داشته باشد)». مرحله به مرحله:

1. <code dir="ltr">score &gt;= 10</code> برابر <code dir="ltr">True</code> است.
2. <code dir="ltr">absences &lt;= 3</code> برابر <code dir="ltr">False</code> است (چهار غیبت).
3. داخل پرانتز: <code dir="ltr">False or has&#95;excuse</code> یعنی <code dir="ltr">False or True</code> که برابر <code dir="ltr">True</code> است.
4. در آخر: <code dir="ltr">True and True</code> که برابر <code dir="ltr">True</code> است.

پرانتز اینجا ضروری است. بدون آن، معنی شرط عوض می‌شود (در بخش ۶ می‌بینید چرا).

### 💡 ارزیابی کوتاه (short-circuit)

پایتون عبارت منطقی را از چپ به راست حساب می‌کند و **به محض اینکه جواب معلوم شد، متوقف می‌شود**:

- در <code dir="ltr">A and B</code> اگر <code dir="ltr">A</code> نادرست باشد، <code dir="ltr">B</code> اصلاً حساب نمی‌شود (جواب حتماً <code dir="ltr">False</code> است).
- در <code dir="ltr">A or B</code> اگر <code dir="ltr">A</code> درست باشد، <code dir="ltr">B</code> اصلاً حساب نمی‌شود (جواب حتماً <code dir="ltr">True</code> است).

این رفتار برای نوشتن شرط‌های **امن** کاربرد دارد:

</div>

````python
count = 0
total = 50
print(count != 0 and total / count > 10)
````

<div dir="rtl">

🔍 چون <code dir="ltr">count != 0</code> نادرست است، پایتون به <code dir="ltr">total / count</code> نمی‌رسد و خطای تقسیم بر صفر رخ نمی‌دهد. اگر ترتیب دو شرط را عوض کنید، <code dir="ltr">ZeroDivisionError</code> می‌گیرید. امتحان کنید!

### ❓ پیش‌بینی کن

کد زیر قرار است بررسی کند که آیا <code dir="ltr">x</code> برابر <code dir="ltr">1</code> یا <code dir="ltr">2</code> است. با <code dir="ltr">x = 5</code> چه چیزی چاپ می‌شود؟

</div>

````python
x = 5
print(x == 1 or 2)
````

<div dir="rtl">

### ⚠️ اشتباه رایج: <code dir="ltr">x == 1 or 2</code>

این یکی از رایج‌ترین باگ‌هاست. پایتون این عبارت را این‌طور می‌خواند: <code dir="ltr">(x == 1) or 2</code>، نه «x برابر 1 یا 2». سمت چپ <code dir="ltr">False</code> است، پس <code dir="ltr">or</code> مقدار سمت راست یعنی <code dir="ltr">2</code> را برمی‌گرداند؛ و <code dir="ltr">2</code> (مثل هر عدد غیرصفر) در شرط‌ها «درست» حساب می‌شود. نتیجه: این شرط **همیشه** برقرار است!

شکل درست:

<div dir="ltr">

```python
x == 1 or x == 2
```

</div>

> **نکته برای آینده:** در NumPy و Pandas برای ترکیب شرط‌ها روی ستون‌های داده از <code dir="ltr">&amp;</code> (به جای <code dir="ltr">and</code>)، <code dir="ltr">|</code> (به جای <code dir="ltr">or</code>) و <code dir="ltr">~</code> (به جای <code dir="ltr">not</code>) استفاده می‌شود. این را در بخش ۳ دوره می‌بینید.

</div>

## ۴. عملگرهای انتساب

### 💡 مفهوم

خیلی وقت‌ها می‌خواهیم مقدار یک متغیر را بر اساس مقدار قبلی‌اش تغییر دهیم. پایتون برای این کار شکل کوتاه‌تری دارد:

| عملگر | مثال | معادل |
|:---:|---|---|
| <code dir="ltr">=</code> | <code dir="ltr">x = 5</code> | — |
| <code dir="ltr">+=</code> | <code dir="ltr">x += 3</code> | <code dir="ltr">x = x + 3</code> |
| <code dir="ltr">-=</code> | <code dir="ltr">x -= 2</code> | <code dir="ltr">x = x - 2</code> |
| <code dir="ltr">&#42;=</code> | <code dir="ltr">x &#42;= 2</code> | <code dir="ltr">x = x &#42; 2</code> |
| <code dir="ltr">/=</code> | <code dir="ltr">x /= 2</code> | <code dir="ltr">x = x / 2</code> |
| <code dir="ltr">//=</code> | <code dir="ltr">x //= 2</code> | <code dir="ltr">x = x // 2</code> |
| <code dir="ltr">%=</code> | <code dir="ltr">x %= 3</code> | <code dir="ltr">x = x % 3</code> |
| <code dir="ltr">&#42;&#42;=</code> | <code dir="ltr">x &#42;&#42;= 2</code> | <code dir="ltr">x = x &#42;&#42; 2</code> |

</div>

````python
balance = 1000
balance += 250     # deposit
balance -= 400     # withdraw
balance *= 1.1     # 10 percent interest
print(balance)
````

<div dir="rtl">

### 🔍 کد را بخوان: دنبال کردن مقدار (trace)

یکی از مهم‌ترین مهارت‌های خواندن کد، نوشتن **جدول دنبال کردن** است: بعد از هر خط، مقدار متغیر را بنویسید.

| خط | مقدار <code dir="ltr">balance</code> | نوع |
|---|---|---|
| <code dir="ltr">balance = 1000</code> | <code dir="ltr">1000</code> | <code dir="ltr">int</code> |
| <code dir="ltr">balance += 250</code> | <code dir="ltr">1250</code> | <code dir="ltr">int</code> |
| <code dir="ltr">balance -= 400</code> | <code dir="ltr">850</code> | <code dir="ltr">int</code> |
| <code dir="ltr">balance &#42;= 1.1</code> | <code dir="ltr">935.0000000000001</code> | <code dir="ltr">float</code> |

دقت کنید که بعد از ضرب در <code dir="ltr">1.1</code> نوع به <code dir="ltr">float</code> تغییر کرد، و چرا آخر عدد یک <code dir="ltr">1</code> اضافه دیده می‌شود؟ (همان مسئله‌ی دقت اعشاری از جلسه‌ی ۲.)

### ❓ پیش‌بینی کن

جدول دنبال کردن را برای کد زیر روی کاغذ بکشید و خروجی را پیش‌بینی کنید.

</div>

````python
x = 10
x += 5
x *= 2
x //= 4
x %= 3
print(x)
````

<div dir="rtl">

### ⚠️ اشتباه رایج

- <code dir="ltr">x =+ 3</code> با <code dir="ltr">x += 3</code> فرق دارد! اولی یعنی <code dir="ltr">x = +3</code> (مقدار ۳ را بده) و خطا هم نمی‌دهد.
- <code dir="ltr">+=</code> روی رشته‌ها هم کار می‌کند و یعنی **چسباندن**: بعد از <code dir="ltr">s = "ab"</code> و <code dir="ltr">s += "c"</code>، مقدار <code dir="ltr">s</code> برابر <code dir="ltr">"abc"</code> است.

</div>

## ۵. عملگر عضویت

### 💡 مفهوم

عملگر <code dir="ltr">in</code> بررسی می‌کند که آیا یک مقدار **داخل** یک مجموعه از مقدارها هست یا نه؛ <code dir="ltr">not in</code> برعکس آن است. نتیجه یک <code dir="ltr">bool</code> است. فعلاً آن را روی رشته‌ها می‌بینیم؛ در جلسه‌های ۸ تا ۱۰ روی لیست، تاپل، مجموعه و دیکشنری هم استفاده می‌شود.

</div>

````python
course = "Python for Data Science"
print("Data" in course)
print("data" in course)
print("R" not in course)
print("" in course)
````

<div dir="rtl">

🔍 <code dir="ltr">in</code> روی رشته‌ها دنبال یک **زیررشته** می‌گردد و به حروف بزرگ و کوچک حساس است؛ برای همین <code dir="ltr">"data"</code> پیدا نمی‌شود. رشته‌ی خالی <code dir="ltr">""</code> جزو هر رشته‌ای حساب می‌شود.

</div>

## ۶. اولویت عملگرها

### 💡 مفهوم

وقتی یک عبارت چند عملگر دارد، پایتون آن‌ها را با یک **ترتیب ثابت** حساب می‌کند؛ مثل ریاضی که ضرب قبل از جمع است. جدول زیر از **بالاترین** اولویت به **پایین‌ترین** است:

| اولویت | عملگر | توضیح |
|:---:|---|---|
| ۱ | <code dir="ltr">( )</code> | پرانتز |
| ۲ | <code dir="ltr">&#42;&#42;</code> | توان (از **راست** به چپ) |
| ۳ | <code dir="ltr">-x</code> | منفی کردن |
| ۴ | <code dir="ltr">&#42;</code>، <code dir="ltr">/</code>، <code dir="ltr">//</code>، <code dir="ltr">%</code> | ضرب و تقسیم (از چپ به راست) |
| ۵ | <code dir="ltr">+</code>، <code dir="ltr">-</code> | جمع و تفریق (از چپ به راست) |
| ۶ | <code dir="ltr">==</code>، <code dir="ltr">!=</code>، <code dir="ltr">&lt;</code>، <code dir="ltr">&gt;</code>، <code dir="ltr">&lt;=</code>، <code dir="ltr">&gt;=</code>، <code dir="ltr">in</code>، <code dir="ltr">not in</code> | مقایسه |
| ۷ | <code dir="ltr">not</code> | نقیض |
| ۸ | <code dir="ltr">and</code> | و |
| ۹ | <code dir="ltr">or</code> | یا |

عملگرهای هم‌اولویت از **چپ به راست** حساب می‌شوند؛ تنها استثنای مهم <code dir="ltr">&#42;&#42;</code> است که از **راست به چپ** حساب می‌شود.

</div>

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(10 / 2 * 5)
print(2 ** 3 ** 2)
print(-3 ** 2)
````

<div dir="rtl">

### 🔍 کد را بخوان

| عبارت | ترتیب محاسبه | نتیجه |
|---|---|---|
| <code dir="ltr">2 + 3 &#42; 4</code> | اول <code dir="ltr">3 &#42; 4</code> | <code dir="ltr">14</code> |
| <code dir="ltr">(2 + 3) &#42; 4</code> | اول پرانتز | <code dir="ltr">20</code> |
| <code dir="ltr">10 / 2 &#42; 5</code> | هم‌اولویت، از چپ: <code dir="ltr">(10 / 2) &#42; 5</code> | <code dir="ltr">25.0</code> |
| <code dir="ltr">2 &#42;&#42; 3 &#42;&#42; 2</code> | از راست: <code dir="ltr">2 &#42;&#42; (3 &#42;&#42; 2)</code> یعنی <code dir="ltr">2 &#42;&#42; 9</code> | <code dir="ltr">512</code> |
| <code dir="ltr">-3 &#42;&#42; 2</code> | توان قبل از منفی: <code dir="ltr">-(3 &#42;&#42; 2)</code> | <code dir="ltr">-9</code> |

### ❓ پیش‌بینی کن

اولویت <code dir="ltr">and</code> از <code dir="ltr">or</code> بیشتر است. حاصل این دو خط چیست؟

</div>

````python
print(True or False and False)
print((True or False) and False)
````

<div dir="rtl">

🔍 در خط اول، <code dir="ltr">and</code> زودتر حساب می‌شود: <code dir="ltr">True or (False and False)</code> یعنی <code dir="ltr">True or False</code> که <code dir="ltr">True</code> است. در خط دوم پرانتز ترتیب را عوض کرده و نتیجه <code dir="ltr">False</code> است.

### ⚠️ توصیه

لازم نیست کل جدول اولویت را حفظ کنید؛ **در کد خودتان از پرانتز استفاده کنید** تا منظورتان برای هر خواننده‌ای روشن باشد. اما برای **خواندن** کد دیگران (و سؤال‌های امتحان!) باید بدانید پایتون یک عبارت بدون پرانتز را چطور حساب می‌کند.

</div>

<div dir="rtl">

## 📌 آنچه باید بلد باشید

- حاصل <code dir="ltr">/</code>، <code dir="ltr">//</code>، <code dir="ltr">%</code> و <code dir="ltr">&#42;&#42;</code> را برای اعداد مثبت و **منفی** پیش‌بینی کنید؛ رابطه‌ی <code dir="ltr">a == (a // b) &#42; b + a % b</code> را بدانید.
- کاربردهای <code dir="ltr">//</code> و <code dir="ltr">%</code>: زوج و فرد بودن، رقم یکان، تبدیل دقیقه به ساعت و دقیقه.
- <code dir="ltr">^</code> توان نیست؛ توان <code dir="ltr">&#42;&#42;</code> است.
- نتیجه‌ی مقایسه‌ها <code dir="ltr">bool</code> است؛ مقایسه‌ی زنجیره‌ای و مقایسه‌ی رشته‌ها (از جمله <code dir="ltr">"10" &lt; "9"</code>) را بفهمید.
- مقدار عبارت‌های ترکیبی با <code dir="ltr">and</code>، <code dir="ltr">or</code> و <code dir="ltr">not</code> را تعیین کنید؛ ارزیابی کوتاه را توضیح دهید.
- باگ <code dir="ltr">x == 1 or 2</code> را تشخیص دهید و درست کنید.
- برای یک دنباله از انتساب‌های ترکیبی (<code dir="ltr">+=</code>، <code dir="ltr">&#42;=</code> و ...) جدول دنبال کردن بنویسید.
- <code dir="ltr">in</code> و <code dir="ltr">not in</code> را روی رشته‌ها به کار ببرید.
- ترتیب محاسبه‌ی یک عبارت بدون پرانتز را بر اساس اولویت عملگرها مشخص کنید.

</div>

## خلاصه‌ی جلسه

- عملگرهای ریاضی: <code dir="ltr">+ - &#42; / // % &#42;&#42;</code>. تقسیم <code dir="ltr">/</code> همیشه <code dir="ltr">float</code> می‌دهد؛ <code dir="ltr">//</code> به پایین گرد می‌کند و <code dir="ltr">%</code> باقی‌مانده‌ی متناظر با آن را می‌دهد.
- مقایسه‌ها <code dir="ltr">bool</code> برمی‌گردانند؛ رشته‌ها حرف به حرف مقایسه می‌شوند.
- <code dir="ltr">and</code>، <code dir="ltr">or</code> و <code dir="ltr">not</code> شرط‌ها را ترکیب می‌کنند؛ ارزیابی از چپ و کوتاه است. مراقب باگ <code dir="ltr">x == 1 or 2</code> باشید.
- <code dir="ltr">x += 1</code> شکل کوتاه <code dir="ltr">x = x + 1</code> است؛ برای خواندن کد، جدول دنبال کردن بنویسید.
- <code dir="ltr">in</code> عضویت را بررسی می‌کند.
- اولویت: پرانتز، توان، ضرب و تقسیم، جمع و تفریق، مقایسه، <code dir="ltr">not</code>، <code dir="ltr">and</code>، <code dir="ltr">or</code>. در کد خودتان پرانتز بگذارید.

**جلسه‌ی بعد:** ورودی و خروجی (<code dir="ltr">input</code>، <code dir="ltr">print</code> و قالب‌بندی با f-string).

</div>
