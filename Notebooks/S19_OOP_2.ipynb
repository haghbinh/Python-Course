## ۱. وراثت و <code dir="ltr">super()</code>

### 💡 مفهوم

در جلسه‌ی ۱۸ کلاس‌ها را از صفر ساختید. اما اغلب کلاس جدید **نوع خاصی** از یک کلاس موجود است: «دانشجو **یک** شخص است»، «لایه‌ی Dense **یک** لایه است». به این رابطه **is-a** می‌گویند و ابزارش **وراثت** (inheritance) است:

- کلاس پایه (**base class** یا **superclass**): کلاس عمومی‌تر، مثلاً <code dir="ltr">Person</code>.
- کلاس فرزند (**subclass** یا **derived class**): کلاس خاص‌تر که همه‌ی ویژگی‌ها و متدهای پایه را **به ارث می‌برد** و می‌تواند چیزهایی **اضافه** یا **عوض** کند.

نحو آن ساده است: نام کلاس پایه را داخل پرانتز بعد از نام کلاس فرزند می‌نویسید: <code dir="ltr">class Student(Person):</code>.

</div>

<p align="center"><img src="images/s19_inheritance.png" width="420" alt="s19_inheritance.png"></p>

<p align="center" dir="rtl"><i>کلاس فرزند همه‌ی ویژگی‌ها و متدهای کلاس پایه را دارد و می‌تواند متدهای خودش را هم اضافه کند.</i></p>

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"I am {self.name}, {self.age} years old."


class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)        # let Person set name and age
        self.student_id = student_id       # the new attribute

    def is_freshman(self):
        return self.student_id.startswith("405")


s = Student("Sara", 19, "4051234")
print(s.introduce())            # inherited from Person
print(s.is_freshman())          # defined in Student
print(s.name, s.student_id)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">Student</code> متد <code dir="ltr">introduce</code> را تعریف نکرده، اما <code dir="ltr">s.introduce()</code> کار می‌کند: پایتون اول در کلاس <code dir="ltr">Student</code> دنبال متد می‌گردد و چون پیدا نمی‌کند، به کلاس پایه <code dir="ltr">Person</code> می‌رود.
- <code dir="ltr">super().&#95;&#95;init&#95;&#95;(name, age)</code> سازنده‌ی کلاس پایه را صدا می‌زند تا <code dir="ltr">name</code> و <code dir="ltr">age</code> را تنظیم کند. بعد <code dir="ltr">Student</code> فقط کار **اضافه‌ی** خودش را انجام می‌دهد. به این ترتیب کد تکرار نمی‌شود.
- <code dir="ltr">super()</code> یعنی «کلاس بعدی در زنجیره‌ی وراثت»؛ لازم نیست نام <code dir="ltr">Person</code> را صریحاً بنویسید.

### ⚠️ اشتباه رایج: فراموش کردن <code dir="ltr">super().&#95;&#95;init&#95;&#95;</code>

</div>

````python
class BadStudent(Person):
    def __init__(self, name, age, student_id):
        self.student_id = student_id       # forgot super().__init__(...)

b = BadStudent("Reza", 20, "4041111")
print(b.student_id)
print(b.introduce())
````

<div dir="rtl">

🔍 وقتی کلاس فرزند <code dir="ltr">&#95;&#95;init&#95;&#95;</code> خودش را دارد، <code dir="ltr">&#95;&#95;init&#95;&#95;</code> پایه **خودبه‌خود** اجرا نمی‌شود. پس <code dir="ltr">self.name</code> هرگز ساخته نشده و <code dir="ltr">introduce</code> خطای <code dir="ltr">AttributeError</code> می‌دهد. خطا هم در خط <code dir="ltr">introduce</code> ظاهر می‌شود، نه جایی که اشتباه رخ داده است؛ برای همین پیدا کردنش سخت است.

### 💡 <code dir="ltr">isinstance</code> و <code dir="ltr">issubclass</code>

چون هر <code dir="ltr">Student</code> یک <code dir="ltr">Person</code> هم هست، <code dir="ltr">isinstance</code> برای کلاس پایه هم <code dir="ltr">True</code> برمی‌گرداند:

</div>

````python
p = Person("Ali", 40)
print(isinstance(s, Student), isinstance(s, Person))
print(isinstance(p, Student))
print(issubclass(Student, Person), issubclass(Person, Student))
print(type(s) == Person)
print(Student.__mro__)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">isinstance(s, Person)</code> درست است (دانشجو یک شخص است)، اما <code dir="ltr">isinstance(p, Student)</code> نادرست است (هر شخصی دانشجو نیست).
- <code dir="ltr">type(s) == Person</code> نادرست است، چون <code dir="ltr">type</code> فقط کلاس **دقیق** را برمی‌گرداند. برای بررسی نوع تقریباً همیشه <code dir="ltr">isinstance</code> درست است، چون فرزندها را هم در نظر می‌گیرد.
- <code dir="ltr">&#95;&#95;mro&#95;&#95;</code> (**Method Resolution Order**) ترتیبی است که پایتون برای پیدا کردن یک متد یا ویژگی کلاس‌ها را می‌گردد: <code dir="ltr">Student</code>، بعد <code dir="ltr">Person</code>، و در انتها <code dir="ltr">object</code>. **همه‌ی** کلاس‌های پایتون در نهایت از <code dir="ltr">object</code> ارث می‌برند؛ متدهایی مثل <code dir="ltr">&#95;&#95;str&#95;&#95;</code> پیش‌فرض از همین‌جا می‌آیند.

### ❓ پیش‌بینی کن

خروجی کد زیر چیست؟ (دنبال کنید هر <code dir="ltr">self.describe()</code> کدام متد را صدا می‌زند.)

</div>

````python
class A:
    def describe(self):
        return "A"

    def show(self):
        return "show " + self.describe()


class B(A):
    def describe(self):
        return "B"


print(A().show())
print(B().show())
````

<div dir="rtl">

🔍 <code dir="ltr">show</code> فقط در <code dir="ltr">A</code> تعریف شده، اما <code dir="ltr">self</code> در فراخوانی دوم یک شیء <code dir="ltr">B</code> است. پس <code dir="ltr">self.describe()</code> اول در <code dir="ltr">B</code> جست‌وجو می‌شود و <code dir="ltr">"B"</code> برمی‌گرداند: خروجی <code dir="ltr">show A</code> و بعد <code dir="ltr">show B</code>. این نکته‌ی بسیار مهمی است: **کلاس پایه می‌تواند متدی را صدا بزند که فرزند تعریف کرده است.** چارچوب‌هایی مثل scikit-learn دقیقاً بر همین اساس ساخته شده‌اند.

## ۲. بازنویسی متد و چندریختی

### 💡 مفهوم

اگر کلاس فرزند متدی با **همان نام** متد پایه تعریف کند، نسخه‌ی فرزند **جایگزین** آن می‌شود (**override**). دو حالت دارد:

- **جایگزین کردن کامل**: متد فرزند کاری کاملاً متفاوت انجام می‌دهد.
- **گسترش دادن**: متد فرزند اول با <code dir="ltr">super().method()</code> نسخه‌ی پایه را اجرا می‌کند و بعد چیزی به آن **اضافه** می‌کند.

</div>

````python
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def introduce(self):                                   # extend
        base = super().introduce()
        return base + f" My student ID is {self.student_id}."


class Professor(Person):
    def __init__(self, name, age, field):
        super().__init__(name, age)
        self.field = field

    def introduce(self):                                   # replace
        return f"Dr. {self.name}, {self.field}."


people = [Person("Ali", 40), Student("Sara", 19, "4051234"), Professor("Haghbin", 45, "Statistics")]
for person in people:
    print(person.introduce())
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">Student.introduce</code> متد پایه را **گسترش** می‌دهد: جمله‌ی کلاس پایه را می‌گیرد و چیزی به آن اضافه می‌کند.
- <code dir="ltr">Professor.introduce</code> متد پایه را کاملاً **جایگزین** می‌کند.
- حلقه‌ی <code dir="ltr">for</code> نمی‌داند و لازم هم ندارد بداند هر عنصر از کدام کلاس است؛ فقط <code dir="ltr">person.introduce()</code> را صدا می‌زند و هر شیء نسخه‌ی **خودش** را اجرا می‌کند. به این **چندریختی** (polymorphism) می‌گویند: یک فراخوانی یکسان، رفتارهای متفاوت.

### 💡 duck typing

در پایتون چندریختی حتی به وراثت نیاز ندارد: هر شیئی که متد <code dir="ltr">introduce</code> داشته باشد در آن حلقه کار می‌کند. به این **duck typing** می‌گویند: «اگر مثل اردک راه می‌رود و مثل اردک صدا می‌دهد، اردک است.» برای همین <code dir="ltr">len</code> روی لیست، رشته، دیکشنری و مجموعه کار می‌کند؛ همه‌ی آن‌ها متد <code dir="ltr">&#95;&#95;len&#95;&#95;</code> دارند (بخش ۳).

### ❓ پیش‌بینی کن

یک کلاس پایه‌ی ساده برای «مدل» و دو فرزند آن. خروجی را پیش‌بینی کنید.

</div>

````python
class Model:
    def predict(self, x):
        return 0

    def predict_all(self, xs):
        return [self.predict(x) for x in xs]


class Doubler(Model):
    def predict(self, x):
        return 2 * x


class Shifter(Doubler):
    def predict(self, x):
        return super().predict(x) + 1


for m in [Model(), Doubler(), Shifter()]:
    print(type(m).__name__, m.predict_all([1, 2, 3]))
````

<div dir="rtl">

🔍 <code dir="ltr">predict&#95;all</code> فقط یک بار در <code dir="ltr">Model</code> نوشته شده و برای هر سه کلاس کار می‌کند، چون <code dir="ltr">self.predict</code> هر بار نسخه‌ی کلاس واقعی شیء را صدا می‌زند. در <code dir="ltr">Shifter</code>، <code dir="ltr">super().predict(x)</code> نسخه‌ی <code dir="ltr">Doubler</code> را اجرا می‌کند (کلاس بعدی در MRO)، نه نسخه‌ی <code dir="ltr">Model</code> را: خروجی <code dir="ltr">[3, 5, 7]</code>.

### ⚠️ اشتباه رایج: وراثت برای چیزی که is-a نیست

اگر جمله‌ی «X یک Y است» درست نیست، وراثت ابزار مناسبی نیست. «ماشین یک موتور است» غلط است؛ ماشین یک موتور **دارد** (رابطه‌ی **has-a**). در این حالت موتور باید یک **ویژگی** ماشین باشد (<code dir="ltr">self.engine = Engine()</code>)، نه کلاس پایه‌ی آن. به این **ترکیب** (composition) می‌گویند (بخش ۵).

</div>

## ۳. متدهای ویژه (۱): نمایش، طول، اندیس و مقایسه

### 💡 مفهوم

در جلسه‌ی ۱۸ دیدید که <code dir="ltr">print</code> یک شیء چیزی شبیه <code dir="ltr">&lt;&#95;&#95;main&#95;&#95;.Point object at 0x7f...&gt;</code> چاپ می‌کند. علت این است که پایتون برای <code dir="ltr">print(obj)</code> در واقع متد <code dir="ltr">obj.&#95;&#95;str&#95;&#95;()</code> را صدا می‌زند و نسخه‌ی پیش‌فرض آن (از کلاس <code dir="ltr">object</code>) همین متن نامفهوم را برمی‌گرداند.

متدهایی که نامشان با دو زیرخط شروع و تمام می‌شود (**متدهای ویژه**، **dunder methods** یا **magic methods**) را خودتان صدا نمی‌زنید؛ **پایتون** وقتی از یک عملگر یا تابع داخلی استفاده می‌کنید آن‌ها را صدا می‌زند:

| شما می‌نویسید | پایتون صدا می‌زند |
|---|---|
| <code dir="ltr">print(x)</code> یا <code dir="ltr">str(x)</code> | <code dir="ltr">x.&#95;&#95;str&#95;&#95;()</code> |
| نمایش در Jupyter یا داخل لیست، <code dir="ltr">repr(x)</code> | <code dir="ltr">x.&#95;&#95;repr&#95;&#95;()</code> |
| <code dir="ltr">len(x)</code> | <code dir="ltr">x.&#95;&#95;len&#95;&#95;()</code> |
| <code dir="ltr">x[i]</code> | <code dir="ltr">x.&#95;&#95;getitem&#95;&#95;(i)</code> |
| <code dir="ltr">item in x</code> | <code dir="ltr">x.&#95;&#95;contains&#95;&#95;(item)</code> |
| <code dir="ltr">for item in x</code> | <code dir="ltr">x.&#95;&#95;iter&#95;&#95;()</code> |
| <code dir="ltr">x == y</code> | <code dir="ltr">x.&#95;&#95;eq&#95;&#95;(y)</code> |
| <code dir="ltr">x &lt; y</code> (و <code dir="ltr">sorted</code>) | <code dir="ltr">x.&#95;&#95;lt&#95;&#95;(y)</code> |

</div>

````python
class Dataset:
    def __init__(self, name, samples):
        self.name = name
        self.samples = list(samples)

    def __str__(self):
        return f"Dataset '{self.name}' with {len(self.samples)} samples"

    def __repr__(self):
        return f"Dataset({self.name!r}, {self.samples!r})"

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, i):
        return self.samples[i]

    def __contains__(self, item):
        return item in self.samples

    def __iter__(self):
        for x in self.samples:
            yield x


d = Dataset("scores", [12, 17.5, 19, 8])
print(d)
print([d])
print(len(d), d[0], d[-1], d[1:3])
print(19 in d, 20 in d)
print(sum(d), max(d), sorted(d))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">print(d)</code> از <code dir="ltr">&#95;&#95;str&#95;&#95;</code> استفاده می‌کند: یک متن **خوانا برای انسان**.
- <code dir="ltr">print([d])</code> لیستی را چاپ می‌کند که <code dir="ltr">d</code> داخل آن است؛ لیست برای نمایش اعضایش از <code dir="ltr">&#95;&#95;repr&#95;&#95;</code> استفاده می‌کند: یک متن **دقیق برای برنامه‌نویس**، ترجیحاً به شکلی که بتوان با آن دوباره همان شیء را ساخت. <code dir="ltr">!r</code> در f-string یعنی <code dir="ltr">repr</code> آن مقدار (برای همین <code dir="ltr">'scores'</code> با نقل‌قول چاپ شده است).
- چون <code dir="ltr">&#95;&#95;getitem&#95;&#95;</code> اندیس را مستقیماً به لیست درونی می‌دهد، **برش** (<code dir="ltr">d[1:3]</code>) و اندیس منفی هم مجانی کار می‌کنند.
- <code dir="ltr">&#95;&#95;iter&#95;&#95;</code> با <code dir="ltr">yield</code> یک generator برمی‌گرداند (جلسه‌ی ۱۴). حالا <code dir="ltr">Dataset</code> یک **iterable** است و همه‌ی ابزارهایی که روی iterable کار می‌کنند (<code dir="ltr">for</code>، <code dir="ltr">sum</code>، <code dir="ltr">max</code>، <code dir="ltr">sorted</code>، comprehension) روی آن هم کار می‌کنند. این همان duck typing است.

> **قاعده:** اگر فقط یکی را می‌نویسید، <code dir="ltr">&#95;&#95;repr&#95;&#95;</code> را بنویسید؛ وقتی <code dir="ltr">&#95;&#95;str&#95;&#95;</code> تعریف نشده باشد، <code dir="ltr">print</code> هم از <code dir="ltr">&#95;&#95;repr&#95;&#95;</code> استفاده می‌کند.

### 💡 برابری و ترتیب: <code dir="ltr">&#95;&#95;eq&#95;&#95;</code> و <code dir="ltr">&#95;&#95;lt&#95;&#95;</code>

به‌طور پیش‌فرض <code dir="ltr">==</code> برای دو شیء همان <code dir="ltr">is</code> است: فقط وقتی <code dir="ltr">True</code> است که هر دو **یک شیء** باشند. برای مقایسه‌ی **محتوا** باید <code dir="ltr">&#95;&#95;eq&#95;&#95;</code> را تعریف کنید. با <code dir="ltr">&#95;&#95;lt&#95;&#95;</code> هم اشیای شما قابل مرتب‌سازی می‌شوند.

</div>

<p align="center"><img src="images/s19_compare_dunder.png" width="620" alt="s19_compare_dunder.png"></p>

<p align="center" dir="rtl"><i>عملگرهای مقایسه و متد ویژه‌ی متناظر هر کدام.</i></p>

````python
class Result:
    def __init__(self, model, accuracy):
        self.model = model
        self.accuracy = accuracy

    def __repr__(self):
        return f"Result({self.model!r}, {self.accuracy})"

    def __eq__(self, other):
        return self.model == other.model and self.accuracy == other.accuracy

    def __lt__(self, other):
        return self.accuracy < other.accuracy


r1 = Result("knn", 0.91)
r2 = Result("knn", 0.91)
print(r1 == r2, r1 is r2)

results = [Result("tree", 0.87), Result("knn", 0.91), Result("logreg", 0.89)]
print(sorted(results))
print(max(results))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">r1 == r2</code> حالا <code dir="ltr">True</code> است (محتوای برابر)، اما <code dir="ltr">r1 is r2</code> همچنان <code dir="ltr">False</code> است (دو شیء جدا در حافظه؛ جلسه‌ی ۸).
- <code dir="ltr">sorted</code> و <code dir="ltr">max</code> فقط به <code dir="ltr">&lt;</code> نیاز دارند؛ چون <code dir="ltr">&#95;&#95;lt&#95;&#95;</code> را تعریف کرده‌ایم، نتایج بر اساس دقت مرتب می‌شوند. ## ۴. متدهای ویژه (۲): عملگرها و <code dir="ltr">&#95;&#95;call&#95;&#95;</code>

### 💡 مفهوم

عملگرهای ریاضی هم متد ویژه دارند. با تعریف آن‌ها می‌توانید بنویسید <code dir="ltr">v + w</code> به جای <code dir="ltr">v.add(w)</code>؛ کدی که بسیار شبیه فرمول‌های ریاضی است. NumPy و PyTorch دقیقاً همین کار را می‌کنند.

</div>

<p align="center"><img src="images/s19_arith_dunder.png" width="700" alt="s19_arith_dunder.png"></p>

<p align="center" dir="rtl"><i>عملگرهای ریاضی و متد ویژه‌ی متناظر هر کدام.</i></p>

````python
class Vector:
    def __init__(self, *values):
        self.values = list(values)

    def __repr__(self):
        return f"Vector{tuple(self.values)}"

    def __len__(self):
        return len(self.values)

    def __add__(self, other):
        return Vector(*[a + b for a, b in zip(self.values, other.values)])

    def __mul__(self, k):
        return Vector(*[k * a for a in self.values])

    def __rmul__(self, k):
        return self * k


v = Vector(1, 2, 3)
w = Vector(10, 20, 30)
print(v + w)
print(v * 2)
print(2 * v)
print(v + w * 0.5)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">v + w</code> به <code dir="ltr">v.&#95;&#95;add&#95;&#95;(w)</code> تبدیل می‌شود و یک <code dir="ltr">Vector</code> **جدید** برمی‌گرداند؛ <code dir="ltr">v</code> و <code dir="ltr">w</code> تغییر نمی‌کنند (مثل اعداد).
- <code dir="ltr">v &#42; 2</code> یعنی <code dir="ltr">v.&#95;&#95;mul&#95;&#95;(2)</code>. اما <code dir="ltr">2 &#42; v</code> اول <code dir="ltr">int.&#95;&#95;mul&#95;&#95;(2, v)</code> را امتحان می‌کند؛ <code dir="ltr">int</code> نمی‌داند با <code dir="ltr">Vector</code> چه کند و <code dir="ltr">NotImplemented</code> برمی‌گرداند. آن‌وقت پایتون نسخه‌ی **وارونه** را روی عملوند راست امتحان می‌کند: <code dir="ltr">v.&#95;&#95;rmul&#95;&#95;(2)</code>. بدون <code dir="ltr">&#95;&#95;rmul&#95;&#95;</code>، عبارت <code dir="ltr">2 &#42; v</code> خطای <code dir="ltr">TypeError</code> می‌داد.
- اولویت عملگرها حفظ می‌شود: در <code dir="ltr">v + w &#42; 0.5</code> اول ضرب انجام می‌شود (جلسه‌ی ۳).
- <code dir="ltr">&#42;values</code> در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> یعنی هر تعداد آرگومان (جلسه‌ی ۷)؛ و <code dir="ltr">&#42;[...]</code> موقع فراخوانی، لیست را باز می‌کند.

### 💡 <code dir="ltr">&#95;&#95;call&#95;&#95;</code>: شیئی که مثل تابع صدا زده می‌شود

اگر کلاسی <code dir="ltr">&#95;&#95;call&#95;&#95;</code> داشته باشد، اشیای آن را می‌توان مثل تابع با پرانتز صدا زد. این الگو برای اشیایی که **وضعیت** (پارامتر) دارند و یک **محاسبه** انجام می‌دهند بسیار رایج است؛ در PyTorch هر لایه را این‌طور صدا می‌زنید: <code dir="ltr">output = layer(x)</code>.

</div>

````python
class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def __call__(self, inputs):
        z = sum(w * x for w, x in zip(self.weights, inputs)) + self.bias
        return max(0, z)                       # ReLU activation


n = Neuron([0.5, -1.0], 0.25)
print(n([2, 1]))
print(n([1, 2]))
print(callable(n), callable(n.bias))
````

<div dir="rtl">

🔍 <code dir="ltr">n([2, 1])</code> یعنی <code dir="ltr">n.&#95;&#95;call&#95;&#95;([2, 1])</code>: <code dir="ltr">0.5×2 - 1×1 + 0.25 = 0.25</code>. در فراخوانی دوم <code dir="ltr">z = 0.5 - 2 + 0.25 = -1.25</code> است و ReLU آن را <code dir="ltr">0</code> می‌کند. <code dir="ltr">callable</code> بررسی می‌کند که آیا چیزی را می‌توان صدا زد؛ این همان نورون جلسه‌ی ۱۸ است، با این تفاوت که حالا مثل یک تابع ریاضی <code dir="ltr">f(x)</code> استفاده می‌شود.

### ⚠️ اشتباه رایج: تغییر دادن <code dir="ltr">self</code> در عملگر

<code dir="ltr">&#95;&#95;add&#95;&#95;</code> باید یک شیء **جدید** برگرداند. اگر <code dir="ltr">self.values</code> را تغییر دهید و <code dir="ltr">self</code> را برگردانید، بعد از <code dir="ltr">u = v + w</code> متغیر <code dir="ltr">v</code> هم عوض شده است؛ رفتاری که هیچ‌کس از <code dir="ltr">+</code> انتظار ندارد. (برای <code dir="ltr">+=</code> متد جدای <code dir="ltr">&#95;&#95;iadd&#95;&#95;</code> وجود دارد.)

## ۵. استثنای سفارشی و ترکیب در برابر وراثت

### 💡 استثنای سفارشی

در جلسه‌ی ۱۵ با <code dir="ltr">raise ValueError(...)</code> خطا ایجاد کردید. خود خطاها هم **کلاس** هستند و در یک سلسله‌مراتب وراثتی قرار دارند: <code dir="ltr">ValueError</code> و <code dir="ltr">KeyError</code> و بقیه، همه از <code dir="ltr">Exception</code> ارث می‌برند. پس می‌توانید خطاهای مخصوص برنامه‌ی خودتان را با وراثت از <code dir="ltr">Exception</code> تعریف کنید:

</div>

````python
class ModelError(Exception):
    pass


class NotFittedError(ModelError):
    pass


class MeanModel:
    def __init__(self):
        self.mean = None

    def fit(self, ys):
        self.mean = sum(ys) / len(ys)
        return self

    def predict(self, n):
        if self.mean is None:
            raise NotFittedError("call fit() before predict()")
        return [self.mean] * n


m = MeanModel()
try:
    m.predict(3)
except ModelError as e:
    print(type(e).__name__, "->", e)

print(m.fit([10, 20, 30]).predict(3))
print(issubclass(NotFittedError, Exception))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">class NotFittedError(ModelError): pass</code> یک کلاس خطای جدید است که هیچ چیز اضافه‌ای ندارد؛ همین **نام** معنادار، ارزشش را دارد. scikit-learn دقیقاً خطایی به همین نام دارد.
- <code dir="ltr">except ModelError</code> خطای <code dir="ltr">NotFittedError</code> را هم می‌گیرد، چون <code dir="ltr">NotFittedError</code> **یک** <code dir="ltr">ModelError</code> است (همان قاعده‌ی <code dir="ltr">isinstance</code>). به این ترتیب کاربر می‌تواند همه‌ی خطاهای کتابخانه‌ی شما را یک‌جا بگیرد، یا فقط یک نوع خاص را.
- <code dir="ltr">fit</code> مقدار <code dir="ltr">self</code> را برمی‌گرداند تا بتوان <code dir="ltr">m.fit(...).predict(...)</code> نوشت (زنجیره‌سازی، جلسه‌ی ۱۸).

### 💡 ترکیب در برابر وراثت

| | وراثت (is-a) | ترکیب (has-a) |
|---|---|---|
| رابطه | <code dir="ltr">Dense</code> **یک** <code dir="ltr">Layer</code> است | شبکه **چند** لایه **دارد** |

| مثال | <code dir="ltr">Student(Person)</code>، <code dir="ltr">NotFittedError(ModelError)</code> | <code dir="ltr">Sequential</code> که لیستی از لایه‌ها را نگه می‌دارد |

قاعده‌ی عملی: وراثت را برای رابطه‌های واقعی «نوعی از» نگه دارید و در بقیه‌ی موارد ترکیب را ترجیح دهید؛ زنجیره‌های طولانی وراثت خواندن و تغییر کد را سخت می‌کنند.

### ⚠️ اشتباه رایج: ترتیب <code dir="ltr">except</code>ها

اگر <code dir="ltr">except ModelError</code> را **قبل از** <code dir="ltr">except NotFittedError</code> بنویسید، شاخه‌ی دوم هرگز اجرا نمی‌شود؛ پایتون اولین شاخه‌ای را که با نوع خطا **سازگار** باشد انتخاب می‌کند. همیشه خطاهای **خاص‌تر** را اول بنویسید.

### سؤال ۳ — شبکه‌ی عصبی کوچک با وراثت و ترکیب

ساختار زیر نسخه‌ی خیلی ساده‌شده‌ی PyTorch است: همه‌ی لایه‌ها از <code dir="ltr">Layer</code> ارث می‌برند و <code dir="ltr">Sequential</code> چند لایه را **نگه می‌دارد**.

```python
import math

class Layer:
    def forward(self, x):
        raise NotImplementedError

    def __call__(self, x):
        return self.forward(x)

    def n_params(self):
        return 0


class Dense(Layer):
    def __init__(self, weights, biases):        # weights: one row per output
        self.weights = weights
        self.biases = biases

    def forward(self, x):
        return [sum(w * xi for w, xi in zip(row, x)) + b
                for row, b in zip(self.weights, self.biases)]

    def n_params(self):
        return sum(len(row) for row in self.weights) + len(self.biases)


class ReLU(Layer):
    def forward(self, x):
        return [max(0, v) for v in x]


class Sigmoid(Layer):
    def forward(self, x):
        return [1 / (1 + math.exp(-v)) for v in x]


class Sequential(Layer):
    def __init__(self, *layers):
        self.layers = list(layers)

    def forward(self, x):
        for layer in self.layers:
            x = layer(x)
        return x

    def __len__(self):
        return len(self.layers)

    def __iter__(self):
        return iter(self.layers)

    def n_params(self):
        return sum(layer.n_params() for layer in self)


net = Sequential(Dense([[1, -1], [2, 0]], [0, -3]), ReLU(), Dense([[1, 1]], [-1]), Sigmoid())
print(len(net), net.n_params())
print(net([2, 1]))
```

class NotFittedError(Exception):
    pass


class BaseRegressor:
    def __init__(self):
        self.fitted = False

    def fit(self, xs, ys):
        self._fit(xs, ys)
        self.fitted = True
        return self

    def predict(self, xs):
        if not self.fitted:
            raise NotFittedError(f"{type(self).__name__} is not fitted")
        return [self._predict_one(x) for x in xs]

    def score(self, xs, ys):                     # mean squared error
        preds = self.predict(xs)
        return sum((p - y) ** 2 for p, y in zip(preds, ys)) / len(ys)


class MeanRegressor(BaseRegressor):
    def _fit(self, xs, ys):
        self.mean = sum(ys) / len(ys)

    def _predict_one(self, x):
        return self.mean


class LinearRegressor(BaseRegressor):
    def _fit(self, xs, ys):
        mx, my = sum(xs) / len(xs), sum(ys) / len(ys)
        sxy = sum((x - mx) * (y - my) for x, y in zip(xs, ys))
        sxx = sum((x - mx) ** 2 for x in xs)
        self.slope = sxy / sxx
        self.intercept = my - self.slope * mx

    def _predict_one(self, x):
        return self.slope * x + self.intercept


xs, ys = [1, 2, 3, 4], [3, 5, 7, 10]
for model in [MeanRegressor(), LinearRegressor()]:
    print(type(model).__name__, round(model.fit(xs, ys).score(xs, ys), 3))

class Trial:
    def __init__(self, params, val_loss):

### سؤال ۷ — سلسله‌مراتب خطاها

```python
class DataError(Exception):
    pass

class MissingValueError(DataError):
    pass

class ShapeError(DataError):
    def __init__(self, expected, got):
        super().__init__(f"expected {expected} columns, got {got}")
        self.expected = expected
        self.got = got


def check_row(row, n_cols):
    if len(row) != n_cols:
        raise ShapeError(n_cols, len(row))
    if None in row:
        raise MissingValueError(f"missing value in {row}")
    return True


rows = [[1, 2, 3], [4, None, 6], [7, 8], "abc", [1, 2, 3, 4]]
for row in rows:
    try:
        check_row(row, 3)
        print("ok")
    except ShapeError as e:
        print("shape:", e.got)
    except DataError as e:
        print("data:", e)
    except Exception as e:
        print("other:", type(e).__name__)
```

ب) اگر ترتیب <code dir="ltr">except ShapeError</code> و <code dir="ltr">except DataError</code> را عوض کنیم، کدام خط‌های خروجی تغییر می‌کنند؟

## خلاصه‌ی جلسه

- وراثت رابطه‌ی is-a را مدل می‌کند: فرزند همه‌چیز را از پایه به ارث می‌برد؛ <code dir="ltr">super()</code> نسخه‌ی پایه را صدا می‌زند.
- بازنویسی متد و چندریختی: کلاس پایه می‌تواند متدی را صدا بزند که فرزند پیاده کرده است (Template Method)؛ این پایه‌ی معماری scikit-learn و PyTorch است.
- متدهای ویژه کاری می‌کنند که اشیای شما با <code dir="ltr">print</code>، <code dir="ltr">len</code>، <code dir="ltr">[]</code>، <code dir="ltr">in</code>، <code dir="ltr">for</code>، <code dir="ltr">==</code>، <code dir="ltr">&lt;</code>، <code dir="ltr">+</code>، <code dir="ltr">&#42;</code> و <code dir="ltr">()</code> کار کنند.
- خطاها هم کلاس‌اند؛ با وراثت از <code dir="ltr">Exception</code> خطاهای معنادار و سلسله‌مراتبی بسازید و خاص‌ترها را اول بگیرید.
- وراثت برای «نوعی از»، ترکیب برای «دارای».
