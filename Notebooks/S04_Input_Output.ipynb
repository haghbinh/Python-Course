## ۱. تابع <code dir="ltr">print()</code>

### 💡 مفهوم

<code dir="ltr">print</code> مقدارهایی را که به آن می‌دهید به **متن** تبدیل و روی خروجی نمایش می‌دهد. در جلسه‌ی ۱ دیدید که می‌تواند چند مقدار بگیرد و با <code dir="ltr">sep</code> جداکننده‌ی آن‌ها را عوض کرد. پارامتر مهم دیگر **<code dir="ltr">end</code>** است: چیزی که **بعد از** آخرین مقدار چاپ می‌شود. مقدار پیش‌فرض <code dir="ltr">end</code> یک **خط جدید** (<code dir="ltr">"&#92;n"</code>) است؛ برای همین هر <code dir="ltr">print</code> در یک خط جدا چاپ می‌شود.

| پارامتر | معنی | مقدار پیش‌فرض |
|---|---|---|
| <code dir="ltr">sep</code> | جداکننده‌ی بین مقدارها | <code dir="ltr">" "</code> (یک فاصله) |
| <code dir="ltr">end</code> | آنچه در انتها چاپ می‌شود | <code dir="ltr">"&#92;n"</code> (خط جدید) |

</div>

````python
a = 5
b = 10
print("a + b =", a + b)

print("Loading", end="")
print("...", end="")
print("done")

print("2026", "10", "30", sep="/")
print()                      # an empty line
print("end of output")
````

<div dir="rtl">

### 🔍 کد را بخوان

- دو <code dir="ltr">print</code> با <code dir="ltr">end=""</code> خط جدید چاپ نمی‌کنند؛ پس <code dir="ltr">Loading</code>، <code dir="ltr">...</code> و <code dir="ltr">done</code> در **یک خط** کنار هم قرار می‌گیرند.
- <code dir="ltr">print()</code> بدون ورودی فقط همان <code dir="ltr">end</code> را چاپ می‌کند؛ یعنی یک خط خالی.

### 💡 کاراکترهای ویژه

داخل رشته، علامت <code dir="ltr">&#92;</code> (backslash) شروع یک **کاراکتر ویژه** است:

| نوشتن | معنی |
|---|---|
| <code dir="ltr">&#92;n</code> | خط جدید |
| <code dir="ltr">&#92;t</code> | فاصله‌ی جدول (tab) |
| <code dir="ltr">&#92;&#92;</code> | خود علامت <code dir="ltr">&#92;</code> |
| <code dir="ltr">&#92;"</code> | علامت <code dir="ltr">"</code> داخل رشته‌ای که با <code dir="ltr">"</code> شروع شده |

</div>

````python
print("Name:\tAli\nScore:\t17")
print("He said \"hello\"")
print("C:\\Users\\Ali")
````

<div dir="rtl">

### 💡 چاپ کردن یا نمایش دادن؟

</div>

````python
"Hello"
````

</div>

## ۲. تابع <code dir="ltr">input()</code>

### 💡 مفهوم

<code dir="ltr">input</code> برنامه را **متوقف** می‌کند، متنی را (اگر داده باشید) به‌عنوان **پیام راهنما** نشان می‌دهد و منتظر می‌ماند تا کاربر چیزی تایپ کند و Enter بزند. نکته‌ی کلیدی:

> **<code dir="ltr">input</code> همیشه یک رشته (<code dir="ltr">str</code>) برمی‌گرداند؛ حتی اگر کاربر عدد تایپ کند.**

### 💡 اجرای <code dir="ltr">input</code> در نوت‌بوک

- در **VS Code** یک کادر ورود متن در **بالای پنجره** باز می‌شود.
- در **Jupyter** یک کادر ورود متن **زیر سلول** ظاهر می‌شود.

تا زمانی که چیزی وارد نکنید، سلول در حال اجرا می‌ماند (علامت <code dir="ltr">[&#42;]</code>) و **هیچ سلول دیگری اجرا نمی‌شود**. اگر گیر کردید، Enter بزنید یا اجرای کرنل را متوقف کنید (Interrupt).

سلول زیر را اجرا کنید و نام خود را (به انگلیسی) وارد کنید:

</div>

````python
name = input("Please enter your name: ")
print("Hello,", name)
print(type(name))
````

<div dir="rtl">

### 🔍 کد را بخوان: عدد گرفتن از کاربر

اگر عدد لازم دارید، باید رشته را خودتان **تبدیل** کنید (جلسه‌ی ۲). کد زیر را بخوانید و بعد اجرا کنید:

</div>

````python
age = int(input("Enter your age: "))
print("Next year you will be", age + 1)
````

<div dir="rtl">

خط اول را از **داخل به بیرون** بخوانید:

1. <code dir="ltr">input(...)</code> متن تایپ‌شده را به‌عنوان رشته برمی‌گرداند؛ مثلاً <code dir="ltr">"25"</code>.
2. <code dir="ltr">int(...)</code> این رشته را به عدد صحیح <code dir="ltr">25</code> تبدیل می‌کند.
3. نام <code dir="ltr">age</code> به عدد <code dir="ltr">25</code> وصل می‌شود.

### ❓ پیش‌بینی کن

سلول زیر را اجرا کنید و دو بار عدد <code dir="ltr">5</code> و <code dir="ltr">3</code> را وارد کنید. **قبل از اجرا** بنویسید چه چیزی چاپ می‌شود.

</div>

````python
a = input("First number: ")
b = input("Second number: ")
print(a + b)
````

<div dir="rtl">

🔍 خروجی <code dir="ltr">53</code> است، نه <code dir="ltr">8</code>! چون <code dir="ltr">a</code> و <code dir="ltr">b</code> رشته‌اند و <code dir="ltr">+</code> روی رشته‌ها یعنی چسباندن. برای جمع عددی باید بنویسید: <code dir="ltr">int(a) + int(b)</code>.

### ⚠️ اشتباه رایج

- **فراموش کردن تبدیل نوع:** همین مثال بالا. اگر به جای <code dir="ltr">+</code> از <code dir="ltr">-</code> یا <code dir="ltr">/</code> استفاده کنید، این بار خطای <code dir="ltr">TypeError</code> می‌گیرید.
- **ورودی نامعتبر:** اگر کاربر برای <code dir="ltr">int(input(...))</code> عبارت <code dir="ltr">abc</code> یا حتی <code dir="ltr">3.5</code> را وارد کند، خطای <code dir="ltr">ValueError</code> رخ می‌دهد. در جلسه‌ی ۱۵ (مدیریت خطا) یاد می‌گیرید چطور این خطا را کنترل کنید.

### 💡 <code dir="ltr">input</code> در علوم داده

در تحلیل داده به‌ندرت داده را با <code dir="ltr">input</code> می‌گیریم؛ داده معمولاً از **فایل** خوانده می‌شود (جلسه‌ی ۱۶ و بخش ۳). اما یک درس <code dir="ltr">input</code> همیشه معتبر است: **داده‌ای که از بیرون برنامه می‌آید معمولاً متن است** و باید نوع آن را بررسی و تبدیل کنید.

</div>

## ۳. f-string

### 💡 مفهوم

**f-string** راحت‌ترین و خواناترین روش برای ساختن متن از ترکیب متن ثابت و مقدار متغیرهاست. کافی است قبل از نقل‌قول حرف <code dir="ltr">f</code> بگذارید و هر **عبارت** پایتون را داخل <code dir="ltr">{ }</code> بنویسید؛ پایتون آن را حساب می‌کند و مقدارش را جایش می‌گذارد.

</div>

````python
name = "Ali"
age = 25
print(f"Hello {name}! You are {age} years old.")
print(f"In 5 years you will be {age + 5}.")
print(f"Your name in capitals: {name.upper()}")
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">{name}</code> با مقدار متغیر جایگزین می‌شود.
- داخل <code dir="ltr">{ }</code> می‌توان **هر عبارتی** نوشت: <code dir="ltr">{age + 5}</code> اول حساب می‌شود (<code dir="ltr">30</code>) و بعد در متن قرار می‌گیرد.
- f-string برخلاف <code dir="ltr">+</code> لازم ندارد عدد را با <code dir="ltr">str()</code> تبدیل کنید.

### 💡 قالب‌بندی اعداد

بعد از عبارت داخل <code dir="ltr">{ }</code> می‌توانید با <code dir="ltr">:</code> یک **قالب (format spec)** بنویسید. پرکاربردترین‌ها در علوم داده:

| قالب | معنی | مثال | خروجی |
|---|---|---|---|
| <code dir="ltr">.2f</code> | دو رقم اعشار | <code dir="ltr">f"{3.14159:.2f}"</code> | <code dir="ltr">3.14</code> |
| <code dir="ltr">,</code> | جداکننده‌ی هزارگان | <code dir="ltr">f"{1250000:,}"</code> | <code dir="ltr">1,250,000</code> |
| <code dir="ltr">,.2f</code> | هر دو با هم | <code dir="ltr">f"{1234.5:,.2f}"</code> | <code dir="ltr">1,234.50</code> |
| <code dir="ltr">.1%</code> | درصد با یک رقم اعشار | <code dir="ltr">f"{0.256:.1%}"</code> | <code dir="ltr">25.6%</code> |
| <code dir="ltr">03d</code> | عدد صحیح با سه رقم، با صفرهای ابتدایی | <code dir="ltr">f"{7:03d}"</code> | <code dir="ltr">007</code> |
| <code dir="ltr">&gt;8</code> / <code dir="ltr">&lt;8</code> / <code dir="ltr">^8</code> | عرض ۸، راست‌چین / چپ‌چین / وسط‌چین | <code dir="ltr">f"[{'ab':&gt;8}]"</code> | <code dir="ltr">[      ab]</code> |

</div>

````python
mean_income = 48753.6789
accuracy = 0.9134
n_samples = 1250000

print(f"Mean income: {mean_income:,.2f}")
print(f"Accuracy:    {accuracy:.1%}")
print(f"Samples:     {n_samples:,}")
````

<div dir="rtl">

### 🔍 کد را بخوان: ساختن یک جدول مرتب

با قالب‌های عرض و جهت می‌توان خروجی را به شکل جدول درآورد؛ کاری که هنگام گزارش نتایج تحلیل زیاد انجام می‌دهید:

</div>

````python
print(f"{'Name':<8}|{'Score':>7}|{'Result':^8}")
print("-" * 25)
print(f"{'Ali':<8}|{17.5:>7.2f}|{'PASS':^8}")
print(f"{'Sara':<8}|{9.25:>7.2f}|{'FAIL':^8}")
print(f"{'Reza':<8}|{12:>7.2f}|{'PASS':^8}")
````

<div dir="rtl">

- <code dir="ltr">&lt;8</code>: متن در ستونی به عرض ۸ کاراکتر، **چپ‌چین**.
- <code dir="ltr">&gt;7.2f</code>: عدد با دو رقم اعشار، در ستونی به عرض ۷، **راست‌چین** (اعداد را معمولاً راست‌چین می‌کنند تا ممیزها زیر هم باشند).
- <code dir="ltr">^8</code>: وسط‌چین.
- <code dir="ltr">"-" &#42; 25</code> یعنی ۲۵ بار تکرار <code dir="ltr">-</code> (جلسه‌ی ۲).

### 💡 ترفند اشکال‌زدایی: <code dir="ltr">{x=}</code>

اگر بعد از عبارت علامت <code dir="ltr">=</code> بگذارید، f-string هم **خود عبارت** و هم **مقدارش** را چاپ می‌کند. برای بررسی سریع مقدار متغیرها بسیار مفید است:

</div>

````python
x = 7
y = 3
print(f"{x=}, {y=}, {x // y=}")
````

<div dir="rtl">

### ❓ پیش‌بینی کن

خروجی هر خط را دقیقاً بنویسید.

</div>

````python
price = 1234567.891
print(f"{price:,.2f}")
print(f"{10 / 3:.3f}")
print(f"{0.05:.0%}")
print(f"[{42:>6}]")
print(f"{2 / 3:.0f}")
````

<div dir="rtl">

🔍 دقت کنید <code dir="ltr">.0f</code> یعنی **بدون رقم اعشار**؛ پس <code dir="ltr">0.666...</code> گرد می‌شود و <code dir="ltr">1</code> نمایش داده می‌شود.

### ⚠️ اشتباه رایج

- **فراموش کردن <code dir="ltr">f</code>:** <code dir="ltr">print("Hello {name}")</code> همان متن را با آکولادها چاپ می‌کند و خطا هم نمی‌دهد.
- **قالب‌بندی، مقدار را تغییر نمی‌دهد:** <code dir="ltr">f"{x:.2f}"</code> فقط یک **رشته** برای نمایش می‌سازد. خود <code dir="ltr">x</code> با همه‌ی رقم‌های اعشارش باقی می‌ماند. اگر واقعاً می‌خواهید عدد را گرد کنید، از <code dir="ltr">round()</code> استفاده کنید.

</div>

````python
x = 3.14159
s = f"{x:.2f}"
print(s, type(s))
print(x)
````

<div dir="rtl">

- **نقل‌قول داخل f-string:** اگر f-string را با <code dir="ltr">"</code> شروع کرده‌اید، داخل <code dir="ltr">{ }</code> از <code dir="ltr">'</code> استفاده کنید؛ مثل <code dir="ltr">f"{'Name':&lt;8}"</code> در مثال جدول.

</div>

## ۴. متد <code dir="ltr">format()</code> و روش‌های قدیمی‌تر

### 💡 مفهوم

f-string از پایتون ۳.۶ اضافه شده است. در کدهای قدیمی‌تر، کتاب‌ها و پاسخ‌های اینترنتی دو روش دیگر را زیاد می‌بینید. لازم نیست با آن‌ها کد بنویسید، اما باید بتوانید آن‌ها را **بخوانید**:

</div>

````python
name = "Zahra"
age = 30

# 1) the format() method: {} are filled in order
print("Hello {}! You are {} years old.".format(name, age))

# 2) the same, with names and a format spec
print("Hello {n}! Score: {s:.1f}".format(n=name, s=17.5))

# 3) the oldest style: % placeholders
print("Hello %s! You are %d years old." % (name, age))

# 4) the modern way
print(f"Hello {name}! You are {age} years old.")
````

<div dir="rtl">

### 🔍 کد را بخوان

| روش | جای خالی | مقدارها کجا می‌آیند؟ |
|---|---|---|
| <code dir="ltr">format()</code> | <code dir="ltr">{}</code> یا <code dir="ltr">{name}</code> | داخل <code dir="ltr">.format(...)</code> بعد از رشته |
| <code dir="ltr">%</code> | <code dir="ltr">%s</code> (رشته)، <code dir="ltr">%d</code> (عدد صحیح)، <code dir="ltr">%.1f</code> (اعشاری) | بعد از علامت <code dir="ltr">%</code> در انتهای رشته |
| f-string | <code dir="ltr">{expression}</code> | مستقیم داخل خود رشته |

قالب‌هایی که برای f-string یاد گرفتید (مثل <code dir="ltr">:.2f</code> یا <code dir="ltr">:,</code>) در <code dir="ltr">format()</code> هم به همان شکل کار می‌کنند.

**توصیه:** در کد خودتان از f-string استفاده کنید؛ خواناتر است و احتمال اشتباه در ترتیب مقدارها کمتر است.

</div>

- تفاوت نمایش مقدار در نوت‌بوک (<code dir="ltr">'Hello'</code>) و چاپ آن (<code dir="ltr">Hello</code>) را بدانید.

- <code dir="ltr">input</code> همیشه <code dir="ltr">str</code> برمی‌گرداند؛ نتیجه‌ی <code dir="ltr">input() + input()</code> را پیش‌بینی کنید و آن را با تبدیل نوع درست کنید.
- عبارت داخل <code dir="ltr">{ }</code> در f-string حساب می‌شود؛ قالب‌های <code dir="ltr">.2f</code>، <code dir="ltr">,</code>، <code dir="ltr">%</code>، <code dir="ltr">03d</code> و عرض/جهت (<code dir="ltr">&lt;</code>، <code dir="ltr">&gt;</code>، <code dir="ltr">^</code>) را بخوانید و بنویسید.
- قالب‌بندی فقط نمایش را تغییر می‌دهد، نه مقدار متغیر را.
- کدی را که با <code dir="ltr">format()</code> یا <code dir="ltr">%</code> نوشته شده بخوانید و به f-string تبدیل کنید.

</div>

### سؤال ۱ — پیش‌بینی خروجی

خروجی کد زیر را **دقیقاً** (با رعایت خط‌ها و فاصله‌ها) بنویسید.

</div>

```python
print("x", "y", "z", sep="")
print("1", "2", end=" ")
print("3")
print("a\tb")
print("Data", "Science", sep="\n")
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

<div dir="ltr">

```text
xyz
1 2 3
a	b
Data
Science
```

</div>

- <code dir="ltr">sep=""</code> یعنی بدون فاصله.
- <code dir="ltr">end=" "</code> به جای خط جدید یک فاصله می‌گذارد؛ پس <code dir="ltr">3</code> در همان خط می‌آید.
- <code dir="ltr">&#92;t</code> یک فاصله‌ی جدول است.
- <code dir="ltr">sep="&#92;n"</code> بین دو مقدار خط جدید می‌گذارد.

</details>

### سؤال ۲ — توضیح کد

```python
x = print("Python")
print(x)
```

### سؤال ۳ — پیدا کردن خطا

برنامه‌ی زیر قرار است میانگین دو نمره را حساب کند. کاربر <code dir="ltr">18</code> و <code dir="ltr">15</code> را وارد کرده، اما برنامه خطا می‌دهد. نوع خطا چیست؟ کد را درست کنید. اگر در خط آخر به جای <code dir="ltr">/ 2</code> نوشته بودیم <code dir="ltr">print(a + b)</code>، چه چیزی چاپ می‌شد؟

</div>

```python
a = input("Score 1: ")
b = input("Score 2: ")
average = (a + b) / 2
print(average)
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

<code dir="ltr">a + b</code> دو رشته را به هم می‌چسباند (<code dir="ltr">"1815"</code>) و تقسیم رشته بر عدد ممکن نیست؛ خطا از نوع <code dir="ltr">TypeError</code> است.

اصلاح:

<div dir="ltr">

```python
a = float(input("Score 1: "))
b = float(input("Score 2: "))
average = (a + b) / 2
print(average)
```

</div>

با <code dir="ltr">print(a + b)</code> (در کد اولیه) خروجی <code dir="ltr">1815</code> می‌شد.

</details>

### سؤال ۴ — قالب‌بندی

خروجی هر خط را بنویسید.

| کد | خروجی؟ |
|---|---|
| <code dir="ltr">f"{1234567.891:,.2f}"</code> | |
| <code dir="ltr">f"{0.4567:.1%}"</code> | |
| <code dir="ltr">f"{7:03d}"</code> | |
| <code dir="ltr">f"[{'ab':&gt;5}]"</code> | |
| <code dir="ltr">f"[{'ab':&lt;5}]"</code> | |
| <code dir="ltr">f"{10 / 4:.1f}"</code> | |
| <code dir="ltr">f"{3 &#42; 4 = }"</code> | |

<details>
<summary><b>پاسخ</b></summary>

| کد | خروجی |
|---|---|
| <code dir="ltr">f"{1234567.891:,.2f}"</code> | <code dir="ltr">1,234,567.89</code> |
| <code dir="ltr">f"{0.4567:.1%}"</code> | <code dir="ltr">45.7%</code> |
| <code dir="ltr">f"{7:03d}"</code> | <code dir="ltr">007</code> |
| <code dir="ltr">f"[{'ab':&gt;5}]"</code> | <code dir="ltr">[   ab]</code> |
| <code dir="ltr">f"[{'ab':&lt;5}]"</code> | <code dir="ltr">[ab   ]</code> |
| <code dir="ltr">f"{10 / 4:.1f}"</code> | <code dir="ltr">2.5</code> |
| <code dir="ltr">f"{3 &#42; 4 = }"</code> | <code dir="ltr">3 &#42; 4 = 12</code> |

در مورد آخر، فاصله‌های اطراف <code dir="ltr">=</code> هم عیناً در خروجی می‌آیند.

</details>

### سؤال ۵ — مفهومی

دانشجویی برای گرد کردن میانگین نوشته است:

</div>

```python
mean = 16.4871
mean_text = f"{mean:.1f}"
total = mean_text * 2
```

<div dir="rtl">

الف) نوع <code dir="ltr">mean&#95;text</code> چیست و مقدار <code dir="ltr">mean</code> بعد از خط دوم چقدر است؟

ب) مقدار <code dir="ltr">total</code> چیست؟ دانشجو انتظار <code dir="ltr">33.0</code> را داشت؛ کد را طوری اصلاح کنید که این نتیجه به دست آید.

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">mean&#95;text</code> یک **رشته** است: <code dir="ltr">"16.5"</code>. مقدار <code dir="ltr">mean</code> تغییر نکرده و همان <code dir="ltr">16.4871</code> است.

ب) <code dir="ltr">"16.5" &#42; 2</code> یعنی تکرار رشته: <code dir="ltr">"16.516.5"</code>. برای گرد کردن **عدد** باید از <code dir="ltr">round</code> استفاده کرد:

<div dir="ltr">

```python
mean_rounded = round(mean, 1)   # 16.5 (float)
total = mean_rounded * 2        # 33.0
```

</div>

</details>

## خلاصه‌ی جلسه

- <code dir="ltr">print</code> با <code dir="ltr">sep</code> و <code dir="ltr">end</code> کنترل می‌شود؛ <code dir="ltr">&#92;n</code> و <code dir="ltr">&#92;t</code> کاراکترهای ویژه‌اند؛ <code dir="ltr">print</code> مقدار <code dir="ltr">None</code> برمی‌گرداند.
- نوت‌بوک مقدار آخرین عبارت را **نمایش** می‌دهد (رشته‌ها با نقل‌قول)؛ <code dir="ltr">print</code> آن را **چاپ** می‌کند.
- <code dir="ltr">input</code> همیشه رشته برمی‌گرداند؛ برای عدد، تبدیل نوع لازم است.
- f-string: <code dir="ltr">f"... {expression:format} ..."</code>؛ قالب‌های <code dir="ltr">.2f</code>، <code dir="ltr">,</code>، <code dir="ltr">%</code>، <code dir="ltr">03d</code>، <code dir="ltr">&lt;</code>، <code dir="ltr">&gt;</code>، <code dir="ltr">^</code> و ترفند <code dir="ltr">{x=}</code>.
- قالب‌بندی فقط نمایش است؛ برای گرد کردن مقدار از <code dir="ltr">round</code> استفاده کنید.
- <code dir="ltr">format()</code> و <code dir="ltr">%</code> روش‌های قدیمی‌ترند؛ آن‌ها را بخوانید، ولی با f-string بنویسید.

**جلسه‌ی بعد:** شرط‌ها (<code dir="ltr">if</code>، <code dir="ltr">elif</code>، <code dir="ltr">else</code>).
