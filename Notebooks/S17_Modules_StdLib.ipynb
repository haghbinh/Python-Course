## ۱. ماژول و انواع <code dir="ltr">import</code>

### 💡 مفهوم

**ماژول** یک فایل پایتون (<code dir="ltr">.py</code>) است که تعدادی تابع، متغیر و ... در آن تعریف شده است. **بسته (package)** پوشه‌ای از چند ماژول است. پایتون همراه خود صدها ماژول آماده دارد که به آن‌ها **کتابخانه‌ی استاندارد** می‌گویند؛ از جلسه‌ی ۷ از دو تا از آن‌ها (<code dir="ltr">math</code> و <code dir="ltr">random</code>) استفاده کرده‌اید و در جلسه‌ی ۱۶ با <code dir="ltr">csv</code>، <code dir="ltr">json</code> و <code dir="ltr">pathlib</code> کار کردید.

دستور <code dir="ltr">import</code> ماژول را بارگذاری می‌کند و یک **نام** در برنامه‌ی شما می‌سازد. چهار شکل رایج:

| شکل | چه نامی ساخته می‌شود؟ | استفاده |
|---|---|---|
| <code dir="ltr">import math</code> | <code dir="ltr">math</code> | <code dir="ltr">math.sqrt(2)</code> |
| <code dir="ltr">import math as m</code> | <code dir="ltr">m</code> (نام مستعار) | <code dir="ltr">m.sqrt(2)</code> |
| <code dir="ltr">from math import sqrt, pi</code> | <code dir="ltr">sqrt</code> و <code dir="ltr">pi</code> | <code dir="ltr">sqrt(2)</code> |
| <code dir="ltr">from math import &#42;</code> | همه‌ی نام‌های عمومی ماژول | ⚠️ توصیه نمی‌شود |

کتابخانه‌های بخش ۳ (مثل NumPy و Pandas) کتابخانه‌ی استاندارد نیستند، اما دقیقاً با همین دستورها وارد می‌شوند؛ مثلاً <code dir="ltr">import numpy as np</code>.

</div>

````python
import math
import random as rnd
from statistics import mean, median

print(math.sqrt(16), math.pi)
print(rnd.randint(1, 1))
print(mean([2, 4, 9]), median([2, 4, 9]))
print(type(math))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">import random as rnd</code> ماژول <code dir="ltr">random</code> را با نام <code dir="ltr">rnd</code> در دسترس می‌گذارد؛ نام <code dir="ltr">random</code> در این حالت تعریف **نشده** است.
- <code dir="ltr">from statistics import mean, median</code> فقط دو نام <code dir="ltr">mean</code> و <code dir="ltr">median</code> را می‌سازد؛ خود <code dir="ltr">statistics</code> تعریف نشده است.
- ماژول هم مثل همه‌چیز در پایتون یک **شیء** است (از نوع <code dir="ltr">module</code>)؛ <code dir="ltr">math.sqrt</code> یعنی «ویژگی <code dir="ltr">sqrt</code> از شیء <code dir="ltr">math</code>».

### 💡 یک ماژول چه چیزهایی دارد؟

<code dir="ltr">dir(module)</code> فهرست نام‌های داخل یک ماژول را برمی‌گرداند و <code dir="ltr">help(name)</code> توضیح (docstring) آن را نشان می‌دهد:

</div>

````python
public = [name for name in dir(math) if not name.startswith("_")]
print(public[:8])
print("comb" in public, "mean" in public)
print(math.comb.__doc__.splitlines()[0])
````

<div dir="rtl">

🔍 نام‌هایی که با <code dir="ltr">&#95;</code> شروع می‌شوند برای استفاده‌ی داخلی‌اند و با comprehension (جلسه‌ی ۱۳) کنار گذاشته شده‌اند. <code dir="ltr">help(math.comb)</code> همان docstring را کامل‌تر نشان می‌دهد؛ در نوت‌بوک می‌توانید <code dir="ltr">math.comb?</code> هم بنویسید.

### ⚠️ اشتباه رایج: <code dir="ltr">from ... import &#42;</code> و سایه انداختن نام‌ها

<code dir="ltr">from math import &#42;</code> ده‌ها نام را بی‌صدا وارد برنامه می‌کند و ممکن است نام‌های شما (یا نام‌های ماژول دیگر) را **بازنویسی** کند. خواننده‌ی کد هم نمی‌فهمد هر نام از کجا آمده است. مشکل مشابه وقتی پیش می‌آید که خودتان روی یک نام آماده **سایه** بیندازید:

</div>

````python
from math import *
e = 0.5                # e was math.e; now it is ours
print(e, exp(1))

values = [3, 4, 5]
sum = 0                # shadows the built-in function sum
for v in values:
    sum += v
print(sum)
print(sum(values))
````

<div dir="rtl">

🔍 بعد از <code dir="ltr">sum = 0</code>، نام <code dir="ltr">sum</code> دیگر به تابع داخلی اشاره نمی‌کند بلکه به عدد <code dir="ltr">12</code>؛ پس <code dir="ltr">sum(values)</code> یعنی «عدد ۱۲ را صدا بزن» و خطای <code dir="ltr">TypeError: 'int' object is not callable</code> می‌دهد. برای برگرداندن تابع، <code dir="ltr">del sum</code> را اجرا کنید. نام‌هایی مثل <code dir="ltr">sum</code>، <code dir="ltr">max</code>، <code dir="ltr">min</code>، <code dir="ltr">list</code>، <code dir="ltr">dict</code>، <code dir="ltr">str</code>، <code dir="ltr">input</code> و <code dir="ltr">id</code> را هرگز برای متغیرها استفاده نکنید.

### ⚠️ اشتباه رایج: فایلی هم‌نام با یک ماژول

اگر در پوشه‌ی پروژه فایلی به نام <code dir="ltr">random.py</code> یا <code dir="ltr">math.py</code> بسازید، <code dir="ltr">import random</code> **فایل شما** را بارگذاری می‌کند، نه ماژول استاندارد را؛ چون پایتون اول در پوشه‌ی جاری جست‌وجو می‌کند. نتیجه، خطاهای عجیبی مثل <code dir="ltr">AttributeError: module 'random' has no attribute 'randint'</code> است. نام فایل‌هایتان را منحصربه‌فرد انتخاب کنید (مثلاً <code dir="ltr">my&#95;random&#95;tools.py</code>).

</div>

````python
del sum
print(sum([3, 4, 5]))
````

## ۲. ساخت ماژول

### 💡 مفهوم

هر فایل <code dir="ltr">.py</code> یک ماژول است. وقتی توابعی را در چند نوت‌بوک یا برنامه لازم دارید (مثلاً <code dir="ltr">tokenize</code> جلسه‌ی ۱۱ یا تابع‌های آماری)، به جای کپی کردن آن‌ها، در یک فایل بنویسید و <code dir="ltr">import</code> کنید.

در نوت‌بوک، دستور جادویی <code dir="ltr">%%writefile name.py</code> در **اولین خط** یک سلول، محتوای سلول را در فایلی به همان نام (کنار نوت‌بوک) ذخیره می‌کند:

</div>

````python
%%writefile mytools.py
"""Small helper functions for data work."""

print("mytools is being loaded")

def mean(values):
    return sum(values) / len(values)

def normalize(values):
    lo, hi = min(values), max(values)
    return [(v - lo) / (hi - lo) for v in values]

if __name__ == "__main__":
    print("running mytools as a script")
    print(normalize([2, 4, 6]))
````

````python
import mytools

print(mytools.mean([2, 4, 9]))
print(mytools.normalize([10, 15, 20]))
print(mytools.__name__)
````

<div dir="rtl">

### 🔍 کد را بخوان

- با <code dir="ltr">import mytools</code>، پایتون فایل <code dir="ltr">mytools.py</code> را **از بالا تا پایین اجرا می‌کند**: <code dir="ltr">print</code> اول اجرا می‌شود، توابع تعریف می‌شوند و همه‌ی این نام‌ها داخل شیء <code dir="ltr">mytools</code> قرار می‌گیرند.
- <code dir="ltr">mean</code> این ماژول با <code dir="ltr">mean</code> ماژول <code dir="ltr">statistics</code> تداخلی ندارد، چون با نام کامل <code dir="ltr">mytools.mean</code> صدا زده می‌شود؛ این مزیت اصلی <code dir="ltr">import mytools</code> نسبت به <code dir="ltr">from mytools import &#42;</code> است.
- هر ماژول یک متغیر ویژه به نام <code dir="ltr">&#95;&#95;name&#95;&#95;</code> دارد. وقتی ماژول **import** می‌شود، مقدار آن نام ماژول (<code dir="ltr">'mytools'</code>) است؛ وقتی فایل **مستقیماً اجرا** می‌شود (مثلاً <code dir="ltr">python mytools.py</code> در خط فرمان)، مقدار آن <code dir="ltr">'&#95;&#95;main&#95;&#95;'</code> است. پس بلوک <code dir="ltr">if &#95;&#95;name&#95;&#95; == "&#95;&#95;main&#95;&#95;":</code> فقط در اجرای مستقیم اجرا می‌شود؛ جای مناسبی برای کد آزمایشی است.

### ❓ پیش‌بینی کن

سلول زیر چه چاپ می‌کند؟ آیا پیام <code dir="ltr">mytools is being loaded</code> دوباره ظاهر می‌شود؟

</div>

````python
import mytools
import mytools as mt
print(mt is mytools)
````

<div dir="rtl">

🔍 هیچ پیامی از ماژول چاپ نمی‌شود و خروجی فقط <code dir="ltr">True</code> است. پایتون هر ماژول را فقط **یک بار** در هر جلسه‌ی اجرا (kernel) بارگذاری می‌کند و نتیجه را **کش** می‌کند؛ <code dir="ltr">import</code>‌های بعدی همان شیء قبلی را برمی‌گردانند. <code dir="ltr">mt</code> و <code dir="ltr">mytools</code> دو نام برای **یک** شیء‌اند.

### ⚠️ اشتباه رایج: تغییر فایل ماژول و نتیجه‌ی قدیمی

به همین دلیل، اگر فایل <code dir="ltr">mytools.py</code> را ویرایش کنید و دوباره <code dir="ltr">import</code> کنید، نسخه‌ی **قدیمی** را می‌بینید. یا kernel را دوباره راه بیندازید، یا از <code dir="ltr">importlib.reload</code> استفاده کنید:

</div>

````python
%%writefile mytools.py
"""Small helper functions for data work (version 2)."""

def mean(values):
    return round(sum(values) / len(values), 2)
````

````python
import importlib

import mytools
print(mytools.mean([1, 2, 2]))

mytools = importlib.reload(mytools)
print(mytools.mean([1, 2, 2]))
print(hasattr(mytools, "normalize"))
````

<div dir="rtl">

🔍 قبل از <code dir="ltr">reload</code>، تابع قدیمی (<code dir="ltr">1.6666666666666667</code>) اجرا می‌شود و بعد از آن، تابع جدید (<code dir="ltr">1.67</code>). خط آخر یک نکته‌ی ظریف را نشان می‌دهد: <code dir="ltr">reload</code> فایل را دوباره در **همان** شیء ماژول اجرا می‌کند و نام‌هایی را که در نسخه‌ی جدید حذف شده‌اند پاک نمی‌کند؛ پس <code dir="ltr">normalize</code> هنوز وجود دارد (<code dir="ltr">True</code>). برای اطمینان کامل، kernel را دوباره راه بیندازید.

</div>

## ۳. <code dir="ltr">math</code> و <code dir="ltr">random</code>

### 💡 ماژول <code dir="ltr">math</code>

| تابع / ثابت | کار | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">sqrt</code>، <code dir="ltr">exp</code>، <code dir="ltr">log</code> | جذر، <code dir="ltr">e</code> به توان، لگاریتم طبیعی | <code dir="ltr">math.log(math.e)</code> | <code dir="ltr">1.0</code> |
| <code dir="ltr">log(x, base)</code>، <code dir="ltr">log2</code>، <code dir="ltr">log10</code> | لگاریتم در پایه‌ی دلخواه | <code dir="ltr">math.log2(8)</code> | <code dir="ltr">3.0</code> |
| <code dir="ltr">floor</code>، <code dir="ltr">ceil</code> | گرد کردن به پایین / بالا | <code dir="ltr">math.ceil(2.1)</code> | <code dir="ltr">3</code> |
| <code dir="ltr">factorial</code>، <code dir="ltr">comb</code> | فاکتوریل، ترکیب <code dir="ltr">n</code> از <code dir="ltr">k</code> | <code dir="ltr">math.comb(5, 2)</code> | <code dir="ltr">10</code> |
| <code dir="ltr">isclose(a, b)</code> | تقریباً برابر؟ | <code dir="ltr">math.isclose(0.1 + 0.2, 0.3)</code> | <code dir="ltr">True</code> |
| <code dir="ltr">pi</code>، <code dir="ltr">e</code>، <code dir="ltr">inf</code> | ثابت‌ها؛ <code dir="ltr">inf</code> بی‌نهایت | <code dir="ltr">-math.inf &lt; -1e300</code> | <code dir="ltr">True</code> |

</div>

````python
import math

print(0.1 + 0.2 == 0.3, math.isclose(0.1 + 0.2, 0.3))
print(math.floor(-2.5), math.ceil(-2.5), int(-2.5), round(2.5))
print(math.comb(10, 3), math.factorial(5))

best = math.inf
for loss in [0.9, 0.42, 0.57]:
    if loss < best:
        best = loss
print(best)
````

<div dir="rtl">

### 🔍 کد را بخوان

- مقایسه‌ی مستقیم اعداد اعشاری با <code dir="ltr">==</code> به خاطر خطای گرد کردن (جلسه‌ی ۲) قابل اعتماد نیست؛ <code dir="ltr">math.isclose</code> با یک تلورانس نسبی مقایسه می‌کند.
- برای اعداد منفی، <code dir="ltr">floor</code> به سمت **کوچک‌تر** (<code dir="ltr">-3</code>) و <code dir="ltr">int</code> به سمت **صفر** (<code dir="ltr">-2</code>) گرد می‌کند. <code dir="ltr">round(2.5)</code> برابر <code dir="ltr">2</code> است: پایتون نیمه‌ها را به نزدیک‌ترین عدد **زوج** گرد می‌کند.
- مقدار اولیه‌ی <code dir="ltr">math.inf</code> برای پیدا کردن کمینه (مثلاً کمترین **خطا** در آموزش یک مدل) از هر عددی بزرگ‌تر است؛ پس اولین مقدار حتماً جایگزینش می‌شود.

### 💡 ماژول <code dir="ltr">random</code> و تکرارپذیری

اعداد <code dir="ltr">random</code> در واقع **شبه‌تصادفی**‌اند: از یک مقدار اولیه به نام **seed** با یک فرمول قطعی تولید می‌شوند. با <code dir="ltr">random.seed(n)</code> همیشه همان دنباله تولید می‌شود؛ برای همین در علوم داده (تقسیم داده، مقداردهی اولیه‌ی وزن‌های شبکه‌ی عصبی) seed را ثابت می‌کنیم تا نتایج **تکرارپذیر** باشند.

| تابع | کار |
|---|---|
| <code dir="ltr">random()</code> | عدد اعشاری در بازه‌ی <code dir="ltr">[0, 1)</code> |
| <code dir="ltr">randint(a, b)</code> | عدد صحیح بین <code dir="ltr">a</code> و <code dir="ltr">b</code> (**هر دو شامل**) |
| <code dir="ltr">uniform(a, b)</code> | عدد اعشاری بین <code dir="ltr">a</code> و <code dir="ltr">b</code> |
| <code dir="ltr">gauss(mu, sigma)</code> | عدد از توزیع نرمال |
| <code dir="ltr">choice(seq)</code> | یک عضو تصادفی |
| <code dir="ltr">choices(seq, weights=..., k=n)</code> | <code dir="ltr">n</code> عضو **با جای‌گذاری** (تکرار مجاز)، با وزن دلخواه |
| <code dir="ltr">sample(seq, k)</code> | <code dir="ltr">k</code> عضو **بدون جای‌گذاری** (بدون تکرار) |
| <code dir="ltr">shuffle(lst)</code> | به هم ریختن خود لیست (درجا؛ <code dir="ltr">None</code> برمی‌گرداند) |

</div>

````python
import random

random.seed(42)
print(random.random())
random.seed(42)
print(random.random())

random.seed(0)
print(random.sample(range(10), 3))
nums = [0, 1, 2, 3, 4]
random.shuffle(nums)
print(nums)
print(round(random.gauss(0, 1), 3))
````

<div dir="rtl">

### 🔍 کد را بخوان

- دو بار <code dir="ltr">random.seed(42)</code> و هر بار **همان** عدد <code dir="ltr">0.6394267984578837</code>؛ این همان تکرارپذیری است. (این اعداد روی هر کامپیوتری با همان نسخه‌ی پایتون یکسان‌اند.)
- <code dir="ltr">sample</code> سه عدد **متفاوت** برمی‌گرداند و <code dir="ltr">shuffle</code> خود لیست را تغییر می‌دهد (مثل <code dir="ltr">list.sort</code> در جلسه‌ی ۸).

### ❓ پیش‌بینی کن

یک داده‌ی نامتوازن شبیه‌سازی می‌کنیم: هر نمونه با وزن ۱ به ۴، <code dir="ltr">"spam"</code> یا <code dir="ltr">"ham"</code> است. از هزار نمونه، تقریباً چند تا <code dir="ltr">"spam"</code> خواهد بود؟

</div>

````python
random.seed(1)
labels = random.choices(["spam", "ham"], weights=[1, 4], k=1000)
print(labels[:6])
print(labels.count("spam"), labels.count("ham"))
````

<div dir="rtl">

🔍 وزن‌ها نسبی‌اند: احتمال <code dir="ltr">"spam"</code> برابر <code dir="ltr">1 / (1 + 4) = 0.2</code> است؛ پس انتظار حدود ۲۰۰ نمونه را داریم. عدد واقعی کمی با ۲۰۰ فرق دارد (تصادف)، اما با seed ثابت، هر بار دقیقاً همان عدد را می‌بینید.

### ⚠️ اشتباه رایج: seed در جای نادرست

<code dir="ltr">random.seed</code> را **یک بار** در ابتدای آزمایش صدا بزنید. اگر آن را داخل حلقه (قبل از هر نمونه‌گیری) قرار دهید، همه‌ی تکرارها **یک عدد** تولید می‌کنند و دیگر تصادفی در کار نیست.

</div>

## ۴. <code dir="ltr">datetime</code>

### 💡 مفهوم

داده‌های واقعی پر از تاریخ و زمان‌اند: تاریخ ثبت‌نام، زمان تراکنش، سری زمانی دما. ماژول <code dir="ltr">datetime</code> سه نوع اصلی دارد:

- <code dir="ltr">date</code>: فقط تاریخ (سال، ماه، روز)،
- <code dir="ltr">datetime</code>: تاریخ و ساعت،
- <code dir="ltr">timedelta</code>: **فاصله‌ی زمانی** (مثلاً ۳ روز یا ۳۶ ساعت).

تفریق دو تاریخ یک <code dir="ltr">timedelta</code> می‌دهد و جمع یک تاریخ با <code dir="ltr">timedelta</code> یک تاریخ جدید. (این ماژول تقویم میلادی دارد؛ برای تاریخ شمسی کتابخانه‌های جداگانه وجود دارد.)

</div>

````python
from datetime import date, datetime, timedelta

start = date(2026, 9, 22)
end = date(2027, 3, 21)
print(end - start, (end - start).days)
print(start + timedelta(days=100))
print(start.year, start.month, start.weekday())

t = datetime(2026, 11, 9, 14, 30)
print(t + timedelta(hours=36))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">end - start</code> یک <code dir="ltr">timedelta</code> است که چاپش <code dir="ltr">180 days, 0:00:00</code> می‌شود؛ ویژگی <code dir="ltr">.days</code> عدد صحیح <code dir="ltr">180</code> را می‌دهد.
- <code dir="ltr">weekday()</code> روز هفته را با شماره برمی‌گرداند: دوشنبه <code dir="ltr">0</code> تا یکشنبه <code dir="ltr">6</code>. (۲۲ سپتامبر ۲۰۲۶ سه‌شنبه است، پس <code dir="ltr">1</code>.)
- با اضافه کردن ۳۶ ساعت، تاریخ هم به‌طور خودکار عوض می‌شود.

### 💡 تبدیل رشته و تاریخ

در فایل‌ها، تاریخ به شکل **رشته** ذخیره می‌شود. <code dir="ltr">strptime</code> (p برای parse) رشته را با یک **الگو** به <code dir="ltr">datetime</code> تبدیل می‌کند و <code dir="ltr">strftime</code> (f برای format) برعکس:

| کد | معنی | مثال |
|---|---|---|
| <code dir="ltr">%Y</code> | سال چهاررقمی | <code dir="ltr">2026</code> |
| <code dir="ltr">%m</code> | ماه دورقمی | <code dir="ltr">09</code> |
| <code dir="ltr">%d</code> | روز دورقمی | <code dir="ltr">22</code> |
| <code dir="ltr">%H</code>، <code dir="ltr">%M</code> | ساعت و دقیقه | <code dir="ltr">14</code>، <code dir="ltr">30</code> |
| <code dir="ltr">%A</code>، <code dir="ltr">%B</code> | نام روز و ماه (انگلیسی) | <code dir="ltr">Tuesday</code>، <code dir="ltr">September</code> |

</div>

````python
s = "22/09/2026 08:15"
t = datetime.strptime(s, "%d/%m/%Y %H:%M")
print(t)
print(t.strftime("%Y-%m-%d"), t.strftime("%A, %B %d"))
print(t.date() == date(2026, 9, 22))
````

<div dir="rtl">

🔍 الگو باید **دقیقاً** با شکل رشته جور باشد (حتی <code dir="ltr">/</code> و فاصله‌ها)؛ وگرنه <code dir="ltr">ValueError</code> رخ می‌دهد که می‌توانید آن را با <code dir="ltr">try</code> / <code dir="ltr">except</code> (جلسه‌ی ۱۵) مدیریت کنید. قالب <code dir="ltr">"%Y-%m-%d"</code> (استاندارد ISO) بهترین قالب برای ذخیره‌ی تاریخ است، چون مرتب کردن رشته‌ها همان مرتب کردن تاریخ‌هاست.

### ⚠️ اشتباه رایج: <code dir="ltr">%m</code> و <code dir="ltr">%M</code>

<code dir="ltr">%m</code> ماه است و <code dir="ltr">%M</code> دقیقه. این دو را جابه‌جا بنویسید، خطایی نمی‌گیرید ولی تاریخ‌ها **بی‌صدا** اشتباه می‌شوند. همین‌طور <code dir="ltr">datetime.now()</code> و <code dir="ltr">date.today()</code> هر بار مقدار متفاوتی دارند؛ در کدی که باید تکرارپذیر باشد، تاریخ مرجع را صریحاً بنویسید.

## ۵. <code dir="ltr">collections</code>

### 💡 مفهوم

ماژول <code dir="ltr">collections</code> نسخه‌های تخصصی‌تری از دیکشنری، تاپل و لیست دارد که الگوهای رایج جلسه‌های ۸ تا ۱۰ را کوتاه‌تر و سریع‌تر می‌کنند:

| نوع | جایگزین چه الگویی؟ |
|---|---|
| <code dir="ltr">Counter</code> | شمارش با <code dir="ltr">d[k] = d.get(k, 0) + 1</code> |
| <code dir="ltr">defaultdict</code> | گروه‌بندی با «اگر کلید نیست، لیست خالی بساز» |
| <code dir="ltr">namedtuple</code> | تاپلی که اعضایش **نام** دارند |
| <code dir="ltr">deque</code> | صفی که از هر دو سر سریع اضافه و حذف می‌کند |

(<code dir="ltr">OrderedDict</code> هم هست، اما از پایتون ۳٫۷ دیکشنری معمولی ترتیب درج را حفظ می‌کند و کمتر به آن نیاز دارید.)

### 💡 <code dir="ltr">Counter</code>

</div>

````python
from collections import Counter

tokens = "the cat sat on the mat the end".split()
counts = Counter(tokens)
print(counts)
print(counts["the"], counts["dog"])
print(counts.most_common(2))

a = Counter({"x": 3, "y": 1})
b = Counter({"x": 1, "z": 2})
print(a + b, a - b)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">Counter(tokens)</code> در یک خط همان کار حلقه‌ی شمارش جلسه‌ی ۱۰ را انجام می‌دهد.
- برای کلید ناموجود، <code dir="ltr">counts["dog"]</code> خطا نمی‌دهد و <code dir="ltr">0</code> برمی‌گرداند (برخلاف دیکشنری معمولی).
- <code dir="ltr">most&#95;common(k)</code> لیستی از <code dir="ltr">k</code> زوج پرتکرار را مرتب‌شده برمی‌گرداند؛ ساختن **واژگان** یک مدل متنی معمولاً با همین شروع می‌شود.
- <code dir="ltr">a - b</code> شمارش‌های صفر و منفی را حذف می‌کند؛ پس <code dir="ltr">y</code> می‌ماند و <code dir="ltr">x</code> با <code dir="ltr">2</code> باقی می‌ماند و <code dir="ltr">z</code> اصلاً ظاهر نمی‌شود.

### 💡 <code dir="ltr">defaultdict</code>

<code dir="ltr">defaultdict(factory)</code> وقتی کلیدی وجود ندارد، با صدا زدن <code dir="ltr">factory()</code> یک مقدار پیش‌فرض می‌سازد: <code dir="ltr">list()</code> یک لیست خالی و <code dir="ltr">int()</code> عدد صفر.

</div>

````python
from collections import defaultdict

records = [("A", 15), ("B", 12), ("A", 18), ("C", 9), ("B", 14)]
groups = defaultdict(list)
for group, score in records:
    groups[group].append(score)
print(dict(groups))

totals = defaultdict(int)
for group, score in records:
    totals[group] += score
print(dict(totals))
````

<div dir="rtl">

🔍 الگوی گروه‌بندی جلسه‌ی ۱۰ (<code dir="ltr">if key not in d: d[key] = []</code>) حذف شده است. <code dir="ltr">dict(groups)</code> فقط برای چاپ خواناتر است.

### ⚠️ اشتباه رایج: خواندن، کلید می‌سازد

در <code dir="ltr">defaultdict</code>، حتی **خواندن** یک کلید ناموجود (<code dir="ltr">groups["Z"]</code>) آن کلید را با مقدار پیش‌فرض **اضافه** می‌کند؛ پس بعد از چند بررسی، دیکشنری کلیدهای خالی ناخواسته پیدا می‌کند. برای بررسی وجود از <code dir="ltr">in</code> استفاده کنید.

### 💡 <code dir="ltr">namedtuple</code> و <code dir="ltr">deque</code>

</div>

````python
from collections import namedtuple, deque

Student = namedtuple("Student", ["name", "city", "gpa"])
s = Student("Sara", "Bushehr", 17.25)
print(s)
print(s.name, s[2])
s2 = s._replace(gpa=18.0)
print(s.gpa, s2.gpa)

window = deque(maxlen=3)
for x in [10, 20, 30, 40, 50]:
    window.append(x)
    print(list(window), sum(window) / len(window))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">namedtuple</code> یک نوع تاپل جدید می‌سازد که اعضایش هم با **نام** (<code dir="ltr">s.name</code>) و هم با اندیس (<code dir="ltr">s[2]</code>) در دسترس‌اند؛ برای رکوردهای داده خواناتر از <code dir="ltr">s[0]</code> است. مثل هر تاپلی **تغییرناپذیر** است؛ <code dir="ltr">&#95;replace</code> یک تاپل **جدید** می‌سازد و <code dir="ltr">s</code> دست‌نخورده می‌ماند.
- <code dir="ltr">deque(maxlen=3)</code> وقتی پر است، با هر <code dir="ltr">append</code> عضو **قدیمی‌ترین** را از سر دیگر بیرون می‌اندازد؛ یک **پنجره‌ی لغزان** آماده برای میانگین متحرک سری‌های زمانی.

### ❓ پیش‌بینی کن

سطر آخر خروجی حلقه‌ی بالا چیست؟ اگر به جای <code dir="ltr">deque</code> از لیست معمولی و <code dir="ltr">lst.pop(0)</code> استفاده می‌کردیم، نتیجه فرق می‌کرد؟

🔍 سطر آخر <code dir="ltr">[30, 40, 50] 40.0</code> است. نتیجه با لیست هم همین است، اما <code dir="ltr">pop(0)</code> در لیست باید همه‌ی عضوهای بعدی را یک خانه جابه‌جا کند (برای لیست‌های بزرگ کند)، در حالی که <code dir="ltr">deque</code> از هر دو سر در زمان ثابت حذف و اضافه می‌کند.

### سؤال ۲ — واژگان یک مدل متنی

یک مدل متنی فقط کلمه‌هایی را در **واژگان** نگه می‌دارد که دست‌کم <code dir="ltr">min&#95;freq</code> بار تکرار شده‌اند؛ بقیه به توکن ویژه‌ی <code dir="ltr">"&lt;unk&gt;"</code> (ناشناخته) تبدیل می‌شوند.

</div>

```python
from collections import Counter

corpus = ["the cat sat", "the dog sat", "a cat ran", "the cat ate"]
counts = Counter()
for sentence in corpus:
    counts.update(sentence.split())

min_freq = 2
vocab = ["<unk>"] + sorted(w for w, c in counts.items() if c >= min_freq)
word2idx = {w: i for i, w in enumerate(vocab)}

def encode(sentence):
    return [word2idx.get(w, 0) for w in sentence.split()]
```

<div dir="rtl">

الف) <code dir="ltr">counts.most&#95;common(3)</code>، <code dir="ltr">vocab</code> و <code dir="ltr">word2idx</code> را بنویسید.

ب) <code dir="ltr">encode("the dog ate a fish")</code> چیست؟ چرا <code dir="ltr">get</code> با مقدار پیش‌فرض <code dir="ltr">0</code> به کار رفته است؟

ج) اگر <code dir="ltr">min&#95;freq</code> را <code dir="ltr">1</code> کنیم، اندازه‌ی واژگان چند می‌شود؟ بالا بردن <code dir="ltr">min&#95;freq</code> چه مزیت و چه عیبی دارد؟

ج) <code dir="ltr">8</code> (هفت کلمه‌ی متمایز به‌علاوه‌ی <code dir="ltr">&lt;unk&gt;</code>). <code dir="ltr">min&#95;freq</code> بالاتر واژگان را کوچک می‌کند (مدل کوچک‌تر و تخمین‌های پایدارتر برای کلمه‌های پرتکرار)، اما اطلاعات کلمه‌های کم‌تکرار ولی مهم را از بین می‌برد و <code dir="ltr">&lt;unk&gt;</code> زیاد می‌شود.

### سؤال ۳ — گروه‌بندی و میانگین گروه‌ها

from collections import defaultdict

data = [("Bushehr", 31.5), ("Shiraz", 22.0), ("Bushehr", 33.5), ("Tehran", 18.0), ("Shiraz", 24.0)]

### سؤال ۴ — میانگین متحرک

میانگین متحرک با پنجره‌ی <code dir="ltr">k</code>، نوسان‌های کوتاه‌مدت یک سری زمانی را صاف می‌کند:

</div>

```python
from collections import deque

def moving_average(series, k):
    window = deque(maxlen=k)
    result = []
    for x in series:
        window.append(x)
        if len(window) == k:
            result.append(sum(window) / k)
    return result
```

<div dir="rtl">

الف) <code dir="ltr">moving&#95;average([3, 5, 7, 5, 3, 9], 3)</code> چیست؟ طول خروجی برای یک سری <code dir="ltr">n</code> عضوی چند است؟

ب) هر بار <code dir="ltr">sum(window)</code> کل پنجره را جمع می‌زند. تابع را طوری بازنویسی کنید که یک متغیر <code dir="ltr">total</code> نگه دارد و در هر گام فقط عضو ورودی را اضافه و عضو خروجی را کم کند. پیچیدگی زمانی دو نسخه برای سری <code dir="ltr">n</code> عضوی و پنجره‌ی <code dir="ltr">k</code> چیست؟

ج) اگر شرط <code dir="ltr">if len(window) == k</code> حذف شود و <code dir="ltr">sum(window) / len(window)</code> ذخیره شود، دو عضو اول خروجی (برای ورودی الف) چه می‌شوند؟

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">[5.0, 5.666666666666667, 5.0, 5.666666666666667]</code>؛ یعنی <code dir="ltr">(3+5+7)/3</code>، <code dir="ltr">(5+7+5)/3</code>، <code dir="ltr">(7+5+3)/3</code> و <code dir="ltr">(5+3+9)/3</code>. طول خروجی <code dir="ltr">n - k + 1</code> است (مثل n-gramها در جلسه‌ی ۱۱).

ب)

<div dir="ltr">

```python
def moving_average(series, k):
    window = deque()
    total = 0
    result = []
    for x in series:
        window.append(x)
        total += x
        if len(window) > k:
            total -= window.popleft()
        if len(window) == k:
            result.append(total / k)
    return result
```

</div>

نسخه‌ی اول <code dir="ltr">O(n·k)</code> و نسخه‌ی دوم <code dir="ltr">O(n)</code> است. (در عمل، جمع و تفریق مکرر اعداد اعشاری کمی خطای گرد کردن انباشته می‌کند.)

ج) <code dir="ltr">[3.0, 4.0, 5.0, ...]</code>: میانگین پنجره‌های ناقص <code dir="ltr">[3]</code> و <code dir="ltr">[3, 5]</code>. Pandas همین رفتار را با <code dir="ltr">rolling(k, min&#95;periods=1)</code> دارد.

</details>

### سؤال ۵ — بوت‌استرپ

**بوت‌استرپ (bootstrap)** روشی برای تخمین عدم قطعیت یک آماره است: از داده، بارها نمونه‌ای **هم‌اندازه و با جای‌گذاری** می‌گیریم و آماره را روی هر نمونه حساب می‌کنیم.

</div>

```python
import random

scores = [12, 15, 14, 18, 9, 16, 17, 11]

def bootstrap_means(data, n_boot, seed=0):
    random.seed(seed)
    means = []
    for _ in range(n_boot):
        sample = random.choices(data, k=len(data))
        means.append(sum(sample) / len(sample))
    return means

means = bootstrap_means(scores, 1000)
means.sort()
print(round(sum(means) / len(means), 2), means[25], means[974])
```

<div dir="rtl">

الف) چرا به جای <code dir="ltr">random.choices</code> نمی‌توان از <code dir="ltr">random.sample(data, len(data))</code> استفاده کرد؟ میانگین نمونه‌ی حاصل از <code dir="ltr">sample</code> چه می‌شد؟

ب) <code dir="ltr">means[25]</code> و <code dir="ltr">means[974]</code> چه چیزی را تخمین می‌زنند؟ (کد را اجرا کنید و عددها را گزارش دهید.)

ج) در یک نمونه‌ی بوت‌استرپ از <code dir="ltr">n</code> عضو، احتمال اینکه یک عضو مشخص **هیچ‌وقت** انتخاب نشود <code dir="ltr">(1 - 1/n)</code> به توان <code dir="ltr">n</code> است. وقتی <code dir="ltr">n</code> بزرگ است، این مقدار به چه عددی میل می‌کند؟ یعنی هر نمونه‌ی بوت‌استرپ تقریباً چند درصد داده‌ی **متمایز** را شامل می‌شود؟ (این همان ایده‌ی داده‌ی **out-of-bag** در جنگل تصادفی است.)

د) اگر خط <code dir="ltr">random.seed(seed)</code> را به داخل حلقه منتقل کنیم، خروجی چه می‌شود؟

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">sample</code> بدون جای‌گذاری است و نمونه‌ای هم‌اندازه با داده، فقط **جایگشتی** از همان داده است؛ میانگینش همیشه دقیقاً میانگین داده (<code dir="ltr">14.0</code>) می‌شد و هیچ تغییری نشان نمی‌داد.

ب) آن‌ها صدک‌های ۲٫۵ و ۹۷٫۵ از توزیع میانگین‌های بوت‌استرپ‌اند؛ یعنی یک **بازه‌ی اطمینان ۹۵٪ تقریبی** برای میانگین. با این کد خروجی <code dir="ltr">14.02 12.0 16.0</code> است؛ یعنی میانگین نمره‌ها با اطمینان تقریبی ۹۵٪ بین ۱۲ و ۱۶ است.

ج) به <code dir="ltr">1/e ≈ 0.368</code> میل می‌کند؛ پس هر نمونه‌ی بوت‌استرپ حدود <code dir="ltr">63.2%</code> داده را (به شکل متمایز) شامل می‌شود و حدود ۳۶٫۸٪ بیرون می‌ماند.

د) در هر تکرار seed دوباره تنظیم می‌شود و <code dir="ltr">choices</code> همان نمونه را تولید می‌کند؛ هر ۱۰۰۰ میانگین برابر می‌شوند و بازه‌ی اطمینان طول صفر پیدا می‌کند (و نتیجه بی‌معنی است).

</details>

### سؤال ۶ — سری زمانی با تاریخ‌های گمشده

raw = ["2026-07-01,34.5", "2026-07-02,35.0", "2026-07-05,36.5", "2026-07-06,35.5", "2026-07-09,37.0"]
