## ۱. رشته به‌عنوان دنباله

### 💡 مفهوم

در جلسه‌ی ۲ دیدید که رشته (<code dir="ltr">str</code>) برای نگه‌داری متن است. حالا که لیست و تاپل را می‌شناسید، می‌توانیم دقیق‌تر بگوییم: رشته یک **دنباله‌ی مرتب و تغییرناپذیر** از کاراکترهاست؛ پس اندیس، برش، <code dir="ltr">len</code>، <code dir="ltr">in</code>، <code dir="ltr">+</code>، <code dir="ltr">&#42;</code> و حلقه‌ی <code dir="ltr">for</code> دقیقاً مثل تاپل کار می‌کنند.

</div>

````python
text = "Python"
print(text[0], text[-1])
print(text[0:4], text[2:], text[:3])
print(text[::-1])
print(len(text), "th" in text)

s1 = "Hello"
s2 = "World"
print(s1 + " " + s2, s1 * 3)

for ch in "abc":
    print(ch, end="-")
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">text[0:4]</code> چهار کاراکتر اول را برمی‌دارد (<code dir="ltr">stop</code> شامل نیست؛ مثل <code dir="ltr">range</code> و برش لیست).
- <code dir="ltr">text[::-1]</code> رشته را برعکس می‌کند.
- <code dir="ltr">"th" in text</code> بررسی می‌کند آیا <code dir="ltr">"th"</code> **زیررشته**‌ی <code dir="ltr">text</code> است (جلسه‌ی ۳).

### ⚠️ رشته تغییرناپذیر است

</div>

````python
word = "Jython"
word[0] = "P"
````

<div dir="rtl">

🔍 مثل تاپل، نمی‌توان یک کاراکتر را عوض کرد. باید یک رشته‌ی **جدید** ساخت: <code dir="ltr">word = "P" + word[1:]</code>.

### 💡 کاراکترها در واقع عدد هستند

هر کاراکتر یک کد عددی در استاندارد **یونیکد (Unicode)** دارد. <code dir="ltr">ord</code> کد یک کاراکتر را می‌دهد و <code dir="ltr">chr</code> برعکس عمل می‌کند. همین کدها هستند که ترتیب مقایسه‌ی رشته‌ها را تعیین می‌کنند (جلسه‌ی ۳):

</div>

````python
print(ord("A"), ord("a"), ord("0"))
print(chr(98))
print("Z" < "a")
````

<div dir="rtl">

🔍 کد <code dir="ltr">"Z"</code> (۹۰) از کد <code dir="ltr">"a"</code> (۹۷) کوچک‌تر است؛ برای همین همه‌ی حروف بزرگ انگلیسی قبل از حروف کوچک مرتب می‌شوند. چون رشته‌های پایتون یونیکد هستند، متن فارسی هم در داده‌ها بدون مشکل نگه‌داری و پردازش می‌شود.

</div>

## ۲. متدهای رشته

### 💡 مفهوم

رشته ده‌ها متد دارد. نکته‌ی کلیدی: چون رشته تغییرناپذیر است، **هیچ متدی خود رشته را عوض نمی‌کند**؛ همه یک رشته (یا مقدار) **جدید** برمی‌گردانند. (برخلاف <code dir="ltr">list.sort</code> یا <code dir="ltr">list.append</code> در جلسه‌ی ۸.)

| متد | کار | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">lower()</code> / <code dir="ltr">upper()</code> | حروف کوچک / بزرگ | <code dir="ltr">"Data".lower()</code> | <code dir="ltr">'data'</code> |
| <code dir="ltr">title()</code> | حرف اول هر کلمه بزرگ | <code dir="ltr">"data science".title()</code> | <code dir="ltr">'Data Science'</code> |
| <code dir="ltr">strip()</code> | حذف فاصله (و <code dir="ltr">&#92;n</code>) از دو طرف | <code dir="ltr">"  hi &#92;n".strip()</code> | <code dir="ltr">'hi'</code> |
| <code dir="ltr">lstrip()</code> / <code dir="ltr">rstrip()</code> | فقط از چپ / راست | <code dir="ltr">"xxhi".lstrip("x")</code> | <code dir="ltr">'hi'</code> |
| <code dir="ltr">startswith(p)</code> / <code dir="ltr">endswith(p)</code> | آیا با <code dir="ltr">p</code> شروع / تمام می‌شود؟ | <code dir="ltr">"data.csv".endswith(".csv")</code> | <code dir="ltr">True</code> |
| <code dir="ltr">isdigit()</code> | فقط رقم؟ | <code dir="ltr">"2026".isdigit()</code> | <code dir="ltr">True</code> |
| <code dir="ltr">isalpha()</code> | فقط حرف؟ | <code dir="ltr">"abc1".isalpha()</code> | <code dir="ltr">False</code> |
| <code dir="ltr">isalnum()</code> | فقط حرف و رقم؟ | <code dir="ltr">"abc1".isalnum()</code> | <code dir="ltr">True</code> |
| <code dir="ltr">isspace()</code> | فقط فاصله؟ | <code dir="ltr">"  ".isspace()</code> | <code dir="ltr">True</code> |

</div>

````python
s = "  Python Programming  "
print(s.lower())
print(s.upper())
print(s.strip())
print(s.strip().lower().replace(" ", "_"))     # methods can be chained
print("[" + s + "]")                           # s itself is unchanged
````

<div dir="rtl">

### 🔍 کد را بخوان

- در خط چهارم چند متد **زنجیر** شده‌اند و از چپ به راست اجرا می‌شوند: اول فاصله‌های دو طرف حذف، بعد حروف کوچک، بعد فاصله‌ی وسط با <code dir="ltr">&#95;</code> جایگزین می‌شود. این کار برای ساختن **نام ستون‌های تمیز** در داده بسیار رایج است.
- خط آخر نشان می‌دهد <code dir="ltr">s</code> بعد از همه‌ی این فراخوانی‌ها دست‌نخورده باقی مانده است.

### ⚠️ اشتباه رایج: فراموش کردن انتساب

<code dir="ltr">name.strip()</code> به‌تنهایی هیچ اثری ندارد؛ نتیجه‌اش دور ریخته می‌شود. باید بنویسید <code dir="ltr">name = name.strip()</code>.

### ❓ پیش‌بینی کن

متدهای <code dir="ltr">is...</code> برای **اعتبارسنجی** داده‌ی ورودی پرکاربردند. کدام خط‌ها <code dir="ltr">True</code> چاپ می‌کنند؟ (به عدد منفی و عدد اعشاری دقت کنید.)

</div>

````python
print("2026".isdigit())
print("-12".isdigit())
print("3.5".isdigit())
print("".isdigit())
print(" 42 ".strip().isdigit())
````

<div dir="rtl">

🔍 <code dir="ltr">isdigit</code> فقط وقتی <code dir="ltr">True</code> است که رشته **غیرخالی** باشد و **همه‌ی** کاراکترها رقم باشند؛ پس <code dir="ltr">"-12"</code> (علامت منفی) و <code dir="ltr">"3.5"</code> (نقطه) نادرست‌اند. بنابراین <code dir="ltr">isdigit</code> برای تشخیص «آیا این رشته یک عدد است؟» کافی نیست؛ در جلسه‌ی ۱۵ راه درست را (تلاش برای تبدیل با <code dir="ltr">float</code> و مدیریت خطا) می‌بینیم.

</div>

## ۳. جست‌وجو و جایگزینی

### 💡 مفهوم

| متد | کار | اگر پیدا نشود |
|---|---|---|
| <code dir="ltr">s.find(sub)</code> | اندیس **اولین** وقوع <code dir="ltr">sub</code> | <code dir="ltr">-1</code> |
| <code dir="ltr">s.index(sub)</code> | مثل <code dir="ltr">find</code> | خطای <code dir="ltr">ValueError</code> |
| <code dir="ltr">s.count(sub)</code> | تعداد وقوع‌ها (بدون هم‌پوشانی) | <code dir="ltr">0</code> |
| <code dir="ltr">s.replace(old, new)</code> | جایگزینی **همه‌ی** وقوع‌ها | همان رشته |
| <code dir="ltr">s.replace(old, new, k)</code> | جایگزینی <code dir="ltr">k</code> وقوع اول | همان رشته |

</div>

````python
email = "student.ali@pgu.ac.ir"
at = email.find("@")
print(at)
print("user:", email[:at])
print("domain:", email[at + 1:])
print(email.count("."), email.find("#"))
print(email.replace(".", "[dot]"))
````

<div dir="rtl">

### 🔍 کد را بخوان

الگوی «پیدا کن و برش بزن» بسیار رایج است: <code dir="ltr">find</code> موقعیت <code dir="ltr">@</code> را می‌دهد و با برش، بخش قبل و بعد از آن جدا می‌شود. چون <code dir="ltr">@</code> خودش جزو هیچ‌کدام نیست، بخش دوم از <code dir="ltr">at + 1</code> شروع می‌شود.

### ❓ پیش‌بینی کن

</div>

````python
s = "banana"
print(s.find("an"), s.find("an", 2))
print(s.count("ana"))
print(s.replace("a", "A", 2))
print(s.index("x"))
````

<div dir="rtl">

🔍 آرگومان دوم <code dir="ltr">find</code> محل **شروع** جست‌وجو است؛ پس <code dir="ltr">s.find("an", 2)</code> وقوع دوم (<code dir="ltr">3</code>) را پیدا می‌کند. <code dir="ltr">count("ana")</code> برابر <code dir="ltr">1</code> است، نه <code dir="ltr">2</code>: وقوع‌های **هم‌پوشان** (<code dir="ltr">b[ana]na</code> و <code dir="ltr">ban[ana]</code>) فقط یک بار شمرده می‌شوند. <code dir="ltr">replace</code> با آرگومان سوم فقط دو <code dir="ltr">a</code> اول را عوض می‌کند. خط آخر <code dir="ltr">ValueError</code> می‌دهد.

### ⚠️ اشتباه رایج: <code dir="ltr">if s.find(x):</code>

<code dir="ltr">find</code> در صورت پیدا نکردن <code dir="ltr">-1</code> برمی‌گرداند که در شرط **درست** حساب می‌شود، و اگر زیررشته در **اول** رشته باشد، <code dir="ltr">0</code> برمی‌گرداند که **نادرست** حساب می‌شود؛ یعنی دقیقاً برعکس منظور شما! برای بررسی وجود، از <code dir="ltr">in</code> استفاده کنید.

## ۴. <code dir="ltr">split</code> و <code dir="ltr">join</code>

### 💡 مفهوم

در جلسه‌ی ۸ با این دو آشنا شدید. حالا جزئیات مهم‌تر:

- <code dir="ltr">s.split()</code> (بدون آرگومان): روی **هر تعداد فاصله، tab یا خط جدید** می‌شکند و رشته‌های خالی را دور می‌اندازد.
- <code dir="ltr">s.split(sep)</code>: دقیقاً روی <code dir="ltr">sep</code> می‌شکند؛ جداکننده‌های پشت سر هم، رشته‌ی **خالی** تولید می‌کنند.
- <code dir="ltr">s.split(sep, maxsplit)</code>: حداکثر <code dir="ltr">maxsplit</code> بار می‌شکند.
- <code dir="ltr">sep.join(list&#95;of&#95;strings)</code>: اعضای لیست را با <code dir="ltr">sep</code> به هم می‌چسباند.

</div>

````python
messy = "  the   cat\tsat \n on  the mat "
print(messy.split())
print("a,,b".split(","))
print("key=value=with=equals".split("=", 1))

words = ["2026", "10", "30"]
print("-".join(words))
print("".join(["P", "y", "t", "h", "o", "n"]))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">messy.split()</code> فاصله‌های اضافه، tab و خط جدید را همه با هم نادیده می‌گیرد؛ معمولاً همین را برای جدا کردن کلمات می‌خواهید.
- <code dir="ltr">"a,,b".split(",")</code> بین دو کامای پشت سر هم یک رشته‌ی خالی تولید می‌کند: <code dir="ltr">['a', '', 'b']</code>. در فایل‌های CSV یعنی **مقدار گمشده**.
- با <code dir="ltr">maxsplit=1</code>، فقط اولین <code dir="ltr">=</code> جداکننده است؛ بقیه جزو مقدار می‌مانند.

### 💡 تجزیه‌ی یک خط داده

ترکیب <code dir="ltr">split</code>، <code dir="ltr">strip</code> و تبدیل نوع، اساس خواندن داده‌ی متنی است:

</div>

````python
line = "Ali, 17.5 ,Bushehr\n"
fields = line.split(",")
print(fields)

name = fields[0].strip()
score = float(fields[1])
city = fields[2].strip()
print(name, score, city)
````

<div dir="rtl">

🔍 بعد از <code dir="ltr">split</code>، فیلدها فاصله‌های اضافه و <code dir="ltr">&#92;n</code> انتهای خط را دارند. <code dir="ltr">float</code> خودش فاصله‌های دو طرف را تحمل می‌کند، اما برای رشته‌ها باید صریحاً <code dir="ltr">strip</code> کنید؛ وگرنه <code dir="ltr">"Bushehr&#92;n"</code> با <code dir="ltr">"Bushehr"</code> برابر نیست و مثلاً در شمارش (جلسه‌ی ۱۰) دو شهر جدا حساب می‌شوند.

### ⚠️ اشتباه رایج: ساختن رشته با <code dir="ltr">+=</code> در حلقه

چون رشته تغییرناپذیر است، هر <code dir="ltr">result += piece</code> یک رشته‌ی **کاملاً جدید** می‌سازد و همه‌ی محتوای قبلی را کپی می‌کند. برای چند ده قطعه مهم نیست، اما برای هزاران قطعه بسیار کند است. الگوی درست: قطعه‌ها را در یک **لیست** جمع کنید و در آخر یک بار <code dir="ltr">join</code> کنید.

</div>

````python
parts = []
for i in range(5):
    parts.append(str(i ** 2))
result = ", ".join(parts)
print(result)
````

## ۵. پیش‌پردازش متن و نکته‌های قالب‌بندی

### 💡 خط لوله‌ی پیش‌پردازش

قبل از اینکه متن وارد هر مدل یادگیری ماشین شود، باید **یکنواخت** شود: «Data»، «data» و «data,» باید یک کلمه حساب شوند. یک خط لوله‌ی ساده:

1. همه‌ی حروف کوچک شوند،
2. علامت‌های نگارشی با فاصله جایگزین شوند،
3. متن به کلمه‌ها (**توکن**ها) شکسته شود،
4. کلمه‌های بسیار رایج و کم‌اطلاعات (**stop words**) حذف شوند.

ماژول <code dir="ltr">string</code> (از کتابخانه‌ی استاندارد) فهرست آماده‌ای از علامت‌های نگارشی انگلیسی دارد:

</div>

````python
import string
print(string.punctuation)

def tokenize(text):
    text = text.lower()
    clean = []
    for ch in text:
        if ch in string.punctuation:
            clean.append(" ")
        else:
            clean.append(ch)
    return "".join(clean).split()

stop_words = {"the", "a", "an", "on", "is", "and"}
review = "The movie was GREAT, and the music... the music is AMAZING!"

tokens = tokenize(review)
print(tokens)

content = []
for t in tokens:
    if t not in stop_words:
        content.append(t)
print(content)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">tokenize</code> کاراکترها را یکی‌یکی بررسی می‌کند و علامت‌های نگارشی را با فاصله عوض می‌کند. قطعه‌ها در یک لیست جمع و در آخر <code dir="ltr">join</code> می‌شوند (الگوی بخش ۴).
- <code dir="ltr">split()</code> بدون آرگومان، فاصله‌های اضافه‌ای را که از حذف علامت‌ها به وجود آمده نادیده می‌گیرد.
- <code dir="ltr">stop&#95;words</code> یک **مجموعه** است؛ بررسی <code dir="ltr">t not in stop&#95;words</code> فوری است (جلسه‌ی ۹).

نتیجه، فهرستی از کلمه‌های پرمعناست که حالا می‌توان آن را شمرد (جلسه‌ی ۱۰)، به عدد تبدیل کرد (<code dir="ltr">word2idx</code>) و به یک مدل داد.

### 💡 یادآوری قالب‌بندی و دو نکته‌ی تازه

قالب‌بندی با f-string را در جلسه‌ی ۴ کامل دیدید (<code dir="ltr">:.2f</code>، <code dir="ltr">:,</code>، <code dir="ltr">:&gt;8</code> و ...). دو نکته‌ی تکمیلی:

**رشته‌ی خام (raw string):** در مسیرهای ویندوز، <code dir="ltr">&#92;n</code> و <code dir="ltr">&#92;t</code> معنی خط جدید و tab دارند و مسیر را خراب می‌کنند. با گذاشتن <code dir="ltr">r</code> قبل از نقل‌قول، علامت <code dir="ltr">&#92;</code> معمولی در نظر گرفته می‌شود:

</div>

````python
print("C:\new\table.csv")
print(r"C:\new\table.csv")
````

<div dir="rtl">

**رشته‌ی چندخطی** با سه نقل‌قول، شکسته شدن خط‌ها را حفظ می‌کند؛ برای متن‌های طولانی و docstringها (جلسه‌ی ۷) استفاده می‌شود.

</div>

## 📌 آنچه باید بلد باشید

- رشته دنباله‌ای تغییرناپذیر است؛ اندیس، برش، <code dir="ltr">in</code>، <code dir="ltr">+</code>، <code dir="ltr">&#42;</code> و حلقه مثل تاپل؛ انتساب به <code dir="ltr">s[i]</code> خطاست.
- <code dir="ltr">ord</code> و <code dir="ltr">chr</code> و ترتیب یونیکد.

### سؤال ۳ — n-gram

بسیاری از مدل‌های زبانی به جای تک‌کلمه‌ها، دنباله‌های <code dir="ltr">n</code> تایی از توکن‌ها (**n-gram**) را می‌شمارند:

</div>

```python
def ngrams(tokens, n):
    result = []
    for i in range(len(tokens) - n + 1):
        result.append(tuple(tokens[i:i + n]))
    return result
```

<div dir="rtl">

الف) خروجی <code dir="ltr">ngrams(["i", "love", "data", "science"], 2)</code> و <code dir="ltr">ngrams(["a", "b"], 5)</code> چیست؟ چرا حلقه تا <code dir="ltr">len(tokens) - n + 1</code> است؟ اگر به جای آن <code dir="ltr">len(tokens)</code> بود، چه می‌شد؟

ب) <code dir="ltr">ngrams(list("data"), 3)</code> چه چیزی برمی‌گرداند؟ (به این‌ها **character n-gram** می‌گویند.)

ج) چرا n-gramها به شکل **تاپل** ذخیره شده‌اند و نه لیست؟ (راهنمایی: قدم بعدی، شمردن آن‌ها با یک دیکشنری است.)

د) یک متن <code dir="ltr">N</code> توکنی چند <code dir="ltr">n</code>-gram دارد؟ اگر واژگان <code dir="ltr">V</code> کلمه داشته باشد، حداکثر چند bigram **متفاوت** ممکن است وجود داشته باشد؟ این چه مشکلی برای مدل‌های مبتنی بر شمارش ایجاد می‌کند؟

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">[('i', 'love'), ('love', 'data'), ('data', 'science')]</code> و <code dir="ltr">[]</code>. آخرین <code dir="ltr">n</code>-gram از اندیس <code dir="ltr">len - n</code> شروع می‌شود؛ پس <code dir="ltr">range</code> باید تا <code dir="ltr">len - n + 1</code> باشد (<code dir="ltr">stop</code> شامل نیست). با <code dir="ltr">len(tokens)</code>، برش‌های آخر کوتاه‌تر از <code dir="ltr">n</code> می‌شدند (مثلاً <code dir="ltr">('science',)</code>)؛ بدون هیچ خطایی، n-gramهای ناقص تولید می‌شد.

def is_palindrome(s):
    clean = []
    for ch in s.lower():
        if ch.isalnum():

    c = "".join(clean)
    return c == c[::-1]

def bag_of_words(text):

    dot = 0
    for w in a:
        dot += a[w] * ______
    norm_a = math.sqrt(sum(v * v for v in a.values()))
    norm_b = math.sqrt(sum(v * v for v in b.values()))

## خلاصه‌ی جلسه

- رشته دنباله‌ی تغییرناپذیری از کاراکترهای یونیکد است؛ اندیس، برش و حلقه مثل تاپل.
- متدهای رشته همیشه مقدار جدید برمی‌گردانند؛ نتیجه را انتساب کنید و برای خوانایی آن‌ها را زنجیر کنید.
- <code dir="ltr">find</code> / <code dir="ltr">index</code> / <code dir="ltr">count</code> / <code dir="ltr">replace</code> برای جست‌وجو و جایگزینی؛ برای بررسی وجود از <code dir="ltr">in</code> استفاده کنید.
- <code dir="ltr">split</code> و <code dir="ltr">join</code> پل بین رشته و لیست‌اند؛ برای ساختن رشته‌های بزرگ، لیست بسازید و یک بار <code dir="ltr">join</code> کنید.
- پیش‌پردازش متن: حروف کوچک، حذف علامت‌ها، توکن‌سازی و حذف stop words؛ پایه‌ی هر کار NLP.

**جلسه‌ی بعد:** پروژه‌ی ۱، تحلیل کارنامه‌ی دانشجویان یک دانشکده با همه‌ی ابزارهای بخش اول.
