## ۱. چرا شیءگرایی؟

### 💡 مفهوم

تا اینجا داده را در لیست، تاپل و دیکشنری نگه داشتید و کار روی داده را با **تابع**‌های جدا انجام دادید. مثلاً یک مستطیل را می‌شود این‌طور نشان داد:

</div>

````python
rect = {"length": 3, "width": 4}

def area(r):
    return r["length"] * r["width"]

def perimeter(r):
    return 2 * (r["length"] + r["width"])

print(area(rect), perimeter(rect))

rect["lenght"] = 10          # typo: a new key is silently created
print(area(rect), rect)
````

<div dir="rtl">

### 🔍 کد را بخوان

این روش کار می‌کند، اما سه ضعف دارد:

- **داده و رفتار جدا هستند**: هیچ چیزی نمی‌گوید تابع <code dir="ltr">area</code> مخصوص این نوع دیکشنری است؛ هر دیکشنری دیگری را هم می‌شود به آن داد.
- **ساختار تضمین نمی‌شود**: یک غلط تایپی (<code dir="ltr">"lenght"</code>) بی‌صدا یک کلید تازه می‌سازد و مساحت عوض نمی‌شود.
- **با بزرگ شدن برنامه**، ده‌ها تابع و دیکشنری با ساختارهای مختلف کنار هم قرار می‌گیرند و ردگیری اینکه کدام تابع با کدام داده کار می‌کند سخت می‌شود.

**برنامه‌نویسی شیءگرا (OOP)** داده و تابع‌های مربوط به آن را در یک واحد به نام **کلاس** جمع می‌کند. هر **شیء** (object) ساخته‌شده از کلاس، داده‌ی خودش را دارد و تابع‌هایش (که به آن‌ها **متد** می‌گوییم) را با خودش حمل می‌کند.

### 💡 شما از اول دوره با شیءها کار کرده‌اید

در پایتون **همه چیز شیء است**: عدد، رشته، لیست، دیکشنری و حتی تابع. هر شیء یک **نوع** (کلاس) دارد و متدهای آن کلاس را می‌شناسد:

</div>

````python
nums = [3, 1, 2]
nums.append(0)               # a method of the list class
print(nums, type(nums))
print("data".upper(), type("data"))
print(type(5), type(2.5), type(True), type(len))
print(isinstance(5, int), isinstance(5, str))
````

<div dir="rtl">

🔍 <code dir="ltr">nums.append(0)</code> یعنی «متد <code dir="ltr">append</code> شیء <code dir="ltr">nums</code> را صدا بزن». <code dir="ltr">type</code> کلاس یک شیء را نشان می‌دهد و <code dir="ltr">isinstance(obj, C)</code> بررسی می‌کند که آیا <code dir="ltr">obj</code> نمونه‌ای از کلاس <code dir="ltr">C</code> است. در این جلسه یاد می‌گیرید **کلاس‌های خودتان** را بسازید. کتابخانه‌هایی مثل NumPy، Pandas و scikit-learn دقیقاً همین‌طور ساخته شده‌اند: <code dir="ltr">DataFrame</code> یک کلاس است و <code dir="ltr">LinearRegression().fit(X, y)</code> یعنی ساختن یک شیء و صدا زدن متد <code dir="ltr">fit</code> آن.

</div>

## ۲. کلاس و شیء

### 💡 مفهوم

**کلاس** نقشه یا قالب است؛ **شیء** (یا **نمونه**، instance) چیزی است که از روی آن نقشه ساخته می‌شود. از یک کلاس می‌توان هر تعداد شیء ساخت و هر کدام مستقل از بقیه است. ساده‌ترین کلاس ممکن:

</div>

````python
class Rectangle:
    pass                     # an empty class body

r1 = Rectangle()             # calling the class creates a new object
r2 = Rectangle()

print(type(r1))
print(isinstance(r1, Rectangle))
print(r1 is r2)
print(r1)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">class Rectangle:</code> یک کلاس تازه تعریف می‌کند. طبق قرارداد، نام کلاس با حرف بزرگ و به شکل <code dir="ltr">CamelCase</code> نوشته می‌شود (برخلاف نام تابع و متغیر که <code dir="ltr">snake&#95;case</code> است).
- <code dir="ltr">pass</code> یعنی «فعلاً هیچ چیزی»؛ بدنه‌ی کلاس نمی‌تواند خالی باشد.
- **صدا زدن کلاس** مثل یک تابع (<code dir="ltr">Rectangle()</code>) یک شیء تازه می‌سازد. <code dir="ltr">r1 is r2</code> نادرست است: دو شیء جدا در حافظه‌اند.
- چاپ یک شیء فعلاً چیز زیبایی نشان نمی‌دهد: نام کلاس و آدرس حافظه (<code dir="ltr">0x...</code>). در جلسه‌ی ۱۹ با متد ویژه‌ی <code dir="ltr">&#95;&#95;str&#95;&#95;</code> این را درست می‌کنیم.

</div>

## ۳. متد <code dir="ltr">&#95;&#95;init&#95;&#95;</code> و <code dir="ltr">self</code>

### 💡 مفهوم

کلاس خالی به کاری نمی‌آید. می‌خواهیم هر مستطیل هنگام ساخته شدن **طول** و **عرض** خودش را بگیرد و بتواند مساحتش را حساب کند:

</div>

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r1 = Rectangle(3, 4)
r2 = Rectangle(5, 2)
print(r1.length, r1.width, r1.area(), r1.perimeter())
print(r2.length, r2.width, r2.area())
print(Rectangle.area(r1))
````

<div dir="rtl">

### 🔍 کد را بخوان

وقتی می‌نویسید <code dir="ltr">r1 = Rectangle(3, 4)</code>، این سه اتفاق به ترتیب می‌افتد:

1. پایتون یک شیء **خالی** از کلاس <code dir="ltr">Rectangle</code> می‌سازد.
2. متد <code dir="ltr">&#95;&#95;init&#95;&#95;</code> را صدا می‌زند و آن شیء خالی را به‌عنوان **اولین آرگومان** به آن می‌دهد: <code dir="ltr">&#95;&#95;init&#95;&#95;(obj, 3, 4)</code>. درون <code dir="ltr">&#95;&#95;init&#95;&#95;</code> این شیء با نام <code dir="ltr">self</code> در دسترس است و <code dir="ltr">self.length = length</code> یک **ویژگی** (attribute) به نام <code dir="ltr">length</code> روی همین شیء می‌سازد.
3. شیء ساخته‌شده به نام <code dir="ltr">r1</code> نسبت داده می‌شود.

به <code dir="ltr">&#95;&#95;init&#95;&#95;</code> (با دو زیرخط در هر طرف) **سازنده** یا initializer می‌گویند. لازم نیست آن را مستقیماً صدا بزنید؛ با هر بار ساختن شیء خودکار اجرا می‌شود.

**متد** همان تابعی است که داخل کلاس تعریف شده. فراخوانی <code dir="ltr">r1.area()</code> در واقع خلاصه‌ی <code dir="ltr">Rectangle.area(r1)</code> است: پایتون شیء سمت چپ نقطه را خودش به‌عنوان <code dir="ltr">self</code> می‌فرستد. برای همین <code dir="ltr">area</code> در تعریف یک پارامتر (<code dir="ltr">self</code>) دارد ولی هنگام صدا زدن بدون آرگومان نوشته می‌شود. <code dir="ltr">self</code> یک کلمه‌ی کلیدی نیست، فقط یک قرارداد بسیار محکم است؛ همیشه از همین نام استفاده کنید.

</div>

### ⚠️ اشتباه رایج: فراموش کردن <code dir="ltr">self</code>

دو خطای بسیار رایج: (۱) نوشتن متد بدون پارامتر <code dir="ltr">self</code>، و (۲) انتساب بدون <code dir="ltr">self.</code> در <code dir="ltr">&#95;&#95;init&#95;&#95;</code>:

</div>

````python
class Broken:
    def __init__(self, length):
        length = length          # only a local variable; nothing is stored on the object

    def show():                  # forgot self
        print("hello")

b = Broken(3)
b.show()
````

````python
print(b.length)
````

<div dir="rtl">

🔍 فراخوانی <code dir="ltr">b.show()</code> در واقع <code dir="ltr">Broken.show(b)</code> است؛ اما <code dir="ltr">show</code> هیچ پارامتری ندارد، پس خطای <code dir="ltr">TypeError: ... takes 0 positional arguments but 1 was given</code> می‌گیریم. این پیام عجیب («من که آرگومانی ندادم!») تقریباً همیشه یعنی <code dir="ltr">self</code> را فراموش کرده‌اید. در سلول دوم، <code dir="ltr">length = length</code> فقط یک متغیر **محلی** می‌سازد که با پایان <code dir="ltr">&#95;&#95;init&#95;&#95;</code> از بین می‌رود (دامنه‌ی متغیرها، جلسه‌ی ۷)؛ پس شیء ویژگی <code dir="ltr">length</code> ندارد و <code dir="ltr">AttributeError</code> می‌گیریم.

### ❓ پیش‌بینی کن

با کلاس <code dir="ltr">Rectangle</code> بالا، هر خط چه چاپ می‌کند یا چه خطایی می‌دهد؟

</div>

````python
sq = Rectangle(2, 2)
print(sq.area() + Rectangle(1, 3).area())
print(Rectangle.perimeter(sq))
print(Rectangle(3))
````

<div dir="rtl">

🔍 <code dir="ltr">4 + 3 = 7</code>؛ سپس <code dir="ltr">8</code>. خط آخر <code dir="ltr">TypeError</code> می‌دهد: <code dir="ltr">&#95;&#95;init&#95;&#95;</code> به دو آرگومان (<code dir="ltr">length</code> و <code dir="ltr">width</code>) نیاز دارد و فقط یکی داده شده. پیام خطا می‌گوید <code dir="ltr">missing 1 required positional argument: 'width'</code>؛ یعنی <code dir="ltr">self</code> را خود پایتون فرستاده است.

## ۴. ویژگی‌ها: ویژگی نمونه و ویژگی کلاس

### 💡 ویژگی نمونه

ویژگی‌هایی که با <code dir="ltr">self.name = ...</code> ساخته می‌شوند **ویژگی نمونه** (instance attribute) هستند: هر شیء نسخه‌ی خودش را دارد. می‌توان آن‌ها را خواند، تغییر داد و حتی بعداً ویژگی تازه اضافه کرد. تابع <code dir="ltr">vars</code> ویژگی‌های یک شیء را به شکل یک دیکشنری نشان می‌دهد:

</div>

````python
r1 = Rectangle(3, 4)
r2 = Rectangle(5, 2)
r1.width = 10                # change one object only
r1.color = "red"             # a brand-new attribute on r1 only
print(r1.area(), r2.area())
print(vars(r1))
print(vars(r2))
````

<div dir="rtl">

🔍 هر شیء در درون، یک دیکشنری از ویژگی‌هایش دارد؛ پس شیء چیزی شبیه دیکشنری بخش ۱ است، با این تفاوت که **متدها** همراهش هستند و ساختار اولیه‌اش را <code dir="ltr">&#95;&#95;init&#95;&#95;</code> تعیین می‌کند. اینکه می‌شود بعداً ویژگی تازه (<code dir="ltr">color</code>) اضافه کرد امکان است، نه توصیه: همه‌ی ویژگی‌ها را در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> بسازید تا همه‌ی شیءهای یک کلاس ساختار یکسانی داشته باشند.

### 💡 ویژگی کلاس

متغیری که مستقیماً در بدنه‌ی کلاس (بیرون از هر متد) تعریف شود **ویژگی کلاس** (class attribute) است: یک نسخه دارد که بین **همه‌ی** شیءها مشترک است. برای ثابت‌ها و مقدارهای مشترک مناسب است:

</div>

````python
class Product:
    discount = 0.1                       # class attribute: shared

    def __init__(self, name, price):
        self.name = name                 # instance attributes
        self.price = price

    def final_price(self):
        return self.price * (1 - self.discount)

p1 = Product("Laptop", 1500)
p2 = Product("Phone", 1000)
print(p1.final_price(), p2.final_price())

Product.discount = 0.2                   # change it on the class: affects everyone
print(p1.final_price(), p2.final_price())

p1.discount = 0.5                        # creates an INSTANCE attribute on p1
print(p1.final_price(), p2.final_price(), Product.discount)
print(vars(p1))
````

<div dir="rtl">

### 🔍 کد را بخوان

وقتی می‌نویسید <code dir="ltr">obj.x</code>، پایتون به این ترتیب جست‌وجو می‌کند:

1. آیا خود **شیء** ویژگی <code dir="ltr">x</code> دارد؟ اگر بله، همان.
2. وگرنه آیا **کلاس** آن ویژگی <code dir="ltr">x</code> دارد؟ اگر بله، همان.
3. وگرنه <code dir="ltr">AttributeError</code>.

پس <code dir="ltr">self.discount</code> در ابتدا از کلاس خوانده می‌شود و تغییر <code dir="ltr">Product.discount</code> روی هر دو محصول اثر دارد. اما **انتساب** <code dir="ltr">p1.discount = 0.5</code> ویژگی کلاس را عوض نمی‌کند؛ یک ویژگی **نمونه** با همان نام روی <code dir="ltr">p1</code> می‌سازد که ویژگی کلاس را برای <code dir="ltr">p1</code> **می‌پوشاند**. <code dir="ltr">vars(p1)</code> حالا کلید <code dir="ltr">discount</code> را دارد ولی <code dir="ltr">vars(p2)</code> ندارد.

### ⚠️ اشتباه رایج: ویژگی کلاسِ تغییرپذیر

اگر ویژگی کلاس یک لیست یا دیکشنری باشد و آن را **درجا** تغییر دهید (مثلاً با <code dir="ltr">append</code>)، هیچ انتسابی رخ نمی‌دهد؛ پس همه‌ی شیءها همان یک لیست مشترک را تغییر می‌دهند:

</div>

````python
class Basket:
    items = []                           # WRONG: one list shared by all baskets

    def add(self, item):
        self.items.append(item)

a = Basket()
b = Basket()
a.add("apple")
print(a.items, b.items, a.items is b.items)
````

<div dir="rtl">

🔍 <code dir="ltr">self.items</code> چون روی شیء پیدا نمی‌شود، از کلاس خوانده می‌شود و <code dir="ltr">append</code> همان لیست مشترک را تغییر می‌دهد؛ سبد <code dir="ltr">b</code> بدون اینکه چیزی به آن اضافه شده باشد، سیب دارد! این همان مشکل ارجاع در لیست‌هاست (جلسه‌ی ۸). راه درست: هر داده‌ی تغییرپذیری که مال یک شیء است در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> ساخته شود:

<div dir="ltr">

```python
class Basket:
    def __init__(self):
        self.items = []          # a new list for every basket
```

## ۵. متدها و شیء به‌عنوان ارجاع

### 💡 متدهایی که وضعیت را تغییر می‌دهند یا مقدار برمی‌گردانند

ویژگی‌های یک شیء روی هم **وضعیت** (state) آن را می‌سازند. متدها دو نوع کلی دارند: متدهایی که وضعیت را **تغییر می‌دهند** (مثل <code dir="ltr">list.append</code>) و متدهایی که از روی وضعیت یک مقدار **حساب می‌کنند و برمی‌گردانند** (مثل <code dir="ltr">list.count</code>). کلاس زیر آمار یک جریان از داده را بدون نگه داشتن خود داده‌ها به‌روز می‌کند:

</div>

````python
class RunningStats:
    def __init__(self):
        self.n = 0
        self.total = 0.0
        self.largest = None

    def add(self, x):                    # changes the state
        self.n += 1
        self.total += x
        if self.largest is None or x > self.largest:
            self.largest = x
        return self                      # allows chaining

    def mean(self):                      # only computes and returns
        return self.total / self.n

    def summary(self):
        return f"n={self.n}, mean={self.mean():.2f}, max={self.largest}"

s = RunningStats()
s.add(4).add(10).add(1)
print(s.summary())
print(s.n, s.total)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">add</code> ویژگی‌های شیء را تغییر می‌دهد. چون در آخر <code dir="ltr">return self</code> دارد، حاصل <code dir="ltr">s.add(4)</code> خود <code dir="ltr">s</code> است و می‌توان روی آن دوباره <code dir="ltr">.add(10)</code> صدا زد؛ به این **زنجیر کردن متدها** (method chaining) می‌گویند. در scikit-learn متد <code dir="ltr">fit</code> دقیقاً همین کار را می‌کند و برای همین <code dir="ltr">model.fit(X, y).predict(X&#95;new)</code> رایج است.
- <code dir="ltr">summary</code> متد دیگری از **همان شیء** را با <code dir="ltr">self.mean()</code> صدا می‌زند. بدون <code dir="ltr">self.</code>، پایتون دنبال یک تابع سراسری به نام <code dir="ltr">mean</code> می‌گشت.
- این کلاس فقط سه عدد نگه می‌دارد، هر قدر هم داده بیاید؛ ایده‌ای که در آموزش شبکه‌های عصبی روی دسته‌های داده (جلسه‌ی ۱۴) هم به کار می‌رود.

### سؤال ۳ — نورون و لایه

یک نورون مصنوعی، ترکیب خطی ورودی‌ها را حساب می‌کند و از یک تابع فعال‌سازی عبور می‌دهد (جلسه‌های ۷ و ۸):

</div>

```python
import math

class Neuron:
    def __init__(self, weights, bias, activation="step"):
        self.weights = weights
        self.bias = bias
        self.activation = activation

    def forward(self, x):
        z = sum(w * xi for w, xi in zip(self.weights, x)) + self.bias
        if self.activation == "step":
            return 1 if z > 0 else 0
        if self.activation == "relu":
            return max(0, z)
        return 1 / (1 + math.exp(-z))         # sigmoid

class Layer:
    def __init__(self, neurons):
        self.neurons = neurons

    def forward(self, x):
        return [n.forward(x) for n in self.neurons]

    def n_params(self):
        return sum(len(n.weights) + 1 for n in self.neurons)
```

<div dir="rtl">

الف) برای <code dir="ltr">n = Neuron([0.5, -1.0], 0.25, "relu")</code>، حاصل <code dir="ltr">n.forward([2, 1])</code> و <code dir="ltr">n.forward([1, 2])</code> چیست؟ اگر فعال‌سازی <code dir="ltr">"sigmoid"</code> بود، اولی (با سه رقم اعشار) چه می‌شد؟

ب) وزن‌ها و بایاسی برای یک نورون <code dir="ltr">"step"</code> پیدا کنید که دروازه‌ی منطقی **AND** را روی ورودی‌های <code dir="ltr">(0, 0)</code>، <code dir="ltr">(0, 1)</code>، <code dir="ltr">(1, 0)</code> و <code dir="ltr">(1, 1)</code> پیاده کند. همین کار را برای **OR** و **NAND** انجام دهید.

ج) XOR همان «OR و NAND» است:

ج) <code dir="ltr">w ≈ 2.0</code> و <code dir="ltr">b ≈ 1.0</code> (داده دقیقاً روی خط <code dir="ltr">y = 2x + 1</code> است)، و <code dir="ltr">predict([5, 10])</code> تقریباً <code dir="ltr">[11.0, 21.0]</code>.

د) تا بتوان نوشت <code dir="ltr">model = LinearRegression().fit(X, y)</code> یا <code dir="ltr">LinearRegression().fit(X, y).predict(X&#95;new)</code>؛ همان قرارداد scikit-learn. فراخوانی <code dir="ltr">predict</code> قبل از <code dir="ltr">fit</code> خطای <code dir="ltr">ValueError</code> با پیام روشن می‌دهد (جلسه‌ی ۱۵). بدون این بررسی، <code dir="ltr">self.w &#42; xi</code> با <code dir="ltr">None</code> خطای مبهم <code dir="ltr">TypeError</code> می‌داد؛ و اگر <code dir="ltr">w</code> به‌جای <code dir="ltr">None</code> با <code dir="ltr">0</code> شروع شده بود، مدل بی‌صدا پیش‌بینی‌های صفر (یعنی فقط <code dir="ltr">b = 0</code>) برمی‌گرداند؛ بدترین نوع خطا، چون دیده نمی‌شود.

ه) **واگرایی**: نرخ یادگیری آن‌قدر بزرگ است که هر قدم از کمینه رد می‌شود و به نقطه‌ای با خطای بیشتر می‌رسد؛ با هر دور، فاصله بزرگ‌تر می‌شود (همان رفتاری که در جلسه‌ی ۷ برای کاهش گرادیان یک‌متغیره دیدید). باید <code dir="ltr">lr</code> را کم کرد یا ویژگی‌ها را مقیاس‌بندی کرد (سؤال ۵).

</details>

### سؤال ۵ — استانداردسازی ویژگی‌ها

بیشتر الگوریتم‌های یادگیری ماشین وقتی ویژگی‌ها هم‌مقیاس باشند بهتر کار می‌کنند. **استانداردسازی** هر مقدار را به <code dir="ltr">(x − mean) / std</code> تبدیل می‌کند:

</div>

```python
import math

class StandardScaler:
    def fit(self, data):
        n = len(data)
        self.mean = sum(data) / n
        self.std = math.sqrt(sum((v - self.mean) ** 2 for v in data) / n)
        return self

    def transform(self, data):
        return [(v - self.mean) / self.std for v in data]

    def fit_transform(self, data):
        return self.fit(data).transform(data)

train = [2, 4, 4, 4, 5, 5, 7, 9]
test = [1, 5, 11]
```

<div dir="rtl">

الف) بعد از <code dir="ltr">scaler = StandardScaler().fit(train)</code>، مقدار <code dir="ltr">scaler.mean</code> و <code dir="ltr">scaler.std</code> و حاصل <code dir="ltr">scaler.transform(test)</code> چیست؟

ب) چرا این کلاس **دو** متد جدا <code dir="ltr">fit</code> و <code dir="ltr">transform</code> دارد و نه یک تابع که هر داده‌ای را مستقیماً استاندارد کند؟ دانشجویی برای داده‌ی آزمون نوشته است <code dir="ltr">StandardScaler().fit&#95;transform(test)</code>. چه ایرادی دارد؟

ج) اگر همه‌ی مقدارهای <code dir="ltr">train</code> برابر باشند (مثلاً <code dir="ltr">[3, 3, 3]</code>)، چه اتفاقی می‌افتد؟ کلاس را طوری تغییر دهید که این حالت را تحمل کند.

د) چرا <code dir="ltr">transform</code> قبل از <code dir="ltr">fit</code> خطای <code dir="ltr">AttributeError</code> می‌دهد؟ با الهام از سؤال ۴، این را به یک خطای واضح‌تر تبدیل کنید.

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">mean = 5.0</code>، <code dir="ltr">std = 2.0</code> (واریانس <code dir="ltr">32/8 = 4</code>) و <code dir="ltr">transform(test) = [-2.0, 0.0, 3.0]</code>.

ب) میانگین و انحراف معیار **پارامترهایی** هستند که باید فقط از داده‌ی **آموزش** یاد گرفته شوند (<code dir="ltr">fit</code>) و بعد روی هر داده‌ی دیگری با **همان** پارامترها اعمال شوند (<code dir="ltr">transform</code>). شیء این پارامترها را بین دو فراخوانی **نگه می‌دارد**؛ کاری که یک تابع تنها نمی‌تواند. <code dir="ltr">fit&#95;transform(test)</code> میانگین و انحراف معیار خود داده‌ی آزمون را حساب می‌کند (<code dir="ltr">5.67</code> و <code dir="ltr">4.11</code>)؛ پس آزمون با مقیاسی متفاوت از آموزش تبدیل می‌شود (<code dir="ltr">[-1.136, -0.162, 1.298]</code> به‌جای <code dir="ltr">[-2, 0, 3]</code>) و عدد <code dir="ltr">5</code> که در آموزش دقیقاً میانگین بود، دیگر صفر نیست. علاوه بر این، اطلاعات داده‌ی آزمون در پیش‌پردازش استفاده شده است (**نشت داده**، جلسه‌ی ۹).

ج) <code dir="ltr">std = 0</code> و <code dir="ltr">transform</code> خطای <code dir="ltr">ZeroDivisionError</code> می‌دهد. یک راه (همان کاری که scikit-learn می‌کند):

<div dir="ltr">

```python
        if self.std == 0:
            self.std = 1.0
```

</div>

در انتهای <code dir="ltr">fit</code>، قبل از <code dir="ltr">return self</code>. ویژگی ثابت اطلاعاتی ندارد و همه‌ی مقدارهایش به صفر تبدیل می‌شوند.

د) ویژگی‌های <code dir="ltr">mean</code> و <code dir="ltr">std</code> فقط درون <code dir="ltr">fit</code> ساخته می‌شوند. بهتر است در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> آن‌ها را <code dir="ltr">None</code> قرار دهیم و در <code dir="ltr">transform</code> بررسی کنیم:

<div dir="ltr">

```python
    def __init__(self):
        self.mean = None
        self.std = None

    def transform(self, data):
        if self.mean is None:
            raise ValueError("call fit before transform")
        return [(v - self.mean) / self.std for v in data]
```

</div>

این همان قاعده‌ی بخش ۴ است: همه‌ی ویژگی‌ها را در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> بسازید.

</details>

### سؤال ۶ — کلاس واژگان

این کلاس، کلمه‌ها را به شماره تبدیل می‌کند (جلسه‌ی ۱۱):

</div>

```python
class Vocabulary:
    def __init__(self):
        self.word2idx = {"<pad>": 0, "<unk>": 1}

    def fit(self, texts):
        for text in texts:
            for w in text.lower().split():
                if w not in self.word2idx:
                    self.word2idx[w] = len(self.word2idx)
        return self

    def encode(self, text, length=None):
        ids = [self.word2idx.get(w, 1) for w in text.lower().split()]
        if length is not None:
            ids = ids[:length] + [0] * (length - len(ids))
        return ids

v = Vocabulary().fit(["the cat sat", "the dog ran"])
```

<div dir="rtl">

الف) <code dir="ltr">v.word2idx</code> را بنویسید. حاصل <code dir="ltr">v.encode("The cat ran fast")</code>، <code dir="ltr">v.encode("a dog", length=4)</code> و <code dir="ltr">v.encode("the cat sat on the mat", length=4)</code> چیست؟

ب) چرا <code dir="ltr">len(self.word2idx)</code> شماره‌ی کلمه‌ی تازه است؟ دو توکن ویژه‌ی <code dir="ltr">&lt;pad&gt;</code> و <code dir="ltr">&lt;unk&gt;</code> چه کاری می‌کنند و چرا <code dir="ltr">&lt;pad&gt;</code> شماره‌ی <code dir="ltr">0</code> دارد؟

ج) اگر بعد از کد بالا بنویسیم <code dir="ltr">v.fit(["a cat"])</code>، اندازه‌ی واژگان و حاصل <code dir="ltr">v.encode("a dog")</code> چه می‌شود؟ کدام متد وضعیت شیء را تغییر می‌دهد و کدام نه؟ چرا مهم است که <code dir="ltr">fit</code> فقط روی داده‌ی آموزش صدا زده شود؟

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">{'&lt;pad&gt;': 0, '&lt;unk&gt;': 1, 'the': 2, 'cat': 3, 'sat': 4, 'dog': 5, 'ran': 6}</code>. سپس <code dir="ltr">[2, 3, 6, 1]</code> (<code dir="ltr">fast</code> ناشناخته است)، <code dir="ltr">[1, 5, 0, 0]</code> (<code dir="ltr">a</code> ناشناخته؛ با <code dir="ltr">0</code> تا طول ۴ پر می‌شود) و <code dir="ltr">[2, 3, 4, 1]</code> (بعد از برش به ۴ توکن؛ <code dir="ltr">on</code> ناشناخته است).

ب) شماره‌ها از صفر و پشت‌سرهم داده می‌شوند؛ در لحظه‌ی اضافه کردن، تعداد کلمه‌های موجود دقیقاً اولین شماره‌ی آزاد است. <code dir="ltr">&lt;unk&gt;</code> جای هر کلمه‌ای است که در آموزش دیده نشده، تا <code dir="ltr">encode</code> هرگز <code dir="ltr">KeyError</code> ندهد. <code dir="ltr">&lt;pad&gt;</code> برای هم‌طول کردن دنباله‌ها است (مدل‌ها ورودی را به شکل یک دسته‌ی مستطیلی با طول ثابت می‌گیرند) و شماره‌ی <code dir="ltr">0</code> قراردادی است که کنار گذاشتن آن در محاسبه‌ها را ساده می‌کند.

ج) واژگان ۸ کلمه می‌شود (<code dir="ltr">a</code> شماره‌ی <code dir="ltr">7</code> می‌گیرد) و <code dir="ltr">encode("a dog")</code> برابر <code dir="ltr">[7, 5]</code>. <code dir="ltr">fit</code> وضعیت را تغییر می‌دهد و <code dir="ltr">encode</code> فقط از وضعیت می‌خواند. اگر <code dir="ltr">fit</code> روی داده‌ی آزمون هم صدا زده شود، کلمه‌های آزمون وارد واژگان می‌شوند و مدل به‌ظاهر «هیچ کلمه‌ی ناشناخته‌ای» نمی‌بیند؛ ارزیابی خوش‌بینانه می‌شود (نشت داده، مثل سؤال ۵).

</details>

### سؤال ۷ — اشتراک ناخواسته‌ی شیءها

با کلاس زیر:

</div>

```python
class Counter:
    def __init__(self):
        self.n = 0

    def inc(self):
        self.n += 1
        return self

def reset(c):
    c.n = 0

def replace(c):
    c = Counter()
    c.n = 99
```

<div dir="rtl">

الف) خروجی این دو تکه کد چیست؟ چرا متفاوت‌اند؟

</div>

```python
cs = [Counter()] * 3
cs[0].inc()
print([c.n for c in cs])

cs2 = [Counter() for _ in range(3)]
cs2[0].inc()
print([c.n for c in cs2])
```

<div dir="rtl">

ب) خروجی کد زیر چیست؟ چرا <code dir="ltr">replace</code> روی <code dir="ltr">k</code> اثری ندارد ولی <code dir="ltr">reset</code> دارد؟

</div>

```python
k = Counter().inc().inc().inc()
print(k.n)
reset(k)
print(k.n)
k.inc()
replace(k)
print(k.n)
```

<div dir="rtl">

ج) دانشجویی برای ساختن یک شبکه‌ی سه‌لایه نوشته است <code dir="ltr">layers = [Layer(neurons)] &#42; 3</code> (با <code dir="ltr">Layer</code> سؤال ۳). اگر بعداً وزن‌های لایه‌ی اول را در آموزش به‌روز کند، چه اتفاقی برای دو لایه‌ی دیگر می‌افتد؟ درستش را بنویسید.

<details>
<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">[1, 1, 1]</code> و <code dir="ltr">[1, 0, 0]</code>. <code dir="ltr">[Counter()] &#42; 3</code> یک شیء می‌سازد و لیستی با **سه ارجاع به همان شیء**؛ مثل <code dir="ltr">[[0] &#42; 3] &#42; 3</code> در جلسه‌ی ۸. comprehension در هر دور <code dir="ltr">Counter()</code> را دوباره صدا می‌زند و سه شیء مستقل می‌سازد.

ب) <code dir="ltr">3</code>، <code dir="ltr">0</code>، <code dir="ltr">1</code>. <code dir="ltr">reset</code> از راه پارامتر به **همان شیء** دسترسی دارد و ویژگی‌اش را تغییر می‌دهد. <code dir="ltr">replace</code> نام محلی <code dir="ltr">c</code> را به یک شیء **تازه** نسبت می‌دهد؛ این انتساب فقط نام محلی را عوض می‌کند و <code dir="ltr">k</code> همچنان به شیء قبلی اشاره می‌کند (تغییر شیء در برابر تغییر نام؛ جلسه‌های ۷ و ۸).

ج) هر سه عنصر یک شیء‌اند؛ پس «سه لایه» در واقع یک لایه است که سه بار پشت سر هم استفاده می‌شود و هر به‌روزرسانی روی هر سه دیده می‌شود. حتی بدتر: هر سه لایه به یک لیست <code dir="ltr">neurons</code> اشاره می‌کنند. درست، ساختن شیءهای مستقل با نورون‌های مستقل است:

<div dir="ltr">

```python
layers = [Layer([Neuron([0.0, 0.0], 0.0) for _ in range(2)]) for _ in range(3)]
```

</div>

(در شبکه‌های عصبی واقعی، گاهی **عمداً** یک لایه را در چند جا به اشتراک می‌گذارند تا وزن‌ها مشترک باشند؛ اما این باید یک تصمیم آگاهانه باشد، نه نتیجه‌ی یک <code dir="ltr">&#42; 3</code>.)

</details>

## خلاصه‌ی جلسه

- کلاس، داده (ویژگی‌ها) و رفتار (متدها) را در یک واحد جمع می‌کند؛ هر شیء ساخته‌شده از کلاس، وضعیت خودش را دارد.
- صدا زدن کلاس یک شیء تازه می‌سازد و <code dir="ltr">&#95;&#95;init&#95;&#95;</code> آن را مقداردهی می‌کند؛ <code dir="ltr">self</code> همان شیئی است که متد روی آن صدا زده شده.
- ویژگی نمونه مال یک شیء است و ویژگی کلاس بین همه مشترک؛ پایتون اول در شیء و بعد در کلاس جست‌وجو می‌کند؛ داده‌ی تغییرپذیرِ هر شیء را در <code dir="ltr">&#95;&#95;init&#95;&#95;</code> بسازید.
- متدها وضعیت را تغییر می‌دهند یا از آن مقدار حساب می‌کنند؛ <code dir="ltr">return self</code> زنجیر کردن را ممکن می‌کند (<code dir="ltr">fit(...).predict(...)</code>).
- شیءها ارجاع‌اند؛ انتساب یا فرستادن به تابع کپی نمی‌سازد.

**جلسه‌ی بعد:** شیءگرایی ۲؛ وراثت، بازنویسی متد، و متدهای ویژه مثل <code dir="ltr">&#95;&#95;str&#95;&#95;</code> و <code dir="ltr">&#95;&#95;len&#95;&#95;</code> که شیءهای شما را مثل انواع داخلی پایتون رفتار می‌دهند.
