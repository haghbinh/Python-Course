## ۱. نمودارهای پایه

### 💡 مفهوم

**Matplotlib** پرکاربردترین کتابخانه‌ی رسم نمودار در پایتون است؛ Pandas و Seaborn (جلسه‌ی بعد) هم در پشت صحنه از آن استفاده می‌کنند. (اگر هنوز آن را نصب نکرده‌اید، دستور نصب در جلسه‌ی ۲۱ آمده است.) معمولاً فقط زیرماژول <code dir="ltr">pyplot</code> را با نام کوتاه <code dir="ltr">plt</code> وارد می‌کنیم.

هر نمودار دو جزء اصلی دارد:

- **Figure**: کل «صفحه» یا قاب؛ می‌تواند یک یا چند نمودار داشته باشد.
- **Axes**: یک نمودار مشخص، با محور x و y، داده‌ها، عنوان و راهنما. (مراقب باشید: Axes با axis، یعنی یک «محور»، فرق دارد.)

دو سبک برای کار با Matplotlib وجود دارد:

| سبک | نمونه | کاربرد |
|---|---|---|
| <code dir="ltr">pyplot</code> (مبتنی بر وضعیت) | <code dir="ltr">plt.plot(x, y)</code> | نمودار سریع و یک‌خطی |
| شیءگرا | <code dir="ltr">fig, ax = plt.subplots()</code> و بعد <code dir="ltr">ax.plot(x, y)</code> | هر کار جدی؛ مخصوصاً وقتی چند نمودار دارید |

در سبک <code dir="ltr">pyplot</code>، دستورها روی «نمودار جاری» اعمال می‌شوند که پایتون پنهانی آن را دنبال می‌کند. در سبک شیءگرا صریحاً می‌گویید روی **کدام** Axes کار می‌کنید (همان ایده‌ی شیء و متد از جلسه‌ی ۱۸). در این دوره از سبک شیءگرا استفاده می‌کنیم.

سلول زیر کتابخانه‌ها و داده‌ی پنگوئن‌ها را بارگذاری می‌کند (داده‌ای که در چند جلسه‌ی این بخش استفاده می‌کنیم؛ ۳۴۴ پنگوئن از سه گونه با اندازه‌های بدن):

</div>

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

LOCAL = Path("data/penguins.csv")
URL = "https://haghbinh.github.io/Python-Course/Notebooks/data/penguins.csv"
penguins = pd.read_csv(LOCAL if LOCAL.exists() else URL)
print(penguins.shape)
````

<div dir="rtl">

### نمودار خطی

برای داده‌ای که ترتیب دارد (زمان، شماره‌ی epoch، مقدار x پیوسته) از نمودار خطی استفاده می‌کنیم:

</div>

````python
x = np.linspace(0, 2 * np.pi, 100)

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Two line plots on one Axes")
ax.legend()
plt.show()
````

### 🔍 کد را بخوان

- <code dir="ltr">plt.subplots()</code> یک Figure و یک Axes می‌سازد و هر دو را برمی‌گرداند (باز کردن تاپل، جلسه‌ی ۹). <code dir="ltr">figsize</code> اندازه‌ی شکل به اینچ است.
- هر <code dir="ltr">ax.plot</code> یک منحنی به **همان** Axes اضافه می‌کند؛ رنگ‌ها خودکار عوض می‌شوند.
- <code dir="ltr">label</code> متنی است که در راهنما نمایش داده می‌شود؛ خود راهنما فقط با <code dir="ltr">ax.legend()</code> ظاهر می‌شود.
- متدهای تنظیم در سبک شیءگرا با <code dir="ltr">set&#95;</code> شروع می‌شوند: <code dir="ltr">set&#95;xlabel</code>، <code dir="ltr">set&#95;title</code> و ...
- <code dir="ltr">plt.show()</code> شکل را نمایش می‌دهد. در Jupyter و VS Code معمولاً بدون آن هم نمایش داده می‌شود، اما نوشتنش عادت خوبی است (در اسکریپت معمولی ضروری است).

همین نمودار در سبک <code dir="ltr">pyplot</code>:

</div>

````python
x = np.linspace(0, 2 * np.pi, 100)
plt.plot(x, np.sin(x), label="sin(x)")
plt.title("Same idea, pyplot style")
plt.legend()
plt.show()
````

<div dir="rtl">

🔍 کوتاه‌تر است، اما معلوم نیست <code dir="ltr">plt.title</code> روی کدام نمودار اثر می‌گذارد؛ وقتی چند نمودار دارید، این ابهام منشأ خطاست. نام توابع هم کمی فرق دارد: <code dir="ltr">plt.title</code> در برابر <code dir="ltr">ax.set&#95;title</code>، و <code dir="ltr">plt.xlabel</code> در برابر <code dir="ltr">ax.set&#95;xlabel</code>.

> 💡 **درباره‌ی متن فارسی:** Matplotlib به‌طور پیش‌فرض حروف فارسی را جدا از هم و از چپ به راست نشان می‌دهد. نمایش درست فارسی به فونت مناسب و کتابخانه‌های جانبی برای اتصال حروف و راست‌به‌چپ کردن متن نیاز دارد. برای همین در این دوره همه‌ی برچسب‌ها و عنوان‌های نمودار را انگلیسی می‌نویسیم.

### نمودار پراکندگی

برای دیدن **رابطه‌ی دو متغیر عددی** از نمودار پراکندگی (<code dir="ltr">scatter</code>) استفاده می‌کنیم. اینجا هر گونه را جداگانه و با رنگ خودش رسم می‌کنیم:

</div>

````python
fig, ax = plt.subplots(figsize=(6, 4))
for species in ["Adelie", "Chinstrap", "Gentoo"]:
    part = penguins[penguins["species"] == species]
    ax.scatter(part["flipper_length_mm"], part["body_mass_g"],
               label=species, alpha=0.7, s=20)
ax.set_xlabel("Flipper length (mm)")
ax.set_ylabel("Body mass (g)")
ax.set_title("Penguins: flipper length vs body mass")
ax.legend(title="Species")
plt.show()
````

### 🔍 کد را بخوان

- حلقه روی سه گونه می‌چرخد و برای هر کدام با فیلتر بولی (جلسه‌ی ۲۳) زیرمجموعه‌ی آن را جدا می‌کند و یک <code dir="ltr">scatter</code> جداگانه با <code dir="ltr">label</code> خودش رسم می‌کند.
- <code dir="ltr">alpha=0.7</code> نقطه‌ها را نیمه‌شفاف می‌کند تا نقطه‌های روی هم دیده شوند؛ <code dir="ltr">s</code> اندازه‌ی نقطه است.
- ردیف‌هایی که مقدار گمشده (<code dir="ltr">NaN</code>) دارند بی‌صدا رسم نمی‌شوند.

### 💡 کدام نمودار برای کدام پرسش؟

| پرسش | نمودار | تابع |
|---|---|---|
| یک مقدار در طول زمان یا یک ترتیب چطور تغییر می‌کند؟ | خطی | <code dir="ltr">ax.plot</code> |
| دو متغیر عددی چه رابطه‌ای دارند؟ | پراکندگی | <code dir="ltr">ax.scatter</code> |
| مقدار یا تعداد در چند گروه چه فرقی دارد؟ | میله‌ای | <code dir="ltr">ax.bar</code> / <code dir="ltr">ax.barh</code> |
| یک متغیر عددی چه **توزیعی** دارد؟ | هیستوگرام | <code dir="ltr">ax.hist</code> |
| توزیع یک متغیر در چند گروه چه فرقی دارد؟ | جعبه‌ای | <code dir="ltr">ax.boxplot</code> |
| سهم هر بخش از کل چقدر است؟ | دایره‌ای (با احتیاط) | <code dir="ltr">ax.pie</code> |

نمودار میله‌ای را می‌توان مستقیم از نتیجه‌ی یک <code dir="ltr">groupby</code> (جلسه‌ی ۲۴) ساخت. <code dir="ltr">barh</code> میله‌ها را افقی می‌کشد؛ برای برچسب‌های بلند خواناتر است:

</div>

````python
mean_mass = penguins.groupby("species")["body_mass_g"].mean().round(1)
print(mean_mass)

fig, ax = plt.subplots(figsize=(5, 2.5))
ax.barh(mean_mass.index, mean_mass.values, color="tab:green")
ax.set_xlabel("Mean body mass (g)")
plt.show()
````

<div dir="rtl">

🔍 Pandas هم یک میان‌بر دارد: متد <code dir="ltr">plot</code> روی Series و DataFrame، که در واقع همین توابع Matplotlib را صدا می‌زند و یک Axes برمی‌گرداند. پس می‌توانید بعد از آن با متدهای <code dir="ltr">ax</code> نمودار را تنظیم کنید:

</div>

````python
ax = mean_mass.plot(kind="bar", figsize=(5, 3), color="tab:purple", rot=0)
ax.set_ylabel("Mean body mass (g)")
plt.show()
````

### ❓ پیش‌بینی کن

در بخش بعد، داده‌ی هر گونه را با <code dir="ltr">dropna()</code> به <code dir="ltr">boxplot</code> می‌دهیم. فرض کنید <code dir="ltr">dropna()</code> را فراموش کنیم. ستون <code dir="ltr">body&#95;mass&#95;g</code> دو مقدار گمشده دارد (یکی در <code dir="ltr">Adelie</code> و یکی در <code dir="ltr">Gentoo</code>). سلول زیر چه چاپ می‌کند و جعبه‌ی <code dir="ltr">Adelie</code> چه شکلی می‌شود؟

</div>

````python
fig, ax = plt.subplots()
adelie = penguins.loc[penguins["species"] == "Adelie", "body_mass_g"]
parts = ax.boxplot([adelie])
print(adelie.isna().sum(), parts["medians"][0].get_ydata())
plt.close(fig)
````

<div dir="rtl">

<code dir="ltr">plt.close(fig)</code> شکل را می‌بندد تا نمایش داده نشود؛ اینجا فقط به عددها نیاز داشتیم.

## ۲. subplot

### 💡 مفهوم

برای مقایسه، اغلب چند نمودار را کنار هم در یک شکل می‌خواهیم. <code dir="ltr">plt.subplots(nrows, ncols)</code> یک Figure و یک **آرایه‌ی NumPy** از Axesها برمی‌گرداند که با اندیس دوبعدی (جلسه‌ی ۲۱) به هر کدام دسترسی دارید: <code dir="ltr">axes[0, 1]</code> یعنی سطر اول، ستون دوم.

| آرگومان یا متد | کار |
|---|---|
| <code dir="ltr">figsize=(w, h)</code> | اندازه‌ی کل شکل (اینچ) |
| <code dir="ltr">sharex=True</code> / <code dir="ltr">sharey=True</code> | محور مشترک بین نمودارها (برای مقایسه‌ی منصفانه) |
| <code dir="ltr">constrained&#95;layout=True</code> یا <code dir="ltr">fig.tight&#95;layout()</code> | تنظیم خودکار فاصله‌ها تا عنوان‌ها روی هم نیفتند |
| <code dir="ltr">fig.suptitle(...)</code> | عنوان کل شکل |

</div>

````python
mass = penguins["body_mass_g"].dropna()
counts = penguins["species"].value_counts()
groups = [penguins.loc[penguins["species"] == s, "body_mass_g"].dropna()
          for s in counts.index]

fig, axes = plt.subplots(2, 2, figsize=(9, 6.5), constrained_layout=True)
print(type(axes), axes.shape)

axes[0, 0].hist(mass, bins=20, color="tab:blue", edgecolor="white")
axes[0, 0].set_title("hist: distribution of body mass")

axes[0, 1].bar(counts.index, counts.values, color="tab:orange")
axes[0, 1].set_title("bar: number of penguins per species")

axes[1, 0].boxplot(groups)
axes[1, 0].set_xticks([1, 2, 3])
axes[1, 0].set_xticklabels(counts.index)
axes[1, 0].set_title("boxplot: body mass per species")

axes[1, 1].scatter(penguins["bill_length_mm"], penguins["bill_depth_mm"], s=10, alpha=0.6)
axes[1, 1].set_title("scatter: bill length vs bill depth")

fig.suptitle("Four basic chart types", fontsize=14)
plt.show()
````

### 🔍 کد را بخوان

- خط <code dir="ltr">print</code> نشان می‌دهد <code dir="ltr">axes</code> یک <code dir="ltr">numpy.ndarray</code> با شکل <code dir="ltr">(2, 2)</code> است.
- <code dir="ltr">boxplot</code> یک **لیست** از داده‌ها می‌گیرد و برای هر عضو یک جعبه می‌کشد؛ جعبه‌ها در مکان‌های <code dir="ltr">1، 2، 3</code> قرار می‌گیرند. برای همین برچسب‌ها را با <code dir="ltr">set&#95;xticks</code> و <code dir="ltr">set&#95;xticklabels</code> روی همین مکان‌ها گذاشته‌ایم.
- <code dir="ltr">groups</code> با یک list comprehension (جلسه‌ی ۱۳) ساخته شده و ترتیبش با ترتیب <code dir="ltr">counts.index</code> یکی است؛ پس برچسب‌ها درست‌اند.
- هیستوگرام وزن **دو قله** دارد (حدود ۳۷۰۰ و ۵۰۰۰ گرم)؛ نمودار جعبه‌ای توضیحش را می‌دهد: <code dir="ltr">Gentoo</code> به‌تنهایی قله‌ی دوم را می‌سازد.

### 💡 یک نمودار، دو داستان

به نمودار پراکندگی پایین سمت راست نگاه کنید: در نگاه اول، با بلندتر شدن نوک، **عمق** آن کمتر می‌شود. اما بیایید ضریب همبستگی را یک بار برای کل داده و یک بار برای هر گونه حساب کنیم:

</div>

````python
print(round(penguins["bill_length_mm"].corr(penguins["bill_depth_mm"]), 3))
for species, group in penguins.groupby("species"):
    print(species, round(group["bill_length_mm"].corr(group["bill_depth_mm"]), 3))
````

<div dir="rtl">

🔍 همبستگی کل داده **منفی** (<code dir="ltr">-0.235</code>) است، اما درون **هر** گونه **مثبت** (بین <code dir="ltr">0.39</code> و <code dir="ltr">0.65</code>)! به این پدیده **پارادوکس سیمپسون** می‌گویند: یک متغیر پنهان (گونه) رابطه را وارونه نشان می‌دهد. درس مهم: قبل از نتیجه‌گیری از یک نمودار پراکندگی، گروه‌ها را با رنگ جدا کنید (مثل نمودار پراکندگی بخش ۱). در جلسه‌ی بعد Seaborn این کار را با یک آرگومان (<code dir="ltr">hue</code>) انجام می‌دهد.

### ❓ پیش‌بینی کن

خروجی هر خط چیست؟

</div>

````python
fig, axes = plt.subplots(1, 3)
print(axes.shape)
fig, axes = plt.subplots(3, 1)
print(axes.shape)
fig, ax = plt.subplots()
print(type(ax).__name__)
fig, axes = plt.subplots(2, 2)
print(axes.flatten().shape)
plt.close("all")
````

<div dir="rtl">

🔍 وقتی فقط یک سطر **یا** یک ستون دارید، آرایه **یک‌بعدی** است: <code dir="ltr">(3,)</code> در هر دو حالت. با <code dir="ltr">subplots()</code> بدون آرگومان، خود یک شیء <code dir="ltr">Axes</code> برگردانده می‌شود، نه آرایه. <code dir="ltr">flatten</code> (جلسه‌ی ۲۱) آرایه‌ی ۲ در ۲ را به یک آرایه‌ی ۴تایی تبدیل می‌کند؛ برای حلقه زدن روی همه‌ی نمودارها راحت است: <code dir="ltr">for ax in axes.flatten():</code>.

<code dir="ltr">plt.close("all")</code> همه‌ی شکل‌های باز را می‌بندد؛ اینجا نمی‌خواستیم چهار شکل خالی نمایش داده شود.

مهم‌ترین نموداری که در یادگیری ماشین می‌کشید، **منحنی آموزش** است: مقدار loss روی داده‌ی آموزش و اعتبارسنجی در طول epochها (gradient descent جلسه‌های ۷ و ۱۸). داده‌ی زیر ساختگی است، اما شکلش واقعی است:

</div>

````python
epochs = np.arange(1, 41)
train_loss = 2.0 * np.exp(-epochs / 8) + 0.10
val_loss = 2.0 * np.exp(-epochs / 8) + 0.15 + 0.0015 * (epochs - 12).clip(min=0) ** 2
best = epochs[np.argmin(val_loss)]
print("best epoch:", best, " val loss:", round(val_loss.min(), 3))

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(epochs, train_loss, color="tab:blue", linestyle="-", label="train")
ax.plot(epochs, val_loss, color="tab:red", linestyle="--", marker="o",
        markersize=3, label="validation")
ax.axvline(best, color="gray", linestyle=":", label=f"best epoch = {best}")
ax.annotate("overfitting starts", xy=(30, val_loss[29]), xytext=(22, 1.2),
            arrowprops=dict(arrowstyle="->"))
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss")
ax.set_title("Training curves")
ax.set_ylim(0, 2.2)
ax.grid(alpha=0.3)
ax.legend()
plt.show()
````

- دو منحنی با **رنگ و سبک خط** متفاوت رسم شده‌اند تا در چاپ سیاه‌وسفید هم قابل تشخیص باشند.
- <code dir="ltr">axvline</code> یک خط عمودی در بهترین epoch می‌کشد و <code dir="ltr">label</code> آن با f-string ساخته شده است.

- <code dir="ltr">set&#95;ylim(0, 2.2)</code> محور y را از صفر شروع می‌کند تا اندازه‌ی loss گمراه‌کننده نباشد.

**تفسیر:** تا epoch ۱۹ هر دو منحنی پایین می‌آیند؛ مدل در حال یادگیری الگوهای کلی است. بعد از آن loss آموزش همچنان کم می‌شود اما loss اعتبارسنجی بالا می‌رود: مدل جزئیات و نویز داده‌ی آموزش را **حفظ** می‌کند (**بیش‌برازش**، overfitting). بهترین کار، نگه داشتن مدل epoch ۱۹ است (**توقف زودهنگام**، early stopping؛ جلسه‌ی ۱۴).

الف)

د) با محور مشترک، **سطح** loss سه مدل مستقیم قابل مقایسه است؛ مثلاً می‌بینیم بدترین نقطه‌ی A از بهترین نقطه‌ی B پایین‌تر است. بدون آن، Matplotlib برای هر نمودار محدوده‌ی جداگانه انتخاب می‌کند و منحنی تقریباً صاف B ممکن است پرشیب و «خوب» به نظر برسد.

### سؤال ۷ — رفع اشکال: رسم توابع فعال‌سازی

<summary><b>پاسخ</b></summary>

الف)

- <code dir="ltr">max(0, x)</code>: تابع <code dir="ltr">max</code> پایتون دو مقدار را مقایسه می‌کند و نمی‌داند مقایسه‌ی یک عدد با یک **آرایه** یعنی چه؛ خطای <code dir="ltr">ValueError: The truth value of an array with more than one element is ambiguous</code> می‌دهد. درستش: <code dir="ltr">np.maximum(0, x)</code> که عضوبه‌عضو کار می‌کند (جلسه‌ی ۲۲).
- <code dir="ltr">ax.legend</code> بدون پرانتز: فقط به متد **اشاره** می‌کند و آن را اجرا نمی‌کند؛ خطایی هم نمی‌دهد و راهنما نمایش داده نمی‌شود.
- <code dir="ltr">plt.title = "..."</code>: تابع <code dir="ltr">plt.title</code> را با یک رشته **جایگزین** می‌کند (انتساب، نه فراخوانی). عنوانی گذاشته نمی‌شود و از این به بعد هر جای نوت‌بوک که <code dir="ltr">plt.title(...)</code> بنویسید، خطای <code dir="ltr">TypeError: 'str' object is not callable</code> می‌گیرید تا kernel را دوباره راه‌اندازی کنید (همان سایه‌اندازی نام، جلسه‌ی ۱۷).
- <code dir="ltr">plt.plot</code> در کنار <code dir="ltr">ax.plot</code>: اینجا کار می‌کند چون Axes جاری همان <code dir="ltr">ax</code> است، اما مخلوط کردن دو سبک گیج‌کننده است.

## خلاصه‌ی جلسه

- هر نمودار در یک **Axes** درون یک **Figure** رسم می‌شود؛ با <code dir="ltr">fig, ax = plt.subplots()</code> صریحاً روی Axes مورد نظر کار کنید.
- برای هر پرسش، نمودار مناسب را انتخاب کنید: خطی برای روند، پراکندگی برای رابطه، میله‌ای برای مقایسه‌ی گروه‌ها، هیستوگرام و جعبه‌ای برای توزیع.
- <code dir="ltr">subplots</code> چند نمودار را کنار هم می‌چیند؛ به شکل آرایه‌ی <code dir="ltr">axes</code> و محورهای مشترک دقت کنید.
- برچسب، راهنما، رنگ، سبک خط، محدوده‌ی محور و حاشیه‌نویسی نمودار را بی‌نیاز از توضیح می‌کنند؛ مراقب نمودارهای گمراه‌کننده باشید.
- منحنی‌های loss آموزش و اعتبارسنجی بیش‌برازش، کم‌برازش و نرخ یادگیری نامناسب را نشان می‌دهند.
- با <code dir="ltr">fig.savefig</code> (قبل از نمایش) و قالب و <code dir="ltr">dpi</code> مناسب ذخیره کنید و شکل‌ها را در حلقه ببندید.
