## ۱. Series

### 💡 مفهوم

<code dir="ltr">Series</code> یک ستون داده است: یک آرایه‌ی یک‌بعدی NumPy (<code dir="ltr">values</code>) به‌علاوه‌ی یک **اندیس** (<code dir="ltr">index</code>) که به هر خانه یک برچسب می‌دهد. می‌توانید آن را ترکیبی از **لیست** (ترتیب دارد، عملیات برداری) و **دیکشنری** (دسترسی با برچسب؛ جلسه‌ی ۱۰) بدانید.

</div>

````python
prices = pd.Series([850, 420, 310, 190], index=["laptop", "phone", "tablet", "monitor"])
print(prices)
print(prices.values, type(prices.values))
print(prices.index)
print(prices["phone"], "tablet" in prices)

stock = pd.Series({"laptop": 12, "phone": 30, "tablet": 0, "monitor": 7})
print(stock * prices)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">prices.values</code> یک آرایه‌ی NumPy است؛ همه‌ی چیزهایی که در جلسه‌ی ۲۲ یاد گرفتید (برداری‌سازی، تابع‌های آماری) روی آن کار می‌کند.
- <code dir="ltr">prices["phone"]</code> مثل دیکشنری با **برچسب** دسترسی دارد، و <code dir="ltr">"tablet" in prices</code> (درست مثل دیکشنری) در **اندیس** جست‌وجو می‌کند، نه در مقدارها.
- <code dir="ltr">stock</code> از یک دیکشنری ساخته شده است: کلیدها اندیس و مقدارها داده می‌شوند.
- <code dir="ltr">stock &#42; prices</code> ارزش موجودی هر کالاست؛ ضرب **خانه به خانه** و بدون حلقه انجام می‌شود (مثل NumPy).

### 💡 هم‌ترازی اندیس (index alignment)

تفاوت مهم <code dir="ltr">Series</code> با آرایه‌ی NumPy این است: در عملیات بین دو <code dir="ltr">Series</code>، خانه‌ها بر اساس **برچسب** جفت می‌شوند، نه بر اساس **موقعیت**. اگر برچسبی فقط در یکی باشد، نتیجه برای آن برچسب <code dir="ltr">NaN</code> («عدد نیست»؛ نشانه‌ی **مقدار گمشده**) است.

### ❓ پیش‌بینی کن

خروجی سلول زیر چیست؟ نتیجه چند سطر دارد و ترتیب سطرها چیست؟

</div>

````python
week1 = pd.Series([5, 3, 8], index=["laptop", "phone", "tablet"])
week2 = pd.Series([2, 4, 1], index=["phone", "tablet", "mouse"])

print(week1 + week2)
print(week1.add(week2, fill_value=0))
````

<div dir="rtl">

🔍 نتیجه‌ی <code dir="ltr">week1 + week2</code> چهار سطر دارد (اجتماع برچسب‌ها) و Pandas آن‌ها را مرتب می‌کند: <code dir="ltr">laptop</code> و <code dir="ltr">mouse</code> فقط در یکی از دو Series هستند، پس <code dir="ltr">NaN</code> می‌شوند؛ <code dir="ltr">phone</code> برابر <code dir="ltr">3 + 2 = 5</code> و <code dir="ltr">tablet</code> برابر <code dir="ltr">8 + 4 = 12</code> است. دقت کنید که نوع داده <code dir="ltr">float64</code> شده است، چون <code dir="ltr">NaN</code> یک عدد اعشاری است.

با متد <code dir="ltr">add</code> و <code dir="ltr">fill&#95;value=0</code>، برچسبِ غایب در هر طرف صفر فرض می‌شود: <code dir="ltr">laptop</code> برابر <code dir="ltr">5</code> و <code dir="ltr">mouse</code> برابر <code dir="ltr">1</code>.

### ⚠️ اشتباه رایج: <code dir="ltr">s[0]</code> برای «اولین عنصر»

وقتی اندیس عدد صحیح باشد، <code dir="ltr">s[0]</code> یعنی «عنصری با **برچسب** <code dir="ltr">0</code>»، نه «اولین عنصر». برای موقعیت همیشه از <code dir="ltr">s.iloc[0]</code> و برای برچسب از <code dir="ltr">s.loc[...]</code> استفاده کنید (بخش ۴)؛ این‌طوری کد شما در هیچ حالتی مبهم نیست.

</div>

## ۲. DataFrame

### 💡 مفهوم

<code dir="ltr">DataFrame</code> یک جدول است: مجموعه‌ای از ستون‌ها (هر ستون یک <code dir="ltr">Series</code>) که **یک اندیس مشترک** دارند. هر ستون نوع داده‌ی خودش را دارد (یکی عدد صحیح، یکی اعشاری، یکی متن)؛ برخلاف آرایه‌ی دوبعدی NumPy که همه‌ی خانه‌هایش هم‌نوع است.

سه راه رایج ساختن:

| ورودی | هر عنصر یعنی |
|---|---|
| دیکشنری از لیست‌ها | یک **ستون** |
| لیستی از دیکشنری‌ها | یک **سطر** (رکورد) |
| آرایه‌ی دوبعدی NumPy | کل جدول؛ نام ستون‌ها را جدا می‌دهید |

</div>

````python
products = pd.DataFrame({
    "product": ["laptop", "phone", "tablet", "monitor", "mouse"],
    "category": ["computer", "mobile", "mobile", "computer", "accessory"],
    "price": [850, 420, 310, 190, 25],
    "stock": [12, 30, 0, 7, 150],
})
print(products)

records = [{"city": "Bushehr", "temp": 31.5}, {"city": "Tabriz", "temp": 12.0}]
print(pd.DataFrame(records))

rng = np.random.default_rng(0)
print(pd.DataFrame(rng.integers(0, 10, size=(3, 2)), columns=["f1", "f2"]))
````

<div dir="rtl">

### 🔍 کد را بخوان

- اندیس پیش‌فرض اعداد <code dir="ltr">0</code> تا <code dir="ltr">n - 1</code> است.
- <code dir="ltr">records</code> همان شکلی است که از یک فایل JSON (جلسه‌ی ۱۶) یا یک API می‌گیرید: هر دیکشنری یک سطر. کلیدها نام ستون‌ها می‌شوند.
- آرایه‌ی NumPy فقط عدد دارد و نام ستون‌ها را با <code dir="ltr">columns</code> می‌دهیم.

### 💡 شناختن یک جدول

اولین کار با هر داده‌ای این است که بفهمید چه شکلی دارد:

</div>

````python
print(products.shape)
print(products.columns.tolist())
print(products.dtypes)
print(products.head(2))
products.info()
print(products.describe())
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">shape</code> برابر <code dir="ltr">(5, 4)</code> است: ۵ سطر (نمونه) و ۴ ستون (ویژگی)؛ مثل <code dir="ltr">shape</code> در NumPy.
- <code dir="ltr">dtypes</code> نوع هر ستون را نشان می‌دهد: <code dir="ltr">price</code> و <code dir="ltr">stock</code> از نوع <code dir="ltr">int64</code> و دو ستون دیگر متنی‌اند.
- <code dir="ltr">head(n)</code> و <code dir="ltr">tail(n)</code> چند سطر اول و آخر را نشان می‌دهند؛ برای جدول‌های بزرگ همیشه از این‌ها شروع کنید، نه از <code dir="ltr">print(df)</code>.
- <code dir="ltr">info()</code> تعداد مقدارهای **غیرگمشده** و نوع هر ستون و حافظه را می‌دهد. خودش چاپ می‌کند و چیزی برنمی‌گرداند (مثل <code dir="ltr">list.sort</code> در جلسه‌ی ۸).
- <code dir="ltr">describe()</code> برای ستون‌های عددی خلاصه‌ی آماری می‌دهد: تعداد، میانگین، انحراف معیار، کمینه، چارک‌ها و بیشینه. مثلاً میانگین قیمت <code dir="ltr">359</code> است اما میانه (<code dir="ltr">50%</code>) فقط <code dir="ltr">310</code>؛ یک کالای گران (<code dir="ltr">laptop</code>) میانگین را بالا کشیده است. (آمار توصیفی را در جلسه‌ی ۲۹ دقیق‌تر می‌بینیم.)

### 💡 انتخاب، افزودن و حذف ستون

</div>

````python
print(type(products["price"]), type(products[["price"]]))
print(products[["product", "price"]])

products["value"] = products["price"] * products["stock"]
products["in_stock"] = products["stock"] > 0
print(products)

smaller = products.drop(columns=["value", "in_stock"])
print(smaller.columns.tolist(), products.columns.tolist())
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">products["price"]</code> (یک نام) یک **Series** برمی‌گرداند؛ <code dir="ltr">products[["price"]]</code> (لیستی از نام‌ها، حتی با یک عضو) یک **DataFrame** تک‌ستونه. این تفاوت بعداً در شکل داده‌ها (<code dir="ltr">(5,)</code> در برابر <code dir="ltr">(5, 1)</code>) مهم می‌شود.
- ستون جدید با انتساب ساخته می‌شود و محاسبه‌اش **برداری** است؛ هیچ حلقه‌ای لازم نیست.
- <code dir="ltr">drop(columns=...)</code> یک جدول **جدید** برمی‌گرداند و <code dir="ltr">products</code> را تغییر نمی‌دهد؛ نتیجه را باید انتساب کنید.

### ❓ پیش‌بینی کن

جدول <code dir="ltr">products</code> حالا چند ستون دارد؟ <code dir="ltr">products["in&#95;stock"].sum()</code> چه عددی می‌دهد؟ (راهنمایی: <code dir="ltr">True</code> در جمع، <code dir="ltr">1</code> حساب می‌شود.)

<summary><b>پاسخ</b></summary>

</details>

### ⚠️ اشتباه رایج: <code dir="ltr">products.price</code> برای ساختن ستون

خواندن ستون با نقطه (<code dir="ltr">products.price</code>) کار می‌کند، اما **ساختن** ستون جدید با <code dir="ltr">products.new = ...</code> ستون نمی‌سازد (فقط یک ویژگی معمولی به شیء اضافه می‌کند). همچنین اگر نام ستون فاصله داشته باشد یا با نام یک متد یکی باشد (مثلاً ستونی به نام <code dir="ltr">count</code>)، نقطه کار نمی‌کند. برای اطمینان همیشه از <code dir="ltr">df["name"]</code> استفاده کنید.

## ۳. خواندن CSV و Excel

### 💡 نوشتن و خواندن CSV

در جلسه‌ی ۱۶ فایل CSV را با ماژول <code dir="ltr">csv</code> و حلقه خواندید. Pandas همه‌ی این کار را در یک خط انجام می‌دهد: <code dir="ltr">pd.read&#95;csv</code> فایل را می‌خواند، ستون‌ها را جدا می‌کند و **نوع هر ستون را حدس می‌زند**. <code dir="ltr">to&#95;csv</code> برعکس عمل می‌کند. ابتدا یک پوشه برای فایل‌های این جلسه می‌سازیم (<code dir="ltr">pathlib</code>، جلسه‌ی ۱۶):

</div>

````python
from pathlib import Path

data_dir = Path("s23_data")
data_dir.mkdir(exist_ok=True)

products.to_csv(data_dir / "products.csv", index=False)
print((data_dir / "products.csv").read_text())

back = pd.read_csv(data_dir / "products.csv")
print(back.dtypes)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">index=False</code> یعنی اندیس (<code dir="ltr">0</code> تا <code dir="ltr">4</code>) در فایل نوشته نشود. اگر آن را فراموش کنید، دفعه‌ی بعد که فایل را بخوانید یک ستون اضافه‌ی بی‌نام (<code dir="ltr">Unnamed: 0</code>) می‌بینید.
- <code dir="ltr">read&#95;csv</code> نوع ستون‌ها را درست حدس زده است: <code dir="ltr">price</code> عدد صحیح و <code dir="ltr">in&#95;stock</code> بولی.

### 💡 گزینه‌های مهم <code dir="ltr">read&#95;csv</code>

فایل‌های واقعی همیشه تمیز نیستند. فایل زیر را (که با ابزارهای جلسه‌ی ۱۶ می‌نویسیم) در نظر بگیرید: جداکننده <code dir="ltr">;</code> است، اعشار با **ویرگول** نوشته شده (رایج در بسیاری از کشورها)، مقدار گمشده با <code dir="ltr">?</code> مشخص شده و کد پستی با صفر شروع می‌شود.

</div>

````python
messy_text = """station;city;temp;humidity;zip
1;Bushehr;31,5;70;07531
2;Shiraz;?;35;71345
3;Tehran;24,0;NA;01234
4;Tabriz;12,5;48;?
"""
(data_dir / "weather.csv").write_text(messy_text)

naive = pd.read_csv(data_dir / "weather.csv")
print(naive.shape)
print(naive.columns.tolist())
````

<div dir="rtl">

🔍 بدون گزینه‌ها، <code dir="ltr">read&#95;csv</code> فرض می‌کند جداکننده ویرگول است. سطر عنوان هیچ ویرگولی ندارد و کل آن **یک** نام ستون می‌شود (<code dir="ltr">shape</code> برابر <code dir="ltr">(4, 1)</code>)، و ویرگول‌های اعشاری هم داده‌ها را در جای اشتباه می‌شکنند؛ نتیجه جدولی بی‌معنی است. حالا با گزینه‌های درست:

</div>

````python
weather = pd.read_csv(
    data_dir / "weather.csv",
    sep=";",
    decimal=",",
    na_values=["?"],
    dtype={"zip": str},
    index_col="station",
)
print(weather)
print(weather.dtypes)
print(weather.isna().sum())
````

<div dir="rtl">

### 🔍 کد را بخوان

| گزینه | کار |
|---|---|
| <code dir="ltr">sep=";"</code> | جداکننده‌ی ستون‌ها |
| <code dir="ltr">decimal=","</code> | <code dir="ltr">31,5</code> را عدد <code dir="ltr">31.5</code> بخوان |
| <code dir="ltr">na&#95;values=["?"]</code> | <code dir="ltr">?</code> را مقدار گمشده (<code dir="ltr">NaN</code>) حساب کن؛ رشته‌هایی مثل <code dir="ltr">NA</code> و خانه‌ی خالی به‌طور پیش‌فرض گمشده‌اند |
| <code dir="ltr">dtype={"zip": str}</code> | کد پستی را متن نگه دار تا صفر اولش (<code dir="ltr">07531</code>) حذف نشود |
| <code dir="ltr">index&#95;col="station"</code> | یک ستون را اندیس جدول کن |
| <code dir="ltr">usecols=[...]</code> | فقط این ستون‌ها را بخوان (برای فایل‌های بزرگ، صرفه‌جویی در حافظه) |
| <code dir="ltr">header=None</code> | فایل سطر عنوان ندارد |
| <code dir="ltr">parse&#95;dates=[...]</code> | این ستون‌ها را تاریخ بخوان (جلسه‌ی ۲۵) |

<code dir="ltr">isna()</code> برای هر خانه می‌گوید گمشده است یا نه و <code dir="ltr">sum()</code> آن‌ها را در هر ستون می‌شمارد: <code dir="ltr">temp</code>، <code dir="ltr">humidity</code> و <code dir="ltr">zip</code> هر کدام یک مقدار گمشده دارند. اینکه با مقدارهای گمشده **چه کنیم** موضوع جلسه‌ی ۲۶ است.

### ⚠️ اشتباه رایج: ستون عددی که متنی خوانده شده

### 💡 مفهوم

برای انتخاب سطر و ستون دو ابزار دقیق داریم:

| | <code dir="ltr">loc</code> | <code dir="ltr">iloc</code> |

| بر اساس | **برچسب** (نام سطر و ستون) | **موقعیت** (شماره‌ی صحیح، از <code dir="ltr">0</code>) |

| برش <code dir="ltr">a:b</code> | **شامل** <code dir="ltr">b</code> | **بدون** <code dir="ltr">b</code> (مثل لیست) |

## ۵. فیلتر کردن

### 💡 مفهوم

فیلتر کردن در Pandas همان **ماسک بولی** NumPy است (جلسه‌ی ۲۱): یک مقایسه روی ستون، یک Series از <code dir="ltr">True</code> و <code dir="ltr">False</code> می‌سازد و <code dir="ltr">df[mask]</code> فقط سطرهای <code dir="ltr">True</code> را نگه می‌دارد. برای ترکیب شرط‌ها:

| عملگر | معنی | به جای |
|---|---|---|
| <code dir="ltr">&amp;</code> | و | <code dir="ltr">and</code> |
| <code dir="ltr">&#124;</code> | یا | <code dir="ltr">or</code> |
| <code dir="ltr">~</code> | نقیض | <code dir="ltr">not</code> |

**هر شرط باید داخل پرانتز باشد**، چون <code dir="ltr">&amp;</code> و <code dir="ltr">&#124;</code> در پایتون اولویتی **بالاتر** از <code dir="ltr">&gt;</code> و <code dir="ltr">==</code> دارند (جلسه‌ی ۳).

</div>

````python
heavy = penguins["body_mass_g"] > 5000
print(heavy.head(3))
print(heavy.sum())

adelie_heavy = penguins[(penguins["species"] == "Adelie") & (penguins["body_mass_g"] > 4000)]
print(adelie_heavy.shape)

gentoo_or_light = (penguins["species"] == "Gentoo") | (penguins["body_mass_g"] < 3000)
print(gentoo_or_light.sum())
print((~(penguins["sex"] == "male")).sum(), (penguins["sex"] == "female").sum())
````

- فقط ۳۵ پنگوئن Adelie وزن بیشتر از ۴۰۰۰ گرم دارند.
- خط آخر یک نکته‌ی ظریف دارد: «نه نر» (<code dir="ltr">176</code>) با «ماده» (<code dir="ltr">165</code>) برابر نیست، چون ۱۱ پنگوئن جنسیت **گمشده** دارند و مقایسه‌ی <code dir="ltr">NaN == "male"</code> نادرست است؛ پس آن‌ها جزو «نه نر» شمرده می‌شوند. وقتی داده‌ی گمشده دارید، نقیض یک شرط لزوماً شرط مخالف نیست.

## خلاصه‌ی جلسه

- <code dir="ltr">Series</code> آرایه‌ای با برچسب و <code dir="ltr">DataFrame</code> جدولی از Seriesها با اندیس مشترک است؛ عملیات بر اساس **برچسب** هم‌تراز می‌شوند و برچسب‌های غایب <code dir="ltr">NaN</code> می‌دهند.
- با <code dir="ltr">shape</code>، <code dir="ltr">dtypes</code>، <code dir="ltr">head</code>، <code dir="ltr">info</code> و <code dir="ltr">describe</code> هر جدول را بشناسید؛ بعد از <code dir="ltr">read&#95;csv</code> همیشه <code dir="ltr">dtypes</code> را بررسی کنید.
- <code dir="ltr">read&#95;csv</code> با <code dir="ltr">sep</code>، <code dir="ltr">decimal</code>، <code dir="ltr">na&#95;values</code>، <code dir="ltr">dtype</code> و <code dir="ltr">index&#95;col</code> فایل‌های نامرتب را درست می‌خواند؛ <code dir="ltr">to&#95;csv(index=False)</code>؛ Excel با <code dir="ltr">read&#95;excel</code> و <code dir="ltr">sheet&#95;name</code>.
- <code dir="ltr">loc</code> برچسبی (برش شامل انتها) و <code dir="ltr">iloc</code> موقعیتی است؛ تغییر مقدار فقط با یک <code dir="ltr">loc[rows, cols]</code>، و <code dir="ltr">.copy()</code> برای زیرجدول مستقل.
- فیلتر = ماسک بولی با <code dir="ltr">&amp;</code>، <code dir="ltr">&#124;</code>، <code dir="ltr">~</code> و **پرانتز**؛ <code dir="ltr">isin</code>، <code dir="ltr">between</code>، <code dir="ltr">str.contains</code> و <code dir="ltr">value&#95;counts</code> ابزارهای روزمره‌اند.

**جلسه‌ی بعد:** Pandas ۲؛ مرتب‌سازی، گروه‌بندی و تجمیع با <code dir="ltr">groupby</code>، <code dir="ltr">apply</code> و جدول محوری (<code dir="ltr">pivot&#95;table</code>).
