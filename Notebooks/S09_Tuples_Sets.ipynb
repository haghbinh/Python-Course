## ۱. تاپل و ویژگی‌های آن

### 💡 مفهوم

**تاپل (tuple)** مثل لیست یک دنباله‌ی **مرتب** از مقدارهاست، با یک تفاوت اساسی: **تغییرناپذیر (immutable)** است. بعد از ساختن یک تاپل، نمی‌توان عضوی را عوض، اضافه یا حذف کرد. تاپل را با پرانتز می‌نویسیم:

</div>

````python
t1 = (1, 2, 3, 4)
t2 = ("apple", "banana", "cherry")
t3 = (1, "hello", 3.14, True)
print(t1, t2, t3)

t = (10, 20, 30, 40, 20)
print(t[0], t[-1], t[1:3])           # indexing and slicing, like lists
print(len(t), max(t), min(t))
print(t.count(20), t.index(30))      # the only two tuple methods
print(t1 + (5, 6), (0,) * 3)         # concatenation and repetition
````

<div dir="rtl">

### 🔍 کد را بخوان

- اندیس‌گذاری، برش، <code dir="ltr">len</code>، <code dir="ltr">max</code>، <code dir="ltr">min</code>، <code dir="ltr">in</code>، <code dir="ltr">+</code> و <code dir="ltr">&#42;</code> دقیقاً مثل لیست کار می‌کنند (جلسه‌ی ۸).
- تاپل فقط **دو** متد دارد: <code dir="ltr">count</code> و <code dir="ltr">index</code>. متدهایی مثل <code dir="ltr">append</code>، <code dir="ltr">remove</code> یا <code dir="ltr">sort</code> وجود ندارند، چون همه‌ی آن‌ها شیء را تغییر می‌دادند.
- <code dir="ltr">t1 + (5, 6)</code> تاپل <code dir="ltr">t1</code> را تغییر نمی‌دهد؛ یک تاپل **جدید** می‌سازد.

### ⚠️ تلاش برای تغییر

</div>

````python
t = (1, 2, 3)
t[0] = 100
````

<div dir="rtl">

🔍 خطای <code dir="ltr">TypeError: 'tuple' object does not support item assignment</code> یعنی تاپل اجازه‌ی انتساب به عضو را نمی‌دهد.

### ⚠️ اشتباه رایج: تاپل یک‌عضوی

**کاما** است که تاپل می‌سازد، نه پرانتز. پرانتز تنها فقط برای گروه‌بندی عبارت‌هاست:

### ❓ پیش‌بینی کن

نوع هر مقدار چیست؟

</div>

````python
a = (5)
b = (5,)
c = 5, 6
d = ()
print(type(a), type(b), type(c), type(d))
````

<div dir="rtl">

🔍 <code dir="ltr">(5)</code> همان عدد <code dir="ltr">5</code> است. برای تاپل یک‌عضوی باید بعد از عضو کاما بگذارید: <code dir="ltr">(5,)</code>. حتی بدون پرانتز، <code dir="ltr">5, 6</code> یک تاپل است. <code dir="ltr">()</code> تاپل خالی است.

### 💡 تغییرناپذیری «سطحی» است

تاپل نمی‌گذارد **خانه‌هایش** به شیء دیگری اشاره کنند؛ اما اگر یکی از اعضا خودش **تغییرپذیر** باشد (مثلاً یک لیست)، محتوای آن عضو را می‌توان تغییر داد:

</div>

````python
config = ("relu", [64, 32])
config[1].append(16)          # allowed: we change the list, not the tuple
print(config)
````

<div dir="rtl">

🔍 تاپل همچنان به **همان** لیست اشاره می‌کند (پیکان‌ها جابه‌جا نشده‌اند؛ شکل جلسه‌ی ۸ را به خاطر بیاورید)، اما خود لیست تغییر کرده است. پس «تاپل تغییرناپذیر است» به این معنی نیست که هر چیزی داخل آن ثابت می‌ماند.

</div>

## ۲. باز کردن تاپل (unpacking)

### 💡 مفهوم

اعضای یک تاپل (یا هر دنباله‌ای) را می‌توان در یک خط در چند متغیر ریخت. این کار را از جلسه‌ی ۲ (انتساب چندگانه) و جلسه‌ی ۷ (برگرداندن چند مقدار) دیده‌اید؛ حالا می‌دانید پشت آن یک تاپل است:

</div>

````python
point = (3, 4)
x, y = point
print(x, y)

def min_max(values):
    return min(values), max(values)      # returns a tuple

lo, hi = min_max([4, 7, 1, 9, 3])
print(lo, hi)

a, b = 1, 2
a, b = b, a                              # swap: the right side is a tuple
print(a, b)
````

<div dir="rtl">

### 💡 باز کردن با <code dir="ltr">&#42;</code>

اگر تعداد متغیرها با تعداد اعضا برابر نباشد، خطای <code dir="ltr">ValueError</code> می‌گیرید. با <code dir="ltr">&#42;</code> می‌توانید «بقیه‌ی اعضا» را در یک **لیست** جمع کنید:

### ❓ پیش‌بینی کن

</div>

````python
first, *rest = (10, 20, 30, 40)
print(first, rest)

a, *middle, z = range(6)
print(a, middle, z)

header, *rows = ["name,score", "Ali,17", "Sara,19"]
print(header, rows)
````

<div dir="rtl">

🔍 <code dir="ltr">&#42;rest</code> همه‌ی اعضای باقی‌مانده را به شکل لیست می‌گیرد. مثال آخر الگوی رایجی برای جدا کردن **سطر عنوان** یک فایل CSV از سطرهای داده است.

### 💡 باز کردن در حلقه

وقتی روی **لیستی از تاپل‌ها** حلقه می‌زنید، می‌توانید هر تاپل را مستقیماً در متغیر حلقه باز کنید (همان کاری که با <code dir="ltr">enumerate</code> و <code dir="ltr">zip</code> در جلسه‌ی ۸ کردید):

</div>

````python
records = [("Ali", 17.5), ("Sara", 19), ("Reza", 12)]
for name, score in records:
    print(f"{name:<6}{score:>6.2f}")
````

## ۳. کاربردهای تاپل

### 💡 چه وقت تاپل و چه وقت لیست؟

| | لیست | تاپل |
|---|---|---|
| تغییرپذیر؟ | بله | خیر |
| کاربرد معمول | مجموعه‌ای از **چیزهای هم‌نوع** که تعدادشان تغییر می‌کند (نمره‌های یک کلاس) | یک **رکورد** با ساختار ثابت که هر جایگاهش معنی خاصی دارد (نام، سن، شهر) |
| مثال | <code dir="ltr">scores = [17, 12, 19]</code> | <code dir="ltr">student = ("Ali", 21, "Bushehr")</code> |
| کلید دیکشنری یا عضو مجموعه | نمی‌تواند باشد | می‌تواند باشد |

یک قاعده‌ی سرانگشتی: اگر می‌توانید جمله‌ی «**هر جایگاه** این دنباله معنی مشخصی دارد» را بگویید، تاپل مناسب‌تر است. در علوم داده هم شکل یک ماتریس یا جدول به شکل تاپل گزارش می‌شود؛ مثلاً NumPy و Pandas ابعاد داده را به شکل <code dir="ltr">(1000, 20)</code> (۱۰۰۰ سطر، ۲۰ ستون) می‌دهند.

### 💡 مرتب‌سازی تاپل‌ها

تاپل‌ها (و لیست‌ها) **عضو به عضو** مقایسه می‌شوند، مثل کلمات در فرهنگ لغت (جلسه‌ی ۳): اول عضو اول؛ اگر برابر بود، عضو دوم؛ و ... . این ویژگی مرتب‌سازی رکوردها را ساده می‌کند:

</div>

````python
print((1, 9) < (2, 0))
print((2, "b") < (2, "a"))

results = [(17, "Sara"), (15, "Ali"), (17, "Reza"), (19, "Mina")]
for score, name in sorted(results, reverse=True):
    print(name, score)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">(1, 9) &lt; (2, 0)</code> درست است، چون عضو اول تعیین‌کننده است و <code dir="ltr">1 &lt; 2</code>.
- در <code dir="ltr">results</code> نمره را **اول** گذاشته‌ایم تا مرتب‌سازی بر اساس نمره انجام شود. دو نفر با نمره‌ی <code dir="ltr">17</code> بر اساس عضو **دوم** (نام) مرتب می‌شوند؛ با <code dir="ltr">reverse=True</code>، <code dir="ltr">Sara</code> قبل از <code dir="ltr">Reza</code> می‌آید.

> مرتب کردن بر اساس یک عضو دلخواه (بدون جابه‌جا کردن ترتیب اعضای تاپل) را در جلسه‌ی ۱۳ با پارامتر <code dir="ltr">key</code> می‌بینیم.

### ❓ پیش‌بینی کن

خروجی کد زیر چیست؟ چرا «بیشترین» و «کمترین» در این مثال معنی خاصی دارند؟

</div>

````python
readings = [("2026-10-28", 31.5), ("2026-10-29", 29.0), ("2026-10-30", 33.2)]
print(max(readings))
print(min(readings))
````

<div dir="rtl">

🔍 <code dir="ltr">max</code> و <code dir="ltr">min</code> هم عضو به عضو مقایسه می‌کنند؛ پس اینجا بر اساس **تاریخ** (رشته‌ای که به ترتیب سال، ماه، روز نوشته شده) انتخاب می‌کنند، نه بر اساس دما. <code dir="ltr">max</code> آخرین روز و <code dir="ltr">min</code> اولین روز را برمی‌گرداند؛ گرم‌ترین روز (<code dir="ltr">33.2</code>) تصادفاً همان آخرین روز است، ولی کمترین دما (<code dir="ltr">29.0</code>) مال <code dir="ltr">min(readings)</code> نیست. اگر دما را جایگاه اول می‌گذاشتیم، نتیجه فرق می‌کرد.

</div>

## ۴. مجموعه (set)

### 💡 مفهوم

**مجموعه (set)** گردآوری‌ای از مقدارهاست که:

- **تکراری ندارد:** هر مقدار حداکثر یک بار در آن هست.
- **ترتیب ندارد:** اندیس ندارد و <code dir="ltr">s[0]</code> معنی ندارد.
- **عضویت را بسیار سریع بررسی می‌کند:** <code dir="ltr">x in s</code> تقریباً مستقل از اندازه‌ی مجموعه، فوراً جواب می‌دهد.

مجموعه را با <code dir="ltr">{ }</code> یا با تابع <code dir="ltr">set()</code> می‌سازیم:

</div>

````python
s1 = {1, 2, 3, 4}
s2 = set([1, 2, 2, 3, 4, 4, 4])     # duplicates removed
s3 = set("banana")
print(s1, s2, s3)
print(len(s2))

empty = set()                        # NOT {}  -- that is an empty dictionary
print(type(empty), type({}))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">set([1, 2, 2, 3, 4, 4, 4])</code> تکراری‌ها را حذف می‌کند: <code dir="ltr">{1, 2, 3, 4}</code>.
- <code dir="ltr">set("banana")</code> حروف **متمایز** کلمه است. ترتیب نمایش آن ممکن است با ترتیب حروف کلمه فرق کند؛ مجموعه ترتیب ندارد و نباید به ترتیب نمایش آن تکیه کنید.
- <code dir="ltr">{}</code> یک **دیکشنری** خالی است (جلسه‌ی ۱۰)، نه مجموعه‌ی خالی.

### 💡 افزودن و حذف

</div>

````python
fruits = {"apple", "banana"}
fruits.add("cherry")
fruits.add("apple")            # already there: nothing happens
print(fruits)

fruits.remove("banana")        # KeyError if not present
fruits.discard("kiwi")         # no error if not present
print(fruits)
print("apple" in fruits, "kiwi" not in fruits)
````

<div dir="rtl">

### 💡 کاربرد رایج: مقدارهای یکتا

در علوم داده بارها می‌پرسیم «این ستون چند مقدار **متمایز** دارد؟» یا «چه کلاس‌هایی در داده هست؟»:

</div>

````python
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
classes = set(labels)
print(classes, len(classes))
print(sorted(classes))           # sorted() returns a list
````

<div dir="rtl">

🔍 <code dir="ltr">sorted</code> روی مجموعه یک **لیست** مرتب برمی‌گرداند؛ این راه معمول برای داشتن یک ترتیب ثابت از اعضای یک مجموعه است.

### ⚠️ چه چیزی می‌تواند عضو مجموعه باشد؟

مجموعه برای جست‌وجوی سریع، از هر عضو یک «اثر انگشت» عددی (**hash**) می‌سازد. این اثر انگشت فقط برای مقدارهای **تغییرناپذیر** معتبر می‌ماند؛ پس اعداد، رشته‌ها و تاپل‌ها (از مقدارهای تغییرناپذیر) می‌توانند عضو مجموعه باشند، اما **لیست نمی‌تواند**:

</div>

````python
points = {(0, 0), (1, 2), (0, 0)}
print(points)

bad = {[1, 2], [3, 4]}
````

<div dir="rtl">

🔍 خطای <code dir="ltr">TypeError: unhashable type: 'list'</code> یعنی لیست اثر انگشت ندارد. برای ذخیره‌ی نقطه‌ها یا رکوردها در مجموعه، از **تاپل** استفاده کنید. (همین قاعده برای **کلیدهای دیکشنری** در جلسه‌ی ۱۰ هم برقرار است.)

### ⚠️ اشتباه رایج: حذف تکراری‌ها با از دست دادن ترتیب

<code dir="ltr">list(set(seq))</code> تکراری‌ها را حذف می‌کند، اما **ترتیب** اولیه را تضمین نمی‌کند. اگر ترتیب مهم است، از الگوی «دیده‌شده‌ها» استفاده کنید (سؤال ۵).

</div>

## ۵. عملیات روی مجموعه‌ها

### 💡 مفهوم

مجموعه‌های پایتون همان مجموعه‌های ریاضی هستند و عملیات آن‌ها را پشتیبانی می‌کنند:

</div>

<p align="center"><img src="images/s09_set_ops.png" width="760" alt="s09_set_ops.png"></p>

<p align="center" dir="rtl"><i>چهار عمل اصلی مجموعه‌ها: اجتماع، اشتراک، تفاضل و تفاضل متقارن (بخش رنگی، نتیجه است).</i></p>

<div dir="rtl">

| عملگر | متد | معنی |
|:---:|---|---|
| <code dir="ltr">A &#124; B</code> | <code dir="ltr">A.union(B)</code> | اجتماع: عضوهای A **یا** B |
| <code dir="ltr">A &amp; B</code> | <code dir="ltr">A.intersection(B)</code> | اشتراک: عضوهای A **و** B |
| <code dir="ltr">A - B</code> | <code dir="ltr">A.difference(B)</code> | تفاضل: در A هست ولی در B نیست |
| <code dir="ltr">A ^ B</code> | <code dir="ltr">A.symmetric&#95;difference(B)</code> | تفاضل متقارن: دقیقاً در یکی از دو مجموعه |
| <code dir="ltr">A &lt;= B</code> | <code dir="ltr">A.issubset(B)</code> | آیا A زیرمجموعه‌ی B است؟ |

</div>

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B, B - A)
print("Symmetric difference:", A ^ B)
print({1, 2} <= A, A <= B)
````

<div dir="rtl">

### 🔍 کد را بخوان: کاربرد در داده

فرض کنید شناسه‌ی دانشجویانی را که در دو درس ثبت‌نام کرده‌اند دارید:

</div>

````python
python_course = {"s01", "s02", "s03", "s05", "s08"}
stats_course = {"s02", "s03", "s04", "s08", "s09"}

both = python_course & stats_course
only_python = python_course - stats_course
everyone = python_course | stats_course

print("both:", sorted(both))
print("only python:", sorted(only_python))
print("total students:", len(everyone))
````

<div dir="rtl">

- اشتراک: چه کسانی **هر دو** درس را گرفته‌اند.
- تفاضل: چه کسانی **فقط** پایتون را گرفته‌اند (ترتیب مهم است: <code dir="ltr">A - B</code> با <code dir="ltr">B - A</code> فرق دارد).
- اجتماع: تعداد کل دانشجویان **متمایز**؛ اگر دو لیست را با <code dir="ltr">+</code> به هم می‌چسباندیم، دانشجویان مشترک دو بار شمرده می‌شدند.

### ❓ پیش‌بینی کن

بدون اجرا، خروجی هر خط را بنویسید.

</div>

````python
X = set("data")
Y = set("science")
print(sorted(X & Y))
print(len(X | Y))
print(sorted(Y - X))
print(X ^ Y == (X | Y) - (X & Y))
````

<div dir="rtl">

🔍 <code dir="ltr">X = {'d', 'a', 't'}</code> و <code dir="ltr">Y = {'s', 'c', 'i', 'e', 'n'}</code> (حرف تکراری فقط یک بار حساب می‌شود). اشتراک خالی است (<code dir="ltr">[]</code>)، اجتماع <code dir="ltr">8</code> عضو دارد، <code dir="ltr">Y - X</code> همان کل <code dir="ltr">Y</code> است، و خط آخر یک اتحاد همیشه درست را بررسی می‌کند: <code dir="ltr">True</code>.

</div>

## 📌 آنچه باید بلد باشید

- تاپل مرتب و تغییرناپذیر است؛ فقط <code dir="ltr">count</code> و <code dir="ltr">index</code> دارد؛ <code dir="ltr">+</code> و <code dir="ltr">&#42;</code> تاپل جدید می‌سازند.
- <code dir="ltr">(5)</code> عدد است و <code dir="ltr">(5,)</code> تاپل؛ کاما تاپل می‌سازد.
- تاپلی که لیست دارد: خود لیست قابل تغییر است.
- باز کردن تاپل، از جمله با <code dir="ltr">&#42;</code>، و باز کردن در حلقه‌ی <code dir="ltr">for</code>.
- انتخاب بین لیست و تاپل؛ مقایسه و مرتب‌سازی تاپل‌ها **عضو به عضو**.
- مجموعه: بدون تکرار، بدون ترتیب، بدون اندیس، عضویت سریع؛ <code dir="ltr">set()</code> برای مجموعه‌ی خالی.
- <code dir="ltr">add</code>، <code dir="ltr">remove</code> (خطا اگر نباشد)، <code dir="ltr">discard</code> (بدون خطا).
- فقط مقدارهای تغییرناپذیر (hashable) می‌توانند عضو مجموعه باشند.
- <code dir="ltr">&#124;</code>، <code dir="ltr">&amp;</code>، <code dir="ltr">-</code>، <code dir="ltr">^</code>، <code dir="ltr">&lt;=</code> و کاربرد آن‌ها روی داده‌ی واقعی.

</div>

### سؤال ۲ — k نزدیک‌ترین همسایه (kNN)

الگوریتم **k نزدیک‌ترین همسایه** برای دسته‌بندی یک نقطه‌ی جدید، <code dir="ltr">k</code> نقطه‌ی نزدیک‌تر از داده‌ی آموزش را پیدا می‌کند و برچسبی را انتخاب می‌کند که بین آن‌ها **بیشترین تکرار** را دارد. داده‌ی آموزش، لیستی از تاپل‌های <code dir="ltr">(features, label)</code> است:

</div>

```python
train = [((1, 1), "A"), ((2, 1), "A"), ((4, 4), "B"),
         ((5, 4), "B"), ((1, 2), "A"), ((4, 5), "B")]

def distance(p, q):
    s = 0
    for a, b in zip(p, q):
        s += (a - b) ** 2
    return s ** 0.5

def knn_predict(train, query, k):
    pairs = []
    for features, label in train:
        pairs.append((distance(features, query), label))
    nearest = sorted(pairs)[:k]
    votes = []
    for d, label in nearest:
        votes.append(label)
    best = None
    for c in sorted(set(votes)):
        if best is None or votes.count(c) > votes.count(best):
            best = c
    return best
```

<div dir="rtl">

الف) برای <code dir="ltr">query = (3, 3)</code> فاصله‌ی هر نقطه‌ی آموزش را حساب کنید (با سه رقم اعشار) و لیست مرتب‌شده‌ی <code dir="ltr">pairs</code> را بنویسید.

ب) خروجی <code dir="ltr">knn&#95;predict(train, (3, 3), k)</code> را برای <code dir="ltr">k = 1</code>، <code dir="ltr">k = 3</code> و <code dir="ltr">k = 5</code> بنویسید.

ج) نتیجه‌ی <code dir="ltr">k = 3</code> به یک «تصادف» بستگی دارد. آن تصادف چیست؟ (راهنمایی: چهار نقطه فاصله‌ی **دقیقاً برابر** دارند. <code dir="ltr">sorted</code> بین آن‌ها چطور تصمیم می‌گیرد؟) اگر برچسب‌ها <code dir="ltr">"A"</code> و <code dir="ltr">"B"</code> با هم عوض شوند (یعنی کلاس اول <code dir="ltr">"B"</code> نام بگیرد)، پیش‌بینی برای <code dir="ltr">k = 3</code> چه می‌شود؟ چه نتیجه‌ای می‌گیرید؟

<details>
<summary><b>پاسخ</b></summary>

الف) فاصله تا <code dir="ltr">(3, 3)</code>: <code dir="ltr">(1, 1)</code>: <code dir="ltr">√8 ≈ 2.828</code>؛ <code dir="ltr">(2, 1)</code>: <code dir="ltr">√5 ≈ 2.236</code>؛ <code dir="ltr">(4, 4)</code>: <code dir="ltr">√2 ≈ 1.414</code>؛ <code dir="ltr">(5, 4)</code>: <code dir="ltr">√5</code>؛ <code dir="ltr">(1, 2)</code>: <code dir="ltr">√5</code>؛ <code dir="ltr">(4, 5)</code>: <code dir="ltr">√5</code>.

<div dir="ltr">

```text
[(1.414, 'B'), (2.236, 'A'), (2.236, 'A'), (2.236, 'B'), (2.236, 'B'), (2.828, 'A')]
```

</div>

ب) <code dir="ltr">k = 1</code>: <code dir="ltr">B</code>. <code dir="ltr">k = 3</code>: رأی‌ها <code dir="ltr">B, A, A</code> پس <code dir="ltr">A</code>. <code dir="ltr">k = 5</code>: رأی‌ها <code dir="ltr">B, A, A, B, B</code> پس <code dir="ltr">B</code>.

ج) چهار نقطه با فاصله‌ی برابر <code dir="ltr">√5</code> وجود دارد و فقط دو تای آن‌ها در سه همسایه‌ی نزدیک جا می‌شوند. چون تاپل‌ها عضو به عضو مقایسه می‌شوند، در فاصله‌ی برابر، **برچسب** تعیین‌کننده است و <code dir="ltr">"A"</code> (به خاطر ترتیب الفبایی) جلوتر می‌آید. اگر نام کلاس‌ها عوض شود، این بار دو نقطه‌ی گروه دیگر (<code dir="ltr">(5, 4)</code> و <code dir="ltr">(4, 5)</code>) جلو می‌افتند و پیش‌بینی برای <code dir="ltr">k = 3</code> به گروه نقطه‌ی <code dir="ltr">(4, 4)</code> تغییر می‌کند. یعنی پیش‌بینی به **اسم** کلاس‌ها بستگی دارد که اصلاً نباید اهمیتی داشته باشد! پیاده‌سازی‌های واقعی برای حالت تساوی قاعده‌ی صریح دارند (مثلاً در نظر گرفتن ترتیب داده، یا انتخاب <code dir="ltr">k</code> فرد و وزن‌دهی بر اساس فاصله).

</details>

### سؤال ۳ — واژگان و شباهت متن

برای کار با متن در یادگیری ماشین، اول باید **واژگان (vocabulary)** را بسازیم:

</div>

```python
docs = ["the cat sat", "the cat ran", "a dog ran"]
```

<div dir="rtl">

الف) با یک حلقه و عملگر <code dir="ltr">&#124;</code>، مجموعه‌ی همه‌ی کلمات (<code dir="ltr">vocab</code>) را بسازید. اندازه‌ی آن چند است؟ <code dir="ltr">sorted(vocab)</code> چیست؟

ب) **شباهت ژاکارد** دو مجموعه برابر است با اندازه‌ی اشتراک تقسیم بر اندازه‌ی اجتماع. تابع <code dir="ltr">jaccard(a, b)</code> را بنویسید و شباهت هر سه جفت سند را (بر اساس مجموعه‌ی کلماتشان) حساب کنید.

ج) با استفاده از <code dir="ltr">sorted(vocab)</code> به‌عنوان ترتیب ثابت ستون‌ها، هر سند را به یک بردار **کیسه‌ی کلمات (bag of words)** تبدیل کنید: عضو <code dir="ltr">j</code> ام بردار، تعداد دفعات تکرار کلمه‌ی <code dir="ltr">j</code> ام در سند است. بردار سند اول چیست؟ چرا ترتیب ثابت ستون‌ها لازم است و چرا نمی‌توانستیم مستقیم از ترتیب اعضای <code dir="ltr">vocab</code> استفاده کنیم؟

<details>
<summary><b>پاسخ</b></summary>

الف)

<div dir="ltr">

```python
vocab = set()
for d in docs:
    vocab = vocab | set(d.split())
```

</div>

اندازه <code dir="ltr">6</code>؛ <code dir="ltr">['a', 'cat', 'dog', 'ran', 'sat', 'the']</code>.

ب)

<div dir="ltr">

```python
def jaccard(a, b):
    return len(a & b) / len(a | b)
```

</div>

| جفت | اشتراک | اجتماع | ژاکارد |
|---|---|---|---|
| سند ۱ و ۲ | <code dir="ltr">{the, cat}</code> | ۴ کلمه | <code dir="ltr">0.5</code> |
| سند ۱ و ۳ | تهی | ۶ کلمه | <code dir="ltr">0.0</code> |
| سند ۲ و ۳ | <code dir="ltr">{ran}</code> | ۵ کلمه | <code dir="ltr">0.2</code> |

ج)

<div dir="ltr">

```python

vector = []
for w in columns:
    vector.append(docs[0].split().count(w))
```

</div>

بردار سند اول: <code dir="ltr">[0, 1, 0, 0, 1, 1]</code>. هر عضو بردار باید در **همه‌ی** سندها به یک کلمه‌ی ثابت اشاره کند تا بردارها با هم قابل مقایسه باشند (مثلاً برای ضرب داخلی یا شباهت کسینوسی جلسه‌ی ۸). مجموعه ترتیب تضمین‌شده ندارد و حتی اندیس‌گذاری را پشتیبانی نمی‌کند.

</details>

### سؤال ۴ — نشت داده

train_ids = ["p01", "p02", "p03", "p04", "p05", "p03", "p06"]

با عملیات مجموعه‌ها، به این سؤال‌ها جواب دهید (کد و نتیجه):

الف) آیا نمونه‌ای هم در آموزش و هم در آزمون آمده است؟ (به این مشکل **نشت داده** می‌گویند و ارزیابی مدل را بیش از حد خوش‌بینانه می‌کند.)

ب) آیا داده‌ی آموزش نمونه‌ی **تکراری** دارد؟ چند تا؟

ج) کدام کلاس در آزمون هست ولی مدل هرگز در آموزش آن را ندیده است؟ چرا این مشکل است؟

</details>

### سؤال ۵ — حذف تکراری‌ها با حفظ ترتیب

دو تابع زیر را مقایسه کنید:

</div>

```python
def unique_a(seq):
    result = []
    for x in seq:
        if x not in result:
            result.append(x)
    return result

def unique_b(seq):
    seen = set()
    result = []
    for x in seq:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result
```

<div dir="rtl">

الف) خروجی هر دو تابع برای <code dir="ltr">[3, 1, 3, 2, 1, 5]</code> چیست؟ خروجی <code dir="ltr">list(set([3, 1, 3, 2, 1, 5]))</code> چه فرقی با آن‌ها می‌تواند داشته باشد؟

ب) هر دو تابع درست‌اند، اما برای یک لیست با یک میلیون عضو متمایز، <code dir="ltr">unique&#95;a</code> عملاً تمام نمی‌شود و <code dir="ltr">unique&#95;b</code> در کسری از ثانیه تمام می‌شود. چرا؟ تعداد تقریبی مقایسه‌ها را برای هر کدام تخمین بزنید.

<details>
<summary><b>پاسخ</b></summary>

الف) هر دو <code dir="ltr">[3, 1, 2, 5]</code>: اولین وقوع هر مقدار، به ترتیب ظاهر شدن. <code dir="ltr">list(set(...))</code> همان اعضا را دارد اما ترتیبشان تضمین‌شده نیست (مثلاً ممکن است <code dir="ltr">[1, 2, 3, 5]</code> باشد).

ب) در <code dir="ltr">unique&#95;a</code>، عبارت <code dir="ltr">x not in result</code> روی یک **لیست** اجرا می‌شود و باید عضوها را یکی‌یکی بررسی کند؛ برای <code dir="ltr">n</code> عضو متمایز، حدود <code dir="ltr">1 + 2 + ... + n</code> یعنی حدود <code dir="ltr">n² / 2</code> مقایسه (برای یک میلیون، حدود ۵ × ۱۰¹¹). در <code dir="ltr">unique&#95;b</code>، بررسی عضویت در **مجموعه** تقریباً یک قدم است؛ پس کل کار حدود <code dir="ltr">n</code> قدم است. انتخاب ساختار داده‌ی درست، سرعت را میلیون‌ها برابر تغییر می‌دهد.

</details>

### سؤال ۶ — چه چیزی hashable است؟

کدام خط‌ها اجرا می‌شوند و کدام خطا می‌دهند؟ برای خطاها دلیل بیاورید. اندازه‌ی مجموعه‌هایی را که ساخته می‌شوند بنویسید.

</div>

```python
s1 = {(1, 2), (2, 1), (1, 2)}
s2 = {"ab", "ba", "a" + "b"}
s3 = {(1, [2])}
s4 = {1, 1.0, True}
s5 = set([[1, 2], [3]])
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

| خط | نتیجه |
|---|---|
| <code dir="ltr">s1</code> | اجرا می‌شود؛ ۲ عضو (<code dir="ltr">(1, 2)</code> تکراری است؛ <code dir="ltr">(2, 1)</code> ترتیب متفاوتی دارد) |
| <code dir="ltr">s2</code> | اجرا می‌شود؛ ۲ عضو (<code dir="ltr">"a" + "b"</code> همان <code dir="ltr">"ab"</code> است) |
| <code dir="ltr">s3</code> | <code dir="ltr">TypeError: unhashable type: 'list'</code>؛ تاپلی که لیست دارد، hashable نیست |
| <code dir="ltr">s4</code> | اجرا می‌شود؛ **۱** عضو! <code dir="ltr">1</code>، <code dir="ltr">1.0</code> و <code dir="ltr">True</code> از نظر پایتون برابرند (<code dir="ltr">1 == 1.0 == True</code>) و یکی حساب می‌شوند |
| <code dir="ltr">s5</code> | <code dir="ltr">TypeError</code>: اعضای لیست ورودی، خودشان لیست‌اند |

نکته‌ی <code dir="ltr">s4</code> در داده‌ی واقعی دردسرساز است: اگر ستونی شامل <code dir="ltr">1</code> و <code dir="ltr">True</code> باشد، شمارش مقدارهای یکتا با مجموعه آن‌ها را یکی می‌کند.

</details>

### سؤال ۷ — کدگذاری برچسب‌ها

مدل‌های یادگیری ماشین با عدد کار می‌کنند، نه با رشته. جاهای خالی را کامل کنید تا برچسب‌های رشته‌ای به **عدد صحیح** (label encoding) و سپس به بردار **one-hot** تبدیل شوند.

</div>

```python
labels = ["cat", "dog", "cat", "bird", "dog"]

classes = ______(set(labels))        # fixed, reproducible order
encoded = []
for lbl in labels:
    encoded.append(classes.______(lbl))

one_hot = []
for e in encoded:
    row = [0] * ______
    row[e] = 1
    one_hot.append(row)

print(classes)
print(encoded)
print(one_hot)
```

<div dir="rtl">

بعد از کامل کردن: خروجی را بنویسید. سپس توضیح دهید چرا به یک مدل **نباید** بگوییم <code dir="ltr">bird = 0</code>، <code dir="ltr">cat = 1</code>، <code dir="ltr">dog = 2</code> و این اعداد را مثل یک ویژگی عددی معمولی به آن بدهیم، و one-hot چه مشکلی را حل می‌کند.

<details>
<summary><b>پاسخ</b></summary>

<div dir="ltr">

```python
classes = sorted(set(labels))
encoded.append(classes.index(lbl))
row = [0] * len(classes)
```

</div>

خروجی:

<div dir="ltr">

```text
['bird', 'cat', 'dog']
[1, 2, 1, 0, 2]
[[0, 1, 0], [0, 0, 1], [0, 1, 0], [1, 0, 0], [0, 0, 1]]
```

</div>

کدگذاری عددی یک **ترتیب و فاصله‌ی ساختگی** ایجاد می‌کند: مدل خطی فکر می‌کند <code dir="ltr">dog</code> «دو برابر» <code dir="ltr">cat</code> است یا <code dir="ltr">cat</code> بین <code dir="ltr">bird</code> و <code dir="ltr">dog</code> قرار دارد. در one-hot هر کلاس یک بُعد جداگانه دارد و فاصله‌ی همه‌ی کلاس‌ها از هم برابر است. (خروجی softmax در جلسه‌ی ۸ هم دقیقاً با بردار one-hot برچسب واقعی مقایسه می‌شود.)

</details>

</div>

## خلاصه‌ی جلسه

- تاپل یک دنباله‌ی مرتب و تغییرناپذیر است؛ برای رکوردهایی با ساختار ثابت و برگرداندن چند مقدار استفاده می‌شود. تاپل یک‌عضوی کاما لازم دارد.
- باز کردن تاپل (با <code dir="ltr">&#42;</code> برای «بقیه») در انتساب و در حلقه بسیار پرکاربرد است.
- تاپل‌ها عضو به عضو مقایسه و مرتب می‌شوند.
- مجموعه: بدون تکرار، بدون ترتیب، با عضویت سریع؛ فقط مقدارهای hashable عضو آن می‌شوند.
