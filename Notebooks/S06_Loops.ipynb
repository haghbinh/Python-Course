## ۱. حلقه‌ی <code dir="ltr">for</code>

### 💡 مفهوم

**حلقه (loop)** یعنی اجرای چندباره‌ی یک بلوک کد. حلقه‌ی <code dir="ltr">for</code> روی **اعضای یک دنباله** حرکت می‌کند و بدنه را **یک بار برای هر عضو** اجرا می‌کند:

<div dir="ltr">

```python
for item in sequence:
    body
```

</div>

- در هر دور، <code dir="ltr">item</code> به عضو بعدی دنباله وصل می‌شود (به آن **متغیر حلقه** می‌گوییم).
- بدنه مثل <code dir="ltr">if</code> با دونقطه و تورفتگی مشخص می‌شود.
- وقتی اعضا تمام شدند، اجرا از اولین خط بعد از حلقه ادامه پیدا می‌کند.

یک رشته دنباله‌ای از کاراکترهاست؛ پس می‌توان روی آن حلقه زد:

</div>

````python
for ch in "Data":
    print(ch)
print("loop finished")
````

<div dir="rtl">

### 🔍 کد را بخوان

حلقه ۴ بار اجرا می‌شود؛ در هر دور <code dir="ltr">ch</code> یکی از حروف <code dir="ltr">"D"</code>، <code dir="ltr">"a"</code>، <code dir="ltr">"t"</code> و <code dir="ltr">"a"</code> است. <code dir="ltr">print("loop finished")</code> تورفتگی ندارد، پس **یک بار** و بعد از پایان حلقه اجرا می‌شود.

حلقه روی **لیست** هم کار می‌کند (لیست‌ها را در جلسه‌ی ۸ مفصل می‌خوانیم؛ فعلاً بدانید لیست مجموعه‌ای از مقدارها داخل <code dir="ltr">[ ]</code> است):

</div>

````python
my_list = [1, 2, 3, 4, 5]
for i in my_list:
    print(i)
````

<div dir="rtl">

### 💡 الگوی جمع‌کننده (accumulator)

مهم‌ترین کاربرد حلقه در علوم داده، **خلاصه کردن** مجموعه‌ای از مقدارهاست: جمع، شمارش، میانگین، بیشینه و ... . الگوی آن همیشه سه قدم دارد:

1. **قبل از حلقه:** یک متغیر با مقدار اولیه بسازید (برای جمع <code dir="ltr">0</code>).
2. **داخل حلقه:** آن را با هر عضو به‌روز کنید (<code dir="ltr">total += x</code>).
3. **بعد از حلقه:** از نتیجه استفاده کنید.

</div>

````python
scores = [17, 12.5, 19, 15]

total = 0
count = 0
for s in scores:
    total += s
    count += 1

print("sum =", total)
print("mean =", total / count)
````

<div dir="rtl">

### 🔍 کد را بخوان: جدول دنبال کردن

مثل جلسه‌ی ۳، مقدار متغیرها را **بعد از هر دور** حلقه بنویسید:

| دور | <code dir="ltr">s</code> | <code dir="ltr">total</code> | <code dir="ltr">count</code> |
|:---:|:---:|:---:|:---:|
| قبل از حلقه | — | <code dir="ltr">0</code> | <code dir="ltr">0</code> |
| ۱ | <code dir="ltr">17</code> | <code dir="ltr">17</code> | <code dir="ltr">1</code> |
| ۲ | <code dir="ltr">12.5</code> | <code dir="ltr">29.5</code> | <code dir="ltr">2</code> |
| ۳ | <code dir="ltr">19</code> | <code dir="ltr">48.5</code> | <code dir="ltr">3</code> |
| ۴ | <code dir="ltr">15</code> | <code dir="ltr">63.5</code> | <code dir="ltr">4</code> |

بعد از حلقه: میانگین <code dir="ltr">63.5 / 4</code> یعنی <code dir="ltr">15.875</code>؛ همان عددی که در جلسه‌ی ۱ با <code dir="ltr">sum</code> و <code dir="ltr">len</code> به دست آوردیم. (<code dir="ltr">sum</code> و <code dir="ltr">len</code> در واقع همین حلقه را برای شما اجرا می‌کنند.)

### ⚠️ اشتباه رایج: مقدار اولیه داخل حلقه

اگر <code dir="ltr">total = 0</code> را **داخل** حلقه بنویسید، در هر دور دوباره صفر می‌شود و در آخر فقط آخرین عضو را نگه می‌دارد. اگر <code dir="ltr">print</code> را داخل حلقه بگذارید، به جای یک نتیجه، در هر دور یک بار چاپ می‌شود. **همیشه بررسی کنید هر خط داخل حلقه است یا بیرون آن.**

</div>

## ۲. تابع <code dir="ltr">range()</code>

### 💡 مفهوم

برای تکرار روی **اعداد**، از <code dir="ltr">range</code> استفاده می‌کنیم. <code dir="ltr">range</code> دنباله‌ای از اعداد صحیح می‌سازد:

| شکل | اعداد تولیدشده | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">range(stop)</code> | از <code dir="ltr">0</code> تا <code dir="ltr">stop - 1</code> | <code dir="ltr">range(5)</code> | <code dir="ltr">0 1 2 3 4</code> |
| <code dir="ltr">range(start, stop)</code> | از <code dir="ltr">start</code> تا <code dir="ltr">stop - 1</code> | <code dir="ltr">range(2, 6)</code> | <code dir="ltr">2 3 4 5</code> |
| <code dir="ltr">range(start, stop, step)</code> | از <code dir="ltr">start</code> با گام <code dir="ltr">step</code>، قبل از رسیدن به <code dir="ltr">stop</code> | <code dir="ltr">range(0, 10, 2)</code> | <code dir="ltr">0 2 4 6 8</code> |

**قاعده‌ی طلایی:** <code dir="ltr">stop</code> **هیچ‌وقت** جزو نتیجه نیست. <code dir="ltr">range(5)</code> پنج عدد دارد، اما عدد <code dir="ltr">5</code> جزو آن‌ها نیست.

</div>

````python
for i in range(5):
    print("Iteration:", i)
````

````python
for i in range(5, 0, -1):     # counting down
    print(i, end=" ")
print("Go!")
````

<div dir="rtl">

### 🔍 کد را بخوان

- در سلول اول، <code dir="ltr">i</code> از <code dir="ltr">0</code> تا <code dir="ltr">4</code> تغییر می‌کند: ۵ دور.
- در سلول دوم گام **منفی** است؛ پس شمارش رو به پایین است و قبل از رسیدن به <code dir="ltr">0</code> متوقف می‌شود: <code dir="ltr">5 4 3 2 1</code>.
- <code dir="ltr">end=" "</code> (جلسه‌ی ۴) باعث می‌شود اعداد در یک خط چاپ شوند.

### ❓ پیش‌بینی کن

جمع اعداد ۱ تا ۱۰۰ را با یک حلقه حساب می‌کنیم. **قبل از اجرا** دو سؤال را جواب دهید: چرا <code dir="ltr">stop</code> برابر <code dir="ltr">101</code> است؟ اگر <code dir="ltr">range(100)</code> می‌نوشتیم، جواب چقدر می‌شد؟

</div>

````python
total = 0
for k in range(1, 101):
    total += k
print(total)
````

<div dir="rtl">

🔍 برای اینکه خود <code dir="ltr">100</code> هم جمع شود، <code dir="ltr">stop</code> باید <code dir="ltr">101</code> باشد. <code dir="ltr">range(100)</code> اعداد <code dir="ltr">0</code> تا <code dir="ltr">99</code> را می‌دهد و جمع آن‌ها <code dir="ltr">4950</code> است؛ یعنی <code dir="ltr">100</code> کمتر از جواب درست.

### ⚠️ اشتباه رایج: خطای «یکی کم یا زیاد» (off-by-one)

رایج‌ترین خطای حلقه‌ها همین است: یک دور کمتر یا بیشتر از مقدار لازم. هر وقت <code dir="ltr">range</code> می‌نویسید، از خودتان بپرسید: **اولین** عدد چیست؟ **آخرین** عدد چیست؟ **چند** عدد؟ (تعداد اعداد <code dir="ltr">range(a, b)</code> برابر <code dir="ltr">b - a</code> است.)

</div>

## ۳. حلقه‌ی <code dir="ltr">while</code>

### 💡 مفهوم

حلقه‌ی <code dir="ltr">while</code> تا وقتی **یک شرط درست باشد** تکرار می‌شود:

<div dir="ltr">

```python
while condition:
    body
```

</div>

قبل از **هر** دور، شرط بررسی می‌شود. اگر درست بود بدنه اجرا می‌شود و دوباره سراغ شرط می‌رود؛ اگر نادرست بود حلقه تمام می‌شود.

</div>

````python
i = 0
while i < 5:
    print("Iteration:", i)
    i += 1
````

<div dir="rtl">

### 🔍 کد را بخوان

این حلقه دقیقاً همان کار <code dir="ltr">for i in range(5)</code> را می‌کند، اما سه بخش آن جداگانه نوشته شده است: **مقدار اولیه** (<code dir="ltr">i = 0</code>)، **شرط** (<code dir="ltr">i &lt; 5</code>) و **به‌روزرسانی** (<code dir="ltr">i += 1</code>). اگر هر کدام از این سه را اشتباه بنویسید، حلقه درست کار نمی‌کند.

### 💡 چه وقت <code dir="ltr">while</code>؟

<code dir="ltr">while</code> وقتی مناسب است که **از قبل نمی‌دانید** حلقه چند بار باید تکرار شود. مثال: پس‌اندازی با سود سالانه‌ی ۱۸٪ چند سال طول می‌کشد تا دو برابر شود؟

</div>

````python
money = 1000
years = 0
while money < 2000:
    money *= 1.18
    years += 1

print(f"{years} years, final amount = {money:.0f}")
````

<div dir="rtl">

### 🔍 کد را بخوان: ارقام یک عدد

کد زیر از <code dir="ltr">%</code> و <code dir="ltr">//</code> (جلسه‌ی ۳) استفاده می‌کند تا ارقام یک عدد را یکی‌یکی جدا کند. اول جدول دنبال کردن را روی کاغذ کامل کنید، بعد اجرا کنید:

</div>

````python
n = 2026
digit_sum = 0
while n > 0:
    digit_sum += n % 10    # take the last digit
    n //= 10               # remove the last digit
print(digit_sum)
````

<div dir="rtl">

| دور | <code dir="ltr">n % 10</code> | <code dir="ltr">digit&#95;sum</code> | <code dir="ltr">n</code> (بعد از دور) |
|:---:|:---:|:---:|:---:|
| قبل از حلقه | — | <code dir="ltr">0</code> | <code dir="ltr">2026</code> |
| ۱ | <code dir="ltr">6</code> | <code dir="ltr">6</code> | <code dir="ltr">202</code> |
| ۲ | <code dir="ltr">2</code> | <code dir="ltr">8</code> | <code dir="ltr">20</code> |
| ۳ | <code dir="ltr">0</code> | <code dir="ltr">8</code> | <code dir="ltr">2</code> |
| ۴ | <code dir="ltr">2</code> | <code dir="ltr">10</code> | <code dir="ltr">0</code> |

بعد از دور چهارم <code dir="ltr">n</code> صفر می‌شود، شرط <code dir="ltr">n &gt; 0</code> نادرست است و حلقه تمام می‌شود. جواب: <code dir="ltr">10</code>.

### ⚠️ اشتباه رایج: حلقه‌ی بی‌پایان

اگر شرط <code dir="ltr">while</code> هیچ‌وقت نادرست نشود، حلقه **تا ابد** اجرا می‌شود. معمولاً علت این است که فراموش کرده‌اید متغیر شرط را داخل حلقه تغییر دهید:

<div dir="ltr">

```python
i = 0
while i < 5:
    print(i)       # i never changes: infinite loop!
```

</div>

اگر سلولی بی‌پایان اجرا شد (علامت <code dir="ltr">[&#42;]</code> کنار آن ماند)، اجرای کرنل را متوقف کنید: در Jupyter دکمه‌ی ■ (Interrupt) و در VS Code دکمه‌ی مربعی کنار سلول.

### مقایسه‌ی <code dir="ltr">for</code> و <code dir="ltr">while</code>

| | <code dir="ltr">for</code> | <code dir="ltr">while</code> |
|---|---|---|
| تکرار بر اساس | اعضای یک دنباله | یک شرط |
| تعداد دورها | از قبل معلوم | ممکن است از قبل معلوم نباشد |
| خطر حلقه‌ی بی‌پایان | تقریباً ندارد | دارد |
| مثال | جمع همه‌ی نمره‌ها | تکرار تا رسیدن به یک هدف |

</div>

## ۴. <code dir="ltr">break</code> و <code dir="ltr">continue</code>

### 💡 مفهوم

- **<code dir="ltr">break</code>**: حلقه را **همین الان** کاملاً متوقف می‌کند و اجرا از بعد از حلقه ادامه پیدا می‌کند.
- **<code dir="ltr">continue</code>**: **بقیه‌ی همین دور** را رد می‌کند و به **دور بعد** می‌رود.

هر دو معمولاً داخل یک <code dir="ltr">if</code> می‌آیند.

</div>

````python
for i in range(10):
    if i == 5:
        break
    print("Value:", i)
````

````python
for i in range(5):
    if i == 2:
        continue
    print("Number:", i)
````

<div dir="rtl">

### 🔍 کد را بخوان

- در سلول اول، وقتی <code dir="ltr">i</code> به <code dir="ltr">5</code> می‌رسد <code dir="ltr">break</code> اجرا می‌شود؛ پس فقط <code dir="ltr">0</code> تا <code dir="ltr">4</code> چاپ می‌شوند و دورهای <code dir="ltr">5</code> تا <code dir="ltr">9</code> اصلاً اجرا نمی‌شوند.
- در سلول دوم، وقتی <code dir="ltr">i</code> برابر <code dir="ltr">2</code> است <code dir="ltr">continue</code> اجرا می‌شود؛ <code dir="ltr">print</code> آن دور رد می‌شود ولی حلقه ادامه پیدا می‌کند: <code dir="ltr">0 1 3 4</code>.

### 💡 جست‌وجو با <code dir="ltr">break</code>: آیا عدد اول است؟

یک کاربرد رایج <code dir="ltr">break</code>، **جست‌وجو** است: به محض پیدا شدن جواب، دیگر لازم نیست ادامه دهیم. کد زیر بررسی می‌کند که <code dir="ltr">n</code> عدد اول است یا نه:

</div>

````python
n = 91
is_prime = True
for d in range(2, n):
    if n % d == 0:
        is_prime = False
        print(n, "is divisible by", d)
        break

print("prime?", is_prime)
````

<div dir="rtl">

🔍 متغیر <code dir="ltr">is&#95;prime</code> یک **پرچم (flag)** است: فرض می‌کنیم عدد اول است و اگر مقسوم‌علیهی پیدا شد، پرچم را <code dir="ltr">False</code> می‌کنیم. برای <code dir="ltr">91</code> در دور <code dir="ltr">d = 7</code> مقسوم‌علیه پیدا می‌شود و <code dir="ltr">break</code> بقیه‌ی دورها (۸ تا ۹۰) را حذف می‌کند. مقدار <code dir="ltr">n</code> را به <code dir="ltr">97</code> تغییر دهید و نتیجه را ببینید.

> **برای خواندن کد دیگران:** پایتون ساختاری به نام <code dir="ltr">for ... else</code> هم دارد: بلوک <code dir="ltr">else</code> حلقه فقط وقتی اجرا می‌شود که حلقه **بدون <code dir="ltr">break</code>** تمام شده باشد. با آن می‌توان همین برنامه را بدون پرچم نوشت. لازم نیست از آن استفاده کنید، ولی اگر در کدی دیدید، معنی‌اش را بدانید.

### ❓ پیش‌بینی کن

خروجی سلول زیر را دقیقاً بنویسید. (برای هر <code dir="ltr">i</code> بپرسید: اول کدام <code dir="ltr">if</code> بررسی می‌شود؟)

</div>

````python
for i in range(10):
    if i % 3 == 0:
        continue
    if i > 7:
        break
    print(i, end=" ")
````

<div dir="rtl">

🔍 مضرب‌های ۳ (<code dir="ltr">0</code>، <code dir="ltr">3</code>، <code dir="ltr">6</code>، <code dir="ltr">9</code>) با <code dir="ltr">continue</code> رد می‌شوند. برای <code dir="ltr">i = 8</code> شرط <code dir="ltr">i &gt; 7</code> درست است و حلقه با <code dir="ltr">break</code> تمام می‌شود. خروجی: <code dir="ltr">1 2 4 5 7</code>.

### ⚠️ اشتباه رایج: <code dir="ltr">continue</code> در <code dir="ltr">while</code>

در حلقه‌ی <code dir="ltr">while</code>، اگر <code dir="ltr">continue</code> **قبل از** به‌روزرسانی شمارنده بیاید، به‌روزرسانی رد می‌شود و حلقه بی‌پایان می‌شود:

<div dir="ltr">

```python
i = 0
while i < 5:
    if i == 3:
        continue     # i stays 3 forever!
    print(i)
    i += 1
```

</div>

راه حل: شمارنده را **قبل از** <code dir="ltr">continue</code> به‌روز کنید (یا از <code dir="ltr">for</code> استفاده کنید).

</div>

## ۵. حلقه‌های تودرتو

### 💡 مفهوم

بدنه‌ی یک حلقه می‌تواند خودش یک حلقه باشد. حلقه‌ی **درونی** برای **هر** دور حلقه‌ی **بیرونی**، از اول تا آخر اجرا می‌شود.

</div>

````python
for i in [1, 2]:
    for j in [10, 20, 30]:
        print(i, j)
````

<div dir="rtl">

### 🔍 کد را بخوان

حلقه‌ی بیرونی ۲ دور و حلقه‌ی درونی در هر دور ۳ دور دارد؛ پس <code dir="ltr">print</code> در مجموع **۲ × ۳ = ۶** بار اجرا می‌شود. ترتیب: اول <code dir="ltr">i = 1</code> با همه‌ی <code dir="ltr">j</code>ها، بعد <code dir="ltr">i = 2</code> با همه‌ی <code dir="ltr">j</code>ها.

یک مثال کاربردی‌تر: جدول ضرب. به نقش f-string و <code dir="ltr">end</code> (جلسه‌ی ۴) در مرتب کردن خروجی دقت کنید:

</div>

````python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()          # new line after each row
````

<div dir="rtl">

- <code dir="ltr">{row &#42; col:4}</code> هر عدد را در ستونی به عرض ۴ قرار می‌دهد (اعداد به‌طور پیش‌فرض راست‌چین‌اند).
- <code dir="ltr">print()</code> بعد از حلقه‌ی درونی (اما داخل حلقه‌ی بیرونی) در پایان هر سطر خط را عوض می‌کند. اگر تورفتگی آن را یک سطح بیشتر کنید، چه اتفاقی می‌افتد؟ امتحان کنید.

### ❓ پیش‌بینی کن

در کد زیر، حلقه‌ی درونی به متغیر حلقه‌ی **بیرونی** وابسته است. چه شکلی چاپ می‌شود؟ و <code dir="ltr">print</code> داخلی در مجموع چند بار اجرا می‌شود؟

</div>

````python
for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
````

<div dir="rtl">

🔍 برای <code dir="ltr">i = 1</code> حلقه‌ی درونی ۱ بار اجرا می‌شود، برای <code dir="ltr">i = 2</code> دو بار و ... . پس یک مثلث چهارسطری چاپ می‌شود و <code dir="ltr">print("&#42;")</code> در مجموع <code dir="ltr">1 + 2 + 3 + 4 = 10</code> بار اجرا می‌شود.

**چرا شمردن تعداد دورها مهم است؟** در علوم داده با داده‌های بزرگ کار می‌کنید. دو حلقه‌ی تودرتو روی یک میلیون ردیف یعنی **یک میلیون میلیون** دور، که عملاً تمام نمی‌شود. یکی از دلایل اصلی استفاده از NumPy و Pandas (بخش ۳) همین است: آن‌ها حلقه‌ها را با سرعت بسیار بیشتری برای شما اجرا می‌کنند.

</div>

<div dir="rtl">

## 📌 آنچه باید بلد باشید

- ساختار <code dir="ltr">for</code> روی رشته، لیست و <code dir="ltr">range</code>؛ اینکه کدام خط‌ها داخل حلقه‌اند و کدام بیرون.
- الگوی جمع‌کننده: مقدار اولیه **قبل از** حلقه، به‌روزرسانی **داخل** حلقه، استفاده **بعد از** حلقه.
- اعداد تولیدشده توسط <code dir="ltr">range(start, stop, step)</code> را دقیقاً بنویسید (<code dir="ltr">stop</code> جزو نتیجه نیست؛ گام منفی؛ <code dir="ltr">range</code> خالی).
- سه بخش حلقه‌ی <code dir="ltr">while</code> (مقدار اولیه، شرط، به‌روزرسانی) و علت حلقه‌ی بی‌پایان.
- برای یک حلقه‌ی <code dir="ltr">for</code> یا <code dir="ltr">while</code> جدول دنبال کردن بنویسید.
- اثر <code dir="ltr">break</code> و <code dir="ltr">continue</code> بر خروجی؛ خطای <code dir="ltr">continue</code> قبل از به‌روزرسانی در <code dir="ltr">while</code>.
- الگوی جست‌وجو با پرچم و <code dir="ltr">break</code>.
- تعداد اجرای بدنه در حلقه‌های تودرتو (حتی وقتی حلقه‌ی درونی به حلقه‌ی بیرونی وابسته است).

</div>

### سؤال ۲ — جدول دنبال کردن

جدول دنبال کردن کد زیر را کامل کنید و خروجی را بنویسید. این کد در کل چه کاری انجام می‌دهد؟

</div>

```python
n = 1234
result = 0
while n > 0:
    result = result * 10 + n % 10
    n //= 10
print(result)
```

<details>

| ۱ | <code dir="ltr">4</code> | <code dir="ltr">4</code> | <code dir="ltr">123</code> |

خروجی: <code dir="ltr">4321</code>. این کد ارقام عدد را **برعکس** می‌کند.

### سؤال ۳ — <code dir="ltr">break</code> و <code dir="ltr">continue</code>

خروجی کد زیر را دقیقاً بنویسید.

</div>

```python
total = 0
for k in range(1, 20):
    if k % 2 == 0:
        continue
    total += k
    if total > 20:
        break
print(k, total)
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

<div dir="ltr">

```text
9 25
```

</div>

فقط اعداد فرد جمع می‌شوند: <code dir="ltr">1</code>، <code dir="ltr">4</code> (=1+3)، <code dir="ltr">9</code>، <code dir="ltr">16</code>، <code dir="ltr">25</code>. بعد از اضافه شدن <code dir="ltr">9</code>، مقدار <code dir="ltr">total</code> برابر <code dir="ltr">25</code> و بیشتر از <code dir="ltr">20</code> است؛ پس حلقه در <code dir="ltr">k = 9</code> با <code dir="ltr">break</code> تمام می‌شود. دقت کنید متغیر حلقه (<code dir="ltr">k</code>) بعد از حلقه هم آخرین مقدارش را نگه می‌دارد.

### سؤال ۴ — پیدا کردن خطا

هدف: چاپ اعداد ۰ تا ۵ **به‌جز ۳**. هر دو کد زیر اشتباه‌اند. مشکل هر کدام را توضیح دهید و درستش کنید.

</div>

```python
# A
i = 0
while i <= 5:
    if i == 3:
        continue
    print(i)
    i += 1

# B
for i in range(5):
    if i == 3:
        break
    print(i)
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

**الف)** وقتی <code dir="ltr">i</code> برابر <code dir="ltr">3</code> می‌شود، <code dir="ltr">continue</code> قبل از <code dir="ltr">i += 1</code> اجرا می‌شود؛ <code dir="ltr">i</code> هیچ‌وقت تغییر نمی‌کند و حلقه بی‌پایان می‌شود. اصلاح:

<div dir="ltr">

```python
i = 0
while i <= 5:
    if i == 3:
        i += 1
        continue
    print(i)
    i += 1
```

</div>

**ب)** دو اشکال دارد: <code dir="ltr">break</code> حلقه را در <code dir="ltr">3</code> کاملاً متوقف می‌کند (باید <code dir="ltr">continue</code> باشد)، و <code dir="ltr">range(5)</code> عدد <code dir="ltr">5</code> را شامل نمی‌شود (باید <code dir="ltr">range(6)</code> باشد). اصلاح:

<div dir="ltr">

```python
for i in range(6):
    if i == 3:
        continue
    print(i)
```

</div>

## خلاصه‌ی جلسه

- <code dir="ltr">for</code> روی اعضای یک دنباله (رشته، لیست، <code dir="ltr">range</code>) تکرار می‌کند؛ <code dir="ltr">while</code> تا وقتی شرط درست است.
- <code dir="ltr">range(start, stop, step)</code>: <code dir="ltr">stop</code> جزو نتیجه نیست؛ مراقب خطای یکی کم یا زیاد باشید.
- الگوی جمع‌کننده: مقدار اولیه قبل از حلقه، به‌روزرسانی داخل حلقه.
- در <code dir="ltr">while</code> حتماً متغیر شرط را تغییر دهید؛ وگرنه حلقه بی‌پایان می‌شود.
- <code dir="ltr">break</code> حلقه را تمام می‌کند؛ <code dir="ltr">continue</code> به دور بعد می‌رود.
- در حلقه‌های تودرتو، حلقه‌ی درونی برای هر دور بیرونی کامل اجرا می‌شود؛ تعداد دورها را بشمارید.

**جلسه‌ی بعد:** توابع (تعریف تابع، پارامتر، مقدار بازگشتی).
