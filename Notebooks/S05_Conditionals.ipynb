## ۱. دستور <code dir="ltr">if</code>

### 💡 مفهوم

تا اینجا همه‌ی برنامه‌ها خط به خط، از بالا به پایین اجرا می‌شدند. **کنترل جریان (control flow)** یعنی برنامه بتواند **تصمیم بگیرد** کدام خط‌ها اجرا شوند. ساده‌ترین ابزار آن <code dir="ltr">if</code> است:

<div dir="ltr">

```python
if condition:
    statement_1
    statement_2
next_statement
```

</div>

- بعد از <code dir="ltr">if</code> یک **شرط** می‌آید (معمولاً یک مقایسه از جلسه‌ی ۳) و بعد از آن **دونقطه** (<code dir="ltr">:</code>).
- خط‌هایی که **تورفتگی** دارند (معمولاً ۴ فاصله) **بدنه‌ی** <code dir="ltr">if</code> هستند و فقط وقتی شرط درست باشد اجرا می‌شوند.
- اولین خط **بدون تورفتگی** (<code dir="ltr">next&#95;statement</code>) دیگر جزو <code dir="ltr">if</code> نیست و در هر حالت اجرا می‌شود.

در پایتون، تورفتگی **بخشی از دستور زبان** است، نه فقط برای زیبایی. پایتون از روی تورفتگی می‌فهمد یک **بلوک** کد کجا شروع و کجا تمام می‌شود.

</div>

````python
temperature = 38.2

if temperature > 37.5:
    print("Fever detected")
    print("Please see a doctor")
print("Check finished")
````

<div dir="rtl">

### 🔍 کد را بخوان

- شرط <code dir="ltr">temperature &gt; 37.5</code> برابر <code dir="ltr">True</code> است، پس **هر دو** خط تورفته اجرا می‌شوند.
- <code dir="ltr">print("Check finished")</code> تورفتگی ندارد؛ پس بیرون از <code dir="ltr">if</code> است و همیشه اجرا می‌شود.

مقدار <code dir="ltr">temperature</code> را به <code dir="ltr">36.6</code> تغییر دهید و دوباره اجرا کنید. کدام خط‌ها چاپ می‌شوند؟

### ❓ پیش‌بینی کن

تنها فرق دو سلول زیر **تورفتگی خط آخر** است. خروجی هر کدام چیست؟

</div>

````python
x = 3
if x > 5:
    print("big")
    print("done")
````

````python
x = 3
if x > 5:
    print("big")
print("done")
````

<div dir="rtl">

🔍 در سلول اول، <code dir="ltr">print("done")</code> جزو بدنه‌ی <code dir="ltr">if</code> است و چون شرط نادرست است، **هیچ چیزی** چاپ نمی‌شود. در سلول دوم، <code dir="ltr">done</code> بیرون از <code dir="ltr">if</code> است و چاپ می‌شود. یک تورفتگی، رفتار برنامه را عوض کرد!

### 💡 شرط‌هایی که بولی نیستند

بعد از <code dir="ltr">if</code> لازم نیست حتماً یک مقایسه بیاید. پایتون هر مقداری را مثل <code dir="ltr">bool()</code> (جلسه‌ی ۲) به درست یا نادرست تبدیل می‌کند: **صفر، رشته‌ی خالی و <code dir="ltr">None</code>** نادرست‌اند و بقیه درست.

### ❓ پیش‌بینی کن

کدام خط‌ها چاپ می‌شوند؟

</div>

````python
name = ""
count = 0
city = "0"

if name:
    print("name is set")
if count:
    print("count is non-zero")
if city:
    print("city is set")
````

<div dir="rtl">

🔍 فقط <code dir="ltr">city is set</code> چاپ می‌شود: <code dir="ltr">""</code> و <code dir="ltr">0</code> نادرست‌اند، اما <code dir="ltr">"0"</code> یک رشته‌ی **غیرخالی** است و درست حساب می‌شود.

### ⚠️ اشتباه رایج

- **فراموش کردن دونقطه** یا **استفاده از <code dir="ltr">=</code> به جای <code dir="ltr">==</code>** در شرط: هر دو <code dir="ltr">SyntaxError</code> می‌دهند.
- **تورفتگی ناهماهنگ:** همه‌ی خط‌های یک بلوک باید دقیقاً یک اندازه تورفتگی داشته باشند. سلول زیر را اجرا کنید و پیام خطا را بخوانید:

</div>

````python
score = 18
if score >= 17:
    print("Excellent")
      print("Well done")
````

<div dir="rtl">

🔍 خطای <code dir="ltr">IndentationError: unexpected indent</code> یعنی خط دوم بدنه، تورفتگی بیشتری از خط قبلی دارد. **همیشه از ۴ فاصله استفاده کنید** (VS Code با زدن کلید Tab به‌طور خودکار ۴ فاصله می‌گذارد).

</div>

## ۲. <code dir="ltr">else</code> و <code dir="ltr">elif</code>

### 💡 مفهوم

- **<code dir="ltr">else</code>** («در غیر این صورت») بلوکی است که وقتی شرط <code dir="ltr">if</code> **نادرست** باشد اجرا می‌شود. <code dir="ltr">else</code> شرط ندارد.
- **<code dir="ltr">elif</code>** (مخفف «else if») یک شرط **دیگر** را بررسی می‌کند؛ فقط اگر همه‌ی شرط‌های قبلی نادرست بوده باشند.

در یک زنجیره‌ی <code dir="ltr">if</code> / <code dir="ltr">elif</code> / ... / <code dir="ltr">else</code>، پایتون شرط‌ها را **از بالا به پایین** بررسی می‌کند و **فقط اولین** شاخه‌ای را که شرطش درست باشد اجرا می‌کند؛ بقیه‌ی شرط‌ها اصلاً بررسی نمی‌شوند. پس **دقیقاً یک** شاخه اجرا می‌شود (اگر <code dir="ltr">else</code> وجود داشته باشد).

</div>

````python
x = 15
if x > 20:
    print("x is greater than 20")
elif x == 15:
    print("x is exactly 15")
else:
    print("x is less than 15")
````

<div dir="rtl">

### 🔍 کد را بخوان: تبدیل نمره به رتبه

نمودار زیر مسیر اجرای یک زنجیره‌ی <code dir="ltr">elif</code> را برای <code dir="ltr">score = 15</code> نشان می‌دهد. کد متناظر آن پایین آمده است. مسیر را با انگشت دنبال کنید:

</div>

<p align="center"><img src="images/s05_elif_flow.png" width="520" alt="s05_elif_flow.png"></p>

<p align="center" dir="rtl"><i>زنجیره‌ی <code dir="ltr">if</code> / <code dir="ltr">elif</code> / <code dir="ltr">else</code>: شرط‌ها از بالا بررسی می‌شوند و اولین شرط درست، مسیر را تعیین می‌کند.</i></p>

````python
score = 15

if score >= 17:
    grade = "A"
elif score >= 14:
    grade = "B"
elif score >= 10:
    grade = "C"
else:
    grade = "F"

print(grade)
````

<div dir="rtl">

- <code dir="ltr">15 &gt;= 17</code> نادرست است، پس می‌رویم سراغ <code dir="ltr">elif</code> بعدی.
- <code dir="ltr">15 &gt;= 14</code> درست است؛ <code dir="ltr">grade = "B"</code> اجرا می‌شود و **بقیه‌ی زنجیره نادیده گرفته می‌شود**؛ با اینکه <code dir="ltr">15 &gt;= 10</code> هم درست است.
- دقت کنید در شاخه‌ی دوم لازم نیست بنویسیم <code dir="ltr">14 &lt;= score &lt; 17</code>؛ چون اگر به این شاخه رسیده‌ایم، **حتماً** شرط قبلی (<code dir="ltr">score &gt;= 17</code>) نادرست بوده است.

### ⚠️ اشتباه رایج: ترتیب نادرست شرط‌ها

اگر شرط **عمومی‌تر** را بالاتر بگذارید، شرط‌های خاص‌تر هیچ‌وقت به نوبت نمی‌رسند:

</div>

````python
score = 19

if score >= 10:
    grade = "C"
elif score >= 14:
    grade = "B"
elif score >= 17:
    grade = "A"
else:
    grade = "F"

print(grade)
````

<div dir="rtl">

🔍 نمره‌ی <code dir="ltr">19</code> رتبه‌ی <code dir="ltr">C</code> گرفت! چون <code dir="ltr">19 &gt;= 10</code> درست است و زنجیره همان‌جا تمام می‌شود. در زنجیره‌ی <code dir="ltr">elif</code> همیشه از **سخت‌گیرانه‌ترین** شرط شروع کنید.

### ❓ پیش‌بینی کن: چند <code dir="ltr">if</code> یا یک زنجیره؟

دو سلول زیر فقط در کلمه‌ی <code dir="ltr">elif</code> فرق دارند. خروجی هر کدام چیست؟

</div>

````python
x = 15
if x > 10:
    print("A")
if x > 5:
    print("B")
if x > 0:
    print("C")
````

````python
x = 15
if x > 10:
    print("A")
elif x > 5:
    print("B")
elif x > 0:
    print("C")
````

<div dir="rtl">

🔍 در سلول اول سه <code dir="ltr">if</code> **مستقل** داریم و هر کدام جداگانه بررسی می‌شود؛ هر سه درست‌اند و <code dir="ltr">A</code>، <code dir="ltr">B</code> و <code dir="ltr">C</code> چاپ می‌شوند. در سلول دوم یک **زنجیره** داریم و فقط اولین شاخه‌ی درست اجرا می‌شود: فقط <code dir="ltr">A</code>.

**قاعده:** اگر حالت‌ها **ناسازگار** هستند (فقط یکی از آن‌ها باید رخ دهد)، از <code dir="ltr">elif</code> استفاده کنید. اگر هر شرط باید **جداگانه** بررسی شود، از <code dir="ltr">if</code>های مستقل.

</div>

## ۳. شرط‌های تودرتو

### 💡 مفهوم

بدنه‌ی یک <code dir="ltr">if</code> می‌تواند خودش یک <code dir="ltr">if</code> دیگر داشته باشد. به این حالت **شرط تودرتو (nested)** می‌گویند. هر سطح، یک تورفتگی بیشتر دارد.

</div>

````python
age = 22
has_id = False

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Please show your ID")
else:
    print("Entry not allowed")
````

<div dir="rtl">

### 🔍 کد را بخوان

تورفتگی نشان می‌دهد هر <code dir="ltr">else</code> مال کدام <code dir="ltr">if</code> است:

- <code dir="ltr">else</code> اول (با ۴ فاصله) مال <code dir="ltr">if has&#95;id</code> است.
- <code dir="ltr">else</code> دوم (بدون تورفتگی) مال <code dir="ltr">if age &gt;= 18</code> است.

برای <code dir="ltr">age = 22</code> و <code dir="ltr">has&#95;id = False</code>: شرط بیرونی درست است، وارد آن می‌شویم؛ شرط درونی نادرست است، پس <code dir="ltr">Please show your ID</code> چاپ می‌شود.

### 💡 ساده کردن با <code dir="ltr">and</code>

وقتی شاخه‌ی <code dir="ltr">else</code> درونی لازم نیست، دو <code dir="ltr">if</code> تودرتو را می‌توان با <code dir="ltr">and</code> یکی کرد. این دو کد **هم‌ارز** هستند:

</div>

```python
# nested
if age >= 18:
    if has_id:
        print("Entry allowed")

# flat
if age >= 18 and has_id:
    print("Entry allowed")
```

<div dir="rtl">

کد **کم‌عمق‌تر** معمولاً خواناتر است. سعی کنید بیشتر از دو سطح تودرتو ننویسید.

### ❓ پیش‌بینی کن

برای هر یک از چهار حالت زیر، خروجی کد را بنویسید. بعد مقدارها را در سلول تغییر دهید و جواب‌هایتان را بررسی کنید.

| <code dir="ltr">is&#95;member</code> | <code dir="ltr">total</code> | خروجی؟ |
|:---:|:---:|---|
| <code dir="ltr">True</code> | <code dir="ltr">120</code> | |
| <code dir="ltr">True</code> | <code dir="ltr">80</code> | |
| <code dir="ltr">False</code> | <code dir="ltr">120</code> | |
| <code dir="ltr">False</code> | <code dir="ltr">80</code> | |

</div>

````python
is_member = True
total = 80

if is_member:
    if total >= 100:
        discount = 0.20
    else:
        discount = 0.10
else:
    if total >= 100:
        discount = 0.05
    else:
        discount = 0.0

print(f"discount = {discount:.0%}")
````

<div dir="rtl">

**نوشتن جدول برای همه‌ی حالت‌ها** بهترین راه فهمیدن (و آزمودن) یک کد شرطی است. در امتحان هم زیاد از شما خواسته می‌شود.

</div>

## ۴. عبارت شرطی یک‌خطی

### 💡 مفهوم

وقتی می‌خواهید بر اساس یک شرط، **یکی از دو مقدار** را انتخاب کنید، پایتون شکل کوتاه‌تری دارد (به آن **عبارت شرطی** یا **ternary** هم می‌گویند):

<div dir="ltr">

```python
value = A if condition else B
```

</div>

یعنی: «اگر <code dir="ltr">condition</code> درست است، مقدار <code dir="ltr">A</code>؛ وگرنه مقدار <code dir="ltr">B</code>». این دو کد هم‌ارزند:

</div>

````python
x = 7

# full form
if x % 2 == 0:
    y = x ** 2
else:
    y = x ** 3
print(y)

# one-line form
y = x ** 2 if x % 2 == 0 else x ** 3
print(y)
````

<div dir="rtl">

### 🔍 کد را بخوان

عبارت یک‌خطی را این‌طور بخوانید: «<code dir="ltr">y</code> برابر است با <code dir="ltr">x &#42;&#42; 2</code>، **اگر** <code dir="ltr">x</code> زوج باشد، **وگرنه** <code dir="ltr">x &#42;&#42; 3</code>». چون <code dir="ltr">7</code> فرد است، هر دو روش <code dir="ltr">343</code> می‌دهند.

یک مثال دیگر از برچسب زدن به داده، که در علوم داده بسیار رایج است:

</div>

````python
age = 27
group = "young" if age < 30 else "old"
print("You are", group)
````

<div dir="rtl">

> **نکته برای آینده:** در NumPy و Pandas، برچسب زدن به **همه‌ی ردیف‌های یک ستون** با یک شرط، با تابعی به نام <code dir="ltr">np.where</code> انجام می‌شود که دقیقاً همین منطق را دارد: <code dir="ltr">np.where(condition, A, B)</code>.

### ⚠️ اشتباه رایج

- عبارت یک‌خطی **همیشه** به <code dir="ltr">else</code> نیاز دارد؛ <code dir="ltr">y = 1 if x &gt; 0</code> خطای <code dir="ltr">SyntaxError</code> می‌دهد.
- برای منطق پیچیده (بیش از دو حالت) از عبارت یک‌خطی استفاده نکنید؛ <code dir="ltr">if</code> / <code dir="ltr">elif</code> معمولی خواناتر است.

### ⚠️ درباره‌ی <code dir="ltr">eval(input(...))</code>

در بعضی کتاب‌ها و کدهای قدیمی برای گرفتن عدد از کاربر این را می‌بینید:

<div dir="ltr">

```python
x = eval(input("Enter x: "))
```

</div>

<code dir="ltr">eval</code> **هر کد پایتونی** را که کاربر تایپ کند اجرا می‌کند؛ پس کاربر می‌تواند به جای عدد، دستوری خطرناک وارد کند. **از <code dir="ltr">eval</code> استفاده نکنید**؛ به جای آن <code dir="ltr">int(input(...))</code> یا <code dir="ltr">float(input(...))</code> بنویسید (جلسه‌ی ۴).

</div>

<div dir="rtl">

## 📌 آنچه باید بلد باشید

- ساختار <code dir="ltr">if</code> / <code dir="ltr">elif</code> / <code dir="ltr">else</code>، نقش دونقطه، و اینکه تورفتگی مرز بلوک‌ها را تعیین می‌کند.
- برای یک مقدار داده‌شده، مشخص کنید کدام شاخه‌ی یک زنجیره اجرا می‌شود؛ در زنجیره فقط **اولین** شرط درست اجرا می‌شود.
- اهمیت **ترتیب** شرط‌ها در زنجیره‌ی <code dir="ltr">elif</code> (از سخت‌گیرانه‌ترین به عمومی‌ترین).
- تفاوت خروجی چند <code dir="ltr">if</code> مستقل و یک زنجیره‌ی <code dir="ltr">elif</code>.
- خواندن شرط‌های تودرتو، تشخیص اینکه هر <code dir="ltr">else</code> مال کدام <code dir="ltr">if</code> است، و ساده کردن آن‌ها با <code dir="ltr">and</code>.
- کامل کردن جدول خروجی برای همه‌ی حالت‌های ورودی.
- عبارت شرطی یک‌خطی <code dir="ltr">A if condition else B</code>.
- <code dir="ltr">""</code>، <code dir="ltr">0</code> و <code dir="ltr">None</code> در شرط نادرست‌اند؛ <code dir="ltr">"0"</code> درست است.

</div>

## 📝 سؤالات مرور / نمونه سؤال امتحان

### سؤال ۱ — پیش‌بینی خروجی

### سؤال ۳ — پیدا کردن خطا

کد زیر قرار است بر اساس شاخص توده‌ی بدنی (BMI) دسته‌بندی کند، اما برای <code dir="ltr">bmi = 32</code> خروجی <code dir="ltr">normal</code> می‌دهد (انتظار: <code dir="ltr">obese</code>). علت را توضیح دهید و کد را درست کنید.

</div>

```python
if bmi >= 18.5:
    category = "normal"
elif bmi >= 25:
    category = "overweight"
elif bmi >= 30:
    category = "obese"
else:
    category = "underweight"
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

چون <code dir="ltr">32 &gt;= 18.5</code> درست است، زنجیره در همان شاخه‌ی اول تمام می‌شود و شرط‌های بعدی هرگز بررسی نمی‌شوند. مشکل، **ترتیب شرط‌ها** است: شرط عمومی‌تر بالاتر آمده است. با این کد، هیچ مقداری هرگز <code dir="ltr">overweight</code> یا <code dir="ltr">obese</code> نمی‌گیرد.

اصلاح (از سخت‌گیرانه‌ترین شرط):

<div dir="ltr">

```python
if bmi >= 30:
    category = "obese"
elif bmi >= 25:
    category = "overweight"
elif bmi >= 18.5:
    category = "normal"
else:
    category = "underweight"
```

</div>

</details>

### سؤال ۴ — هم‌ارزی

آیا کد **الف** و کد **ب** برای همه‌ی مقدارهای <code dir="ltr">x</code> (عدد) خروجی یکسانی دارند؟ اگر نه، یک مقدار برای <code dir="ltr">x</code> مثال بزنید که خروجی‌ها فرق کنند.

</div>

```python
# A
if x > 0:
    if x < 10:
        print("small positive")
    else:
        print("other")

# B
if x > 0 and x < 10:
    print("small positive")
else:
    print("other")
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

**خیر.** در کد الف، برای <code dir="ltr">x &lt;= 0</code> هیچ چیزی چاپ نمی‌شود (چون <code dir="ltr">else</code> درونی فقط برای <code dir="ltr">x &gt; 0</code> اجرا می‌شود). در کد ب، برای <code dir="ltr">x &lt;= 0</code> عبارت <code dir="ltr">other</code> چاپ می‌شود. مثال: <code dir="ltr">x = -5</code>. برای <code dir="ltr">x</code> بین ۰ تا ۱۰ و برای <code dir="ltr">x &gt;= 10</code> خروجی دو کد یکسان است.

</details>

ب) **بزرگ‌ترین** مقدار بین <code dir="ltr">a</code> و <code dir="ltr">b</code> (همان کاری که تابع <code dir="ltr">max(a, b)</code> انجام می‌دهد).

### سؤال ۶ — پیدا کردن خطاهای نحوی

کد زیر چند خطای **نحوی (syntax)** دارد. همه را پیدا کنید و نسخه‌ی درست را بنویسید.

</div>

```python
x = 10
if x = 10
    print("ten")
else if x > 10:
    print("more")
else:
print("less")
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

1. <code dir="ltr">if x = 10</code> باید <code dir="ltr">if x == 10</code> باشد (مقایسه، نه انتساب).
2. بعد از شرط <code dir="ltr">if</code> دونقطه (<code dir="ltr">:</code>) نیامده است.
3. <code dir="ltr">else if</code> در پایتون وجود ندارد؛ باید <code dir="ltr">elif</code> باشد.
4. خط آخر تورفتگی ندارد؛ بدنه‌ی <code dir="ltr">else</code> باید تورفتگی داشته باشد.

<div dir="ltr">

```python
x = 10
if x == 10:
    print("ten")
elif x > 10:
    print("more")
else:
    print("less")
```

</details>

### سؤال ۷ — تکمیل کد

یک فروشگاه هزینه‌ی ارسال را این‌طور حساب می‌کند: سفارش‌های **۵۰۰ هزار تومان یا بیشتر** ارسال رایگان دارند؛ برای بقیه، اگر مقصد **همان شهر** باشد هزینه‌ی ارسال ۳۰ هزار تومان و در غیر این صورت ۶۰ هزار تومان است. جاهای خالی را کامل کنید.

</div>

```python
total = 420       # thousand tomans
same_city = False

if ______:
    shipping = 0
elif ______:
    shipping = 30
else:
    shipping = ______

print(f"Shipping: {shipping}")
```

<div dir="rtl">

<details>
<summary><b>پاسخ</b></summary>

<div dir="ltr">

```python
if total >= 500:
    shipping = 0
elif same_city:
    shipping = 30
else:
    shipping = 60
```

</div>

برای <code dir="ltr">total = 420</code> و <code dir="ltr">same&#95;city = False</code> خروجی <code dir="ltr">Shipping: 60</code> است. دقت کنید <code dir="ltr">elif same&#95;city</code> کافی است؛ لازم نیست بنویسید <code dir="ltr">same&#95;city == True</code>.

</details>

## خلاصه‌ی جلسه

- <code dir="ltr">if</code> بر اساس یک شرط تصمیم می‌گیرد؛ دونقطه و تورفتگی (۴ فاصله) بلوک را مشخص می‌کنند.
- در زنجیره‌ی <code dir="ltr">if</code> / <code dir="ltr">elif</code> / <code dir="ltr">else</code> فقط **اولین** شاخه‌ی درست اجرا می‌شود؛ پس ترتیب شرط‌ها مهم است.
- چند <code dir="ltr">if</code> مستقل ممکن است چند شاخه را اجرا کنند؛ زنجیره‌ی <code dir="ltr">elif</code> حداکثر یکی.
- شرط‌های تودرتو را با تورفتگی بخوانید و در صورت امکان با <code dir="ltr">and</code> ساده کنید؛ برای فهم کد، جدول همه‌ی حالت‌ها را بنویسید.
- <code dir="ltr">A if condition else B</code> برای انتخاب بین دو مقدار است.
- از <code dir="ltr">eval(input())</code> استفاده نکنید.

**جلسه‌ی بعد:** حلقه‌ها (<code dir="ltr">for</code>، <code dir="ltr">while</code>، <code dir="ltr">range</code>، <code dir="ltr">break</code> و <code dir="ltr">continue</code>).
