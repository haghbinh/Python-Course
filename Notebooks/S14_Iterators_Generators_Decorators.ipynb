## ۱. iterable و iterator

### 💡 مفهوم

از جلسه‌ی ۶ روی لیست، رشته، <code dir="ltr">range</code> و دیکشنری حلقه زده‌اید. به هر چیزی که بشود رویش حلقه‌ی <code dir="ltr">for</code> زد **iterable** می‌گویند. اما خود حلقه مستقیماً با لیست کار نمی‌کند؛ اول از آن یک **iterator** می‌گیرد:

- <code dir="ltr">iter(x)</code>: از یک iterable، یک iterator تازه می‌سازد.
- <code dir="ltr">next(it)</code>: مقدار **بعدی** را از iterator می‌گیرد. iterator جای خودش را **به خاطر می‌سپارد**.
- وقتی مقداری نمانده باشد، <code dir="ltr">next</code> خطای <code dir="ltr">StopIteration</code> می‌دهد.

</div>

<p align="center"><img src="images/s14_iter_protocol.png" width="820" alt="s14_iter_protocol.png"></p>

<p align="center" dir="rtl"><i>پروتکل پیمایش: <code dir="ltr">iter()</code> از iterable یک iterator می‌سازد و <code dir="ltr">next()</code> مقدارها را یکی‌یکی تحویل می‌دهد.</i></p>

````python
nums = [10, 20, 30]
it = iter(nums)
print(type(nums).__name__, type(it).__name__)
print(next(it))
print(next(it))
print(next(it))
````

````python
print(next(it))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">it</code> یک <code dir="ltr">list&#95;iterator</code> است؛ یک «نشانگر» روی لیست که می‌داند تا کجا جلو رفته است.
- هر <code dir="ltr">next</code> یک قدم جلو می‌رود. بعد از سه مقدار، <code dir="ltr">next</code> چهارم خطای <code dir="ltr">StopIteration</code> می‌دهد؛ این خطا یعنی «تمام شد».

### 💡 حلقه‌ی <code dir="ltr">for</code> در پشت صحنه

حلقه‌ی <code dir="ltr">for x in nums:</code> در واقع معادل کد زیر است (<code dir="ltr">try</code> و <code dir="ltr">except</code> را در جلسه‌ی ۱۵ می‌خوانید؛ اینجا فقط یعنی «وقتی <code dir="ltr">StopIteration</code> رخ داد، از حلقه بیرون برو»):

<div dir="ltr">

```python
it = iter(nums)
while True:
    try:
        x = next(it)
    except StopIteration:
        break
    print(x)
```

</div>

پس هر چیزی که بتواند به <code dir="ltr">iter</code> جواب دهد، در <code dir="ltr">for</code>، <code dir="ltr">list()</code>، <code dir="ltr">sum()</code>، <code dir="ltr">max()</code>، <code dir="ltr">sorted()</code>، <code dir="ltr">zip()</code> و ... قابل استفاده است.

### 💡 iterable چندبار مصرف، iterator یک‌بار مصرف

| نوع | مثال‌ها | چند بار می‌شود پیمود؟ |
|---|---|---|
| iterable معمولی | <code dir="ltr">list</code>، <code dir="ltr">tuple</code>، <code dir="ltr">str</code>، <code dir="ltr">dict</code>، <code dir="ltr">set</code>، <code dir="ltr">range</code> | هر چند بار؛ هر <code dir="ltr">for</code> یک iterator تازه می‌گیرد |
| iterator | خروجی <code dir="ltr">iter()</code>، <code dir="ltr">zip</code>، <code dir="ltr">enumerate</code>، <code dir="ltr">map</code>، <code dir="ltr">filter</code>، <code dir="ltr">reversed</code>، generatorها | فقط **یک بار** |

</div>

````python
pairs = zip(["a", "b", "c"], [1, 2, 3])
print(type(pairs).__name__)
print(list(pairs))
print(list(pairs))
````

<div dir="rtl">

🔍 <code dir="ltr">zip</code> (جلسه‌ی ۸) یک iterator برمی‌گرداند، نه لیست. اولین <code dir="ltr">list(pairs)</code> همه‌ی مقدارها را مصرف می‌کند و دومی چیزی پیدا نمی‌کند و لیست **خالی** می‌دهد؛ بدون هیچ خطایی.

### ❓ پیش‌بینی کن

قبل از اجرا، خروجی هر خط را بنویسید.

</div>

````python
letters = iter("data")
print(next(letters), next(letters))
for ch in letters:
    print(ch, end=" ")
print()
print(list(letters))

r = range(3)
print(list(r), list(r))
````

<div dir="rtl">

🔍 دو <code dir="ltr">next</code> اول، <code dir="ltr">d</code> و <code dir="ltr">a</code> را برمی‌دارند؛ حلقه‌ی <code dir="ltr">for</code> روی **همان** iterator ادامه می‌دهد و فقط <code dir="ltr">t a</code> را می‌بیند؛ بعد از آن iterator خالی است و <code dir="ltr">list(letters)</code> برابر <code dir="ltr">[]</code> است. اما <code dir="ltr">range</code> یک iterable است (نه iterator): هر بار یک iterator تازه ساخته می‌شود و دو بار <code dir="ltr">[0, 1, 2]</code> چاپ می‌شود.

### ⚠️ اشتباه رایج: دو بار مصرف کردن یک iterator

اگر خروجی <code dir="ltr">zip</code>، <code dir="ltr">map</code> یا <code dir="ltr">enumerate</code> را در یک متغیر بگذارید و دو بار رویش حلقه بزنید، بار دوم **بی‌صدا** هیچ کاری انجام نمی‌شود. اگر داده را چند بار لازم دارید، یک بار آن را به <code dir="ltr">list</code> تبدیل کنید. همچنین <code dir="ltr">next(nums)</code> روی خود لیست خطای <code dir="ltr">TypeError</code> می‌دهد: لیست iterable است، iterator نیست.

</div>

## ۲. تابع generator و <code dir="ltr">yield</code>

### 💡 مفهوم

**تابع generator** تابعی است که به جای <code dir="ltr">return</code>، کلمه‌ی کلیدی <code dir="ltr">yield</code> دارد. فراخوانی آن بدنه را **اجرا نمی‌کند**؛ فقط یک شیء **generator** (که یک iterator است) برمی‌گرداند. هر بار <code dir="ltr">next</code> صدا زده شود:

1. بدنه از جایی که قبلاً متوقف شده بود ادامه پیدا می‌کند،
2. به اولین <code dir="ltr">yield</code> که رسید، آن مقدار را تحویل می‌دهد و **مکث** می‌کند (همه‌ی متغیرهای محلی حفظ می‌شوند)،
3. وقتی بدنه تمام شود، <code dir="ltr">StopIteration</code> رخ می‌دهد.

</div>

````python
def countdown(n):
    print("start")
    while n > 0:
        yield n
        n -= 1
    print("done")

g = countdown(3)
print(type(g).__name__)
print(next(g))
print(next(g))
for x in g:
    print("loop got", x)
````

<div dir="rtl">

### 🔍 کد را بخوان

- خط <code dir="ltr">g = countdown(3)</code> هیچ چیزی چاپ نمی‌کند؛ <code dir="ltr">"start"</code> فقط با اولین <code dir="ltr">next</code> چاپ می‌شود.
- بعد از <code dir="ltr">yield 3</code>، تابع مکث می‌کند. <code dir="ltr">next</code> بعدی از خط <code dir="ltr">n -= 1</code> ادامه می‌دهد؛ مقدار <code dir="ltr">n</code> حفظ شده است.
- حلقه‌ی <code dir="ltr">for</code> بقیه را مصرف می‌کند: فقط <code dir="ltr">1</code> مانده است. بعد از آن حلقه‌ی <code dir="ltr">while</code> تمام می‌شود، <code dir="ltr">"done"</code> چاپ و generator تمام می‌شود و <code dir="ltr">for</code> بی‌صدا متوقف می‌شود.

### 💡 generator بی‌پایان

چون مقدارها فقط وقتی خواسته شوند تولید می‌شوند، generator می‌تواند **بی‌پایان** باشد؛ به شرط اینکه مصرف‌کننده خودش جایی متوقف شود:

</div>

````python
def naturals(start=1):
    n = start
    while True:
        yield n
        n += 1

for i, n in zip(range(5), naturals(10)):
    print(n, end=" ")
````

<div dir="rtl">

🔍 <code dir="ltr">zip</code> به محض تمام شدن کوتاه‌ترین ورودی (<code dir="ltr">range(5)</code>) متوقف می‌شود؛ پس از generator بی‌پایان فقط ۵ مقدار گرفته می‌شود. (بدون چنین حدی، <code dir="ltr">list(naturals())</code> هرگز تمام نمی‌شود.)

### 💡 کاربرد: تقسیم داده به mini-batch

در آموزش شبکه‌های عصبی، داده را یکجا به مدل نمی‌دهیم؛ آن را به **دسته‌های کوچک (mini-batch)** تقسیم می‌کنیم. generator برای این کار ایده‌آل است، چون هر دسته فقط وقتی ساخته می‌شود که مدل آن را بخواهد:

</div>

````python
def batches(data, batch_size):
    for start in range(0, len(data), batch_size):
        yield data[start:start + batch_size]

samples = list(range(10))
for b in batches(samples, 4):
    print(b)
````

<div dir="rtl">

🔍 <code dir="ltr">range(0, 10, 4)</code> شروع‌های <code dir="ltr">0</code>، <code dir="ltr">4</code> و <code dir="ltr">8</code> را می‌دهد. دسته‌ی آخر فقط دو عضو دارد، چون برش لیست از انتهای لیست جلوتر نمی‌رود (جلسه‌ی ۸).

### ❓ پیش‌بینی کن

خروجی را **به ترتیب** بنویسید. آیا اول همه‌ی <code dir="ltr">computing</code>ها چاپ می‌شوند؟

</div>

````python
def squares(n):
    for i in range(n):
        print("computing", i)
        yield i * i

for s in squares(3):
    print("got", s)
````

<div dir="rtl">

🔍 نه. خط‌ها **یک در میان** چاپ می‌شوند: <code dir="ltr">computing 0</code>، <code dir="ltr">got 0</code>، <code dir="ltr">computing 1</code>، <code dir="ltr">got 1</code>، ... هر مقدار درست قبل از مصرف تولید می‌شود؛ به این رفتار **ارزیابی تنبل (lazy evaluation)** می‌گویند. اگر <code dir="ltr">squares</code> به جای <code dir="ltr">yield</code> یک لیست می‌ساخت و <code dir="ltr">return</code> می‌کرد، اول هر سه <code dir="ltr">computing</code> چاپ می‌شد.

### ⚠️ اشتباه رایج: انتظار اجرای فوری

<code dir="ltr">squares(10 &#42;&#42; 9)</code> در یک لحظه برمی‌گردد و هیچ چیزی محاسبه نمی‌کند؛ کار فقط هنگام پیمایش انجام می‌شود. برعکس، اگر در بدنه‌ی یک generator خطایی باشد، آن خطا هم تا اولین <code dir="ltr">next</code> دیده نمی‌شود.

</div>

## ۳. generator expression

### 💡 مفهوم

در جلسه‌ی ۱۳ list comprehension را دیدید. اگر به جای کروشه از **پرانتز** استفاده کنید، یک **generator expression** دارید: همان نحو، اما به جای ساختن یک لیست کامل در حافظه، یک generator برمی‌گرداند که مقدارها را یکی‌یکی تولید می‌کند.

عبارت <code dir="ltr">sum(x &#42; x for x in v)</code> که در جلسه‌ی ۱۳ دیدید همین است؛ وقتی generator expression تنها آرگومان یک تابع باشد، پرانتز اضافه لازم نیست.

</div>

````python
squares_list = [x * x for x in range(5)]
squares_gen = (x * x for x in range(5))
print(squares_list)
print(type(squares_gen).__name__)
print(sum(squares_gen))
print(sum(squares_gen))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">squares&#95;list</code> یک لیست واقعی است و می‌شود هر چند بار از آن استفاده کرد.
- <code dir="ltr">squares&#95;gen</code> یک generator است؛ اولین <code dir="ltr">sum</code> آن را تا ته مصرف می‌کند (<code dir="ltr">30</code>) و <code dir="ltr">sum</code> دوم چیزی پیدا نمی‌کند (<code dir="ltr">0</code>).

### 💡 چرا generator؟ حافظه

</div>

````python
import sys

big_list = [x * x for x in range(1_000_000)]
big_gen = (x * x for x in range(1_000_000))
print(f"{sys.getsizeof(big_list):,} bytes")
print(f"{sys.getsizeof(big_gen):,} bytes")
print(sum(big_gen) == sum(big_list))
````

<div dir="rtl">

🔍 لیست یک میلیون عضوی حدود ۸ میلیون بایت فقط برای **ارجاع** به عددها جا می‌گیرد (خود عددها جدا حساب می‌شوند)، اما generator فقط چند صد بایت است و هر چقدر هم بزرگ باشد، اندازه‌اش ثابت می‌ماند؛ چون فقط «دستور تولید» و وضعیت فعلی را نگه می‌دارد. (عددهای دقیق به نسخه‌ی پایتون بستگی دارد.) نتیجه‌ی جمع در هر دو یکسان است.

### 💡 خط لوله‌ی تنبل

چند generator را می‌توان پشت سر هم وصل کرد. هر مقدار **یکی‌یکی** از کل خط لوله عبور می‌کند و هیچ لیست میانی ساخته نمی‌شود. وقتی در جلسه‌ی ۱۶ فایل‌های چند گیگابایتی را خط‌به‌خط بخوانید، همین الگو را به کار می‌برید:

</div>

````python
lines = ["3.5", " 7 ", "", "x", "10"]        # imagine millions of lines from a file

stripped = (s.strip() for s in lines)
nonempty = (s for s in stripped if s)
numbers = (float(s) for s in nonempty if s.replace(".", "", 1).isdigit())
print(sum(numbers))
````

<div dir="rtl">

🔍 تا خط آخر هیچ کاری انجام نشده است؛ فقط سه generator تعریف شده‌اند. <code dir="ltr">sum</code> مقدار اول را از <code dir="ltr">numbers</code> می‌خواهد، <code dir="ltr">numbers</code> از <code dir="ltr">nonempty</code> و <code dir="ltr">nonempty</code> از <code dir="ltr">stripped</code>. رشته‌ی خالی و <code dir="ltr">"x"</code> در مسیر حذف می‌شوند و جمع <code dir="ltr">3.5 + 7 + 10 = 20.5</code> است.

### ❓ پیش‌بینی کن

<code dir="ltr">any</code> به محض پیدا کردن اولین مقدار درست متوقف می‌شود. تابع <code dir="ltr">check</code> چند بار صدا زده می‌شود؟

</div>

````python
def check(x):
    print("checking", x)
    return x < 0

data = [3, 8, -1, 5, -7]
print(any(check(x) for x in data))
````

<div dir="rtl">

🔍 سه بار: <code dir="ltr">3</code>، <code dir="ltr">8</code> و <code dir="ltr">-1</code>. با رسیدن به اولین عدد منفی، <code dir="ltr">any</code> مقدار <code dir="ltr">True</code> برمی‌گرداند و بقیه‌ی generator اصلاً تولید نمی‌شود. اگر به جای آن <code dir="ltr">any([check(x) for x in data])</code> می‌نوشتید، list comprehension اول **هر پنج** مقدار را بررسی می‌کرد. (<code dir="ltr">all</code> هم به محض دیدن اولین مقدار نادرست متوقف می‌شود.)

### ⚠️ اشتباه رایج: رفتار با generator مثل لیست

generator طول، اندیس و برش ندارد: <code dir="ltr">len(gen)</code> و <code dir="ltr">gen[0]</code> خطای <code dir="ltr">TypeError</code> می‌دهند و بعد از یک بار مصرف، خالی است. قاعده‌ی سرانگشتی: اگر نتیجه را **یک بار** و **به ترتیب** لازم دارید (مثلاً داخل <code dir="ltr">sum</code>، <code dir="ltr">max</code>، <code dir="ltr">any</code> یا یک حلقه)، generator؛ اگر چند بار یا با اندیس لازم دارید، لیست.

</div>

## ۴. مقدمه‌ای بر decorator

### 💡 تابع هم یک شیء است

در پایتون، تابع مثل عدد و لیست یک **شیء** است: می‌شود آن را در یک متغیر گذاشت، به تابع دیگری داد یا از تابعی برگرداند. (در جلسه‌ی ۱۳ تابع را به <code dir="ltr">sorted</code> و <code dir="ltr">map</code> دادید.)

</div>

````python
def shout(text):
    return text.upper() + "!"

speak = shout
print(speak("hello"))
print(shout.__name__)

def apply_twice(f, x):
    return f(f(x))

print(apply_twice(lambda v: v * 3, 2))
````

<div dir="rtl">

🔍 <code dir="ltr">speak = shout</code> تابع را **صدا نمی‌زند** (پرانتز ندارد)؛ فقط نام دیگری برای همان شیء تابع می‌سازد. <code dir="ltr">apply&#95;twice</code> تابع <code dir="ltr">f</code> را دو بار اعمال می‌کند: <code dir="ltr">2 &#42; 3 = 6</code> و <code dir="ltr">6 &#42; 3 = 18</code>.

### 💡 تابع درونی و closure

تابع را می‌شود **داخل** تابع دیگری تعریف کرد و برگرداند. تابع درونی متغیرهای تابع بیرونی را **به خاطر می‌سپارد**، حتی بعد از اینکه تابع بیرونی تمام شده باشد؛ به این ترکیب **closure** می‌گویند:

</div>

````python
def make_power(p):
    def power(x):
        return x ** p
    return power

square = make_power(2)
cube = make_power(3)
print(square(5), cube(2))
print(square.__name__)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">make&#95;power(2)</code> تابع <code dir="ltr">power</code> را برمی‌گرداند که <code dir="ltr">p = 2</code> را در خودش نگه داشته است؛ <code dir="ltr">make&#95;power(3)</code> یک تابع **جداگانه** با <code dir="ltr">p = 3</code> می‌سازد.
- این الگوی «کارخانه‌ی تابع» در یادگیری ماشین زیاد دیده می‌شود؛ مثلاً ساختن یک تابع زیان یا یک برنامه‌ی تغییر نرخ یادگیری با پارامترهای مشخص.

### 💡 decorator

**decorator** تابعی است که یک تابع می‌گیرد و یک تابع **جدید** (معمولاً یک تابع درونی که تابع اصلی را صدا می‌زند) برمی‌گرداند. نوشتن <code dir="ltr">@timer</code> بالای تعریف یک تابع دقیقاً یعنی <code dir="ltr">slow&#95;sum = timer(slow&#95;sum)</code>.

ماژول <code dir="ltr">time</code> و تابع <code dir="ltr">perf&#95;counter</code> آن (یک ساعت دقیق برای زمان‌سنجی) را در جلسه‌ی ۱۷ کامل‌تر می‌بینید. <code dir="ltr">&#42;args</code> و <code dir="ltr">&#42;&#42;kwargs</code> (جلسه‌ی ۷) باعث می‌شوند <code dir="ltr">wrapper</code> هر نوع آرگومانی را بپذیرد و دست‌نخورده به تابع اصلی بدهد.

</div>

````python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"{func.__name__} took {elapsed:.4f} s")
        return result
    return wrapper

@timer
def slow_sum(n):
    total = 0
    for i in range(n):
        total += i
    return total

print(slow_sum(1_000_000))
````

<div dir="rtl">

### 🔍 کد را بخوان

- با <code dir="ltr">@timer</code>، نام <code dir="ltr">slow&#95;sum</code> حالا به <code dir="ltr">wrapper</code> اشاره می‌کند. صدا زدن <code dir="ltr">slow&#95;sum(1&#95;000&#95;000)</code> در واقع <code dir="ltr">wrapper</code> را اجرا می‌کند که زمان را ثبت می‌کند، تابع اصلی را صدا می‌زند، زمان سپری‌شده را چاپ می‌کند و **نتیجه را برمی‌گرداند**.
- عدد زمان در هر اجرا و روی هر کامپیوتر فرق می‌کند.
- کد <code dir="ltr">slow&#95;sum</code> هیچ تغییری نکرده است؛ این قدرت decorator است: افزودن رفتاری مشترک (زمان‌سنجی، ثبت گزارش، بررسی ورودی) به هر تابعی، فقط با یک خط.

### 💡 شمارش فراخوانی‌ها و حافظه‌ی نهان (memoization)

decorator زیر تعداد فراخوانی‌ها را در یک **ویژگی** روی خود <code dir="ltr">wrapper</code> نگه می‌دارد:

</div>

````python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper

@count_calls
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(20), fib.calls)
````

<div dir="rtl">

🔍 فراخوانی‌های **بازگشتی** داخل <code dir="ltr">fib</code> هم شمرده می‌شوند، چون نام <code dir="ltr">fib</code> حالا به <code dir="ltr">wrapper</code> اشاره می‌کند. برای <code dir="ltr">fib(20)</code> بیش از ۲۱ هزار فراخوانی لازم است؛ چون همان مقدارها بارها و بارها از نو محاسبه می‌شوند (مثلاً <code dir="ltr">fib(18)</code> هم از <code dir="ltr">fib(20)</code> و هم از <code dir="ltr">fib(19)</code> صدا زده می‌شود).

راه حل: نتیجه‌ی هر ورودی را یک بار حساب کنیم و در یک دیکشنری نگه داریم (**memoization**؛ همان الگوی شمارش و گروه‌بندی جلسه‌ی ۱۰):

</div>

````python
def memoize(func):
    cache = {}
    def wrapper(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]
    return wrapper

@count_calls
@memoize
def fib2(n):
    if n < 2:
        return n
    return fib2(n - 1) + fib2(n - 2)

print(fib2(20), fib2.calls)
````

<div dir="rtl">

### 🔍 کد را بخوان

- دو decorator **روی هم** قرار گرفته‌اند و از پایین به بالا اعمال می‌شوند: <code dir="ltr">fib2 = count&#95;calls(memoize(fib2))</code>.
- دیکشنری <code dir="ltr">cache</code> متغیر محلی <code dir="ltr">memoize</code> است، اما چون <code dir="ltr">wrapper</code> به آن ارجاع دارد (closure)، بین فراخوانی‌ها باقی می‌ماند.
- تعداد فراخوانی‌ها از ۲۱٬۸۹۱ به ۳۹ رسید؛ هر <code dir="ltr">fib2(k)</code> فقط یک بار واقعاً محاسبه می‌شود و بقیه‌ی دفعات جواب از <code dir="ltr">cache</code> خوانده می‌شود.

کتابخانه‌ی استاندارد همین کار را آماده دارد: <code dir="ltr">functools.lru&#95;cache</code>.

</div>

````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib3(n):
    if n < 2:
        return n
    return fib3(n - 1) + fib3(n - 2)

print(fib3(100))
````

<div dir="rtl">

🔍 بدون حافظه‌ی نهان، محاسبه‌ی <code dir="ltr">fib(100)</code> با این روش بازگشتی عملاً هرگز تمام نمی‌شود؛ با <code dir="ltr">lru&#95;cache</code> فوری است. آرگومان <code dir="ltr">maxsize</code> حداکثر تعداد نتیجه‌های ذخیره‌شده را تعیین می‌کند (<code dir="ltr">None</code> یعنی نامحدود).

### ⚠️ اشتباه رایج: گم شدن نام و مستندات تابع

بعد از decorate کردن، <code dir="ltr">slow&#95;sum.&#95;&#95;name&#95;&#95;</code> برابر <code dir="ltr">'wrapper'</code> است و docstring تابع اصلی هم دیده نمی‌شود. راه حل استاندارد این است که بالای تعریف <code dir="ltr">wrapper</code> بنویسید <code dir="ltr">@functools.wraps(func)</code>؛ این decorator نام و مستندات تابع اصلی را روی <code dir="ltr">wrapper</code> کپی می‌کند.

</div>

### سؤال ۳ — mini-batch، epoch و یک باگ رایج

تابع زیر داده‌ی آموزش را در هر epoch به‌هم‌ریخته و به دسته‌های <code dir="ltr">batch&#95;size</code> تایی تقسیم می‌کند:

</div>

```python
import random

def minibatches(X, y, batch_size, shuffle=True, seed=0):
    idx = list(range(len(X)))
    if shuffle:
        random.seed(seed)
        random.shuffle(idx)
    for start in range(0, len(idx), batch_size):
        chunk = idx[start:start + batch_size]
        yield [X[i] for i in chunk], [y[i] for i in chunk]
```

### سؤال ۴ — آمار جریانی و توقف زودهنگام

در داده‌های جریانی (streaming)، نمی‌توانیم همه‌ی داده را نگه داریم؛ فقط یک «خلاصه» را به‌روز می‌کنیم:

</div>

```python
def running_mean(stream):
    total = 0
    count = 0
    for x in stream:
        total += x
        count += 1
        yield total / count

def ema(stream, beta=0.9):
    avg = None
    for x in stream:
        avg = x if avg is None else beta * avg + (1 - beta) * x
        yield avg
```

<div dir="rtl">

الف) <code dir="ltr">list(running&#95;mean([4, 8, 6, 2]))</code> و <code dir="ltr">list(ema([10, 0, 0], beta=0.5))</code> را حساب کنید. حافظه‌ی مصرفی هر کدام نسبت به طول جریان چگونه رشد می‌کند؟

ب) **میانگین متحرک نمایی (EMA)** در کجای آموزش شبکه‌های عصبی استفاده می‌شود؟ <code dir="ltr">beta</code> نزدیک به ۱ چه اثری دارد؟

ج) generator زیر زیان آموزش را در هر epoch شبیه‌سازی می‌کند. حلقه‌ی پایین در چه epochی متوقف می‌شود و چه چیزی چاپ می‌شود؟ چرا استفاده از یک generator **بی‌پایان** اینجا مشکلی ندارد؟

</div>

```python
def train_losses():
    loss = 1.0
    epoch = 0
    while True:
        epoch += 1
        loss *= 0.8
        yield epoch, loss

for epoch, loss in train_losses():
    if loss < 0.3:
        print(epoch, round(loss, 4))
        break
```

def clip(lo, hi):

            value = func(*args, **kwargs)

## خلاصه‌ی جلسه

- iterable چیزی است که رویش حلقه می‌زنیم؛ iterator با <code dir="ltr">next</code> مقدارها را یکی‌یکی می‌دهد و با <code dir="ltr">StopIteration</code> تمام می‌شود. iteratorها یک‌بار مصرف‌اند.
- تابع generator با <code dir="ltr">yield</code> مقدارها را تنبل و فقط هنگام نیاز تولید می‌کند و بین <code dir="ltr">next</code>ها مکث می‌کند؛ ابزاری طبیعی برای mini-batch و داده‌های جریانی.
- generator expression همان comprehension با پرانتز است: حافظه‌ی ثابت، یک بار مصرف، بدون طول و اندیس.
- تابع شیء است؛ closure متغیرهای تابع بیرونی را نگه می‌دارد؛ decorator با <code dir="ltr">@</code> رفتاری مثل زمان‌سنجی، شمارش یا حافظه‌ی نهان را بدون تغییر کد تابع اضافه می‌کند.

**جلسه‌ی بعد:** مدیریت خطا؛ اینکه به جای توقف برنامه با خطاهایی مثل <code dir="ltr">StopIteration</code>، <code dir="ltr">ValueError</code> و <code dir="ltr">KeyError</code>، آن‌ها را با <code dir="ltr">try</code> و <code dir="ltr">except</code> بگیریم و مدیریت کنیم.
