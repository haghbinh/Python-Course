## ۱. list comprehension

### 💡 مفهوم

در جلسه‌های ۸ تا ۱۲ بارها این الگو را نوشتید: یک لیست خالی، یک حلقه، و <code dir="ltr">append</code>. پایتون برای این الگو یک نحو کوتاه دارد به نام **list comprehension**:

<div dir="ltr">

```python
result = [expression for item in iterable]
```

</div>

که دقیقاً معادل این است:

<div dir="ltr">

```python
result = []
for item in iterable:
    result.append(expression)
```

</div>

</div>

````python
squares = []
for x in range(6):
    squares.append(x ** 2)
print(squares)

squares = [x ** 2 for x in range(6)]
print(squares)

words = ["Data", "Science", "AI"]
print([w.lower() for w in words])
print([len(w) for w in words])
````

<div dir="rtl">

### 🔍 کد را بخوان

- comprehension را از **وسط** بخوانید: «برای هر <code dir="ltr">x</code> در <code dir="ltr">range(6)</code>، مقدار <code dir="ltr">x &#42;&#42; 2</code> را در لیست بگذار».
- نتیجه همیشه یک لیست **جدید** است و لیست اصلی (<code dir="ltr">words</code>) تغییر نمی‌کند.

### 💡 فیلتر کردن با <code dir="ltr">if</code>

یک <code dir="ltr">if</code> در **انتهای** comprehension فقط عنصرهایی را نگه می‌دارد که شرط برایشان درست است:

</div>

````python
nums = [4, -2, 7, 0, -5, 3]
print([x for x in nums if x > 0])
print([x * 10 for x in nums if x % 2 == 0])
````

<div dir="rtl">

### 💡 انتخاب بین دو مقدار با <code dir="ltr">if ... else</code>

اگر بخواهید **همه‌ی** عنصرها بمانند، اما مقدارشان بسته به شرط فرق کند، از عبارت شرطی یک‌خطی (جلسه‌ی ۵) در **ابتدای** comprehension استفاده کنید. این همان تابع فعال‌سازی **ReLU** در شبکه‌های عصبی است:

</div>

````python
z = [-2.0, 3.5, -0.5, 1.0]
relu = [v if v > 0 else 0.0 for v in z]
print(relu)

labels = ["spam" if p >= 0.5 else "ham" for p in [0.9, 0.2, 0.5]]
print(labels)
````

<div dir="rtl">

### ⚠️ اشتباه رایج: جای <code dir="ltr">if</code>

| شکل | معنی | طول نتیجه |
|---|---|---|
| <code dir="ltr">[x for x in nums if x &gt; 0]</code> | فیلتر | ممکن است کوتاه‌تر شود |
| <code dir="ltr">[x if x &gt; 0 else 0 for x in nums]</code> | انتخاب مقدار | همیشه برابر طول <code dir="ltr">nums</code> |

<code dir="ltr">[x if x &gt; 0 for x in nums]</code> (بدون <code dir="ltr">else</code>) خطای نحوی است: <code dir="ltr">if</code> در ابتدا حتماً <code dir="ltr">else</code> می‌خواهد و <code dir="ltr">if</code> در انتها هرگز <code dir="ltr">else</code> نمی‌گیرد.

### ❓ پیش‌بینی کن

</div>

````python
print([c for c in "Python" if c.isupper()])
print([n % 3 for n in range(7)])
print([w[0] for w in ["map", "filter", "zip"]])
print(["even" if n % 2 == 0 else "odd" for n in [1, 2, 3]])
````

<div dir="rtl">

🔍 comprehension روی **هر** iterable کار می‌کند: رشته (کاراکتر به کاراکتر)، <code dir="ltr">range</code>، لیست، تاپل، مجموعه و دیکشنری.

### 💡 جمع زدن بدون ساختن لیست

در جلسه‌ی ۱۱ عبارت <code dir="ltr">sum(v &#42; v for v in a.values())</code> را دیدید. اگر comprehension را **بدون** کروشه و مستقیم داخل یک تابع مثل <code dir="ltr">sum</code>، <code dir="ltr">max</code> یا <code dir="ltr">min</code> بنویسید، پایتون مقادیر را یکی‌یکی به تابع می‌دهد و لیست واسطه‌ای نمی‌سازد:

</div>

````python
v = [3, 4]
norm = sum(x * x for x in v) ** 0.5
print(norm)

a = [1, 2, 3]
b = [4, 5, 6]
print(sum(x * y for x, y in zip(a, b)))
print(sum(1 for c in "banana" if c == "a"))
````

<div dir="rtl">

🔍 خط دوم **ضرب داخلی** دو بردار است (جلسه‌ی ۸) در یک خط، و خط آخر تعداد <code dir="ltr">a</code>ها را می‌شمارد. نام این ساختار **generator expression** است و در جلسه‌ی ۱۴ دقیق می‌بینید چرا لیست نمی‌سازد و چه مزیتی دارد.

### ⚠️ اشتباه رایج: comprehension برای اثر جانبی

<code dir="ltr">[print(x) for x in items]</code> کار می‌کند، اما یک لیست بی‌مصرف از <code dir="ltr">None</code>ها می‌سازد. comprehension برای **ساختن** یک مجموعه‌ی جدید است؛ برای «کاری انجام بده» (چاپ، نوشتن در فایل، <code dir="ltr">append</code> به لیست دیگر) از حلقه‌ی معمولی استفاده کنید.

</div>

## ۲. comprehensionهای تودرتو، dict و set comprehension

### 💡 دو <code dir="ltr">for</code> در یک comprehension

می‌توان چند <code dir="ltr">for</code> پشت سر هم نوشت. ترتیب آن‌ها **همان ترتیب حلقه‌های تودرتو** است: <code dir="ltr">for</code> اول، حلقه‌ی بیرونی است.

</div>

````python
matrix = [[1, 2, 3],
          [4, 5, 6]]

flat = [x for row in matrix for x in row]
print(flat)

flat = []
for row in matrix:
    for x in row:
        flat.append(x)
print(flat)

pairs = [(i, j) for i in range(3) for j in range(i)]
print(pairs)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">[x for row in matrix for x in row]</code> ماتریس را **صاف (flatten)** می‌کند. اگر آن را با دو حلقه‌ی تودرتو بنویسید، ترتیب <code dir="ltr">for</code>ها دقیقاً همان می‌ماند.
- در <code dir="ltr">pairs</code>، حلقه‌ی درونی از متغیر حلقه‌ی بیرونی (<code dir="ltr">i</code>) استفاده می‌کند؛ برای <code dir="ltr">i = 0</code> هیچ <code dir="ltr">j</code> وجود ندارد.

### 💡 comprehension داخل comprehension

اگر خروجی باید **لیستی از لیست‌ها** باشد (یک ماتریس)، یک comprehension کامل را داخل دیگری می‌گذاریم. این بار comprehension **بیرونی** سطرها را می‌سازد:

</div>

````python
table = [[i * j for j in range(1, 4)] for i in range(1, 3)]
print(table)

transpose = [[row[k] for row in matrix] for k in range(3)]
print(transpose)
````

<div dir="rtl">

### ⚠️ دو شکل را با هم اشتباه نگیرید

| کد | نتیجه |
|---|---|
| <code dir="ltr">[x for row in m for x in row]</code> | یک لیست صاف |
| <code dir="ltr">[[x for x in row] for row in m]</code> | لیستی از لیست‌ها (یک کپی از ماتریس) |

قاعده‌ی خوانایی: اگر comprehension بیش از دو <code dir="ltr">for</code> یا شرط پیچیده دارد، حلقه‌ی معمولی بنویسید.

### 💡 dict comprehension

با <code dir="ltr">{key: value for ...}</code> یک **دیکشنری** ساخته می‌شود. الگوهای جلسه‌ی ۱۰ (ساختن واژگان، وارونه کردن، فیلتر) هر کدام یک خط می‌شوند:

</div>

````python
vocab = ["cat", "dog", "bird"]
word2idx = {w: i for i, w in enumerate(vocab)}
idx2word = {i: w for w, i in word2idx.items()}
print(word2idx)
print(idx2word)

scores = {"Ali": 17.5, "Sara": 19, "Reza": 9.5, "Mina": 12}
passed = {name: s for name, s in scores.items() if s >= 10}
print(passed)
````

<div dir="rtl">

### 💡 set comprehension

با <code dir="ltr">{expression for ...}</code> (بدون <code dir="ltr">:</code>) یک **مجموعه** ساخته می‌شود و مقادیر تکراری خودبه‌خود حذف می‌شوند:

</div>

````python
words = ["data", "ai", "model", "code", "ai"]
print({len(w) for w in words})
print({w[0] for w in words})
````

<div dir="rtl">

### ⚠️ اشتباه رایج: <code dir="ltr">{}</code> خالی

<code dir="ltr">{}</code> یک **دیکشنری** خالی است، نه مجموعه (جلسه‌ی ۹). comprehension با <code dir="ltr">:</code> دیکشنری و بدون <code dir="ltr">:</code> مجموعه می‌سازد.

### ❓ پیش‌بینی کن

خروجی را قبل از اجرا بنویسید. به‌ویژه دقت کنید در dict comprehension اگر یک کلید **دو بار** تولید شود چه می‌شود:

</div>

````python
print({w: len(w) for w in ["data", "ai", "model"] if len(w) > 2})
print({len(w): w for w in ["ab", "cd", "efg"]})
print([[0] * 2 for _ in range(3)])
````

<div dir="rtl">

🔍 در دومی، کلید <code dir="ltr">2</code> دو بار تولید می‌شود و مثل انتساب معمولی، مقدار **آخر** (<code dir="ltr">"cd"</code>) می‌ماند. در سومی، <code dir="ltr">&#95;</code> نامی رایج برای متغیر حلقه‌ای است که از آن استفاده نمی‌کنیم. این روش درست ساختن یک ماتریس صفر است: برخلاف <code dir="ltr">[[0] &#42; 2] &#42; 3</code> (جلسه‌ی ۸)، هر سطر یک لیست **جدا** است.

</div>

## ۳. توابع <code dir="ltr">lambda</code>

### 💡 مفهوم

<code dir="ltr">lambda</code> راهی برای نوشتن یک تابع **کوچک و بی‌نام** در یک عبارت است:

<div dir="ltr">

```python
lambda parameters: expression
```

</div>

بدنه‌ی <code dir="ltr">lambda</code> فقط **یک عبارت** است (نه دستور) و مقدار آن خودبه‌خود برگردانده می‌شود؛ <code dir="ltr">return</code> نمی‌نویسیم.

</div>

````python
def square(x):
    return x ** 2

square2 = lambda x: x ** 2
print(square(5), square2(5))

add = lambda a, b: a + b
print(add(2, 3))

print((lambda x: x * 10)(4))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">square</code> و <code dir="ltr">square2</code> دقیقاً یک کار می‌کنند. تابع در پایتون یک **شیء** است و می‌توان آن را در یک متغیر گذاشت (در جلسه‌ی ۱۴ از همین ویژگی برای ساختن decorator استفاده می‌کنیم).
- خط آخر یک <code dir="ltr">lambda</code> را بلافاصله فراخوانی می‌کند؛ فقط برای نشان دادن اینکه <code dir="ltr">lambda</code> یک مقدار است.

### 💡 کاربرد اصلی: تابع به‌عنوان آرگومان

<code dir="ltr">lambda</code> وقتی می‌درخشد که یک تابع **یک‌بار مصرف** را به تابع دیگری بدهید؛ مثلاً <code dir="ltr">key</code> در <code dir="ltr">sorted</code> (بخش ۶) یا تابع فعال‌سازی یک لایه:

</div>

````python
def apply_all(f, values):
    return [f(v) for v in values]

print(apply_all(lambda v: max(0, v), [-1, 2, -3]))
print(apply_all(lambda v: 1 if v > 0 else 0, [-1, 2, -3]))
````

<div dir="rtl">

🔍 <code dir="ltr">apply&#95;all</code> نمی‌داند <code dir="ltr">f</code> چه می‌کند؛ هر تابعی را روی همه‌ی مقادیر اعمال می‌کند. به تابعی که تابع دیگری را به‌عنوان ورودی بگیرد (یا برگرداند) **تابع مرتبه‌ی بالاتر (higher-order function)** می‌گویند؛ ایده‌ی اصلی برنامه‌نویسی تابعی.

### ⚠️ اشتباه رایج: <code dir="ltr">lambda</code> طولانی

اگر <code dir="ltr">lambda</code> منطق پیچیده دارد یا جای دیگری دوباره استفاده می‌شود، یک تابع معمولی با <code dir="ltr">def</code> و یک نام گویا بنویسید. <code dir="ltr">f = lambda x: ...</code> از نظر سبک توصیه نمی‌شود: تابع نام درستی در پیام خطا ندارد و docstring نمی‌گیرد.

</div>

## ۴. <code dir="ltr">map</code> و <code dir="ltr">filter</code>

### 💡 مفهوم

دو تابع مرتبه‌ی بالاتر کلاسیک:

- <code dir="ltr">map(f, iterable)</code>: تابع <code dir="ltr">f</code> را روی **هر** عنصر اعمال می‌کند،
- <code dir="ltr">filter(f, iterable)</code>: فقط عنصرهایی را نگه می‌دارد که <code dir="ltr">f</code> برایشان درست است.

هر دو یک شیء «تنبل» برمی‌گردانند که مقادیر را تا وقتی خواسته نشوند حساب نمی‌کند؛ برای دیدن نتیجه آن را با <code dir="ltr">list</code> به لیست تبدیل می‌کنیم.

</div>

````python
line = "3 1 4 1 5"
nums = list(map(int, line.split()))
print(nums)

print(list(filter(lambda w: len(w) > 2, ["a", "bee", "sea", "to"])))
print(list(map(lambda a, b: a * b, [1, 2, 3], [4, 5, 6])))

print(map(int, ["1", "2"]))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">list(map(int, line.split()))</code> الگوی بسیار رایجی برای تبدیل یک خط عدد به لیست اعداد است. دقت کنید <code dir="ltr">int</code> را **بدون** پرانتز داده‌ایم: خود تابع را می‌دهیم، نه نتیجه‌ی فراخوانی آن را.
- <code dir="ltr">map</code> با دو iterable، تابع دوآرگومانی را روی جفت‌ها اعمال می‌کند (مثل <code dir="ltr">zip</code>).
- خط آخر نشان می‌دهد خروجی <code dir="ltr">map</code> خودش لیست نیست.

### 💡 map و filter یا comprehension؟

| کار | با <code dir="ltr">map</code> / <code dir="ltr">filter</code> | با comprehension |
|---|---|---|
| تبدیل | <code dir="ltr">list(map(str.upper, ws))</code> | <code dir="ltr">[w.upper() for w in ws]</code> |
| فیلتر | <code dir="ltr">list(filter(lambda x: x &gt; 0, xs))</code> | <code dir="ltr">[x for x in xs if x &gt; 0]</code> |
| هر دو | <code dir="ltr">list(map(lambda x: x &#42; 2, filter(lambda x: x &gt; 0, xs)))</code> | <code dir="ltr">[x &#42; 2 for x in xs if x &gt; 0]</code> |

در پایتون معمولاً comprehension خواناتر است، به‌ویژه وقتی به <code dir="ltr">lambda</code> نیاز باشد. <code dir="ltr">map</code> وقتی تمیز است که یک تابع **آماده** (مثل <code dir="ltr">int</code>، <code dir="ltr">str.strip</code>، <code dir="ltr">float</code>) داشته باشید.

### ⚠️ اشتباه رایج: دوبار استفاده از خروجی <code dir="ltr">map</code>

</div>

````python
m = map(lambda x: x * 2, [1, 2, 3])
print(list(m))
print(list(m))
````

<div dir="rtl">

🔍 بار دوم لیست خالی است! شیء <code dir="ltr">map</code> (و <code dir="ltr">filter</code> و <code dir="ltr">zip</code>) فقط **یک بار** قابل پیمایش است. دلیلش را در جلسه‌ی ۱۴ (iterator) می‌بینید. اگر نتیجه را چند بار لازم دارید، یک بار با <code dir="ltr">list</code> ذخیره‌اش کنید.

</div>

## ۵. <code dir="ltr">zip</code> و <code dir="ltr">enumerate</code>

### 💡 یادآوری و نکته‌های تازه

در جلسه‌ی ۸ با این دو آشنا شدید. چند استفاده‌ی مهم‌تر:

- <code dir="ltr">enumerate(items, start=1)</code>: شماره‌گذاری از عدد دلخواه،
- <code dir="ltr">dict(zip(keys, values))</code>: ساختن دیکشنری از دو لیست هم‌تراز،
- <code dir="ltr">zip(&#42;pairs)</code>: «باز کردن» لیستی از جفت‌ها به دو دنباله‌ی جدا،
- <code dir="ltr">zip</code> در **کوتاه‌ترین** ورودی متوقف می‌شود.

</div>

````python
for rank, name in enumerate(["Sara", "Ali", "Mina"], start=1):
    print(rank, name)

features = ["age", "height", "weight"]
row = [23, 180, 75]
record = dict(zip(features, row))
print(record)

data = [(1.0, "A"), (2.5, "B"), (0.7, "A")]
xs, ys = zip(*data)
print(xs, ys)

print(list(zip([1, 2, 3], ["a", "b"])))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">dict(zip(features, row))</code> یک سطر جدول را به یک **رکورد** با نام ستون‌ها تبدیل می‌کند؛ همان کاری که <code dir="ltr">csv.DictReader</code> در جلسه‌ی ۱۶ برای هر خط فایل انجام می‌دهد.
- در <code dir="ltr">zip(&#42;data)</code>، ستاره عناصر <code dir="ltr">data</code> را به‌عنوان آرگومان‌های جدا به <code dir="ltr">zip</code> می‌دهد (مثل <code dir="ltr">&#42;args</code> در جلسه‌ی ۷)؛ پس <code dir="ltr">zip((1.0, "A"), (2.5, "B"), (0.7, "A"))</code> اجرا می‌شود که عنصرهای اول را با هم و دوم را با هم جفت می‌کند. این روش استاندارد جدا کردن **ویژگی‌ها (X)** و **برچسب‌ها (y)** است. نتیجه تاپل است.
- خط آخر نشان می‌دهد <code dir="ltr">zip</code> عنصر اضافه را **بی‌صدا** دور می‌اندازد.

### ⚠️ اشتباه رایج: طول‌های نابرابر

اگر لیست ویژگی‌ها و برچسب‌ها به اشتباه هم‌طول نباشند، <code dir="ltr">zip</code> خطا نمی‌دهد و بخشی از داده بی‌صدا حذف می‌شود. در پایتون ۳.۱۰ به بعد <code dir="ltr">zip(a, b, strict=True)</code> در این حالت خطا می‌دهد.

## ۶. <code dir="ltr">sorted</code>، <code dir="ltr">min</code> و <code dir="ltr">max</code> با <code dir="ltr">key</code>

### 💡 مفهوم

<code dir="ltr">sorted</code>، <code dir="ltr">min</code> و <code dir="ltr">max</code> (و متد <code dir="ltr">list.sort</code>) آرگومانی به نام <code dir="ltr">key</code> می‌گیرند: تابعی که روی **هر عنصر** اعمال می‌شود و مقایسه بر اساس **خروجی آن** انجام می‌شود، نه خود عنصر. خود عنصرها تغییر نمی‌کنند. <code dir="ltr">reverse=True</code> ترتیب را نزولی می‌کند.

</div>

````python
names = ["Bob", "alice", "Carl"]
print(sorted(names))
print(sorted(names, key=str.lower))
print(sorted(names, key=len, reverse=True))

print(max(["pear", "fig", "banana"], key=len))
print(min([3, -7, 5], key=abs))
````

<div dir="rtl">

### 🔍 کد را بخوان

- بدون <code dir="ltr">key</code>، رشته‌ها با کد یونیکد مقایسه می‌شوند و حروف بزرگ قبل از کوچک می‌آیند (جلسه‌ی ۱۱). با <code dir="ltr">key=str.lower</code> مقایسه روی نسخه‌ی کوچک‌شده انجام می‌شود، اما خروجی همان رشته‌های اصلی است.
- <code dir="ltr">min([3, -7, 5], key=abs)</code> عددی را برمی‌گرداند که **قدرمطلقش** کمترین است؛ یعنی خود <code dir="ltr">3</code>، نه <code dir="ltr">abs(3)</code>.

### 💡 مرتب‌سازی رکوردها با <code dir="ltr">lambda</code>

در داده‌های واقعی، عنصرها تاپل یا دیکشنری‌اند و <code dir="ltr">key</code> معمولاً یک <code dir="ltr">lambda</code> است که فیلد مورد نظر را بیرون می‌کشد:

</div>

````python
students = [("Ali", 17), ("Sara", 19), ("Reza", 17), ("Mina", 15)]

print(sorted(students, key=lambda s: s[1], reverse=True))
print(sorted(students, key=lambda s: (-s[1], s[0])))

scores = {"Ali": 17, "Sara": 19, "Mina": 15}
print(max(scores, key=scores.get))
print(sorted(scores.items(), key=lambda kv: kv[1]))
````

<div dir="rtl">

### 🔍 کد را بخوان

- در خط اول، <code dir="ltr">Ali</code> و <code dir="ltr">Reza</code> نمره‌ی برابر دارند و به همان ترتیبی که در لیست بودند می‌مانند. مرتب‌سازی پایتون **پایدار (stable)** است: عنصرهای با کلید برابر، ترتیب نسبی خود را حفظ می‌کنند.
- در خط دوم، کلید یک **تاپل** است و تاپل‌ها عضو به عضو مقایسه می‌شوند (جلسه‌ی ۹): اول نمره به شکل نزولی (با علامت منفی) و در صورت تساوی، نام به شکل الفبایی. این روش استاندارد مرتب‌سازی **چندمعیاره** است.
- <code dir="ltr">max(scores, key=scores.get)</code> کلیدی را برمی‌گرداند که **مقدارش** بیشترین است: «بهترین دانشجو» در یک خط.

### 💡 argmax و argsort

در یادگیری ماشین مدام لازم است **اندیس** بزرگ‌ترین مقدار را پیدا کنیم (کلاس پیش‌بینی‌شده از روی احتمال‌ها) یا اندیس‌ها را بر اساس مقدارشان مرتب کنیم:

</div>

````python
probs = [0.1, 0.7, 0.2]
argmax = max(range(len(probs)), key=lambda i: probs[i])
print(argmax)

order = sorted(range(len(probs)), key=lambda i: probs[i], reverse=True)
print(order)
````

<div dir="rtl">

🔍 به جای مقادیر، روی **اندیس‌ها** (<code dir="ltr">range(len(probs))</code>) جست‌وجو می‌کنیم و <code dir="ltr">key</code> هر اندیس را به مقدارش نگاشت می‌کند. <code dir="ltr">order</code> همان <code dir="ltr">argsort</code> نزولی است: اول کلاس <code dir="ltr">1</code>، بعد <code dir="ltr">2</code>، بعد <code dir="ltr">0</code>. در جلسه‌ی ۲۱ همین را با <code dir="ltr">np.argmax</code> و <code dir="ltr">np.argsort</code> می‌بینید.

### ⚠️ اشتباه رایج: <code dir="ltr">key=f()</code> به جای <code dir="ltr">key=f</code>

<code dir="ltr">key</code> باید خود تابع باشد: <code dir="ltr">key=len</code> درست است، <code dir="ltr">key=len()</code> خطاست. همچنین <code dir="ltr">list.sort()</code> لیست را درجا مرتب می‌کند و <code dir="ltr">None</code> برمی‌گرداند (جلسه‌ی ۸)؛ <code dir="ltr">x = x.sort(key=...)</code> لیست را از دست می‌دهد.

### ❓ پیش‌بینی کن

</div>

````python
words = ["kiwi", "Apple", "fig", "banana"]
print(sorted(words, key=lambda w: w[-1]))
print(max(words, key=lambda w: w.lower()))
print(sorted([-3, 2, -1, 4], key=lambda x: x * x))
````

<div dir="rtl">

🔍 در اولی مرتب‌سازی بر اساس **آخرین حرف** است (<code dir="ltr">a</code>، <code dir="ltr">e</code>، <code dir="ltr">g</code>، <code dir="ltr">i</code>). در دومی، بیشترین بر اساس نسخه‌ی کوچک‌شده‌ی کلمه است، اما خود کلمه برگردانده می‌شود. در سومی، کلید مربع عدد است؛ <code dir="ltr">-1</code> (مربع ۱) اول می‌آید.

```python
def softmax(scores):
    m = max(scores)
    exps = [math.exp(s - m) for s in scores]
    total = sum(exps)
    return [e / total for e in exps]
```

```python
funcs = [lambda x: x * i for i in range(3)]
print([f(10) for f in funcs])

funcs2 = [lambda x, i=i: x * i for i in range(3)]
print([f(10) for f in funcs2])
```

ب) چرا خروجی خط اول <code dir="ltr">[0, 10, 20]</code> نیست؟

ج) دانشجویی برای ساختن سه تابع فعال‌سازی مختلف یک لایه این کد را نوشته است. چه خطایی در نتیجه رخ می‌دهد و چطور اصلاحش می‌کنید؟

```python
slopes = [0.0, 0.01, 0.2]          # ReLU, leaky ReLU, leaky ReLU
acts = [lambda v: v if v > 0 else a * v for a in slopes]
print([act(-10) for act in acts])
```

<summary><b>پاسخ</b></summary>

الف) <code dir="ltr">[20, 20, 20]</code> و <code dir="ltr">[0, 10, 20]</code>.

### سؤال ۶ — کد را بخوان: k نزدیک‌ترین همسایه

کد زیر دسته‌ی نقطه‌ی <code dir="ltr">q</code> را با الگوریتم **kNN** (جلسه‌ی ۸) پیش‌بینی می‌کند:

</div>

```python
import math

train = [((1, 1), "A"), ((2, 1), "A"), ((4, 4), "B"), ((5, 4), "B"), ((3, 2), "A")]
q = (3, 3)

dists = sorted((math.dist(p, q), label) for p, label in train)
k_labels = [label for _, label in dists[:3]]
prediction = max(set(k_labels), key=k_labels.count)
```

<div dir="rtl">

الف) <code dir="ltr">dists</code> (با فاصله‌های گرد شده تا سه رقم اعشار)، <code dir="ltr">k&#95;labels</code> و <code dir="ltr">prediction</code> را بنویسید. (<code dir="ltr">math.dist</code> فاصله‌ی اقلیدسی دو نقطه است.)

ج) <code dir="ltr">max(set(k&#95;labels), key=k&#95;labels.count)</code> دقیقاً چه می‌کند؟ اگر دو برچسب تعداد برابر داشته باشند، کدام برگردانده می‌شود؟ چرا این رفتار قابل اعتماد نیست؟

## خلاصه‌ی جلسه

- comprehension الگوی «لیست خالی + حلقه + <code dir="ltr">append</code>» را در یک عبارت خوانا خلاصه می‌کند؛ <code dir="ltr">if</code> انتها فیلتر است و <code dir="ltr">if ... else</code> ابتدا انتخاب مقدار.
- dict و set comprehension برای واژگان، وارونه کردن و فیلتر؛ comprehension تودرتو برای ماتریس‌ها.
- generator expression داخل <code dir="ltr">sum</code> و <code dir="ltr">max</code> بدون ساختن لیست محاسبه می‌کند.
- <code dir="ltr">lambda</code> تابعی کوچک و بی‌نام است، بیشتر برای آرگومان‌هایی مثل <code dir="ltr">key</code>؛ مراقب late binding در حلقه باشید.
- <code dir="ltr">map</code> و <code dir="ltr">filter</code> تنبل و یک‌بار مصرف‌اند؛ در بیشتر موارد comprehension خواناتر است.
- <code dir="ltr">zip(&#42;...)</code> برای باز کردن و ترانهاده؛ <code dir="ltr">key</code> در <code dir="ltr">sorted</code> / <code dir="ltr">min</code> / <code dir="ltr">max</code> برای مرتب‌سازی چندمعیاره، argmax و top-k.

**جلسه‌ی بعد:** iterator، generator و decorator؛ می‌بینیم چرا <code dir="ltr">map</code> یک بار مصرف است و generator expression چطور بدون ساختن لیست کار می‌کند.
