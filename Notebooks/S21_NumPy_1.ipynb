## ۱. ساخت آرایه

### 💡 چرا NumPy؟

تا اینجا داده‌ها را در **لیست** نگه داشتیم. لیست انعطاف زیادی دارد (هر عضو می‌تواند هر نوعی باشد)، اما برای محاسبات عددی دو مشکل دارد:

- **کند است**: هر عضو یک شیء جداگانه‌ی پایتون است و هر عمل ریاضی با یک حلقه‌ی پایتونی انجام می‌شود.
- **عملگرهایش ریاضی نیستند**: <code dir="ltr">[1, 2] &#42; 2</code> لیست را تکرار می‌کند و <code dir="ltr">[1, 2] + [3, 4]</code> دو لیست را به هم می‌چسباند؛ جمع برداری نیست.

کتابخانه‌ی **NumPy** (Numerical Python) نوع داده‌ی **<code dir="ltr">ndarray</code>** (آرایه‌ی n-بعدی) را اضافه می‌کند: مجموعه‌ای از عددها که همه **یک نوع** دارند و پشت سر هم در حافظه ذخیره شده‌اند. عملیات روی آن به زبان C و بدون حلقه‌ی پایتونی اجرا می‌شود. تقریباً همه‌ی کتابخانه‌های علوم داده و یادگیری ماشین (Pandas، Matplotlib، scikit-learn، PyTorch) روی همین ایده ساخته شده‌اند.

### نصب کتابخانه‌های بخش ۳

NumPy و کتابخانه‌های بعدی جزو پایتون نیستند و باید یک بار نصب شوند. همان‌طور که در جلسه‌ی ۱ Jupyter را نصب کردید، در Command Prompt دستور زیر را اجرا کنید (همه‌ی کتابخانه‌های بخش ۳ با هم نصب می‌شوند):

بعد از نصب، NumPy را همیشه با نام کوتاه **<code dir="ltr">np</code>** وارد می‌کنیم (قرارداد همه‌ی کدهای علوم داده):

</div>

````python
import numpy as np

nums = [1, 2, 3]
arr = np.array([1, 2, 3])

print(nums * 2)
print(arr * 2)
print(nums + nums)
print(arr + arr)
print(arr ** 2, arr / 2)
print(type(arr))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">np.array(list)</code> از یک لیست، آرایه می‌سازد.
- روی آرایه، <code dir="ltr">&#42;</code>، <code dir="ltr">+</code>، <code dir="ltr">&#42;&#42;</code> و <code dir="ltr">/</code> **عضو به عضو** انجام می‌شوند: <code dir="ltr">arr &#42; 2</code> هر عضو را دو برابر می‌کند (<code dir="ltr">[2 4 6]</code>)، در حالی که <code dir="ltr">nums &#42; 2</code> لیست را دو بار تکرار می‌کند.
- آرایه‌ها بدون کاما چاپ می‌شوند: <code dir="ltr">[2 4 6]</code>. این راه ساده‌ای است برای اینکه در خروجی، آرایه را از لیست تشخیص دهید.
- در جلسه‌ی ۱۹ یک کلاس <code dir="ltr">Vector</code> با <code dir="ltr">&#95;&#95;add&#95;&#95;</code> و <code dir="ltr">&#95;&#95;mul&#95;&#95;</code> نوشتید تا همین رفتار را بسازید؛ آرایه‌ی NumPy همان ایده است، با سرعت و امکانات بسیار بیشتر.

### 💡 مقایسه‌ی سرعت

برای دیدن تفاوت سرعت، مجموع مربع‌های یک میلیون عدد را یک بار با لیست و یک بار با NumPy حساب می‌کنیم (<code dir="ltr">time.perf&#95;counter</code> از جلسه‌ی ۱۴):

</div>

````python
import time

n = 1_000_000
py_list = list(range(n))
np_arr = np.arange(n)

t0 = time.perf_counter()
s1 = sum(x * x for x in py_list)
t1 = time.perf_counter()
s2 = int((np_arr * np_arr).sum())
t2 = time.perf_counter()

print(s1 == s2)
print(f"list: {t1 - t0:.4f} s   numpy: {t2 - t1:.4f} s")
````

<div dir="rtl">

🔍 هر دو روش جواب یکسان می‌دهند (<code dir="ltr">True</code>)، اما نسخه‌ی NumPy روی بیشتر کامپیوترها **ده‌ها برابر** سریع‌تر است (عدد دقیق به کامپیوتر شما بستگی دارد). در جلسه‌ی ۲۲ این روش «بدون حلقه» را با نام **برداری‌سازی (vectorization)** کامل می‌بینیم.

### 💡 توابع ساخت آرایه

به جز تبدیل لیست، NumPy توابع آماده‌ای برای ساختن آرایه‌های رایج دارد:

| تابع | کار | مثال | نتیجه |
|---|---|---|---|
| <code dir="ltr">np.zeros(shape)</code> | همه صفر | <code dir="ltr">np.zeros(3)</code> | <code dir="ltr">[0. 0. 0.]</code> |
| <code dir="ltr">np.ones(shape)</code> | همه یک | <code dir="ltr">np.ones((2, 2))</code> | ماتریس ۲×۲ از یک |
| <code dir="ltr">np.full(shape, v)</code> | همه برابر <code dir="ltr">v</code> | <code dir="ltr">np.full(3, 7)</code> | <code dir="ltr">[7 7 7]</code> |
| <code dir="ltr">np.arange(start, stop, step)</code> | مثل <code dir="ltr">range</code>، اما آرایه (گام اعشاری هم مجاز) | <code dir="ltr">np.arange(0, 1, 0.25)</code> | <code dir="ltr">[0. 0.25 0.5 0.75]</code> |
| <code dir="ltr">np.linspace(start, stop, num)</code> | <code dir="ltr">num</code> عدد با فاصله‌ی **مساوی**، **شامل** <code dir="ltr">stop</code> | <code dir="ltr">np.linspace(0, 1, 5)</code> | <code dir="ltr">[0. 0.25 0.5 0.75 1.]</code> |
| <code dir="ltr">np.eye(n)</code> | ماتریس همانی <code dir="ltr">n×n</code> | <code dir="ltr">np.eye(3)</code> | قطر اصلی یک |
| <code dir="ltr">np.zeros&#95;like(a)</code> | هم‌شکل و هم‌نوع با <code dir="ltr">a</code>، پر از صفر | | |

برای آرایه‌های چندبعدی، شکل را به صورت یک **تاپل** می‌دهیم: <code dir="ltr">np.zeros((2, 3))</code> یعنی ۲ سطر و ۳ ستون.

</div>

````python
print(np.zeros((2, 3)))
print(np.full((2, 2), 0.5))
print(np.arange(10))
print(np.arange(2, 11, 3))
print(np.linspace(0, 1, 5))
print(np.eye(3))

M = np.array([[1, 2, 3],
              [4, 5, 6]])
print(M)
print(np.ones_like(M))
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">np.zeros((2, 3))</code> یک ماتریس ۲ سطری و ۳ ستونی می‌سازد. صفرها با نقطه چاپ می‌شوند (<code dir="ltr">0.</code>)، چون نوع پیش‌فرض این توابع **اعشاری** است.
- <code dir="ltr">np.arange(2, 11, 3)</code> مثل <code dir="ltr">range</code> کار می‌کند و <code dir="ltr">stop</code> را شامل نمی‌شود: <code dir="ltr">[2 5 8]</code>.
- یک **لیستِ لیست‌ها** به آرایه‌ی دوبعدی (ماتریس) تبدیل می‌شود؛ هر لیست درونی یک سطر است.
- <code dir="ltr">np.ones&#95;like(M)</code> آرایه‌ای هم‌شکل با <code dir="ltr">M</code> (و هم‌نوع با آن، یعنی صحیح) می‌سازد.

### ❓ پیش‌بینی کن

خروجی هر خط را قبل از اجرا بنویسید. در خط آخر تعداد اعضا را پیش‌بینی کنید.

</div>

````python
print(np.arange(5, 0, -2))
print(np.linspace(1, 2, 3))
print(np.arange(1, 2, 0.5))
print(np.zeros((3, 1)))
print(len(np.linspace(0, 10, 11)), len(np.arange(0, 10, 1)))
````

<div dir="rtl">

🔍 <code dir="ltr">np.arange(5, 0, -2)</code> برابر <code dir="ltr">[5 3 1]</code> است (گام منفی، مثل <code dir="ltr">range</code>). <code dir="ltr">np.linspace(1, 2, 3)</code> سه عدد با فاصله‌ی مساوی **شامل هر دو سر** می‌دهد: <code dir="ltr">[1. 1.5 2.]</code>. <code dir="ltr">np.arange(1, 2, 0.5)</code> سر دوم را شامل نمی‌شود: <code dir="ltr">[1. 1.5]</code>. <code dir="ltr">np.zeros((3, 1))</code> یک **ستون** سه‌تایی است. در خط آخر، <code dir="ltr">linspace(0, 10, 11)</code> عدد ۱۰ را هم دارد (۱۱ عضو) ولی <code dir="ltr">arange(0, 10, 1)</code> نه (۱۰ عضو).

### ⚠️ اشتباه رایج: <code dir="ltr">arange</code> با گام اعشاری

عددهای اعشاری در کامپیوتر دقیق نیستند (جلسه‌ی ۲)؛ پس آخرین عضو <code dir="ltr">np.arange</code> با گام اعشاری ممکن است کمی با انتظار شما فرق کند و حتی تعداد اعضا یکی کم یا زیاد شود. وقتی **تعداد** نقاط برایتان مهم است (مثلاً برای رسم نمودار)، از <code dir="ltr">linspace</code> استفاده کنید.

## ۲. <code dir="ltr">shape</code> و <code dir="ltr">dtype</code>

### 💡 مفهوم

هر آرایه چند ویژگی (attribute) دارد که باید همیشه بتوانید بخوانید:

| ویژگی | معنی | برای ماتریس ۲×۳ |
|---|---|---|
| <code dir="ltr">a.ndim</code> | تعداد ابعاد (محورها) | <code dir="ltr">2</code> |
| <code dir="ltr">a.shape</code> | اندازه در هر بعد، به شکل تاپل | <code dir="ltr">(2, 3)</code> |
| <code dir="ltr">a.size</code> | تعداد کل اعضا | <code dir="ltr">6</code> |
| <code dir="ltr">a.dtype</code> | نوع داده‌ی **همه‌ی** اعضا | مثلاً <code dir="ltr">int64</code> یا <code dir="ltr">float64</code> |
| <code dir="ltr">a.itemsize</code> / <code dir="ltr">a.nbytes</code> | بایت هر عضو / کل آرایه | <code dir="ltr">8</code> / <code dir="ltr">48</code> |

بعد اول (<code dir="ltr">axis 0</code>) **سطرها** و بعد دوم (<code dir="ltr">axis 1</code>) **ستون‌ها** هستند:

</div>

<p align="center" dir="rtl"><i>یک آرایه‌ی ۳×۴: محور ۰ در جهت سطرها و محور ۱ در جهت ستون‌ها. <code dir="ltr">A[1]</code> یک سطر و <code dir="ltr">A[:, 2]</code> یک ستون است.</i></p>

<div dir="rtl">

### 💡 ماتریس داده

در یادگیری ماشین، داده‌ی جدولی را به شکل یک ماتریس <code dir="ltr">X</code> با شکل <code dir="ltr">(n&#95;samples, n&#95;features)</code> نگه می‌داریم: هر **سطر** یک نمونه (مثلاً یک خانه) و هر **ستون** یک ویژگی (مثلاً متراژ، تعداد اتاق، سن ساختمان) است.

</div>

````python
X = np.array([[120, 3, 10],
              [ 85, 2, 25],
              [200, 4,  2],
              [ 60, 1, 40]])

print(X.ndim, X.shape, X.size)
print(X.dtype, X.itemsize, X.nbytes)

n_samples, n_features = X.shape
print(n_samples, "samples,", n_features, "features")

v = np.array([1.5, 2.0, 3.25])
print(v.shape, v.ndim, v.dtype)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">X</code> چهار سطر (نمونه) و سه ستون (ویژگی) دارد: <code dir="ltr">shape</code> برابر <code dir="ltr">(4, 3)</code> و <code dir="ltr">size</code> برابر <code dir="ltr">12</code>.
- چون همه‌ی عددها صحیح‌اند، <code dir="ltr">dtype</code> برابر <code dir="ltr">int64</code> (عدد صحیح ۶۴ بیتی، یعنی ۸ بایت) است و کل آرایه <code dir="ltr">96</code> بایت حافظه می‌گیرد. (روی بعضی سیستم‌ها ممکن است <code dir="ltr">int32</code> ببینید.)
- با **باز کردن تاپل** (جلسه‌ی ۹)، <code dir="ltr">X.shape</code> را مستقیم در دو متغیر می‌ریزیم؛ الگویی که در کدهای یادگیری ماشین بسیار می‌بینید.
- آرایه‌ی یک‌بعدی <code dir="ltr">v</code> شکل <code dir="ltr">(3,)</code> دارد: یک تاپل **یک‌عضوی** (به کامای آن دقت کنید).

### 💡 نوع داده (dtype) و ارتقای نوع

همه‌ی اعضای آرایه یک نوع دارند. اگر لیست ورودی نوع‌های مختلف داشته باشد، NumPy نوعی را انتخاب می‌کند که **همه** را در خود جا دهد (**ارتقای نوع**، upcasting). با <code dir="ltr">astype</code> می‌توان نوع را صریحاً عوض کرد؛ نتیجه یک آرایه‌ی **جدید** است.

</div>

````python
print(np.array([1, 2, 3]).dtype)
print(np.array([1, 2.5]).dtype)
print(np.array([True, 2]).dtype)
print(np.array([1, "a"]))

f = np.array([3.7, -3.7, 2.5])
print(f.astype(int))
print(np.array([0, 1, 2]).astype(bool))
print(np.zeros(3, dtype=int))
````

<div dir="rtl">

### 🔍 کد را بخوان

- یک عدد اعشاری کافی است تا کل آرایه <code dir="ltr">float64</code> شود. <code dir="ltr">True</code> به <code dir="ltr">1</code> تبدیل می‌شود و آرایه صحیح می‌ماند.
- لیستی از عدد و رشته به آرایه‌ای از **رشته‌ها** تبدیل می‌شود (<code dir="ltr">['1' 'a']</code>)؛ دیگر نمی‌توان روی آن حساب کرد. این یکی از دلایل رایج خطا هنگام خواندن داده‌های کثیف است (جلسه‌ی ۲۶).
- <code dir="ltr">astype(int)</code> بخش اعشاری را **دور می‌ریزد** (به سمت صفر)، گرد نمی‌کند: <code dir="ltr">3.7</code> به <code dir="ltr">3</code> و <code dir="ltr">-3.7</code> به <code dir="ltr">-3</code> تبدیل می‌شود.
- آرگومان <code dir="ltr">dtype</code> در توابع ساخت، نوع را از ابتدا تعیین می‌کند.

### ❓ پیش‌بینی کن

</div>

````python
a = np.array([1, 2, 3])
a[0] = 2.9
print(a)
print(a / 2)
print(a // 2)

small = np.array([127], dtype=np.int8)
print(small + 1)
````

<div dir="rtl">

🔍 چون <code dir="ltr">a</code> از نوع صحیح است، انتساب <code dir="ltr">2.9</code> به یکی از اعضایش **بی‌صدا** آن را به <code dir="ltr">2</code> تبدیل می‌کند (بدون خطا و بدون هشدار): <code dir="ltr">[2 2 3]</code>. تقسیم <code dir="ltr">/</code> همیشه آرایه‌ی اعشاری جدید می‌سازد (<code dir="ltr">[1. 1. 1.5]</code>) و <code dir="ltr">//</code> صحیح می‌ماند (<code dir="ltr">[1 1 1]</code>).

<code dir="ltr">int8</code> فقط عددهای ‎-128 تا 127 را نگه می‌دارد. <code dir="ltr">127 + 1</code> در این نوع **سرریز (overflow)** می‌کند و دور می‌زند: <code dir="ltr">[-128]</code>. برخلاف اعداد صحیح خود پایتون (جلسه‌ی ۲) که اندازه‌ی نامحدود دارند، اعداد NumPy اندازه‌ی ثابت دارند.

### ⚠️ اشتباه رایج: انتساب اعشاری به آرایه‌ی صحیح

اگر آرایه‌ای را با <code dir="ltr">np.array([1, 2, 3])</code> بسازید و بعد بخواهید مقدارهای نرمال‌شده (مثلاً <code dir="ltr">0.25</code>) را در آن بنویسید، همه به <code dir="ltr">0</code> تبدیل می‌شوند. آرایه‌ای که قرار است عدد اعشاری نگه دارد را از اول اعشاری بسازید (<code dir="ltr">np.array([1., 2., 3.])</code> یا <code dir="ltr">np.zeros(n)</code>) یا با <code dir="ltr">astype(float)</code> تبدیل کنید.

## ۳. اندیس‌گذاری و slicing

### 💡 آرایه‌ی یک‌بعدی

اندیس و برش در آرایه‌ی یک‌بعدی دقیقاً مثل لیست است (جلسه‌ی ۸): اندیس از صفر، اندیس منفی از انتها، و برش <code dir="ltr">start:stop:step</code> بدون <code dir="ltr">stop</code>.

</div>

````python
x = np.array([10, 20, 30, 40, 50, 60])
print(x[0], x[-1])
print(x[1:4])
print(x[::2])
print(x[::-1])

x[0] = 99
print(x)
````

<div dir="rtl">

### 💡 آرایه‌ی دوبعدی

در آرایه‌ی دوبعدی، اندیس هر بعد را با **کاما** جدا می‌کنیم: <code dir="ltr">A[i, j]</code> یعنی سطر <code dir="ltr">i</code> و ستون <code dir="ltr">j</code>. علامت <code dir="ltr">:</code> به‌تنهایی یعنی «همه» در آن بعد.

</div>

````python
A = np.arange(12).reshape(3, 4)
print(A)

print(A[1, 2])
print(A[1])
print(A[:, 2])
print(A[0:2, 1:3])
print(A[:, -1])
print(A[::2, ::3])
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">np.arange(12).reshape(3, 4)</code> عددهای ۰ تا ۱۱ را در ۳ سطر و ۴ ستون می‌چیند (بخش ۴). همین آرایه در شکل بالا رسم شده است.
- <code dir="ltr">A[1, 2]</code> عدد <code dir="ltr">6</code> است؛ <code dir="ltr">A[1]</code> کل سطر ۱ (<code dir="ltr">[4 5 6 7]</code>) و <code dir="ltr">A[:, 2]</code> کل ستون ۲ (<code dir="ltr">[2 6 10]</code>).
- <code dir="ltr">A[0:2, 1:3]</code> یک **زیرماتریس** ۲×۲ است: سطرهای ۰ و ۱، ستون‌های ۱ و ۲.
- <code dir="ltr">A[:, -1]</code> ستون آخر و <code dir="ltr">A[::2, ::3]</code> سطرهای زوج و ستون‌های ۰ و ۳ را برمی‌دارد.
- در یک ماتریس داده، <code dir="ltr">X[:, j]</code> همه‌ی مقدارهای **ویژگی** <code dir="ltr">j</code> و <code dir="ltr">X[i]</code> همه‌ی ویژگی‌های **نمونه‌ی** <code dir="ltr">i</code> است.

### ⚠️ <code dir="ltr">A[1][2]</code> در برابر <code dir="ltr">A[1, 2]</code>

<code dir="ltr">A[1][2]</code> هم کار می‌کند (اول سطر ۱ را برمی‌دارد و بعد عضو ۲ آن را)، اما دو مرحله دارد و برای برش‌ها رفتار متفاوتی دارد: <code dir="ltr">A[:][2]</code> یعنی «همه‌ی سطرها، و بعد سطر ۲»، نه ستون ۲! همیشه از شکل کامایی استفاده کنید.

### 💡 ماسک بولی

مقایسه‌ی یک آرایه با یک عدد، آرایه‌ای از <code dir="ltr">True</code>/<code dir="ltr">False</code> هم‌شکل با آن می‌سازد (**ماسک**). دادن ماسک به جای اندیس، فقط اعضایی را برمی‌دارد که ماسکشان <code dir="ltr">True</code> است. این ابزار اصلی **فیلتر کردن** داده است.

</div>

````python
scores = np.array([12.5, 18.0, 9.5, 15.0, 7.0, 19.5])

mask = scores >= 10
print(mask)
print(scores[mask])
print(scores[scores < 10])
print(scores[(scores >= 10) & (scores < 16)])
print(mask.sum(), mask.mean())

scores[scores < 10] = 10
print(scores)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">scores &gt;= 10</code> آرایه‌ی بولی <code dir="ltr">[ True  True False  True False  True]</code> است و <code dir="ltr">scores[mask]</code> فقط عضوهای قبولی را برمی‌دارد.
- برای ترکیب شرط‌ها روی آرایه، از <code dir="ltr">&amp;</code> (و)، <code dir="ltr">&#124;</code> (یا) و <code dir="ltr">~</code> (نقیض) استفاده می‌کنیم و هر شرط را **داخل پرانتز** می‌گذاریم.
- <code dir="ltr">True</code> مثل <code dir="ltr">1</code> و <code dir="ltr">False</code> مثل <code dir="ltr">0</code> حساب می‌شود؛ پس <code dir="ltr">mask.sum()</code> تعداد قبولی‌ها (<code dir="ltr">4</code>) و <code dir="ltr">mask.mean()</code> **نسبت** آن‌ها (<code dir="ltr">0.666...</code>) است.
- با ماسک می‌توان **انتساب** هم کرد: خط آخر همه‌ی نمره‌های زیر ۱۰ را ۱۰ می‌کند.

### ⚠️ اشتباه رایج: <code dir="ltr">and</code> و <code dir="ltr">or</code> روی آرایه

</div>

````python
print(scores[(scores > 10) and (scores < 16)])
````

<div dir="rtl">

🔍 <code dir="ltr">and</code> و <code dir="ltr">or</code> پایتون فقط با یک مقدار درست/نادرست کار می‌کنند و نمی‌دانند «آرایه‌ای از شش مقدار بولی» درست است یا نادرست؛ پس <code dir="ltr">ValueError</code> (با پیام <code dir="ltr">truth value of an array ... is ambiguous</code>) می‌دهند. روی آرایه همیشه <code dir="ltr">&amp;</code> و <code dir="ltr">&#124;</code> را با پرانتز بنویسید. (بدون پرانتز، <code dir="ltr">scores &gt; 10 &amp; scores &lt; 16</code> هم اشتباه است، چون اولویت <code dir="ltr">&amp;</code> از <code dir="ltr">&gt;</code> بیشتر است؛ جلسه‌ی ۳.)

### 💡 اندیس‌گذاری فهرستی (fancy indexing)

به جای یک اندیس، می‌توان **لیست یا آرایه‌ای از اندیس‌ها** داد؛ نتیجه، اعضای آن موقعیت‌ها به همان ترتیب است (تکرار هم مجاز است).

</div>

````python
x = np.array([10, 20, 30, 40, 50])
print(x[[0, 2, 2, 4]])
print(x[[4, 3, 2, 1, 0]])

A = np.arange(12).reshape(3, 4)
print(A[[0, 2]])
print(A[:, [3, 0]])
print(A[[0, 2], [1, 3]])
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">x[[0, 2, 2, 4]]</code> اعضای ۰، ۲، ۲ و ۴ را برمی‌دارد: <code dir="ltr">[10 30 30 50]</code>.
- <code dir="ltr">A[[0, 2]]</code> سطرهای ۰ و ۲ و <code dir="ltr">A[:, [3, 0]]</code> ستون‌های ۳ و ۰ (به همین ترتیب) را برمی‌دارد؛ برای **انتخاب یا جابه‌جا کردن ویژگی‌ها** بسیار کاربردی است.
- <code dir="ltr">A[[0, 2], [1, 3]]</code> **ماتریس** نمی‌دهد؛ اندیس‌ها جفت‌جفت با هم خوانده می‌شوند: <code dir="ltr">A[0, 1]</code> و <code dir="ltr">A[2, 3]</code>، یعنی <code dir="ltr">[1 11]</code>.

### 💡 کاربرد: کدگذاری one-hot

در دسته‌بندی، برچسب کلاس‌ها (مثلاً <code dir="ltr">0</code>، <code dir="ltr">1</code>، <code dir="ltr">2</code>) را به بردارهای **one-hot** تبدیل می‌کنیم: برداری به طول تعداد کلاس‌ها که فقط در جای کلاس درست، <code dir="ltr">1</code> دارد. چون سطر <code dir="ltr">k</code> ماتریس همانی دقیقاً همین بردار است، یک اندیس‌گذاری فهرستی کافی است:

</div>

````python
labels = np.array([2, 0, 1, 2])
Y = np.eye(3)[labels]
print(Y)
print(Y.shape)
````

<div dir="rtl">

🔍 <code dir="ltr">np.eye(3)[labels]</code> سطرهای ۲، ۰، ۱ و ۲ ماتریس همانی را کنار هم می‌گذارد؛ نتیجه یک ماتریس <code dir="ltr">(4, 3)</code> است که هر سطرش برچسب یک نمونه را کد کرده است. همین ماتریس، هدف (target) آموزش یک شبکه‌ی عصبی با خروجی softmax است (جلسه‌ی ۸).

### 💡 view و copy

**مهم‌ترین تفاوت آرایه با لیست:** برش لیست یک **کپی** می‌سازد (جلسه‌ی ۸)، اما برش آرایه یک **view** است؛ یعنی پنجره‌ای به **همان** داده در حافظه. تغییر view، آرایه‌ی اصلی را هم تغییر می‌دهد. (NumPy این کار را برای سرعت و صرفه‌جویی در حافظه انجام می‌دهد.) در مقابل، ماسک بولی و اندیس‌گذاری فهرستی همیشه **کپی** برمی‌گردانند.

</div>

````python
lst = [0, 1, 2, 3, 4]
part = lst[1:3]
part[0] = 100
print(lst)

arr = np.array([0, 1, 2, 3, 4])
view = arr[1:3]
view[0] = 100
print(arr)

fancy = arr[[1, 2]]
fancy[0] = -1
print(arr)

safe = arr[1:3].copy()
safe[0] = -5
print(arr, np.shares_memory(arr, view), np.shares_memory(arr, safe))
````

<div dir="rtl">

### 🔍 کد را بخوان

- در لیست، تغییر <code dir="ltr">part</code> روی <code dir="ltr">lst</code> اثری ندارد.
- در آرایه، <code dir="ltr">view</code> و <code dir="ltr">arr</code> یک حافظه‌ی مشترک دارند: بعد از <code dir="ltr">view[0] = 100</code>، عضو ۱ <code dir="ltr">arr</code> هم <code dir="ltr">100</code> می‌شود.
- <code dir="ltr">fancy</code> کپی است؛ تغییرش روی <code dir="ltr">arr</code> بی‌اثر است.
- <code dir="ltr">.copy()</code> یک کپی مستقل می‌سازد. <code dir="ltr">np.shares&#95;memory</code> بررسی می‌کند دو آرایه حافظه‌ی مشترک دارند یا نه.

### ⚠️ اشتباه رایج: خراب کردن داده‌ی اصلی

فرض کنید <code dir="ltr">X&#95;train = X[:800]</code> و بعد برای نرمال‌سازی، <code dir="ltr">X&#95;train[:, 0] = X&#95;train[:, 0] / 100</code> بنویسید. چون <code dir="ltr">X&#95;train</code> یک view است، ستون اول <code dir="ltr">X</code> اصلی هم تقسیم شده است؛ اگر بعداً دوباره از <code dir="ltr">X</code> استفاده کنید (مثلاً یک بار دیگر نرمال کنید)، داده‌ها دو بار تقسیم شده‌اند. هر وقت قرار است روی بخشی از داده تغییر **درجا** بدهید و داده‌ی اصلی را هم لازم دارید، <code dir="ltr">.copy()</code> بگیرید.

</div>

## ۴. <code dir="ltr">reshape</code>

### 💡 مفهوم

<code dir="ltr">a.reshape(new&#95;shape)</code> همان اعضا را با یک **شکل جدید** برمی‌گرداند. اعضا به ترتیب **سطری** (row-major) خوانده و چیده می‌شوند: اول سطر اول از چپ به راست، بعد سطر دوم و ... . تعداد کل اعضا باید ثابت بماند؛ یکی از ابعاد را می‌توان <code dir="ltr">-1</code> گذاشت تا NumPy خودش آن را حساب کند.

</div>

````python
a = np.arange(6)
print(a.reshape(2, 3))
print(a.reshape(3, 2))
print(a.reshape(3, -1))
print(a.reshape(-1, 1))
print(a.reshape(1, -1), a.reshape(1, -1).shape)
print(a)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">reshape(2, 3)</code> عددهای ۰ تا ۵ را در دو سطر سه‌تایی می‌چیند: <code dir="ltr">[[0 1 2] [3 4 5]]</code>.
- <code dir="ltr">reshape(3, -1)</code>: «سه سطر، تعداد ستون‌ها را خودت حساب کن» (۲).
- <code dir="ltr">reshape(-1, 1)</code> یک **بردار ستونی** با شکل <code dir="ltr">(6, 1)</code> و <code dir="ltr">reshape(1, -1)</code> یک **بردار سطری** با شکل <code dir="ltr">(1, 6)</code> می‌سازد. این دو با آرایه‌ی یک‌بعدی <code dir="ltr">(6,)</code> فرق دارند؛ تفاوتشان در جلسه‌ی ۲۲ (broadcasting) و در scikit-learn (که ورودی را دوبعدی می‌خواهد) مهم می‌شود.
- <code dir="ltr">reshape</code> خود <code dir="ltr">a</code> را تغییر نمی‌دهد و یک آرایه‌ی جدید (معمولاً یک view) برمی‌گرداند.

### 💡 <code dir="ltr">ravel</code>، <code dir="ltr">flatten</code>، <code dir="ltr">T</code> و <code dir="ltr">np.newaxis</code>

| عمل | نتیجه | view یا copy |
|---|---|---|
| <code dir="ltr">a.ravel()</code> | آرایه‌ی یک‌بعدی (صاف شده) | معمولاً view |
| <code dir="ltr">a.flatten()</code> | آرایه‌ی یک‌بعدی | همیشه copy |
| <code dir="ltr">a.T</code> | **ترانهاده** (جای سطر و ستون عوض می‌شود) | view |
| <code dir="ltr">a[:, np.newaxis]</code> | یک بعد با اندازه‌ی ۱ اضافه می‌کند | view |

</div>

````python
M = np.array([[1, 2, 3],
              [4, 5, 6]])

print(M.ravel())
print(M.flatten())
print(M.T)
print(M.T.shape)

v = np.array([1, 2, 3])
print(v[:, np.newaxis].shape, v[np.newaxis, :].shape)
````

<div dir="rtl">

### 🔍 کد را بخوان

- <code dir="ltr">ravel</code> و <code dir="ltr">flatten</code> هر دو <code dir="ltr">[1 2 3 4 5 6]</code> می‌دهند؛ تفاوتشان فقط در view یا copy بودن است (سؤال ۲).
- <code dir="ltr">M.T</code> ماتریس <code dir="ltr">(2, 3)</code> را به <code dir="ltr">(3, 2)</code> تبدیل می‌کند: سطر اول آن <code dir="ltr">[1 4]</code> است، یعنی ستون اول <code dir="ltr">M</code>.
- <code dir="ltr">np.newaxis</code> (که همان <code dir="ltr">None</code> است) یک بعد جدید با اندازه‌ی ۱ می‌سازد: <code dir="ltr">(3,)</code> به <code dir="ltr">(3, 1)</code> یا <code dir="ltr">(1, 3)</code> تبدیل می‌شود.

### 💡 کاربرد: تصویر به‌عنوان آرایه

یک تصویر سیاه‌وسفید یک ماتریس از شدت روشنایی پیکسل‌هاست (معمولاً بین ۰ و ۲۵۵). مثلاً هر تصویر داده‌ی معروف MNIST (رقم‌های دست‌نویس) ماتریسی <code dir="ltr">28 × 28</code> است. یک شبکه‌ی عصبی ساده (MLP) ورودی را **برداری** می‌خواهد؛ پس هر تصویر را به برداری با <code dir="ltr">784</code> عضو صاف می‌کنیم. برای یک **دسته (batch)** از تصاویر با شکل <code dir="ltr">(N, 28, 28)</code>، این کار با یک <code dir="ltr">reshape</code> انجام می‌شود:

</div>

````python
batch = np.zeros((32, 28, 28))
print(batch.ndim, batch.shape)

flat = batch.reshape(32, -1)
print(flat.shape)

flat2 = batch.reshape(len(batch), -1)
print(flat2.shape)

back = flat.reshape(-1, 28, 28)
print(back.shape)
````

<div dir="rtl">

🔍 آرایه‌ی سه‌بعدی <code dir="ltr">(32, 28, 28)</code> یعنی ۳۲ تصویر ۲۸×۲۸. <code dir="ltr">reshape(32, -1)</code> هر تصویر را به یک سطر ۷۸۴تایی تبدیل می‌کند: ماتریس داده‌ای با شکل <code dir="ltr">(n&#95;samples, n&#95;features)</code>. نوشتن <code dir="ltr">len(batch)</code> به جای <code dir="ltr">32</code>، کد را برای هر اندازه‌ی دسته‌ای درست نگه می‌دارد. <code dir="ltr">reshape(-1, 28, 28)</code> همان تصاویر را برمی‌گرداند.

### ❓ پیش‌بینی کن

آیا <code dir="ltr">M.reshape(3, 2)</code> با <code dir="ltr">M.T</code> برابر است؟ (<code dir="ltr">M</code> همان ماتریس <code dir="ltr">[[1 2 3] [4 5 6]]</code> است.) قبل از اجرای سلول زیر جواب دهید.

</div>

````python
print(M.reshape(3, 2))
print(M.T)
````

<div dir="rtl">

🔍 هر دو شکل <code dir="ltr">(3, 2)</code> دارند، اما **برابر نیستند**: <code dir="ltr">reshape</code> همان ترتیب سطری <code dir="ltr">1 2 3 4 5 6</code> را دوتا‌دوتا می‌چیند (<code dir="ltr">[[1 2] [3 4] [5 6]]</code>)، در حالی که <code dir="ltr">T</code> سطرها را به ستون تبدیل می‌کند (<code dir="ltr">[[1 4] [2 5] [3 6]]</code>). اشتباه گرفتن این دو، یکی از رایج‌ترین باگ‌های بی‌صدا در کار با داده است: هیچ خطایی نمی‌دهد، اما داده‌ها را به‌هم می‌ریزد.

### ⚠️ اشتباه رایج: تعداد اعضای ناسازگار

</div>

````python
x = np.arange(10)
x.reshape(-1, 4)
````

<div dir="rtl">

🔍 ده عضو را نمی‌توان در سطرهای چهارتایی چید؛ <code dir="ltr">reshape</code> خطای <code dir="ltr">ValueError</code> می‌دهد. اگر واقعاً می‌خواهید داده را به دسته‌های چهارتایی تقسیم کنید، یا اعضای اضافه را کنار بگذارید (<code dir="ltr">x[:8].reshape(-1, 4)</code>) یا با برش، دسته‌ی آخر را کوچک‌تر بگیرید (سؤال ۷).

</div>

## 📌 آنچه باید بلد باشید

- چرا آرایه‌ی NumPy از لیست سریع‌تر است و عملگرهای <code dir="ltr">+</code>، <code dir="ltr">&#42;</code>، <code dir="ltr">&#42;&#42;</code> و <code dir="ltr">/</code> روی آن عضو به عضو اند.
- ساخت آرایه با <code dir="ltr">array</code>، <code dir="ltr">zeros</code>، <code dir="ltr">ones</code>، <code dir="ltr">full</code>، <code dir="ltr">arange</code>، <code dir="ltr">linspace</code>، <code dir="ltr">eye</code> و <code dir="ltr">&#42;&#95;like</code>؛ تفاوت <code dir="ltr">arange</code> و <code dir="ltr">linspace</code> در شامل بودن سر دوم.
- خواندن <code dir="ltr">ndim</code>، <code dir="ltr">shape</code>، <code dir="ltr">size</code> و <code dir="ltr">dtype</code>؛ شکل <code dir="ltr">(n&#95;samples, n&#95;features)</code> برای داده.
- ارتقای نوع، <code dir="ltr">astype</code> (دور ریختن بخش اعشاری)، انتساب بی‌صدای اعشاری به آرایه‌ی صحیح، سرریز در نوع‌های کوچک.
- اندیس و برش در آرایه‌ی یک‌بعدی و دوبعدی (<code dir="ltr">A[i, j]</code>، <code dir="ltr">A[:, j]</code>، <code dir="ltr">A[i0:i1, j0:j1]</code>).
- ماسک بولی، ترکیب شرط‌ها با <code dir="ltr">&amp;</code> و <code dir="ltr">&#124;</code> و پرانتز، شمردن با <code dir="ltr">mask.sum()</code> و انتساب با ماسک.
- اندیس‌گذاری فهرستی (و رفتار جفتی <code dir="ltr">A[[i...], [j...]]</code>)؛ one-hot با <code dir="ltr">np.eye(K)[labels]</code>.
- برش view است و ماسک و اندیس فهرستی copy؛ <code dir="ltr">.copy()</code> و <code dir="ltr">np.shares&#95;memory</code>.

### سؤال ۲ — view یا copy؟

خروجی کد زیر را بنویسید. برای هر خط بگویید متغیر سمت چپ view است یا copy.

</div>

```python

b[0] = 100
c = a[[1, 2]]
c[1] = -1

d[:] = 0

e[1, 0] = 7

f[5] = 9
g = a.flatten()

```

در یک پروژه‌ی یادگیری ماشین، کدام یک از این رفتارها می‌تواند باعث شود نتیجه‌ی ارزیابی مدل شما اشتباه شود؟ یک مثال بزنید.

<details>
<summary><b>پاسخ</b></summary>

<code dir="ltr">[  0 100   2   7   4   9]</code>

| متغیر | نوع | اثر روی <code dir="ltr">a</code> |
|---|---|---|
| <code dir="ltr">b = a[1:4]</code> | view (برش) | <code dir="ltr">a[1]</code> برابر <code dir="ltr">100</code> می‌شود |
| <code dir="ltr">c = a[[1, 2]]</code> | copy (اندیس فهرستی) | بی‌اثر |

| <code dir="ltr">f = a.ravel()</code> | view (چون <code dir="ltr">a</code> پیوسته است) | <code dir="ltr">a[5]</code> برابر <code dir="ltr">9</code> می‌شود |
| <code dir="ltr">g = a.flatten()</code> | copy | بی‌اثر |

### سؤال ۴ — one-hot و هموارسازی برچسب

الف) با <code dir="ltr">labels = np.array([2, 0, 1, 2])</code> و <code dir="ltr">Y = np.eye(3)[labels]</code>، شکل <code dir="ltr">Y</code> و سطر اول آن را بنویسید. چرا این کد کار می‌کند؟

ب) برعکس: با داشتن <code dir="ltr">Y</code>، برچسب‌ها را چطور برمی‌گردانید؟ خروجی <code dir="ltr">[int(row.argmax()) for row in Y]</code> چیست؟ (<code dir="ltr">argmax</code> اندیس بزرگ‌ترین عضو است؛ جلسه‌ی ۸.)

ج) در **هموارسازی برچسب (label smoothing)**، به جای <code dir="ltr">1</code> و <code dir="ltr">0</code> از مقدارهای «نرم‌تر» استفاده می‌شود تا مدل بیش از حد مطمئن نشود: <code dir="ltr">Y&#95;smooth = Y &#42; (1 - eps) + eps / K</code> که <code dir="ltr">K</code> تعداد کلاس‌هاست. با <code dir="ltr">eps = 0.1</code>، سطر اول <code dir="ltr">Y&#95;smooth</code> و مجموع آن را حساب کنید.

د) اگر یکی از برچسب‌ها اشتباهاً <code dir="ltr">3</code> باشد، <code dir="ltr">np.eye(3)[labels]</code> چه می‌کند؟

<details>
<summary><b>پاسخ</b></summary>

الف) شکل <code dir="ltr">(4, 3)</code>؛ سطر اول <code dir="ltr">[0. 0. 1.]</code>. سطر <code dir="ltr">k</code> ماتریس همانی، برداری است که فقط در موقعیت <code dir="ltr">k</code> یک دارد؛ اندیس‌گذاری فهرستی با <code dir="ltr">labels</code> سطرهای ۲، ۰، ۱ و ۲ را به همین ترتیب کنار هم می‌گذارد.

ب) <code dir="ltr">[2, 0, 1, 2]</code>؛ در هر سطر one-hot، تنها عضو <code dir="ltr">1</code> بزرگ‌ترین است و اندیس آن همان برچسب است.

ج) <code dir="ltr">[0.0333 0.0333 0.9333]</code> (تقریباً)؛ مجموع <code dir="ltr">1.0</code>. هر سطر همچنان یک توزیع احتمال است، چون <code dir="ltr">(1 - eps) &#42; 1 + K &#42; eps / K = 1</code>.

د) خطای <code dir="ltr">IndexError</code> (اندیس ۳ برای محوری با اندازه‌ی ۳ خارج از محدوده است). این خطا مفید است: برچسب نامعتبر را بی‌صدا رد نمی‌کند.

</details>


ب) <code dir="ltr">[(4, 2), (4, 2), (2, 2)]</code>. برش‌ها view هستند؛ نرمال کردن درجای هر دسته، همان سطرهای <code dir="ltr">X</code> را تغییر می‌دهد و در **epoch** بعد، داده‌ها دوباره (و چندباره) نرمال می‌شوند. (اگر قبلش اندیس‌ها را با <code dir="ltr">permutation</code> به هم بریزید و <code dir="ltr">X[idx[start:start + batch&#95;size]]</code> بنویسید، دسته‌ها copy می‌شوند.) در جلسه‌ی ۱۴ همین کار را با یک generator انجام دادید.

ج) ده سطر را نمی‌توان به گروه‌های چهارتایی تقسیم کرد (۲۰ عضو بر <code dir="ltr">4 × 2 = 8</code> بخش‌پذیر نیست)؛ <code dir="ltr">ValueError</code>. <code dir="ltr">X[:8].reshape(-1, 4, 2)</code> شکل <code dir="ltr">(2, 4, 2)</code> دارد: دو دسته‌ی چهارتایی، اما دو نمونه‌ی آخر **کنار گذاشته** شده‌اند (همان کاری که <code dir="ltr">drop&#95;last=True</code> در کتابخانه‌های یادگیری عمیق انجام می‌دهد).

د) <code dir="ltr">B.reshape(3, 2)</code> برابر <code dir="ltr">[[1 2] [3 4] [5 6]]</code> است و ویژگی‌های دو نمونه را با هم قاطی می‌کند (مثلاً ستون اول <code dir="ltr">[1 3 5]</code> هم ویژگی نمونه‌ی اول را دارد و هم دوم را). <code dir="ltr">B.T</code> برابر <code dir="ltr">[[1 4] [2 5] [3 6]]</code> است: هر **ستون** یک نمونه است. پس <code dir="ltr">B.T</code> درست است. <code dir="ltr">reshape</code> فقط اعضا را به ترتیب سطری دوباره می‌چیند و معنای سطر و ستون را نمی‌فهمد.

## خلاصه‌ی جلسه

- آرایه‌ی NumPy مجموعه‌ای هم‌نوع از عددهاست که عملیات روی آن عضو به عضو و بسیار سریع انجام می‌شود؛ پایه‌ی همه‌ی کتابخانه‌های علوم داده.
- آرایه‌ها با <code dir="ltr">array</code>، <code dir="ltr">zeros</code>، <code dir="ltr">ones</code>، <code dir="ltr">full</code>، <code dir="ltr">arange</code>، <code dir="ltr">linspace</code> و <code dir="ltr">eye</code> ساخته می‌شوند؛ <code dir="ltr">shape</code> و <code dir="ltr">dtype</code> را همیشه بررسی کنید.
- نوع داده‌ی ثابت یعنی ارتقای نوع، تبدیل بی‌صدای اعشاری به صحیح و سرریز در نوع‌های کوچک.
- اندیس، برش، ماسک بولی و اندیس‌گذاری فهرستی ابزارهای انتخاب داده‌اند؛ برش view است و تغییرش، داده‌ی اصلی را هم تغییر می‌دهد.
- <code dir="ltr">reshape</code> با ترتیب سطری اعضا را دوباره می‌چیند و با <code dir="ltr">T</code> فرق دارد؛ صاف کردن تصویرها و ساختن دسته‌ها نمونه‌های رایج آن‌اند.

**جلسه‌ی بعد:** NumPy ۲؛ برداری‌سازی، broadcasting، توابع آماری روی محورها، اعداد تصادفی و جبر خطی.
