## ۱. ساخت دیکشنری

### 💡 مفهوم

**دیکشنری (dictionary)** مجموعه‌ای از جفت‌های **کلید: مقدار (key: value)** است. در لیست، به هر عضو با یک **اندیس عددی** (۰، ۱، ۲، ...) دسترسی دارید؛ در دیکشنری، به هر مقدار با یک **کلید معنی‌دار** (مثلاً <code dir="ltr">"name"</code> یا <code dir="ltr">"math"</code>) دسترسی پیدا می‌کنید؛ درست مثل یک لغت‌نامه که با «کلمه» به «معنی» می‌رسید.

</div>

````python
student = {"name": "Ali", "age": 20, "major": "Physics"}
grades = {"math": 18, "physics": 19, "chemistry": 17}
print(student)
print(grades["physics"])
print(len(grades), type(grades))

empty = {}                                   # an empty dict (not a set!)
pairs = dict([("x", 1), ("y", 2)])           # from a list of pairs
zipped = dict(zip(["a", "b", "c"], [1, 2, 3]))
print(empty, pairs, zipped)
````

<div dir="rtl">

### 🔍 کد را بخوان

- هر عضو یک جفت <code dir="ltr">key: value</code> است و جفت‌ها با کاما جدا می‌شوند.
- <code dir="ltr">grades["physics"]</code> مقدار مربوط به کلید <code dir="ltr">"physics"</code> را می‌دهد. اینجا داخل <code dir="ltr">[ ]</code> **کلید** می‌آید، نه اندیس.
- <code dir="ltr">dict(zip(keys, values))</code> دو لیست هم‌طول را به یک دیکشنری تبدیل می‌کند؛ الگویی بسیار رایج.

### 💡 قواعد کلید و مقدار

- **کلیدها یکتا هستند.** اگر یک کلید را دو بار بنویسید، مقدار دوم جایگزین اولی می‌شود.
- **کلید باید hashable باشد** (جلسه‌ی ۹): عدد، رشته یا تاپل؛ اما **نه** لیست، مجموعه یا دیکشنری.
- **مقدار** می‌تواند هر چیزی باشد: عدد، رشته، لیست، دیکشنری دیگر، حتی یک تابع.
- از پایتون ۳.۷ به بعد، دیکشنری **ترتیب افزودن** کلیدها را حفظ می‌کند.

### 💡 چرا دیکشنری سریع است؟

دیکشنری هم مثل مجموعه از **اثر انگشت (hash)** کلیدها استفاده می‌کند؛ پس پیدا کردن مقدار یک کلید، تقریباً مستقل از تعداد اعضا، **فوری** است. پیدا کردن یک عضو در لیست اما ممکن است لازم داشته باشد همه‌ی اعضا را یکی‌یکی بررسی کنید. اگر بارها چیزی را «بر اساس یک شناسه» جست‌وجو می‌کنید، دیکشنری ساختار درست است.

</div>

````python
D = {"Reiazi": 12, "Amar": 17}
D[["Reiazi", "Amar"]]
````

<div dir="rtl">

🔍 کسی که با جدول‌های Pandas کار کرده باشد ممکن است انتظار داشته باشد با یک **لیست** از کلیدها چند مقدار را یک‌جا بگیرد؛ اما در دیکشنری، لیست می‌شود **یک کلید** و چون لیست hashable نیست، خطای <code dir="ltr">TypeError: unhashable type: 'list'</code> رخ می‌دهد.

</div>

<div dir="rtl">

## ۲. دسترسی و به‌روزرسانی

### 💡 خواندن: <code dir="ltr">d[key]</code> یا <code dir="ltr">d.get(key)</code>؟

| روش | اگر کلید **باشد** | اگر کلید **نباشد** |
|---|---|---|
| <code dir="ltr">d[key]</code> | مقدار | خطای <code dir="ltr">KeyError</code> |
| <code dir="ltr">d.get(key)</code> | مقدار | <code dir="ltr">None</code> |
| <code dir="ltr">d.get(key, default)</code> | مقدار | <code dir="ltr">default</code> |

</div>

````python
student = {"name": "Ali", "age": 20}
print(student["name"])
print(student.get("major"))
print(student.get("major", "unknown"))
````

````python
student["major"]
````

<div dir="rtl">

🔍 پیام <code dir="ltr">KeyError: 'major'</code> دقیقاً نام کلید گمشده را نشان می‌دهد. وقتی مطمئن نیستید کلید وجود دارد، از <code dir="ltr">get</code> با یک مقدار پیش‌فرض استفاده کنید.

### 💡 افزودن، تغییر و حذف

انتساب به یک کلید، اگر کلید **نباشد** آن را **اضافه** می‌کند و اگر **باشد** مقدارش را **عوض** می‌کند. دیکشنری مثل لیست **تغییرپذیر** است.

</div>

````python
student = {"name": "Ali", "age": 20}
student["major"] = "Physics"        # add a new key
student["age"] = 21                 # modify an existing key
print(student)

del student["major"]                # remove by key
age = student.pop("age")            # remove and return the value
print(student, "| removed age:", age)

student.update({"age": 22, "city": "Bushehr"})   # add/modify several keys
print(student)
````

<div dir="rtl">

### ❓ پیش‌بینی کن

عملگر <code dir="ltr">in</code> روی دیکشنری چه چیزی را بررسی می‌کند؟ هر خط <code dir="ltr">True</code> است یا <code dir="ltr">False</code>؟

</div>

````python
D = {"Reiazi": 12, "Amar": 17}
print("Amar" in D)
print(17 in D)
print(17 in D.values())
````

<div dir="rtl">

🔍 <code dir="ltr">in</code> فقط **کلیدها** را بررسی می‌کند؛ پس <code dir="ltr">17 in D</code> نادرست است، چون <code dir="ltr">17</code> یک **مقدار** است. برای جست‌وجو در مقدارها باید صریحاً <code dir="ltr">D.values()</code> بنویسید (که جست‌وجوی آن، برخلاف کلیدها، سریع نیست).

### ⚠️ اشتباه رایج

- **دسترسی با اندیس:** <code dir="ltr">grades[0]</code> معنی «اولین عضو» نمی‌دهد؛ دنبال کلیدی به نام <code dir="ltr">0</code> می‌گردد و <code dir="ltr">KeyError</code> می‌دهد.
- **فرض بر وجود کلید:** در داده‌ی واقعی بعضی رکوردها فیلدهایی را ندارند. <code dir="ltr">record["email"]</code> برنامه را متوقف می‌کند؛ <code dir="ltr">record.get("email")</code> نه.

</div>

## ۳. متدها و پیمایش

### 💡 سه نمای دیکشنری

| متد | چه چیزی می‌دهد |
|---|---|
| <code dir="ltr">d.keys()</code> | همه‌ی کلیدها |
| <code dir="ltr">d.values()</code> | همه‌ی مقدارها |
| <code dir="ltr">d.items()</code> | همه‌ی جفت‌های <code dir="ltr">(key, value)</code> به شکل تاپل |

این‌ها لیست نیستند، بلکه **نما (view)** هستند؛ یعنی همیشه وضعیت فعلی دیکشنری را نشان می‌دهند. اگر لیست لازم دارید، <code dir="ltr">list(...)</code> بنویسید.

</div>

````python
grades = {"math": 18, "physics": 19, "chemistry": 17}
print(grades.keys())
print(list(grades.values()))
print(list(grades.items()))

for subject in grades:               # looping over a dict gives its KEYS
    print(subject, end=" ")
print()

for subject, score in grades.items():
    print(f"{subject:<10}{score:>3}")
````

<div dir="rtl">

### 🔍 کد را بخوان

- حلقه‌ی <code dir="ltr">for</code> مستقیم روی دیکشنری، **کلیدها** را می‌دهد.
- <code dir="ltr">items()</code> در هر دور یک تاپل <code dir="ltr">(key, value)</code> می‌دهد که با <code dir="ltr">subject, score</code> باز می‌شود (جلسه‌ی ۹)؛ رایج‌ترین شکل پیمایش.

### ❓ پیش‌بینی کن

توابعی مثل <code dir="ltr">max</code>، <code dir="ltr">sum</code> و <code dir="ltr">sorted</code> روی دیکشنری با کدام بخش کار می‌کنند؟

</div>

````python
grades = {"math": 18, "physics": 19, "chemistry": 17}
print(max(grades))
print(max(grades.values()))
print(sorted(grades))
print(sum(grades.values()) / len(grades))
````

<div dir="rtl">

🔍 <code dir="ltr">max(grades)</code> روی **کلیدها** کار می‌کند و بزرگ‌ترین کلید از نظر الفبایی (<code dir="ltr">"physics"</code>) را می‌دهد، نه درسی که بیشترین نمره را دارد. برای «درس با بیشترین نمره» باید روی <code dir="ltr">items()</code> حلقه بزنید (یا در جلسه‌ی ۱۳ از <code dir="ltr">key</code> استفاده کنید).

### ⚠️ اشتباه رایج: تغییر اندازه‌ی دیکشنری حین پیمایش

</div>

````python
scores = {"Ali": 8, "Sara": 17, "Reza": 9}
for name in scores:
    if scores[name] < 10:
        del scores[name]
````

<div dir="rtl">

🔍 خطای <code dir="ltr">RuntimeError: dictionary changed size during iteration</code>؛ پایتون اجازه نمی‌دهد حین پیمایش، کلیدی اضافه یا حذف کنید (با لیست‌ها در جلسه‌ی ۸ دیدید که بدون خطا نتیجه‌ی غلط می‌گرفتید؛ اینجا حداقل خطا می‌گیرید). راه حل: یک **دیکشنری جدید** از اعضای مطلوب بسازید، یا روی <code dir="ltr">list(scores)</code> حلقه بزنید.

</div>

## ۴. الگوهای پرکاربرد

### 💡 الگوی ۱: شمارش

شمردن تعداد تکرار هر مقدار (فراوانی) شاید رایج‌ترین کار در تحلیل داده باشد:

</div>

````python
text = "the cat sat on the mat the cat ran"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

<div dir="rtl">

🔍 خط کلیدی <code dir="ltr">counts.get(word, 0) + 1</code> است: اگر کلمه قبلاً دیده نشده، <code dir="ltr">get</code> مقدار <code dir="ltr">0</code> را برمی‌گرداند؛ وگرنه شمارش فعلی را. بدون <code dir="ltr">get</code>، اولین بار <code dir="ltr">counts[word]</code> خطای <code dir="ltr">KeyError</code> می‌داد.

### 💡 الگوی ۲: گروه‌بندی

جمع کردن اعضا بر اساس یک ویژگی، در **دیکشنری‌ای از لیست‌ها**:

</div>

````python
students = [("Ali", "A"), ("Sara", "B"), ("Reza", "A"), ("Mina", "C"), ("Hadi", "B")]
by_grade = {}
for name, grade in students:
    if grade not in by_grade:
        by_grade[grade] = []          # a NEW list for each new key
    by_grade[grade].append(name)
print(by_grade)
````

<div dir="rtl">

این همان کاری است که بعدها با <code dir="ltr">groupby</code> در Pandas (جلسه‌ی ۲۴) انجام می‌دهید.

### 💡 الگوی ۳: جدول نگاشت (lookup) و کدگذاری

در جلسه‌ی ۹، برای تبدیل برچسب به عدد از <code dir="ltr">classes.index(label)</code> استفاده کردیم که برای هر برچسب کل لیست را جست‌وجو می‌کند. دیکشنری همین کار را فوری انجام می‌دهد:

</div>

````python
size_mapping = {"XL": 3, "L": 2, "M": 1}
sizes = ["M", "XL", "L", "M"]
encoded = []
for s in sizes:
    encoded.append(size_mapping[s])
print(encoded)
````

<div dir="rtl">

### 💡 الگوی ۴: وارونه‌سازی

برای تبدیل عدد به برچسب (مثلاً برای نمایش خروجی مدل)، دیکشنری را **وارونه** می‌کنیم:

</div>

````python
inv_size_mapping = {}
for k, v in size_mapping.items():
    inv_size_mapping[v] = k
print(inv_size_mapping)
print(inv_size_mapping[3])
````

<div dir="rtl">

### ⚠️ اشتباه رایج: وارونه‌سازی با مقدارهای تکراری

وارونه‌سازی فقط وقتی بدون از دست دادن اطلاعات کار می‌کند که **مقدارها هم یکتا** باشند. اگر دو کلید مقدار یکسانی داشته باشند، در دیکشنری وارونه فقط **آخرین** آن‌ها باقی می‌ماند (چون کلیدها یکتا هستند).

</div>

<div dir="rtl">

## ۵. دیکشنری‌های تودرتو و داده‌ی جدولی

### 💡 مفهوم

مقدار یک دیکشنری می‌تواند لیست یا دیکشنری دیگری باشد. دو شکل استاندارد برای نگه‌داری یک **جدول داده** وجود دارد:

| شکل | ساختار | مثال |
|---|---|---|
| **سطری** (row-oriented) | لیستی از دیکشنری‌ها؛ هر دیکشنری یک رکورد | <code dir="ltr">[{"name": "Ali", "score": 17}, ...]</code> |
| **ستونی** (column-oriented) | دیکشنری‌ای از لیست‌ها؛ هر کلید یک ستون | <code dir="ltr">{"name": ["Ali", ...], "score": [17, ...]}</code> |

شکل سطری همان چیزی است که در فایل‌های **JSON** و پاسخ سرویس‌های وب می‌بینید؛ شکل ستونی تقریباً همان چیزی است که <code dir="ltr">pd.DataFrame</code> در Pandas (جلسه‌ی ۲۳) می‌سازد.

</div>

````python
# row-oriented: a list of records
records = [
    {"name": "Ali", "math": 17, "stat": 15},
    {"name": "Sara", "math": 19, "stat": 18},
    {"name": "Reza", "math": 12, "stat": 16},
]
print(records[1]["math"])

# column-oriented: one list per column
table = {
    "name": ["Ali", "Sara", "Reza"],
    "math": [17, 19, 12],
    "stat": [15, 18, 16],
}
print(table["math"][1])
````

<div dir="rtl">

### 🔍 کد را بخوان

هر دو خط چاپ، نمره‌ی ریاضی سارا (<code dir="ltr">19</code>) را می‌دهند، اما ترتیب دسترسی برعکس است:

- **سطری:** <code dir="ltr">records[1]["math"]</code> یعنی «رکورد دوم، فیلد ریاضی».
- **ستونی:** <code dir="ltr">table["math"][1]</code> یعنی «ستون ریاضی، عضو دوم».

هر شکل برای یک کار راحت‌تر است. محاسبه روی یک **ستون** (مثلاً میانگین ریاضی) در شکل ستونی ساده است:

</div>

````python
means = {}
for column in ["math", "stat"]:
    values = table[column]
    means[column] = sum(values) / len(values)
print(means)
````

<div dir="rtl">

### ❓ پیش‌بینی کن

کد زیر شکل ستونی را به شکل سطری تبدیل می‌کند. خروجی را پیش‌بینی کنید. چرا حلقه‌ی بیرونی روی اندیس سطرهاست؟

</div>

````python
columns = list(table.keys())
n_rows = len(table["name"])

rows = []
for i in range(n_rows):
    row = {}
    for col in columns:
        row[col] = table[col][i]
    rows.append(row)
print(rows[2])
print(rows == records)
````

<div dir="rtl">

🔍 برای هر سطر <code dir="ltr">i</code>، از هر ستون عضو <code dir="ltr">i</code> ام برداشته می‌شود؛ همان ایده‌ی **ترانهاده** در جلسه‌ی ۸. <code dir="ltr">rows[2]</code> رکورد رضاست و چون محتوا دقیقاً با <code dir="ltr">records</code> یکی است، مقایسه‌ی <code dir="ltr">==</code> درست است (دو دیکشنری وقتی برابرند که همان کلیدها با همان مقدارها را داشته باشند).

</div>

<div dir="rtl">

## 📌 آنچه باید بلد باشید

- دیکشنری جفت‌های <code dir="ltr">key: value</code> است؛ کلیدها یکتا و hashable‌اند؛ مقدارها هر چیزی.
- تفاوت <code dir="ltr">d[key]</code> (خطای <code dir="ltr">KeyError</code>) و <code dir="ltr">d.get(key, default)</code>.
- انتساب به کلید: افزودن یا تغییر؛ <code dir="ltr">del</code>، <code dir="ltr">pop</code> و <code dir="ltr">update</code>.
- <code dir="ltr">in</code> فقط کلیدها را بررسی می‌کند؛ <code dir="ltr">max</code>، <code dir="ltr">sorted</code> و حلقه‌ی <code dir="ltr">for</code> هم روی کلیدها کار می‌کنند.
- <code dir="ltr">keys()</code>، <code dir="ltr">values()</code> و <code dir="ltr">items()</code>؛ پیمایش با <code dir="ltr">for k, v in d.items()</code>.
- خطای تغییر اندازه‌ی دیکشنری حین پیمایش.
- الگوهای شمارش (<code dir="ltr">get(k, 0) + 1</code>)، گروه‌بندی (دیکشنری از لیست‌ها)، نگاشت/کدگذاری و وارونه‌سازی.
- جدول داده به دو شکل سطری و ستونی و تبدیل آن‌ها به یکدیگر.

</div>

## 📝 سؤالات مرور / نمونه سؤال امتحان

> اول خودتان جواب دهید (روی کاغذ، بدون اجرای کد)، بعد پاسخ را باز کنید و در آخر کد را اجرا کنید.

### سؤال ۱ — پیش‌بینی

با فرض <code dir="ltr">d = {"a": 1, "b": 2}</code>، حاصل هر خط را بنویسید (هر خط را مستقل، روی همین <code dir="ltr">d</code> اولیه، در نظر بگیرید). اگر خطا می‌دهد، نوع خطا را بنویسید.

| عبارت | حاصل؟ |
|---|---|
| <code dir="ltr">d["c"]</code> | |
| <code dir="ltr">d.get("c", 0)</code> | |
| <code dir="ltr">"1" in d</code> | |
| <code dir="ltr">1 in d.values()</code> | |
| <code dir="ltr">list(d.items())[1]</code> | |

### سؤال ۳ — ماتریس درهم‌ریختگی چندکلاسه

برای یک دسته‌بند سه‌کلاسه:

</div>

```python
y_true = ["cat", "dog", "cat", "bird", "dog", "cat", "bird", "dog"]
y_pred = ["cat", "cat", "cat", "bird", "dog", "dog", "dog", "dog"]
```

الف) چرا کلید <code dir="ltr">cm</code> یک **تاپل** است و نمی‌توانست لیست <code dir="ltr">[t, p]</code> باشد؟ محتوای <code dir="ltr">cm</code> را بنویسید.

ج) چرا <code dir="ltr">cm[("cat", "bird")]</code> خطا می‌دهد؟ چطور بدون تغییر دیکشنری، تعداد آن را درست (یعنی <code dir="ltr">0</code>) بخوانید؟

<details>
<summary><b>پاسخ</b></summary>

الف) کلید باید hashable باشد؛ تاپل هست، لیست نیست.

<code dir="ltr">{('cat', 'cat'): 2, ('dog', 'cat'): 1, ('bird', 'bird'): 1, ('dog', 'dog'): 2, ('cat', 'dog'): 1, ('bird', 'dog'): 1}</code>

ب)

<div dir="ltr">

```python
for k in sorted(set(y_true)):
    tp = cm.get((k, k), 0)
    actual = 0
    predicted = 0
    for (t, p), n in cm.items():
        if t == k:
            actual += n
        if p == k:
            predicted += n
    print(k, tp / actual, tp / predicted)
```

### سؤال ۵ — حافظه‌ی نهان (memoization)

تابع بازگشتی زیر جمله‌ی <code dir="ltr">n</code> ام دنباله‌ی فیبوناچی را حساب می‌کند؛ متغیر <code dir="ltr">calls</code> تعداد فراخوانی‌ها را می‌شمارد:

</div>

```python
calls = 0

def fib(n):
    global calls
    calls += 1
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(20), calls)
```

<div dir="rtl">

الف) خروجی <code dir="ltr">fib(20)</code> برابر <code dir="ltr">6765</code> و تعداد فراخوانی‌ها <code dir="ltr">21891</code> است. چرا این‌قدر زیاد؟ (درخت فراخوانی <code dir="ltr">fib(5)</code> را بکشید و بشمارید <code dir="ltr">fib(2)</code> چند بار حساب می‌شود.)

ب) با یک دیکشنری <code dir="ltr">memo</code> که نتیجه‌ی هر <code dir="ltr">n</code> حساب‌شده را نگه می‌دارد، تابع <code dir="ltr">fib&#95;memo</code> را بنویسید. حالا <code dir="ltr">fib&#95;memo(20)</code> چند بار فراخوانی می‌شود؟

ج) این ایده (ذخیره‌ی نتیجه‌ی محاسبه‌های تکراری) در یادگیری عمیق کجا کاربرد دارد؟ (راهنمایی: محاسبه‌ی گرادیان در یک شبکه‌ی چندلایه به مشتق‌های لایه‌های بعدی نیاز دارد.)

<details>
<summary><b>پاسخ</b></summary>

الف) هر فراخوانی دو فراخوانی دیگر می‌سازد و زیرمسئله‌های یکسان بارها از نو حساب می‌شوند. در درخت <code dir="ltr">fib(5)</code>، <code dir="ltr">fib(3)</code> دو بار، <code dir="ltr">fib(2)</code> سه بار و <code dir="ltr">fib(1)</code> پنج بار حساب می‌شود؛ تعداد کل فراخوانی‌ها تقریباً نمایی رشد می‌کند.

ب)

<div dir="ltr">

```python
memo = {}

def fib_memo(n):
    global calls
    calls += 1
    if n in memo:
        return memo[n]
    if n < 2:
        result = n
    else:
        result = fib_memo(n - 1) + fib_memo(n - 2)
    memo[n] = result
    return result
```

</div>

با <code dir="ltr">calls = 0</code> و <code dir="ltr">memo</code> خالی، <code dir="ltr">fib&#95;memo(20)</code> فقط <code dir="ltr">39</code> فراخوانی دارد (هر <code dir="ltr">n</code> یک بار محاسبه می‌شود و بقیه از <code dir="ltr">memo</code> خوانده می‌شود).

ج) **پس‌انتشار (backpropagation)** دقیقاً همین ایده است: مشتق هر لایه یک بار حساب و ذخیره می‌شود و برای لایه‌های قبلی دوباره استفاده می‌شود. بدون آن، محاسبه‌ی گرادیان در شبکه‌های عمیق عملاً ناممکن بود. (در جلسه‌ی ۱۴ همین حافظه‌ی نهان را با یک decorator به هر تابعی اضافه می‌کنیم.)

</details>

### سؤال ۶ — جست‌وجوی ابرپارامترها

## خلاصه‌ی جلسه

- دیکشنری جفت‌های <code dir="ltr">key: value</code> با کلیدهای یکتا و hashable است و جست‌وجو با کلید را فوری انجام می‌دهد.

- جدول داده را می‌توان سطری (لیستی از دیکشنری‌ها) یا ستونی (دیکشنری از لیست‌ها) نگه داشت؛ پایه‌ی DataFrame در Pandas.
